## 1. Do que se trata

Este notebook substitui o conteúdo que havia sido colado nesta pasta a partir de outro contexto: um notebook de análise lexical dos mesmos seis planos, mas com outros JSONs (da pasta `notebooks/ Novo/`), outro pré-processamento e unigramas e bigramas misturados no mesmo ranking. Por comando do pesquisador, o notebook passa a **dar continuidade às análises dos notebooks individuais desta pasta** (`Análise_<SIGLA>.ipynb`):

1. para cada um dos seis planos, estende até **75 lemas** o recorte de cerca de 30 lemas feito no notebook individual, com os mesmos JSONs, o mesmo pré-processamento e as mesmas exclusões definidas pelo pesquisador naquele notebook. Os primeiros lemas de cada plano são, por isso, exatamente os do notebook individual, o que a seção 11.2 confere;
2. identifica os **termos distintivos** de cada plano em relação aos outros cinco, com o critério da versão anterior deste notebook alinhado à metodologia comparativa da pasta (skill 03).

Toda a análise é feita em **unigramas** (lemas de uma palavra). Palavras compostas com hífen, como *open-source*, *human-centric* ou *AI-driven*, contam como **um único unigrama**. Não há bigramas.

## 2. Finalidade

- conhecer o vocabulário principal de cada plano além das cerca de 30 posições analisadas nos notebooks individuais;
- identificar o que cada plano emprega com ênfase própria, isto é, os termos que o distinguem dos demais planos.

## 3. Insumos e recorte do conteúdo

São utilizados **apenas os resultados da skill 01**, isto é, os seis JSONs desta pasta. Os relatórios de extração (`relatorios_extracao/`) são apenas informativos e **não** são consultados. De cada JSON interessam **somente** o título e o subtítulo (`document.title` e `document.subtitle`), que formam a unidade 0 e entram na contagem; o texto integral e literal de cada unidade (`sections[].text`: títulos internos, parágrafos e itens de lista); e, apenas como filtros, o tipo (`type`) e a categoria estrutural (`category`). O identificador (`id`) serve apenas como índice de posição. **Todo conteúdo periférico fica fora da análise**: numeração, nível, página, caminho estrutural e os demais metadados. Os filtros são os mesmos para todos os planos e os mesmos dos notebooks individuais: por padrão, nenhum retira conteúdo.

Para conferir a continuidade, o notebook lê também, em modo somente leitura, as tabelas dos recortes de cerca de 30 lemas gravadas nos notebooks individuais (seção 11.2). Nenhum número desses notebooks entra nos cálculos: todos os planos são reprocessados aqui, a partir dos JSONs.

## 4. Por que a comparação é relativa

Os planos têm tamanhos muito diferentes: de 2.874 tokens (*AI+*, China, 2025) a 16.391 (PBIA), uma razão de 5,7 vezes. Comparar contagens absolutas favoreceria os planos mais longos. Por isso, toda comparação entre planos usa a **taxa por 1.000 tokens**:

> taxa por 1.000 tokens = (ocorrências do lema no plano ÷ total de tokens do plano) × 1.000

O denominador é o total de tokens (palavras) do plano, **incluindo** as *stopwords* e as palavras retiradas por exclusão do pesquisador. As frequências absolutas aparecem nas tabelas apenas como informação auxiliar.

## 5. Método (visão geral)

| Etapa | Procedimento |
|---|---|
| Verificação | Integridade de cada JSON (skill 01, corpus *Planos Nacionais de IA*, ids sequenciais, unidades não vazias, contagem de unidades, título, idioma `en`) e SHA-256 |
| Recorte | Título e subtítulo como unidade 0; filtros por tipo, categoria e unidade, iguais para todos os planos (padrão: nenhum conteúdo retirado) |
| Pré-processamento | Pipeline dos notebooks individuais: spaCy `en_core_web_sm`, reunião dos compostos com hífen em um token, lematização, classes de conteúdo (substantivos, nomes próprios, verbos, adjetivos e advérbios) e *stopwords* do spaCy. Cada plano recebe as correções de lema e os ajustes de tokenização que o seu notebook individual adotou |
| Exclusões do pesquisador | As definidas nos notebooks individuais: EUA (seção 15: nomes de órgãos e instituições governamentais, designações genéricas de governo, *Federal* ligado a órgão, nome do país e a construção *led by*, com *Trump Administration* mantida) e PBIA (seção 16: *Brazil* e nomes de instituições e órgãos brasileiros). Os outros quatro planos não têm exclusões |
| Equivalência e continuidade | Exame das diferenças de pré-processamento entre os planos (seção 11.1) e conferência dos primeiros lemas com os recortes gravados nos notebooks individuais (seção 11.2) |
| 75 lemas | Frequência absoluta, taxa por 1.000 tokens, alcance e dispersão (DP normalizado de Gries) de cada lema; recorte de 75 por plano, ampliado até incluir os lemas empatados com o 75º |
| Termos distintivos | *Keyness* de cada plano contra o conjunto dos outros cinco (G² de log-verossimilhança e *Log Ratio*, skill 03), restrita aos lemas usados com ênfase pelo plano e por no máximo 3 dos 6 planos (critério da versão anterior); 15 termos por plano, ampliados nos empates |

## 7. Princípios

Todo o conteúdo deste notebook segue a **skill 03** e, por meio dela, a **skill 02**. Os comandos do pesquisador prevalecem: por isso, cada plano é processado com os parâmetros e as exclusões do seu notebook individual, para que a contagem de 75 seja a continuação exata da contagem de cerca de 30. O procedimento é o mesmo para todos os planos (mesmo pipeline, mesmos filtros, mesma medida relativa, mesmos critérios de *keyness*), e nenhum plano é tratado como referência privilegiada, inclusive o PBIA. As diferenças de pré-processamento e de exclusões herdadas dos notebooks individuais não são corrigidas aqui, porque isso romperia a continuidade pedida: elas são **examinadas e declaradas** na seção 11.1 e nas fragilidades.

> As interpretações são redigidas **somente após a execução** do notebook, com base nos valores efetivamente obtidos. Se os JSONs, os filtros ou os parâmetros mudarem, os números mudam e as interpretações precisam ser revistas.

```python
# ---------------------------------------------------------------------------
# Restrição de ambiente: o notebook só roda de dentro da pasta "Planos Nacionais de IA".
# ---------------------------------------------------------------------------
PASTA = Path.cwd()
if unicodedata.normalize("NFC", PASTA.name) != "Planos Nacionais de IA":
    raise RuntimeError(
        "Este notebook deve ser executado de dentro da pasta 'Planos Nacionais de IA' "
        f"(restrição de ambiente da skill). Pasta atual: {PASTA}"
    )

# ---------------------------------------------------------------------------
# Pasta de saída versionada: resultados anteriores nunca são sobrescritos.
# ---------------------------------------------------------------------------
DIR_SAIDA = PASTA / "resultados" / NOME_SAIDA / VERSAO_SAIDA
if SALVAR_SAIDAS:
    if DIR_SAIDA.exists() and any(DIR_SAIDA.iterdir()) and not SOBRESCREVER:
        raise FileExistsError(
            f"A pasta de saída '{DIR_SAIDA.relative_to(PASTA)}' já contém arquivos. "
            "Para preservar a versão anterior, altere VERSAO_SAIDA (ex.: 'v2') na configuração. "
            "Use SOBRESCREVER = True somente se a substituição for intencional."
        )
    DIR_SAIDA.mkdir(parents=True, exist_ok=True)
SAIDAS_GRAVADAS = []   # arquivos gravados nesta execução (vão para o registro de execução)

# ---------------------------------------------------------------------------
# Estilo das figuras (padrão acadêmico: fonte >= 9 pt, 300 dpi, PNG + PDF)
# ---------------------------------------------------------------------------
# Cor fixa por plano (skill 02, seção 3.7), com as cores-base da pasta: Brasil verde, Estados
# Unidos azul-escuro, União Europeia amarelo, China vermelho. Planos do mesmo bloco recebem tons
# da mesma cor: o amarelo-escuro do Apply AI é o do notebook individual dele; nos notebooks
# individuais, os dois planos chineses usam o mesmo vermelho, e o vermelho-escuro do plano de
# 2017 foi definido aqui para distingui-los. Editável pelo pesquisador. Como verde e vermelho se
# confundem para leitores com daltonismo, cada plano é identificado também por rótulo direto.
COR_DOCUMENTO = {
    "AI_Continent_2025": "#EDB120",   # União Europeia (amarelo)
    "Apply_AI_2025": "#B8860B",       # União Europeia (amarelo-escuro): segundo plano do bloco
    "China_2025": "#B2182B",          # China (vermelho)
    "New_Gen_2017": "#67001F",        # China (vermelho-escuro): segundo plano do bloco
    "EUA_2025": "#1F3F8F",            # Estados Unidos (azul-escuro)
    "PBIA_2025": "#2E9E4F",           # Brasil (verde)
}
COR_TINTA = "#0b0b0b"     # texto principal
COR_TINTA_2 = "#52514e"   # texto secundário (rótulos, anotações)
COR_GRADE = "#e4e3df"     # grade e linhas de apoio
COR_REFERENCIA = "#8f8e88"   # marcador da taxa nos demais planos (termos distintivos)

plt.rcParams.update({
    "figure.dpi": 110,
    "savefig.dpi": 300,
    "savefig.bbox": "tight",
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "font.size": 10,
    "axes.titlesize": 11,
    "axes.labelsize": 10,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "legend.fontsize": 9,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.edgecolor": COR_TINTA_2,
    "axes.linewidth": 0.6,
    "axes.grid": True,
    "axes.axisbelow": True,
    "grid.color": COR_GRADE,
    "grid.linewidth": 0.6,
    "grid.linestyle": "-",
    "text.color": COR_TINTA,
    "axes.labelcolor": COR_TINTA,
    "xtick.color": COR_TINTA_2,
    "ytick.color": COR_TINTA_2,
    "legend.frameon": False,
    "pdf.fonttype": 42,
})

# ---------------------------------------------------------------------------
# Funções auxiliares de formatação e gravação
# ---------------------------------------------------------------------------
def num(x, casas=2):
    """Formata número no padrão brasileiro (vírgula decimal, ponto de milhar)."""
    if x is None or (isinstance(x, float) and math.isnan(x)):
        return "—"
    texto = f"{x:,.{casas}f}"
    return texto.replace(",", "§").replace(".", ",").replace("§", ".")


# Rótulos de eixo com vírgula decimal.
FMT_BR = FuncFormatter(lambda v, _: f"{v:g}".replace(".", ","))


def exibir(df, casas=2):
    """Exibe uma tabela com vírgula decimal, sem alterar os valores."""
    display(df.style.format(precision=casas, decimal=",", thousands=".", na_rep="—"))


def salvar_figura(fig, nome):
    """Grava a figura em PNG (300 dpi) e PDF na pasta de saída versionada."""
    if SALVAR_SAIDAS:
        for ext in ("png", "pdf"):
            fig.savefig(DIR_SAIDA / f"{nome}.{ext}")
            SAIDAS_GRAVADAS.append(f"{nome}.{ext}")


def salvar_tabela(df, nome, index=True):
    """Grava a tabela completa em CSV (UTF-8 com BOM, ';', vírgula decimal)."""
    if SALVAR_SAIDAS:
        df.to_csv(DIR_SAIDA / f"{nome}.csv", sep=";", decimal=",", encoding="utf-8-sig", index=index)
        SAIDAS_GRAVADAS.append(f"{nome}.csv")


sem_cor = [sigla for sigla in DOCUMENTOS if sigla not in COR_DOCUMENTO]
if sem_cor:
    raise KeyError(f"Defina a cor dos planos {sem_cor} em COR_DOCUMENTO (skill 02, seção 3.7).")

print(f"Pasta de trabalho: {PASTA}")
print(f"Saídas: {DIR_SAIDA.relative_to(PASTA) if SALVAR_SAIDAS else 'gravação desativada'}")
```

```python
# ===========================================================================
# PIPELINE DE PRÉ-PROCESSAMENTO E MEDIDAS
# Mesmo pipeline dos notebooks individuais da pasta. As poucas escolhas que os notebooks
# individuais fizeram para o seu plano (ajustes de tokenização e correções de lema) entram
# como parâmetros por plano, definidos na configuração (seção 8). Na construção deste
# notebook, verificou-se que o pipeline abaixo reproduz, plano a plano, a tabela de tokens e
# a tabela de frequências de cada notebook individual (a seção 11.2 confere os recortes).
# ===========================================================================

HIFENS = {"-", "‐", "‑"}

# Valores admitidos nos campos usados como filtro (skill 01, seção 4.3).
TIPOS_VALIDOS = {"heading", "paragraph", "list_item"}
CATEGORIAS_VALIDAS = {"Foreword", "Executive summary", "Body", "Box", "Annex"}

COLUNAS_TOKENS = ["unidade", "frase", "i", "posicao", "forma", "lema_spacy", "lema", "classe", "etiqueta"]


def compilar_re_palavra(siglas_com_e_comercial):
    """Token (palavra) = sequência de letras, admitindo hífen interno (multi-stakeholder), sigla
    com pontos (U.S.) e, se siglas_com_e_comercial = True (notebooks dos planos chineses), sigla
    com "&" (R&D). Números, pontuação e símbolos NÃO são tokens."""
    return re.compile(
        r"(?:[^\W\d_]\.){2,}"
        + (r"|[^\W\d_]+(?:&[^\W\d_]+)+" if siglas_com_e_comercial else "")
        + r"|[^\W\d_]+(?:[-‐‑][^\W\d_]+)*"
    )


def carregar_modelo(separar_simbolo_final):
    """Carrega o modelo do spaCy. Se separar_simbolo_final = True (notebooks dos planos chineses),
    separa do fim da palavra o "+" (AI+) e o hífen suspenso ("intelligentization- and"), que o
    tokenizador mantém colados: a palavra é contada e o símbolo não."""
    nlp = spacy.load(MODELO_SPACY)
    if separar_simbolo_final:
        sufixos = list(nlp.Defaults.suffixes) + [r"(?<=[^\W\d_])\+", r"(?<=[^\W\d_])[-‐‑]"]
        nlp.tokenizer.suffix_search = compile_suffix_regex(sufixos).search
    return nlp


def normalizar_hifen(texto):
    """Uniformiza variantes tipográficas do hífen (apenas nas chaves de contagem;
    o texto do JSON não é alterado)."""
    return texto.replace("‐", "-").replace("‑", "-")


def carregar_extracao(caminho):
    """Lê o JSON produzido pela skill 01 e devolve SOMENTE o título, o subtítulo e as
    unidades (id, tipo, categoria e texto integral). Todo o conteúdo periférico (number,
    level, page, path e demais metadados) é descartado aqui.
    Interrompe a execução se o arquivo não passar nas verificações de integridade."""
    if not caminho.exists():
        raise FileNotFoundError(
            f"Arquivo '{caminho.name}' não encontrado na pasta. "
            "Execute antes a skill 01 (skill_extração.md) para gerar o JSON do plano."
        )
    bruto = caminho.read_bytes()
    dados = json.loads(bruto.decode("utf-8"))

    problemas = [f"chave '{c}' ausente" for c in ("document", "extraction", "sections") if c not in dados]
    if problemas:
        raise ValueError(f"Falha na verificação de '{caminho.name}':\n- " + "\n- ".join(problemas))

    extracao, documento, secoes = dados["extraction"], dados["document"], dados["sections"]
    if extracao.get("skill") != "skill_extração":
        problemas.append("o JSON não registra ter sido produzido pela skill 01 (extraction.skill)")
    if extracao.get("corpus") != "Planos Nacionais de IA":
        problemas.append("o JSON não registra o corpus 'Planos Nacionais de IA' (extraction.corpus)")
    if [s.get("id") for s in secoes] != list(range(1, len(secoes) + 1)):
        problemas.append("os ids das unidades não são sequenciais a partir de 1")
    vazias = [s.get("id") for s in secoes if not isinstance(s.get("text"), str) or not s["text"].strip()]
    if vazias:
        problemas.append(f"unidades com texto vazio: {vazias}")
    if extracao.get("units_extracted") != len(secoes):
        problemas.append(
            f"extraction.units_extracted = {extracao.get('units_extracted')}, mas há {len(secoes)} unidades"
        )
    tipos = {s.get("type") for s in secoes} - TIPOS_VALIDOS
    if tipos:
        problemas.append(f"tipos de unidade não previstos pela skill 01: {sorted(map(str, tipos))}")
    categorias = {s.get("category") for s in secoes} - CATEGORIAS_VALIDAS
    if categorias:
        problemas.append(f"categorias não previstas pela skill 01: {sorted(map(str, categorias))}")
    if not documento.get("title"):
        problemas.append("título (document.title) ausente")
    if documento.get("language") != IDIOMA_ESPERADO:
        problemas.append(
            f"idioma declarado no JSON: {documento.get('language')!r}; esperado: {IDIOMA_ESPERADO!r}"
        )
    if problemas:
        raise ValueError(f"Falha na verificação de '{caminho.name}':\n- " + "\n- ".join(problemas))

    return {
        "arquivo": caminho.name,
        "sha256": hashlib.sha256(bruto).hexdigest(),
        "titulo": documento["title"],
        "subtitulo": documento.get("subtitle"),
        "idioma": documento.get("language"),
        # SOMENTE id (posição), tipo e categoria (filtros) e texto (conteúdo analisado)
        "unidades": [
            {"id": s["id"], "tipo": s["type"], "categoria": s["category"], "texto": s["text"]}
            for s in secoes
        ],
    }


def selecionar_unidades(documento, unidades_excluidas):
    """Monta as unidades analisadas a partir dos parâmetros de recorte da configuração. O título
    e o subtítulo formam a unidade 0, separados por uma quebra de linha, que impede a junção da
    última palavra do título com a primeira do subtítulo sem acrescentar pontuação ao texto.
    Cada unidade retirada é atribuída ao primeiro filtro que a retira. Nenhum texto é alterado."""
    inexistentes = sorted(set(unidades_excluidas) - {u["id"] for u in documento["unidades"]})
    if inexistentes:
        raise ValueError(f"UNIDADES_EXCLUIDAS contém ids que não existem no JSON: {inexistentes}")
    sem_motivo = [i for i, motivo in unidades_excluidas.items() if not str(motivo).strip()]
    if sem_motivo:
        raise ValueError(f"Registre o motivo da exclusão das unidades {sem_motivo} em UNIDADES_EXCLUIDAS.")
    categorias_invalidas = sorted(set(CATEGORIAS_EXCLUIDAS) - CATEGORIAS_VALIDAS)
    if categorias_invalidas:
        raise ValueError(f"CATEGORIAS_EXCLUIDAS contém categorias não previstas: {categorias_invalidas}")

    texto_0 = "\n".join(t for t in (documento["titulo"], documento["subtitulo"]) if t)
    candidatas = [{"id": 0, "tipo": "title", "categoria": None, "texto": texto_0}] + documento["unidades"]
    analisadas, retiradas = [], []
    for u in candidatas:
        if u["tipo"] == "title" and not INCLUIR_TITULO_NA_CONTAGEM:
            filtro = "INCLUIR_TITULO_NA_CONTAGEM"
        elif u["tipo"] == "heading" and not INCLUIR_TITULOS_INTERNOS:
            filtro = "INCLUIR_TITULOS_INTERNOS"
        elif u["categoria"] in CATEGORIAS_EXCLUIDAS:
            filtro = "CATEGORIAS_EXCLUIDAS"
        elif u["id"] in unidades_excluidas:
            filtro = "UNIDADES_EXCLUIDAS"
        else:
            analisadas.append(u)
            continue
        retiradas.append({**u, "filtro": filtro})
    return analisadas, retiradas


def montar_stopwords(nlp):
    """Lista de stopwords do spaCy, ajustada pelas listas editáveis da configuração."""
    base = {w.lower() for w in nlp.Defaults.stop_words}
    return (base | {w.lower() for w in STOPWORDS_ADICIONAIS}) - {w.lower() for w in STOPWORDS_MANTIDAS}


def unir_compostos_hifenizados(doc):
    """Reúne em um único token as palavras compostas com hífen (ex.: open-source, multi-stakeholder,
    human-centric, state-of-the-art), que o spaCy separa em vários tokens. Só une quando não há
    espaço em volta do hífen. Por isso, cada composto conta como um único unigrama."""
    trechos, i, n = [], 0, len(doc)
    while i < n:
        j = i
        while (j + 2 < n and doc[j].is_alpha and doc[j].whitespace_ == ""
               and doc[j + 1].text in HIFENS and doc[j + 1].whitespace_ == ""
               and doc[j + 2].is_alpha):
            j += 2
        if j > i:
            trechos.append(doc[i:j + 1])
            i = j + 1
        else:
            i += 1
    with doc.retokenize() as rt:
        for trecho in trechos:
            ultimo = trecho[-1]
            # Lema do composto = forma escrita; se o último elemento estiver no plural,
            # usa-se o lema dele (multi-stakeholders -> multi-stakeholder).
            if ultimo.tag_ in {"NNS", "NNPS"}:
                lema = trecho.text[: len(trecho.text) - len(ultimo.text)] + ultimo.lemma_
            else:
                lema = trecho.text
            rt.merge(trecho, attrs={"LEMMA": lema, "POS": ultimo.pos_, "TAG": ultimo.tag_})
    return doc


def processar_documento(unidades, nlp, stopwords, re_palavra, correcoes_de_lema):
    """Aplica o pré-processamento a cada unidade (título, parágrafo ou item) e devolve uma tabela
    com uma linha por token (palavra), na ordem original do documento. 'lema_spacy' é o lema dado
    pelo spaCy; 'lema' é o lema depois das correções explícitas do plano."""
    linhas, posicao, frases_por_unidade = [], 0, {}
    for unidade, doc in zip(unidades, nlp.pipe([u["texto"] for u in unidades])):
        doc = unir_compostos_hifenizados(doc)
        uid = unidade["id"]
        for frase in doc.sents:
            k = frases_por_unidade.get(uid, 0)
            frases_por_unidade[uid] = k + 1
            for tok in frase:
                if not re_palavra.fullmatch(tok.text):
                    continue
                lema_spacy = normalizar_hifen(tok.lemma_.lower())
                linhas.append({
                    "unidade": uid,
                    "frase": f"{uid}.{k}",
                    "i": tok.i,                      # índice na unidade (detecta adjacência)
                    "posicao": posicao,              # posição no documento inteiro
                    "forma": normalizar_hifen(tok.text.lower()),
                    "lema_spacy": lema_spacy,
                    "lema": correcoes_de_lema.get(lema_spacy, lema_spacy),
                    "classe": tok.pos_,
                    "etiqueta": tok.tag_,
                })
                posicao += 1
    tokens = pd.DataFrame(linhas, columns=COLUNAS_TOKENS)
    tokens["conteudo"] = (
        tokens["classe"].isin(CLASSES_DE_CONTEUDO)
        & ~tokens["lema"].isin(stopwords)
        & (tokens["lema"].str.len() >= TAMANHO_MINIMO_LEMA)
        & tokens["lema"].map(lambda lema: bool(re_palavra.fullmatch(lema)))
    ).astype(bool)
    return tokens


# ---------------------------------------------------------------------------
# Exclusões por comando do pesquisador (funções da seção 15 do notebook dos EUA, com o padrão
# de palavra do plano como parâmetro). Aplicadas à lista do PBIA, reproduzem exatamente o
# resultado da função da seção 16 do notebook do PBIA (verificado na construção deste notebook).
# ---------------------------------------------------------------------------
CATEGORIA_PROTEGIDA = "Expressão protegida (mantida na contagem)"


def formas_da_expressao(texto, nlp, re_palavra):
    """Sequência de formas (minúsculas) de uma expressão, com o mesmo pré-processamento
    do texto: tokenização do spaCy, reunião de compostos com hífen e padrão de palavra."""
    doc = unir_compostos_hifenizados(nlp(texto))
    return [normalizar_hifen(t.text.lower()) for t in doc if re_palavra.fullmatch(t.text)]


def preparar_exclusoes(exclusoes, nlp, re_palavra, protegidas=()):
    """Converte as expressões em padrões de formas, da mais longa para a mais curta.
    Palavras entre colchetes fazem parte do padrão, mas não são retiradas; as expressões
    protegidas são tratadas como inteiramente entre colchetes. Interrompe a execução se uma
    expressão ficar vazia ou repetir outra (as categorias são exclusivas)."""
    entradas, vistos = [], {}
    grupos = dict(exclusoes)
    grupos[CATEGORIA_PROTEGIDA] = [f"[{x}]" for x in protegidas]
    for categoria, expressoes in grupos.items():
        for expressao in expressoes:
            formas, retirar = [], []
            for parte in re.split(r"(\[[^\]]*\])", expressao):
                if not parte.strip():
                    continue
                f = formas_da_expressao(parte.strip("[]"), nlp, re_palavra)
                formas += f
                retirar += [not parte.startswith("[")] * len(f)
            chave = tuple(formas)
            if not chave:
                raise ValueError(f"A expressão '{expressao}' não contém nenhuma palavra.")
            if chave in vistos:
                raise ValueError(f"A expressão '{expressao}' repete '{vistos[chave]}'.")
            vistos[chave] = expressao
            entradas.append({"categoria": categoria, "expressao": expressao,
                             "formas": chave, "retirar": tuple(retirar)})
    return sorted(entradas, key=lambda e: -len(e["formas"]))


def aplicar_exclusoes(tokens, entradas):
    """Marca os tokens retirados da contagem de conteúdo. A busca percorre cada frase da
    esquerda para a direita; em cada posição, vale a expressão mais longa que coincide, e
    cada token é retirado no máximo uma vez. O total de tokens não muda.
    Devolve os tokens (com 'exclusao', 'expressao_excluida', 'conteudo_antes_das_exclusoes' e
    'conteudo' recalculado) e o número de ocorrências de cada expressão."""
    categoria = pd.Series(None, index=tokens.index, dtype=object)
    expressao = pd.Series(None, index=tokens.index, dtype=object)
    ocorrencias = {e["expressao"]: 0 for e in entradas}
    if entradas:
        for _, frase in tokens.groupby("frase", sort=False):
            indices, formas = frase.index.to_list(), frase["forma"].to_list()
            j = 0
            while j < len(formas):
                for e in entradas:
                    n = len(e["formas"])
                    if tuple(formas[j:j + n]) == e["formas"]:
                        for k in range(n):
                            if e["retirar"][k]:
                                categoria[indices[j + k]] = e["categoria"]
                                expressao[indices[j + k]] = e["expressao"]
                        ocorrencias[e["expressao"]] += 1
                        j += n
                        break
                else:
                    j += 1
    resultado = tokens.copy()
    resultado["exclusao"] = categoria
    resultado["expressao_excluida"] = expressao
    resultado["conteudo_antes_das_exclusoes"] = tokens["conteudo"]
    resultado["conteudo"] = tokens["conteudo"] & categoria.isna()
    return resultado, ocorrencias


# ---------------------------------------------------------------------------
# Frequências e recortes
# ---------------------------------------------------------------------------
def tabela_frequencias(tokens):
    """Frequência, taxa por 1.000 tokens, alcance e dispersão (DP normalizado) de
    TODOS os lemas de conteúdo. Ordenação: frequência decrescente; empate -> ordem alfabética."""
    n_tokens = len(tokens)
    unidades = sorted(tokens["unidade"].unique())
    proporcao_unidade = (tokens.groupby("unidade").size().reindex(unidades) / n_tokens).to_numpy()
    conteudo = tokens[tokens["conteudo"]]
    cruz = pd.crosstab(conteudo["lema"], conteudo["unidade"]).reindex(columns=unidades, fill_value=0)
    freq = cruz.sum(axis=1)
    # DP de Gries (2008), normalizado por Lijffijt & Gries (2012):
    # 0 = distribuição proporcional ao tamanho das unidades; 1 = concentração máxima.
    v = cruz.div(freq, axis=0).to_numpy()
    dp = 0.5 * np.abs(v - proporcao_unidade).sum(axis=1)
    dp_norm = dp / (1 - proporcao_unidade.min()) if len(unidades) > 1 else np.full(len(freq), np.nan)
    formas = conteudo.groupby("lema")["forma"].agg(
        lambda s: "; ".join(f"{forma} ({n})" for forma, n in s.value_counts().items())
    )
    tab = pd.DataFrame({
        "frequencia": freq,
        "por_1000_tokens": freq / n_tokens * 1000,
        "unidades_com_ocorrencia": (cruz > 0).sum(axis=1),
        "pct_unidades": (cruz > 0).sum(axis=1) / len(unidades) * 100,
        "DP_norm": dp_norm,
        "formas_observadas": formas,
    })
    tab.index.name = "lema"
    tab = tab.reset_index().sort_values(["frequencia", "lema"], ascending=[False, True])
    return tab.set_index("lema")


def recorte_com_empates(tab, n, coluna="frequencia"):
    """Primeiras n linhas de uma tabela ordenada por `coluna` (decrescente), ampliadas até incluir
    todas as linhas com o mesmo valor da n-ésima, para que o desempate alfabético não decida quem
    aparece (regra definida pelo pesquisador). Devolve o recorte e um resumo do ponto de corte."""
    n_base = min(n, len(tab))
    if n_base == 0:
        return tab.iloc[:0], {"pedido": n, "valor_do_corte": None, "n_recorte": 0,
                              "empatados_acrescentados": [], "primeiro_fora": None}
    valor_corte = tab[coluna].iloc[n_base - 1]
    n_recorte = int((tab[coluna] >= valor_corte).sum())
    recorte = tab.iloc[:n_recorte]
    seguinte = tab.iloc[n_recorte] if len(tab) > n_recorte else None
    # Verificação: o primeiro item fora do recorte tem valor menor que o do último incluído.
    assert seguinte is None or seguinte[coluna] < valor_corte
    return recorte, {
        "pedido": n,
        "valor_do_corte": valor_corte,
        "n_recorte": n_recorte,
        "empatados_acrescentados": list(recorte.index[n_base:]),
        "primeiro_fora": (seguinte.name, seguinte[coluna]) if seguinte is not None else None,
    }


def lemas_com_enfase(tab, k):
    """Lemas entre os k mais frequentes do plano, com todos os empatados com o k-ésimo
    (critério de 'ênfase' da versão anterior deste notebook)."""
    if len(tab) <= k:
        return set(tab.index)
    return set(tab.index[tab["frequencia"] >= tab["frequencia"].iloc[k - 1]])


# ---------------------------------------------------------------------------
# Termos distintivos: keyness (skill 03, seção 3.5)
# ---------------------------------------------------------------------------
def tabela_keyness(freqs, n_tokens):
    """Keyness de cada plano contra a soma dos demais (MODO_KEYNESS = "um_contra_os_demais").
    freqs: lemas x planos (frequências absolutas); n_tokens: total de tokens de cada plano.
    Testam-se os lemas com frequência combinada >= FREQ_MINIMA_COMPARACAO.
    - G² de log-verossimilhança (Dunning, 1993; Rayson & Garside, 2000), 1 grau de liberdade;
    - Log Ratio (Hardie, 2014) = log2(taxa no plano / taxa nos demais), com correção de 0,5
      apenas quando uma das frequências é zero (o termo é marcado como exclusivo);
    - p-valor do G² pela distribuição qui-quadrado com 1 grau de liberdade: p = erfc(sqrt(G²/2)).
    Devolve {sigla: tabela} e o número total de testes (base da correção de Bonferroni)."""
    if MODO_KEYNESS != "um_contra_os_demais":
        raise ValueError("Este notebook implementa apenas MODO_KEYNESS = 'um_contra_os_demais'.")
    testados = freqs[freqs.sum(axis=1) >= FREQ_MINIMA_COMPARACAO]
    n_testes = testados.shape[0] * testados.shape[1]
    tabelas = {}
    for sigla in freqs.columns:
        a = testados[sigla].to_numpy(dtype=float)
        b = testados.drop(columns=sigla).sum(axis=1).to_numpy(dtype=float)
        c = float(n_tokens[sigla])
        d = float(n_tokens.drop(sigla).sum())
        esperado_a = c * (a + b) / (c + d)
        esperado_b = d * (a + b) / (c + d)
        with np.errstate(divide="ignore", invalid="ignore"):
            g2 = 2 * (np.where(a > 0, a * np.log(a / esperado_a), 0.0)
                      + np.where(b > 0, b * np.log(b / esperado_b), 0.0))
        log_ratio = np.log2((np.where(a > 0, a, 0.5) / c) / (np.where(b > 0, b, 0.5) / d))
        tab = pd.DataFrame({
            "freq_plano": a.astype(int),
            "freq_demais": b.astype(int),
            "por_1000_plano": a / c * 1000,
            "por_1000_demais": b / d * 1000,
            "G2": g2,
            "log_ratio": log_ratio,
            "p_valor": [math.erfc(math.sqrt(x / 2)) for x in g2],
        }, index=testados.index)
        tab["exclusivo"] = tab["freq_demais"] == 0
        tab["passa_keyness"] = (tab["G2"] >= LIMIAR_G2) & (tab["log_ratio"] >= LIMIAR_LOG_RATIO)
        tab["resiste_bonferroni"] = tab["passa_keyness"] & (tab["p_valor"] < ALFA_BONFERRONI / n_testes)
        tab.index.name = "lema"
        tabelas[sigla] = tab
    return tabelas, n_testes


# ---------------------------------------------------------------------------
# Equivalência de tratamento entre os planos (skill 03, seção 3.6)
# ---------------------------------------------------------------------------
# Regras e expressões regulares dos ajustes de tokenização dos notebooks dos planos chineses.
REGRAS_DE_TOKENIZACAO = [
    ("'+' colado ao fim da palavra", "separar_simbolo_final", r"[^\W\d_]+\+(?![^\W\d_])"),
    ("hífen suspenso", "separar_simbolo_final", r"[^\W\d_]+[-‐‑](?=\s)"),
    ("sigla com '&'", "siglas_com_e_comercial", r"(?<![^\W\d_])[^\W\d_]+(?:&[^\W\d_]+)+(?![^\W\d_])"),
]


def formas_dos_ajustes_de_tokenizacao(sigla, unidades):
    """Formas do texto do plano que os ajustes de tokenização afetam, com o número de ocorrências e o
    tratamento que recebem NESTE plano (a contagem é feita diretamente no texto das unidades)."""
    linhas = []
    for ajuste, parametro, padrao in REGRAS_DE_TOKENIZACAO:
        ativo = DOCUMENTOS[sigla][parametro]
        contagem = pd.Series([m.group(0) for u in unidades for m in re.finditer(padrao, u["texto"])],
                             dtype=object).value_counts()
        for forma, n in contagem.items():
            linhas.append({"plano": sigla, "ajuste": ajuste, "forma_no_texto": forma, "ocorrencias": int(n),
                           "tratamento": ("contada (ajuste ativo no plano)" if ativo
                                          else "fora da contagem (ajuste inativo no plano)")})
    return pd.DataFrame(linhas, columns=["plano", "ajuste", "forma_no_texto", "ocorrencias", "tratamento"])


def lemas_sem_correcao_de_outro_plano(freq_por_plano):
    """Para cada plano, os lemas que o notebook individual de OUTRO plano corrige (CORRECOES_DE_LEMA)
    e que, neste plano, ficam sem a correção, com a frequência e o lema de destino."""
    linhas = []
    for sigla, tab in freq_por_plano.items():
        proprias = CORRECOES_DE_LEMA.get(sigla, {})
        origens = {}
        for outra, correcoes in CORRECOES_DE_LEMA.items():
            for origem, destino in correcoes.items():
                if outra != sigla and origem not in proprias and origem in tab.index:
                    origens.setdefault((origem, destino), []).append(outra)
        for (origem, destino), outras in sorted(origens.items()):
            linhas.append({"plano": sigla, "lema_sem_correcao": origem,
                           "frequencia": int(tab.loc[origem, "frequencia"]), "lema_de_destino": destino,
                           "correcao_adotada_em": ", ".join(outras)})
    return pd.DataFrame(linhas, columns=["plano", "lema_sem_correcao", "frequencia", "lema_de_destino",
                                         "correcao_adotada_em"])


# ---------------------------------------------------------------------------
# Conferência de continuidade (seção 11.2)
# ---------------------------------------------------------------------------
def ler_recorte_individual(caminho, id_celula):
    """Lê, no notebook individual (somente leitura), a tabela gravada na saída da célula do recorte
    de cerca de 30 lemas. Devolve a tabela (lema, frequência, taxa por 1.000 tokens) ou None se a
    célula ou a tabela não forem encontradas. Nenhum número é usado nos cálculos deste notebook."""
    if not caminho.exists():
        return None
    nb = json.loads(caminho.read_text(encoding="utf-8"))
    celula = next((c for c in nb.get("cells", []) if c.get("id") == id_celula), None)
    if celula is None:
        return None
    for saida in celula.get("outputs", []):
        html = saida.get("data", {}).get("text/html")
        if not html:
            continue
        tab = pd.read_html(io.StringIO("".join(html)), decimal=",", thousands=".", index_col=0)[0]
        if isinstance(tab.columns, pd.MultiIndex):
            tab.columns = [c[0] for c in tab.columns]
        if {"frequencia", "por_1000_tokens"} <= set(tab.columns):
            tab = tab[tab.index.notna()]
            tab.index = tab.index.astype(str)
            return tab[["frequencia", "por_1000_tokens"]].astype({"frequencia": int})
    return None


def versoes_ambiente(nlp):
    return {
        "python": platform.python_version(),
        "spacy": spacy.__version__,
        "modelo_spacy": f"{MODELO_SPACY} {nlp.meta.get('version', '?')}",
        "pandas": pd.__version__,
        "numpy": np.__version__,
        "matplotlib": matplotlib.__version__,
    }
```

## 10. Carregamento e verificação dos insumos

Cada JSON da skill 01 é lido em modo somente leitura. A execução é interrompida se algum arquivo não existir; não tiver sido produzido pela skill 01 para o corpus *Planos Nacionais de IA* (`extraction.skill` e `extraction.corpus`); tiver identificadores não sequenciais, unidade vazia, número de unidades diferente do registrado na extração, ou tipo e categoria não previstos; não tiver título; ou não estiver em inglês. O SHA-256 de cada JSON é exibido para registro. Em seguida, aplicam-se os parâmetros de recorte da configuração.

```python
documentos = {sigla: carregar_extracao(PASTA / cfg["arquivo_json"]) for sigla, cfg in DOCUMENTOS.items()}

insumos = pd.DataFrame([{
    "plano": sigla,
    "arquivo": d["arquivo"],
    "titulo": d["titulo"],
    "subtitulo": d["subtitulo"] or "—",
    "idioma": d["idioma"],
    "unidades_no_json": len(d["unidades"]),
    "titulos_internos": sum(u["tipo"] == "heading" for u in d["unidades"]),
    "paragrafos": sum(u["tipo"] == "paragraph" for u in d["unidades"]),
    "itens_de_lista": sum(u["tipo"] == "list_item" for u in d["unidades"]),
    "sha256": d["sha256"],
} for sigla, d in documentos.items()]).set_index("plano")
exibir(insumos, 0)

print(f"Todos os {len(documentos)} JSONs passaram nas verificações (skill 01, corpus 'Planos Nacionais de IA', "
      f"ids sequenciais, nenhuma unidade vazia, contagem de unidades, título e idioma '{IDIOMA_ESPERADO}').")
print(f"Título e subtítulo incluídos nas contagens (unidade 0): {'sim' if INCLUIR_TITULO_NA_CONTAGEM else 'não'}")
print(f"Títulos internos incluídos: {'sim' if INCLUIR_TITULOS_INTERNOS else 'não'}")
print(f"Categorias excluídas: {', '.join(CATEGORIAS_EXCLUIDAS) if CATEGORIAS_EXCLUIDAS else 'nenhuma'}")
print("Unidades excluídas: " + ("; ".join(f"{s}: {sorted(u)}" for s, u in UNIDADES_EXCLUIDAS.items() if u) or "nenhuma"))
```

## 11. Pré-processamento, exclusões e perfil comparativo

*Visualização prévia: base inicial para a construção de novos gráficos (ver nota no início do notebook).*

```python
modelos = {}        # um modelo do spaCy por configuração de tokenização
processados = {}    # por plano: unidades, tokens, exclusões e frequências
for sigla, cfg in DOCUMENTOS.items():
    if cfg["separar_simbolo_final"] not in modelos:
        modelos[cfg["separar_simbolo_final"]] = carregar_modelo(cfg["separar_simbolo_final"])
    nlp = modelos[cfg["separar_simbolo_final"]]
    re_palavra = compilar_re_palavra(cfg["siglas_com_e_comercial"])
    stopwords = montar_stopwords(nlp)

    analisadas, retiradas = selecionar_unidades(documentos[sigla], UNIDADES_EXCLUIDAS.get(sigla, {}))
    tokens = processar_documento(analisadas, nlp, stopwords, re_palavra, CORRECOES_DE_LEMA.get(sigla, {}))
    entradas = preparar_exclusoes(EXCLUSOES.get(sigla, {}), nlp, re_palavra, EXPRESSOES_PROTEGIDAS.get(sigla, []))
    tokens, ocorrencias = aplicar_exclusoes(tokens, entradas)
    processados[sigla] = {
        "analisadas": analisadas,
        "retiradas": retiradas,
        "tokens": tokens,
        "n_tokens": len(tokens),
        "n_unidades": tokens["unidade"].nunique(),
        "ocorrencias_exclusoes": ocorrencias,
        "freq": tabela_frequencias(tokens),   # base dos recortes e da comparação (com as exclusões)
    }
    salvar_tabela(tokens, f"00_tokens_processados_{sigla}", index=False)
    salvar_tabela(processados[sigla]["freq"], f"00_frequencias_lemas_completa_{sigla}")

freq = {sigla: p["freq"] for sigla, p in processados.items()}
N_TOKENS = pd.Series({sigla: p["n_tokens"] for sigla, p in processados.items()})
STOPWORDS = montar_stopwords(modelos[False])

# (a) Perfil comparativo: extensão e vocabulário de cada plano
perfil = pd.DataFrame({sigla: {
    "Unidades analisadas (com a unidade 0)": len(p["analisadas"]),
    "Unidades retiradas pelos filtros do recorte": len(p["retiradas"]),
    "Tokens (palavras): denominador das taxas": p["n_tokens"],
    "Tamanho em relação ao menor plano (vezes)": p["n_tokens"] / N_TOKENS.min(),
    "Tokens de conteúdo (pipeline)": int(p["tokens"]["conteudo_antes_das_exclusoes"].sum()),
    "Tokens de conteúdo retirados por exclusões do pesquisador": int(
        (p["tokens"]["conteudo_antes_das_exclusoes"] & ~p["tokens"]["conteudo"]).sum()),
    "Tokens de conteúdo considerados": int(p["tokens"]["conteudo"].sum()),
    "Lemas de conteúdo distintos": len(p["freq"]),
    "Lemas de conteúdo com uma única ocorrência": int((p["freq"]["frequencia"] == 1).sum()),
} for sigla, p in processados.items()})
display(Markdown("**(a) Perfil comparativo dos planos**"))
display(perfil.style.format(lambda v: num(v, 0) if float(v).is_integer() else num(v, 2)))
salvar_tabela(perfil, "00_perfil_comparativo")

# (b) Tratamentos de cada plano, reproduzidos dos notebooks individuais
tratamentos = pd.DataFrame({sigla: {
    "Correções de lema: entradas na configuração": len(CORRECOES_DE_LEMA.get(sigla, {})),
    "Correções de lema: entradas com efeito no texto": p["tokens"].loc[
        p["tokens"]["lema_spacy"] != p["tokens"]["lema"], "lema_spacy"].nunique(),
    "Correções de lema: tokens afetados": int((p["tokens"]["lema_spacy"] != p["tokens"]["lema"]).sum()),
    "Separar '+' e hífen suspenso do fim da palavra": "sim" if DOCUMENTOS[sigla]["separar_simbolo_final"] else "não",
    "Siglas com '&' como token": "sim" if DOCUMENTOS[sigla]["siglas_com_e_comercial"] else "não",
    "Exclusões do pesquisador: expressões na lista": sum(len(v) for v in EXCLUSOES.get(sigla, {}).values()),
    "Exclusões do pesquisador: ocorrências localizadas": sum(
        n for e, n in p["ocorrencias_exclusoes"].items() if not e.startswith("[")),
    "Expressões protegidas: ocorrências mantidas": sum(
        n for e, n in p["ocorrencias_exclusoes"].items() if e.startswith("[")),
} for sigla, p in processados.items()})
display(Markdown("**(b) Tratamentos de cada plano, reproduzidos dos notebooks individuais**"))
exibir(tratamentos, 0)
salvar_tabela(tratamentos, "00_tratamentos_por_plano")

print(f"Stopwords aplicadas: {len(STOPWORDS)} (lista do spaCy, {len(STOPWORDS_ADICIONAIS)} adicionada(s), "
      f"{len(STOPWORDS_MANTIDAS)} mantida(s)), as mesmas para todos os planos.")
print(f"Síntese factual: os planos têm de {num(N_TOKENS.min(), 0)} ({N_TOKENS.idxmin()}) a "
      f"{num(N_TOKENS.max(), 0)} ({N_TOKENS.idxmax()}) tokens, uma razão de {num(N_TOKENS.max() / N_TOKENS.min(), 1)} "
      f"vezes; as exclusões do pesquisador retiram "
      + "; ".join(f"{s}: {num(perfil.loc['Tokens de conteúdo retirados por exclusões do pesquisador', s], 0)}"
                  for s in DOCUMENTOS if perfil.loc["Tokens de conteúdo retirados por exclusões do pesquisador", s])
      + " tokens de conteúdo.")
```

**Método e escolhas.** Cada plano é processado com o pipeline dos notebooks individuais: modelo `en_core_web_sm` do spaCy; título e subtítulo como unidade 0, separados por uma quebra de linha; palavras compostas com hífen reunidas em um só token, de modo que *open-source*, *AI-driven* ou *human-computer* contam como **um unigrama**; **token** = palavra formada por letras, com hífen interno ou sigla com pontos (números, pontuação e símbolos não são tokens); **tokens de conteúdo** = substantivos, nomes próprios, verbos, adjetivos e advérbios fora da lista de *stopwords* do spaCy, verificada pelo lema, com pelo menos 2 caracteres. Cada plano recebe, além disso, o que o seu notebook individual fixou (tabela b): as correções de lema; nos dois planos chineses, a separação do "+" e do hífen suspenso no fim da palavra e a contagem das siglas com "&" (*R&D*, *S&T*) como um token; nos EUA e no PBIA, as exclusões definidas pelo pesquisador. As palavras excluídas deixam de ser tokens de conteúdo, mas continuam no total de tokens, que é o denominador de todas as taxas. Por isso, as frequências deste notebook são as mesmas dos recortes de cerca de 30 lemas dos notebooks individuais (seção 11.2). Na construção do notebook, verificou-se que o pipeline reproduz, plano a plano, a tabela de tokens e a tabela de frequências dos notebooks individuais. Há uma única diferença de implementação: o notebook dos EUA processa o título e o subtítulo como dois textos separados, e aqui eles são unidos por uma quebra de linha, como nos demais planos. Essa diferença muda apenas a numeração interna das frases da unidade 0; lemas, classes gramaticais e frequências são idênticos.

**Como ler.** A tabela (a) expõe a disparidade de tamanho entre os planos, que justifica a comparação por 1.000 tokens, e o efeito das exclusões do pesquisador sobre os tokens de conteúdo. A tabela (b) resume os tratamentos herdados dos notebooks individuais.

**Interpretação.** *Observação:* o PBIA (16.391 tokens, 586 unidades) tem 5,7 vezes o tamanho do *AI+* (`China_2025`: 2.874 tokens, 65 unidades); os demais ficam entre 2,6 e 3,9 vezes o menor plano: Apply AI (7.554 tokens), EUA (8.198), AI Continent (8.931) e New Generation (11.331). O número de lemas de conteúdo distintos vai de 728 (*AI+*) a 1.721 (PBIA) e cresce menos que a extensão: o PBIA tem 5,7 vezes os tokens do *AI+*, mas 2,4 vezes os lemas. As exclusões do pesquisador retiram 699 tokens de conteúdo nos EUA (13,8% dos 5.067) e 209 no PBIA (2,1% dos 9.850); nos outros quatro planos, não há exclusões. As correções de lema pesam mais no AI Continent (28 entradas, todas com efeito, em 198 tokens) e no Apply AI (22 das 40 entradas com efeito, em 81 tokens); nos EUA, as 14 entradas afetam 44 tokens; nos dois planos chineses e no PBIA, a única correção (*datum* → *data*) afeta 9, 41 e 70 tokens. *Inferência:* a diferença de tamanho torna indispensável a comparação relativa. O número de lemas distintos depende da extensão do texto e não indica, por si, maior ou menor riqueza vocabular (skill 02, seção 3.4). As exclusões têm peso muito maior nos EUA do que no PBIA, e não existem nos planos da UE e da China: essa assimetria é examinada na seção 11.1 e considerada na seção 13.

**Fragilidades e ajustes possíveis.** Os tratamentos não são idênticos entre os planos, porque reproduzem as decisões de cada notebook individual, como pede o comando de continuidade. A seção 11.1 mede essas diferenças. As exclusões dos EUA e do PBIA retiram nomes de órgãos e do próprio país; nos outros quatro planos, nomes equivalentes (*EU*, *European Commission*, *Member States*, *State Council*, *China*) continuam na contagem, porque não houve comando para retirá-los. Uniformizar esse tratamento é decisão do pesquisador e exigiria acrescentar exclusões nos notebooks individuais da UE e da China, para manter a continuidade.

### 11.1. Equivalência de tratamento entre os planos

*Skill 03, seção 3.6: na comparação, o problema não é só a inflação de um plano, mas a assimetria entre os planos. Esta subseção não altera nenhuma contagem: mede as diferenças de pré-processamento herdadas dos notebooks individuais e marca os lemas afetados por elas.*

```python
# (a) Formas afetadas pelos ajustes de tokenização dos notebooks dos planos chineses, em todos os planos
ajustes = pd.concat([formas_dos_ajustes_de_tokenizacao(sigla, p["analisadas"]) for sigla, p in processados.items()],
                    ignore_index=True)
resumo_ajustes = (ajustes.groupby(["plano", "ajuste", "tratamento"], sort=False)
                  .agg(ocorrencias=("ocorrencias", "sum"),
                       formas=("forma_no_texto", lambda s: "; ".join(
                           f"{f} ({n})" for f, n in zip(s, ajustes.loc[s.index, "ocorrencias"])))))
display(Markdown("**(a) Formas do texto afetadas pelos ajustes de tokenização, em cada plano**"))
exibir(resumo_ajustes, 0)
salvar_tabela(ajustes, "00_equivalencia_ajustes_de_tokenizacao", index=False)

# (b) Lemas que a correção de outro notebook uniria a outro lema e que, neste plano, ficam separados
sem_correcao = lemas_sem_correcao_de_outro_plano(freq)
display(Markdown("**(b) Lemas sem a correção adotada no notebook individual de outro plano**"))
if sem_correcao.empty:
    print("Nenhum lema nesta situação.")
else:
    exibir(sem_correcao.set_index(["plano", "lema_sem_correcao"]), 0)
salvar_tabela(sem_correcao, "00_equivalencia_correcoes_de_lema", index=False)

# Lemas cuja comparação entre planos depende dessas diferenças de pré-processamento: os lemas sem a
# correção, os lemas de destino e as siglas com "&" (tokens apenas nos planos com o ajuste ativo).
LEMAS_SENSIVEIS = {}
for _, linha in sem_correcao.iterrows():
    nota = f"{linha['plano']}: '{linha['lema_sem_correcao']}' ({linha['frequencia']}) fora de '{linha['lema_de_destino']}'"
    for lema in (linha["lema_sem_correcao"], linha["lema_de_destino"]):
        LEMAS_SENSIVEIS.setdefault(lema, []).append(nota)
for sigla, tab in freq.items():
    for lema in tab.index[tab.index.str.contains("&", regex=False)]:
        LEMAS_SENSIVEIS.setdefault(lema, []).append(f"sigla com '&': token só nos planos com o ajuste ativo ({sigla})")
LEMAS_SENSIVEIS = {lema: "; ".join(notas) for lema, notas in sorted(LEMAS_SENSIVEIS.items())}

fora = ajustes[ajustes["tratamento"].str.startswith("fora")].groupby("plano")["ocorrencias"].sum()
print(f"Síntese factual: {int(ajustes['ocorrencias'].sum())} ocorrências de formas afetadas pelos ajustes de "
      f"tokenização, das quais {int(fora.sum())} ficam fora da contagem nos planos sem o ajuste ("
      + "; ".join(f"{s}: {n}" for s, n in fora.items()) + f"). {len(sem_correcao)} lemas, em "
      f"{sem_correcao['plano'].nunique()} planos, ficam sem a correção adotada no notebook de outro plano, somando "
      f"{int(sem_correcao['frequencia'].sum())} ocorrências. {len(LEMAS_SENSIVEIS)} lemas são marcados como sensíveis "
      "a essas diferenças nas tabelas de termos distintivos (seção 13).")
```

**Método e escolhas.** Duas diferenças de pré-processamento entre os planos vêm dos notebooks individuais e foram mantidas, para preservar a continuidade: (a) os ajustes de tokenização dos dois planos chineses (`separar_simbolo_final` e `siglas_com_e_comercial`, em `DOCUMENTOS`), que os outros quatro notebooks não adotam; (b) as listas de correções de lema, levantadas por cada notebook no exame do próprio texto. A tabela (a) conta, diretamente no texto de cada plano, as formas afetadas pelas regras de tokenização dos planos chineses e informa se elas são contadas no plano. A tabela (b) lista, para cada plano, os lemas que a correção do notebook de **outro** plano uniria a um lema de destino e que, neste plano, ficam separados. Os lemas da tabela (b), os seus lemas de destino e as siglas com "&" formam a lista `LEMAS_SENSIVEIS`, marcada na coluna `sensivel_ao_pre_processamento` das tabelas de termos distintivos (seção 13). Nada é recontado aqui.

**Como ler.** Na tabela (a), "fora da contagem" indica formas que não são tokens no plano, porque o ajuste está inativo nele: não entram nem no total de tokens. Na tabela (b), cada linha é um lema que, sob o pré-processamento de outro notebook, seria somado ao lema de destino.

**Interpretação.** *Observação:* os ajustes de tokenização afetam 59 ocorrências nos seis planos; 30 delas ficam fora da contagem nos planos sem o ajuste. A maior parte são siglas com "&": *R&D* (15), *R&D&I* (2) e *ST&I* (1) no PBIA; *R&D* (4) nos EUA; *R&I* (4) e *R&D* (1) no AI Continent. Nos planos chineses, as mesmas siglas são contadas: *R&D* (3) e *S&T* (3) no *AI+*; *R&D* (9) no New Generation. Os demais casos são o "+" colado ao fim da palavra (*AI+*, 9 ocorrências, e *Intelligence+*, 2, no *AI+*; *Internet+*, 1, no New Generation, todas contadas; *Erasmus+*, 1, fora da contagem no Apply AI) e o hífen suspenso (*intelligentization-* e *medium-*, contados no *AI+*; *medium-* e *profile-*, fora da contagem no Apply AI). Na tabela (b), 20 lemas, em 5 planos, ficam sem a correção adotada no notebook de outro plano, somando 51 ocorrências. Os casos com mais ocorrências são *gigafactorie* (6) no AI Continent, que o notebook do Apply AI corrige para *gigafactory*, e *services* (6) no PBIA, que o notebook do Apply AI corrige para *service*; os demais têm de 1 a 5 ocorrências. *Inferência:* as diferenças são pequenas diante do tamanho dos planos, mas criam dois efeitos na comparação. O lema *r&d* só existe nos planos chineses (9 ocorrências no New Generation e 3 no *AI+*), embora o PBIA, os EUA e o AI Continent também escrevam *R&D*; por isso, ele passa nos critérios de distinção do New Generation (seção 13). Além disso, alguns lemas ficam divididos em um plano e unidos em outro (*gigafactory*, *service*, *state*, *center*, *skill*), o que subestima a frequência do lema de destino no plano sem a correção.

**Fragilidades e ajustes possíveis.** Nem toda correção de outro notebook se aplicaria necessariamente a este plano: *medium*, por exemplo, só deve virar *media* quando designa meios de comunicação, e essa verificação, caso a caso, cabe ao notebook individual de cada plano. Uniformizar o pré-processamento (por exemplo, contar as siglas com "&" em todos os planos ou adotar a mesma lista de correções) mudaria os totais de tokens e as frequências dos notebooks individuais e, portanto, só pode ser feito por decisão do pesquisador, nos notebooks individuais e neste notebook ao mesmo tempo.

### 11.2. Conferência de continuidade com os notebooks individuais

*Comando do pesquisador: a contagem de 75 é apenas a continuação da contagem de cerca de 30 dos notebooks individuais.*

```python
conferencia = []
N_RECORTE_INDIVIDUAL = {}   # tamanho do recorte de cada notebook individual (marcado nos gráficos da seção 12)
for sigla, cfg in DOCUMENTOS.items():
    gravado = ler_recorte_individual(PASTA / cfg["notebook"], cfg["recorte_individual"]["celula"])
    linha = {"plano": sigla, "notebook": cfg["notebook"], "secao": cfg["recorte_individual"]["secao"]}
    if gravado is None:
        linha.update({"lemas_no_recorte_gravado": None, "coincidem": "não conferido (tabela não encontrada)",
                      "primeira_divergencia": "—"})
    else:
        n = len(gravado)
        N_RECORTE_INDIVIDUAL[sigla] = n
        atual = freq[sigla].head(n)
        iguais_lema = list(atual.index) == list(gravado.index)
        iguais_freq = iguais_lema and (atual["frequencia"].to_numpy() == gravado["frequencia"].to_numpy()).all()
        iguais_taxa = iguais_lema and np.allclose(atual["por_1000_tokens"].round(2), gravado["por_1000_tokens"])
        divergencia = "—"
        if not (iguais_lema and iguais_freq and iguais_taxa):
            for k, (a, b) in enumerate(zip(atual.itertuples(), gravado.itertuples()), start=1):
                if a.Index != b.Index or a.frequencia != b.frequencia:
                    divergencia = f"posição {k}: {a.Index} ({a.frequencia}) aqui; {b.Index} ({b.frequencia}) no notebook"
                    break
        linha.update({"lemas_no_recorte_gravado": n,
                      "coincidem": "sim" if (iguais_lema and iguais_freq and iguais_taxa) else "NÃO",
                      "primeira_divergencia": divergencia})
    conferencia.append(linha)
conferencia = pd.DataFrame(conferencia).set_index("plano")
exibir(conferencia, 0)
salvar_tabela(conferencia, "00_conferencia_continuidade")

ok = (conferencia["coincidem"] == "sim").sum()
print(f"Síntese factual: em {ok} dos {len(conferencia)} planos, os primeiros lemas deste notebook coincidem, na mesma "
      "ordem, com as mesmas frequências e as mesmas taxas por 1.000 tokens, com o recorte gravado no notebook individual.")
```

**Método e escolhas.** Para cada plano, a célula lê, no notebook individual, a tabela gravada na saída da célula do recorte de cerca de 30 lemas (a seção e a célula estão em `DOCUMENTOS["recorte_individual"]`): AI Continent, seção 15 (30 lemas); Apply AI, seção 16 (30 lemas, ampliados para 32 no empate); *AI+*, seção 16 (34 lemas, o recorte sem empate que sucedeu o de 30 da seção 15); New Generation, seção 15 (30 lemas); EUA, seção 16 (30 lemas, com as exclusões da seção 15); PBIA, seção 16 (31 lemas, sem *Brazil* e sem nomes de instituições e órgãos brasileiros). A leitura é feita em modo somente leitura, e nenhum número lido entra nos cálculos. A comparação é feita lema a lema: mesma posição, mesma frequência e mesma taxa por 1.000 tokens, esta com as duas casas decimais gravadas na tabela do notebook individual.

**Como ler.** "sim" indica que as primeiras posições do recorte de 75 deste notebook são exatamente o recorte do notebook individual. Em caso de divergência, a última coluna mostra a primeira posição diferente.

**Interpretação.** *Observação:* nos seis planos, os primeiros lemas deste notebook coincidem com o recorte gravado no notebook individual: 30 lemas no AI Continent, no New Generation e nos EUA, 31 no PBIA, 32 no Apply AI e 34 no *AI+*, todos na mesma ordem, com as mesmas frequências e as mesmas taxas. Os lemas que os recortes de 30 do AI Continent e dos EUA deixaram de fora pelo desempate alfabético aparecem logo depois do corte, na mesma ordem: *application*, *cloud* e *facilitate* (21 ocorrências cada) nas posições 31 a 33 do AI Continent; *manufacturing*, *risk*, *semiconductor* e *standard* (16 cada) nas posições 31 a 34 dos EUA. *Inferência:* o recorte de 75 é a continuação exata da contagem dos notebooks individuais, como pede o comando.

**Fragilidades e ajustes possíveis.** A conferência depende das saídas gravadas nos notebooks individuais. Se um deles for reexecutado com outros parâmetros, ela aponta a divergência, e os parâmetros por plano da configuração (seção 8) precisam ser atualizados. Só se conferem as posições gravadas (de 30 a 34); as posições seguintes não existem nos notebooks individuais e foram verificadas apenas na construção deste notebook, pela igualdade das tabelas completas de frequência.

## 12. Os 75 lemas de conteúdo mais frequentes de cada plano

*Visualização construída por comando do pesquisador (2026-10-05): continuação, até 75 lemas, dos recortes de cerca de 30 lemas dos notebooks individuais, só com unigramas e com os empates no 75º lugar incluídos.*

```python
recortes, info_recortes = {}, {}
for sigla in DOCUMENTOS:
    recortes[sigla], info_recortes[sigla] = recorte_com_empates(freq[sigla], TOP_N_COMPARACAO)

resumo_recortes = pd.DataFrame([{
    "plano": sigla,
    "recorte_do_notebook_individual": N_RECORTE_INDIVIDUAL.get(sigla),
    f"frequencia_do_{TOP_N_COMPARACAO}o_lema": int(info["valor_do_corte"]),
    "lemas_no_recorte": info["n_recorte"],
    "acrescentados_por_empate": len(info["empatados_acrescentados"]),
    "primeiro_lema_fora": (f"{info['primeiro_fora'][0]} ({int(info['primeiro_fora'][1])})"
                           if info["primeiro_fora"] else "—"),
    "ocorrencias_no_recorte": int(recortes[sigla]["frequencia"].sum()),
    "pct_dos_tokens_de_conteudo": recortes[sigla]["frequencia"].sum() / processados[sigla]["tokens"]["conteudo"].sum() * 100,
} for sigla, info in info_recortes.items()]).set_index("plano")
exibir(resumo_recortes)
salvar_tabela(resumo_recortes, "01_resumo_recortes")


def mostrar_recorte(sigla):
    """Gráfico e tabela do recorte de TOP_N_COMPARACAO lemas (ampliado nos empates) de um plano.
    A linha tracejada superior marca o fim do recorte gravado no notebook individual; a inferior,
    quando há empate no corte, separa os lemas acrescentados por empate."""
    top, info = recortes[sigla], info_recortes[sigla]
    n, n_ind = len(top), N_RECORTE_INDIVIDUAL.get(sigla)
    n_unidades = processados[sigla]["n_unidades"]

    fig, ax = plt.subplots(figsize=(7.5, 0.2 * n + 1.6))
    y = np.arange(n)
    ax.barh(y, top["por_1000_tokens"], height=0.7, color=COR_DOCUMENTO[sigla])
    ax.set_yticks(y, labels=top.index)
    ax.set_ylim(n - 0.5, -0.5)
    xmax = top["por_1000_tokens"].max()
    for yi, (_, linha) in zip(y, top.iterrows()):
        ax.text(linha["por_1000_tokens"] + xmax * 0.015, yi,
                f"{num(linha['por_1000_tokens'])} · {int(linha['frequencia'])} ocorr. · "
                f"{int(linha['unidades_com_ocorrencia'])}/{n_unidades} unid.",
                va="center", fontsize=9, color=COR_TINTA_2)
    ax.set_xlim(0, xmax * 1.75)
    linhas = []
    if n_ind and n_ind < n:
        linhas.append((n_ind, f"acima da linha: recorte de {n_ind} do notebook individual"))
    if info["empatados_acrescentados"]:
        linhas.append((TOP_N_COMPARACAO, f"abaixo da linha: empatados com o {TOP_N_COMPARACAO}º"))
    for pos, texto in linhas:
        ax.axhline(pos - 0.5, color=COR_TINTA_2, lw=0.8, ls="--")
        ax.text(xmax * 1.73, pos - 0.5, texto, ha="right", va="bottom", fontsize=9, color=COR_TINTA_2)
    ax.xaxis.set_major_formatter(FMT_BR)
    ax.set_xlabel("Ocorrências por 1.000 tokens")
    ax.set_title(f"{DOCUMENTOS[sigla]['rotulo']}\n{n} lemas de conteúdo mais frequentes", loc="left")
    ax.grid(axis="y", visible=False)
    salvar_figura(fig, f"01_lemas_mais_frequentes_{sigla}")
    plt.show()

    tabela = top[["frequencia", "por_1000_tokens", "unidades_com_ocorrencia", "pct_unidades", "DP_norm",
                  "formas_observadas"]].copy()
    tabela.insert(0, "posicao", np.arange(1, n + 1))
    tabela.insert(1, "recorte", [
        "notebook individual" if n_ind and k <= n_ind
        else (f"ampliação até {TOP_N_COMPARACAO}" if k <= TOP_N_COMPARACAO else f"empate com o {TOP_N_COMPARACAO}º")
        for k in range(1, n + 1)])
    if sigla in EXCLUSOES:
        t = processados[sigla]["tokens"]
        sem_exclusoes = t.loc[t["conteudo_antes_das_exclusoes"], "lema"].value_counts()
        tabela.insert(3, "frequencia_sem_exclusoes", sem_exclusoes.reindex(tabela.index).astype(int))
    exibir(tabela)
    salvar_tabela(tabela, f"01_lemas_mais_frequentes_{sigla}")

    anteriores, novos = (top.iloc[:n_ind], top.iloc[n_ind:]) if n_ind else (top.iloc[:0], top)
    print(f"Síntese factual: {n} lemas ({TOP_N_COMPARACAO} pedidos"
          + (f" + {len(info['empatados_acrescentados'])} empatados com o {TOP_N_COMPARACAO}º, com "
             f"{int(info['valor_do_corte'])} ocorrências" if info["empatados_acrescentados"]
             else f"; o {TOP_N_COMPARACAO}º tem {int(info['valor_do_corte'])} ocorrências e não há empate no corte")
          + (f"; primeiro lema fora: {info['primeiro_fora'][0]}, com {int(info['primeiro_fora'][1])}" if info["primeiro_fora"] else "")
          + f"). Os {n} lemas somam {num(top['frequencia'].sum(), 0)} ocorrências, "
          f"{num(top['frequencia'].sum() / processados[sigla]['tokens']['conteudo'].sum() * 100, 1)}% dos tokens de "
          f"conteúdo. Posições novas em relação ao notebook individual: {len(novos)}, com frequências de "
          f"{int(novos['frequencia'].max())} a {int(novos['frequencia'].min())} ocorrências. DP_norm mediano: "
          f"{num(anteriores['DP_norm'].median())} nas {len(anteriores)} posições do notebook individual e "
          f"{num(novos['DP_norm'].median())} nas novas; alcance mediano: {num(anteriores['unidades_com_ocorrencia'].median(), 1)} "
          f"e {num(novos['unidades_com_ocorrencia'].median(), 1)} unidades.")
```

**Método e escolhas.** Para cada plano, a tabela de frequências é a tabela completa da seção 11, com as exclusões do pesquisador nos EUA e no PBIA: a mesma base dos recortes de cerca de 30 lemas dos notebooks individuais. O recorte tem 75 lemas (`TOP_N_COMPARACAO`), por comando do pesquisador, e é ampliado até incluir todos os lemas com a mesma frequência do 75º, para que o desempate alfabético não decida quem aparece (regra definida pelo pesquisador para os recortes por ele pedidos). Dentro do recorte, a ordem é a dos notebooks individuais: frequência decrescente e, nos empates, ordem alfabética. Só há unigramas, e cada composto com hífen conta como um unigrama. Nos gráficos, a barra é a **taxa por 1.000 tokens**, medida relativa usada em toda a comparação entre planos (skill 03, seção 3.7); dentro de um plano, a ordem é a mesma da frequência absoluta. O rótulo de cada barra traz a taxa, a frequência absoluta e o alcance (em quantas unidades o lema aparece). A linha tracejada superior marca o fim do recorte gravado no notebook individual (30, 31, 32 ou 34 lemas); a inferior, quando existe, separa os lemas acrescentados por empate com o 75º. Cada tabela traz a posição, a origem da posição (notebook individual, ampliação até 75 ou empate com o 75º), a frequência, a taxa, o alcance, o DP normalizado e as formas observadas; nos EUA e no PBIA, também a frequência sem as exclusões do pesquisador.

**Como ler.** Barras mais longas indicam lemas mais frequentes no plano. Um lema frequente presente em muitas unidades é estruturante no texto; um lema presente em poucas unidades é localizado. A distinção aparece no alcance e na coluna `DP_norm` (0 = espalhado; 1 = concentrado). A escala horizontal muda de um gráfico para outro: as taxas de planos diferentes devem ser comparadas pelos números, e não pelo comprimento das barras em gráficos distintos.

**Interpretação.** *Observação:* em três planos, o 75º lema não está empatado: AI Continent (13 ocorrências; o seguinte, *accelerate*, tem 12), New Generation (22; *level*, 21) e EUA (11; *application*, 10). Nos outros três, o corte cai num empate e o recorte é ampliado: Apply AI, de 75 para 85 lemas (10 ocorrências; o primeiro lema fora, *addition*, tem 9); PBIA, para 80 (23; *implementation*, 22); e *AI+*, para 98 (5; *achievement*, 4), porque 34 lemas desse plano têm exatamente 5 ocorrências. Os recortes cobrem de 36,4% (EUA) a 51,0% (*AI+*) dos tokens de conteúdo: 41,7% no AI Continent, 41,6% no Apply AI, 47,9% no New Generation e 40,9% no PBIA. Em todos os planos, as posições acrescentadas são mais concentradas que as do recorte individual, como mostram as sínteses das seções 12.1 a 12.6: o DP normalizado mediano passa de 0,76 para 0,86 no AI Continent, de 0,79 para 0,88 no Apply AI, de 0,72 para 0,86 no *AI+*, de 0,67 para 0,83 no New Generation, de 0,86 para 0,92 nos EUA e de 0,82 para 0,92 no PBIA, e o alcance mediano cai a cerca da metade (no PBIA, de 49 para 25 unidades). *Inferência:* a ampliação até 75 alcança, em todos os planos, um vocabulário ligado a partes específicas do texto (seções, quadros, anexos ou parágrafos sobre um tema), mais do que vocabulário que atravessa o documento inteiro.

**Fragilidades e ajustes possíveis.** Frequência não equivale a importância política. Nas posições finais, as frequências são próximas (uma ocorrência separa um lema do seguinte) e a ordem não deve ser lida como diferença de ênfase; dentro de cada frequência, a ordem é apenas alfabética. No *AI+*, plano de 2.874 tokens, uma ocorrência vale 0,35 por 1.000 tokens, e o corte no 75º lugar cai num empate de 34 lemas com 5 ocorrências: o recorte chega a 98 lemas, bem acima dos 75 pedidos. A lematização reúne formas de sentidos diferentes (*develop* reúne *developing* e *developed*), e *ai* e *artificial intelligence* não são unificados. As diferenças de pré-processamento herdadas dos notebooks individuais (seção 11.1) valem também aqui.

Entre as posições novas, destacam-se: (1) resultados esperados das ações: *reduction* (37; *reduction of* em 28 ocorrências), *improvement* (33), *efficiency* (31), *quality* (25) e *great* (25; 21 na forma *greater*), concentrados nos anexos: 29 das 37 ocorrências de *reduction* e 28 das 33 de *improvement* estão em unidades da categoria `Annex`; (2) público e valores: *population* (29; 7 vezes precedido de *Brazilian*), *society* (29), *social* (29), *responsible* (31), *sustainable* (28), *benefit* (26) e *opportunity* (25); (3) o próprio plano e a sua estrutura: *pbia* (30), *plan* (31) e *axis* (25, dos quais 10 em títulos internos com o rótulo *Axis N*); (4) formação e capacidades: *training* (35), *capacity* (31), *qualification* (27) e *professional* (23); (5) economia: *industry* (28), *company* (24), *chain* (24) e *economic* (23). Os lemas novos mais concentrados são *axis* (DP 0,96), *professional*, *great* e *industry* (0,95); os mais espalhados, *strategic* e *responsible* (0,86) e *application* (0,87). As exclusões do pesquisador alcançam oito posições novas, com perdas de 1 a 3 ocorrências: *strategic*, *government*, *management*, *center*, *social*, *energy*, *company* e *digital*. *Inferência:* as posições 32 a 80 trazem o vocabulário com que o plano descreve os resultados esperados das ações (redução, melhoria, eficiência, qualidade), concentrado nos anexos, e o vocabulário dos destinatários e dos valores do plano (população, sociedade, responsabilidade, sustentabilidade).

**Fragilidades e ajustes possíveis.** Parte do vocabulário novo é formular e vem da estrutura em lista dos anexos (31 das 37 ocorrências de *reduction* estão em itens de lista); o notebook individual tratou essa repetição como frequência real, e não como inflação (seção 10.2 daquele notebook). *axis* inclui os rótulos *Axis N* dos títulos internos, mantidos pela skill 01. *great* é quase sempre o comparativo *greater*. As frequências dependem das exclusões da seção 16 do notebook do PBIA, cujo efeito aparece na coluna `frequencia_sem_exclusoes`.

## 13. Termos distintivos de cada plano

*Visualização construída por comando do pesquisador (2026-10-05): os termos distintivos da versão anterior deste notebook, atualizados com a metodologia da pasta e restritos a unigramas.*

```python
# Frequências de todos os lemas de conteúdo, plano a plano (com as exclusões do pesquisador), e total de
# tokens de cada plano (denominador das taxas, inclusive stopwords)
freqs = pd.DataFrame({sigla: freq[sigla]["frequencia"] for sigla in DOCUMENTOS}).fillna(0).astype(int)
keyness, N_TESTES = tabela_keyness(freqs, N_TOKENS)

# Critério de especificidade da versão anterior deste notebook: ênfase = estar entre os TOP_K_ENFASE lemas
# mais frequentes do plano (empates incluídos)
ENFASE = {sigla: lemas_com_enfase(freq[sigla], TOP_K_ENFASE) for sigla in DOCUMENTOS}
planos_com_enfase = pd.Series(0, index=freqs.index)
for sigla in DOCUMENTOS:
    planos_com_enfase[list(ENFASE[sigla])] += 1

distintivos, top_distintivos, info_distintivos = {}, {}, {}
for sigla, tab in keyness.items():
    tab["planos_com_enfase"] = planos_com_enfase.reindex(tab.index).astype(int)
    tab["enfase_no_plano"] = tab.index.isin(list(ENFASE[sigla]))
    tab["distintivo"] = tab["passa_keyness"] & tab["enfase_no_plano"] & (tab["planos_com_enfase"] <= MAX_PLANOS_COM_ENFASE)
    tab["sensivel_ao_pre_processamento"] = tab.index.map(lambda lema: LEMAS_SENSIVEIS.get(lema, ""))
    tab["formas_observadas"] = freq[sigla]["formas_observadas"].reindex(tab.index)
    # Ordenação da versão anterior: taxa no próprio plano (decrescente); empate -> ordem alfabética
    d = tab[tab["distintivo"]].reset_index().sort_values(["freq_plano", "lema"], ascending=[False, True]).set_index("lema")
    distintivos[sigla] = d
    top_distintivos[sigla], info_distintivos[sigla] = recorte_com_empates(d, TOP_N_KEYNESS, coluna="freq_plano")
    salvar_tabela(tab, f"02_keyness_completa_{sigla}")

resumo_keyness = pd.DataFrame([{
    "plano": sigla,
    "lemas_testados": len(tab),
    "passam_G2_e_log_ratio": int(tab["passa_keyness"].sum()),
    "com_enfase_no_plano": int((tab["passa_keyness"] & tab["enfase_no_plano"]).sum()),
    f"distintivos_(enfase_em_ate_{MAX_PLANOS_COM_ENFASE}_planos)": len(distintivos[sigla]),
    "distintivos_que_resistem_a_Bonferroni": int(distintivos[sigla]["resiste_bonferroni"].sum()),
    "exibidos": len(top_distintivos[sigla]),
    "exibidos_que_resistem_a_Bonferroni": int(top_distintivos[sigla]["resiste_bonferroni"].sum()),
} for sigla, tab in keyness.items()]).set_index("plano")
exibir(resumo_keyness, 0)
salvar_tabela(resumo_keyness, "02_resumo_termos_distintivos")

n_testados = int((freqs.sum(axis=1) >= FREQ_MINIMA_COMPARACAO).sum())
alfa_corrigido = ALFA_BONFERRONI / N_TESTES
alfa_texto = f"{alfa_corrigido:.2e}".replace(".", ",")   # notação brasileira
print(f"Síntese factual: {num(len(freqs), 0)} lemas de conteúdo distintos nos seis planos; {num(n_testados, 0)} "
      f"têm frequência combinada >= {FREQ_MINIMA_COMPARACAO} e foram testados em cada um dos {len(DOCUMENTOS)} planos "
      f"({num(N_TESTES, 0)} testes; nível de Bonferroni: p < {alfa_texto}). Lemas com ênfase "
      f"(entre os {TOP_K_ENFASE} mais frequentes, empates incluídos; frequência do {TOP_K_ENFASE}º lema entre parênteses): "
      + "; ".join(f"{s}: {len(ENFASE[s])} ({int(freq[s]['frequencia'].iloc[TOP_K_ENFASE - 1])})" for s in DOCUMENTOS) + ".")
```

```python
COLUNAS_DISTINTIVOS = ["freq_plano", "por_1000_plano", "por_1000_demais", "G2", "log_ratio", "exclusivo",
                       "resiste_bonferroni", "planos_com_enfase", "sensivel_ao_pre_processamento", "formas_observadas"]
for sigla in DOCUMENTOS:
    top, info = top_distintivos[sigla], info_distintivos[sigla]
    tabela = top[COLUNAS_DISTINTIVOS].copy()
    tabela.insert(0, "posicao", np.arange(1, len(tabela) + 1))
    display(Markdown(f"**{DOCUMENTOS[sigla]['rotulo']}** (`{sigla}`): {len(tabela)} termos exibidos"
                     + (f" ({TOP_N_KEYNESS} + {len(info['empatados_acrescentados'])} empatados com o {TOP_N_KEYNESS}º)"
                        if info["empatados_acrescentados"] else "")
                     + f" de {len(distintivos[sigla])} distintivos"))
    exibir(tabela)
    salvar_tabela(tabela, f"02_termos_distintivos_{sigla}")
```

```python
n_max = max(len(t) for t in top_distintivos.values())
fig, eixos = plt.subplots(3, 2, figsize=(13, 3 * (0.3 * n_max + 1.3)))
for ax, sigla in zip(eixos.flat, DOCUMENTOS):
    top, info = top_distintivos[sigla], info_distintivos[sigla]
    y = np.arange(len(top))
    ax.barh(y, top["por_1000_plano"], height=0.68, color=COR_DOCUMENTO[sigla])
    # Marcador: taxa do mesmo lema no conjunto dos outros cinco planos (referência da keyness)
    ax.scatter(top["por_1000_demais"], y, marker="|", s=170, linewidths=2, color=COR_REFERENCIA, zorder=3)
    xmax = top["por_1000_plano"].max()
    for yi, (_, linha) in zip(y, top.iterrows()):
        ax.text(linha["por_1000_plano"] + xmax * 0.02, yi, num(linha["por_1000_plano"]),
                va="center", ha="left", fontsize=9, color=COR_TINTA)
    if info["empatados_acrescentados"]:
        ax.axhline(TOP_N_KEYNESS - 0.5, color=COR_TINTA_2, lw=0.8, ls="--")
        ax.text(xmax * 1.2, TOP_N_KEYNESS - 0.5, f"abaixo: empatados com o {TOP_N_KEYNESS}º",
                ha="right", va="bottom", fontsize=9, color=COR_TINTA_2)
    ax.set_yticks(y, labels=top.index)
    ax.set_ylim(n_max - 0.5, -0.5)
    ax.set_xlim(0, xmax * 1.22)
    ax.xaxis.set_major_formatter(FMT_BR)
    ax.set_xlabel("Ocorrências por 1.000 tokens")
    ax.set_title(DOCUMENTOS[sigla]["rotulo"], loc="left", fontsize=10.5)
    ax.grid(axis="y", visible=False)
fig.legend(handles=[Patch(color=COR_TINTA_2, label="barra (cor do plano): taxa no plano"),
                    Line2D([], [], marker="|", ls="", markersize=12, markeredgewidth=2, color=COR_REFERENCIA,
                           label="traço: taxa no conjunto dos outros cinco planos")],
           loc="upper left", bbox_to_anchor=(0.0, 1.0), ncol=2, fontsize=9)
fig.suptitle(f"Termos distintivos de cada plano (até {TOP_N_KEYNESS} por plano, ampliado nos empates)",
             x=0.0, ha="left", y=1.025, fontsize=12)
fig.tight_layout(rect=(0, 0, 1, 0.985))
salvar_figura(fig, "02_termos_distintivos")
plt.show()

linhas_sintese = []
for s in DOCUMENTOS:
    t = top_distintivos[s]
    razao = 2 ** t.loc[~t["exclusivo"], "log_ratio"]
    linhas_sintese.append(
        f"{s}: {len(t)} exibidos; taxa no plano de {num(razao.min(), 1)} a {num(razao.max(), 1)} vezes a dos demais"
        + (f", mais {int(t['exclusivo'].sum())} exclusivo(s)" if t["exclusivo"].any() else ""))
print("Síntese factual: " + "; ".join(linhas_sintese) + ".")
```

**Método e escolhas.** O critério da versão anterior deste notebook foi mantido e combinado com os critérios de distinção da skill 03 (seção 3.5). Um lema é **termo distintivo** de um plano quando atende, ao mesmo tempo, a:

1. **especificidade** (critério da versão anterior): o lema é usado "com ênfase" no próprio plano e em no máximo 3 dos 6 planos (`MAX_PLANOS_COM_ENFASE`). "Com ênfase" quer dizer estar entre os 200 lemas mais frequentes do plano (`TOP_K_ENFASE`), com os empates no 200º lugar incluídos;
2. ***keyness*** (skill 03): comparação de cada plano com a soma dos outros cinco (`MODO_KEYNESS = "um_contra_os_demais"`), entre os lemas com frequência combinada de pelo menos 5 (`FREQ_MINIMA_COMPARACAO`), com **G²** de log-verossimilhança ≥ 6,63 (p < 0,01; 1 grau de liberdade) **e** ***Log Ratio*** ≥ 1, isto é, a taxa por 1.000 tokens no plano é pelo menos o dobro da taxa no conjunto dos demais. Quando o lema não aparece nos demais planos, a correção de 0,5 entra só no cálculo do *Log Ratio*, e o termo é marcado como exclusivo. A coluna `resiste_bonferroni` informa se o p-valor do G² fica abaixo de 0,01 dividido pelo número total de testes (6.930: 1.155 lemas testados em cada um dos 6 planos); é informação, e não filtro.

Como na versão anterior, os termos de cada plano são ordenados pela taxa no próprio plano (o que, dentro de um plano, equivale à frequência), e exibem-se 15 (`TOP_N_KEYNESS`), com a mesma regra de empate dos recortes de 75. As contagens são as da seção 11, com as exclusões do pesquisador nos EUA e no PBIA: as mesmas dos recortes de 75. Só há unigramas, e cada composto com hífen conta como um unigrama. O modo "um contra os demais" foi adotado porque responde à pergunta da versão anterior, isto é, o que distingue cada plano dos outros; o modo por pares (padrão da skill 03) produziria 30 listas (15 pares, nos dois sentidos). A escolha do modo cabe ao pesquisador e pode ser revista. As tabelas completas de *keyness* de cada plano (`02_keyness_completa_<sigla>`) são exportadas para `resultados/` apenas por comando do pesquisador.

**Como ler.** Em cada painel, a barra é a taxa por 1.000 tokens do lema no plano, e o traço cinza é a taxa do mesmo lema no conjunto dos outros cinco planos. A distância entre a ponta da barra e o traço mostra quanto o plano emprega o lema acima do conjunto; quando o traço fica no zero, o lema é exclusivo do plano. Cada painel tem a sua escala. Nas tabelas, `planos_com_enfase` indica em quantos planos o lema está entre os 200 mais frequentes, e `sensivel_ao_pre_processamento` assinala os lemas cuja comparação depende das diferenças de pré-processamento examinadas na seção 11.1.

**Interpretação.** *Observação:* dos 3.629 lemas de conteúdo dos seis planos, 1.155 foram testados. O número de termos distintivos vai de 48 (Apply AI) a 96 (New Generation): 68 no AI Continent, 49 no *AI+*, 55 nos EUA e 72 no PBIA. Entre os exibidos, todos resistem à correção de Bonferroni no AI Continent (15 de 15) e no New Generation (17 de 17); nos outros planos, a proporção é menor: 11 de 18 no PBIA, 8 de 15 no Apply AI, 8 de 19 no *AI+* e 6 de 16 nos EUA. Por plano:

- **AI Continent:** *eu* (11,98 por 1.000 tokens, contra 1,12 nos demais), *european*, *act*, *factory*, *strategy*, *commission*, *europe*, *digital*, *state*, *launch*, *member*, *apply*, *cloud*, *facilitate* e *hub*. Seis dos 15 nomeiam o bloco ou as suas instituições (*eu*, *european*, *europe*, *commission*, *member* e *state*, este quase sempre em *Member States*); *act* e *factory* vêm sobretudo de *AI Act* e *AI Factory*.
- **Apply AI:** *european*, *eu*, *commission*, *strategy*, *solution*, *europe*, *defence*, *digital*, *apply*, *sectoral*, *frontier*, *healthcare*, *uptake*, *deployment* e *space*. Os contrastes mais fortes são *defence* (2,65 contra 0,02; *Log Ratio* 6,98), *sectoral* (5,75) e *uptake* (4,66).
- ***AI+*:** *intelligent* (11,13 contra 3,47), *smart*, *explore*, *compute*, *coordination*, *consumption*, *encourage*, *open-source*, *product*, *agricultural*, *device*, *fully*, *cultural*, *incubate*, *integrate*, *intelligentize*, *refine*, *scenario* e *terminal* (19 termos: 15 mais 4 empatados com 6 ocorrências). *incubate* e *intelligentize* são exclusivos; *open-source* é contado como um unigrama.
- **New Generation:** *intelligent* (15,36 contra 0,91), *artificial* (14,56 contra 1,57), *theory* (5,38 contra 0,09; *Log Ratio* 5,89), *social*, *major*, *construction*, *enterprise*, *learning*, *knowledge*, *basic*, *product*, *group*, *form*, *robot*, *achieve*, *generation* e *smart* (17 termos).
- **EUA:** *program* (3,29 contra 1,17), *control*, *power*, *workforce*, *federal*, *grid*, *critical*, *exist*, *semiconductor*, *standard*, *export*, *adversary*, *compute*, *effort*, *relevant* e *worker* (16 termos). *export* e *adversary* são exclusivos.
- **PBIA:** *brazilian* (4,94; exclusivo), *solution*, *program*, *reduction*, *aim*, *improvement*, *government*, *responsible*, *pbia*, *center*, *population*, *society*, *sustainable*, *qualification*, *axis*, *great*, *opportunity* e *quality* (18 termos). Cinco são exclusivos: *brazilian*, *reduction*, *pbia*, *qualification* e *axis*.

Alguns termos são distintivos dos dois planos do mesmo bloco: *european*, *eu*, *commission*, *europe*, *strategy*, *digital* e *apply* nos dois planos da UE; *intelligent*, *smart* e *product* nos dois planos chineses. Entre planos de blocos diferentes, *program* é distintivo dos EUA e do PBIA; *compute*, do *AI+* e dos EUA; e *solution*, do Apply AI e do PBIA.

*Relação com a versão colada:* dos 81 unigramas que as listas da versão anterior traziam (misturados a bigramas), 57 continuam entre os termos exibidos: 8 de 12 no AI Continent, 9 de 13 no Apply AI, 12 de 14 no *AI+*, 10 de 15 no New Generation, 9 de 13 nos EUA e 9 de 14 no PBIA. As diferenças vêm de três fontes: (1) as exclusões: a versão anterior retirava, em todos os planos, o nome do país ou do bloco e de instituições, enquanto aqui valem apenas as exclusões que o pesquisador definiu nos notebooks individuais; por isso, *eu*, *european*, *commission*, *brazilian* e *pbia* aparecem, e *agency* e *government* saem dos EUA; (2) os critérios de *keyness*, que alguns termos não atendem (*solution* no AI Continent; *act* e *impact* no Apply AI; *government* e *society* no *AI+*; *center* e *expand* nos EUA; *artificial*, *social* e *digital* no PBIA); (3) o novo pré-processamento, com o qual alguns termos passam a ter ênfase em 4 ou mais planos (*investment*, *foster*, *address*, *enable*, *computing*, *country*) ou ficam abaixo do 15º lugar (*gigafactory*; *autonomous*, *breakthrough*, *core* e *method*).

*Inferência:* os termos distintivos indicam o enfoque próprio de cada plano: nos planos da UE, o vocabulário institucional do bloco e as suas iniciativas nomeadas (*AI Act*, *AI Factories*, *hubs*, nuvem) e, no Apply AI, a abordagem setorial (*sectoral*, *healthcare*, *defence*, *space*, *uptake*, *deployment*); no *AI+*, o léxico de *intelligent* e *smart*, a exploração e a coordenação, o consumo, o código aberto e termos de política próprios (*intelligentize*, *incubate*); no New Generation, a base técnico-científica (*theory*, *learning*, *knowledge*, *basic*, *group*) e o vocabulário de planejamento (*construction*, *form*, *achieve*); nos EUA, controles de exportação, adversários, rede elétrica (*grid*, *power*), semicondutores, força de trabalho e programas federais; no PBIA, o adjetivo nacional, a sigla e a estrutura do próprio plano (*pbia*, *axis*), os resultados esperados das ações (*reduction*, *improvement*, *quality*, *great*) e termos sociais (*population*, *society*, *responsible*, *sustainable*, *qualification*). Os termos compartilhados pelos dois planos do mesmo bloco sugerem um vocabulário de bloco, institucional na UE e centrado em *intelligent* e *smart* na China. Frequência relativa maior não equivale, por si, a maior importância política.

**Fragilidades e ajustes possíveis.**

- **Exclusões assimétricas.** Os EUA e o PBIA foram analisados sem os nomes de órgãos e do próprio país (com exceção de *Brazilian* e das menções ao presidente Trump, mantidas por decisão do pesquisador), e os planos da UE e da China, com eles. Por isso, seis dos 15 termos do AI Continent e quatro dos 15 do Apply AI nomeiam o bloco ou as suas instituições e ocupam posições que, com tratamento simétrico, iriam para outros lemas. Retirar esses nomes também nos planos da UE e da China é decisão do pesquisador.
- **Referência do modo "um contra os demais".** A soma dos outros cinco planos pesa mais os planos longos (o PBIA representa de 31% a 37% da referência dos outros planos), e, quando os dois planos de um mesmo bloco usam o mesmo termo, cada um entra na referência do outro, o que reduz o contraste de termos próprios do bloco.
- **Ênfase em planos curtos.** No *AI+*, o 200º lema tem só 2 ocorrências, e o critério de ênfase alcança 309 lemas; nos outros planos, a frequência do 200º lema vai de 5 a 10. No *AI+* e nos EUA, os termos exibidos têm frequências baixas (6 a 32 no *AI+*; 14 a 27 nos EUA), e a maioria não resiste à correção de Bonferroni.
- **Pré-processamento.** Dois termos exibidos estão marcados como sensíveis: *state* (AI Continent), porque o PBIA tem 3 ocorrências de *states* não unidas a *state*, e *center* (PBIA), porque o próprio PBIA tem 3 ocorrências de *centers* não unidas a *center*. Nos dois casos, somar as 3 ocorrências não altera o atendimento aos critérios, como se verificou na construção deste notebook (*state*: G² de 34,10 para 31,47 e *Log Ratio* de 2,20 para 2,08; *center*: G² de 12,71 para 16,31 e *Log Ratio* de 1,40 para 1,55). Fora dos termos exibidos, *r&d* aparece como distintivo do New Generation só porque as siglas com "&" são tokens apenas nos planos chineses (seção 11.1). Nos planos da UE, *factory*, *state* e *hub* reúnem, por correção do notebook individual, os plurais escritos como nome próprio (*AI Factories*, *Member States*, *Hubs*).
- **Forma de redação.** O New Generation escreve *artificial intelligence* por extenso muito mais que os outros planos, que usam sobretudo *AI*; *artificial* (e *intelligence*, abaixo do 15º lugar) distingue esse plano pela forma de escrever o termo, e não necessariamente pelo conteúdo.
- **Recorte e ordem.** A ordenação pela taxa privilegia os termos distintivos mais frequentes; termos com contraste maior, mas menos frequentes, ficam nas tabelas completas. Dentro de uma mesma frequência, a ordem é alfabética.

## 14. Registro de execução e síntese

### 15.2. Composição dos grupos e conferência do codebook

Lemas de cada grupo, posição de cada grupo na figura e conferências: lema repetido em dois grupos (interrompe a execução), lema dos recortes de 75 sem grupo, lema do codebook sem ocorrência e lemas sensíveis às diferenças de pré-processamento examinadas na seção 11.1.

```python
# ---------------------------------------------------------------------------
# Índice lema -> grupo e conferências do codebook
# ---------------------------------------------------------------------------
GRUPOS_ATIVOS = [g for g in GRUPOS_QUADRANTE if g != GRUPO_DESCARTADO]
if sorted(GRUPOS_ATIVOS) != sorted(EIXO_X + EIXO_Y):
    raise ValueError(f"Os grupos dos eixos ({EIXO_X + EIXO_Y}) devem ser todos os grupos fora do descarte "
                     f"({GRUPOS_ATIVOS}).")

LEMA_GRUPO, repetidos = {}, []
for grupo, cfg in GRUPOS_QUADRANTE.items():
    for lema in cfg["lemas"]:
        chave = lema.strip().lower()
        if chave in LEMA_GRUPO:
            repetidos.append(f"'{chave}' em {LEMA_GRUPO[chave]} e em {grupo}")
        else:
            LEMA_GRUPO[chave] = grupo
if repetidos:
    raise ValueError("Cada lema deve estar em um único grupo. Repetições: " + "; ".join(repetidos))

# Lemas dos recortes de 75 da seção 12 (união dos seis planos, com os empates incluídos)
LEMAS_RECORTES = sorted(set().union(*(set(t.index) for t in recortes.values())))
sem_grupo = [lema for lema in LEMAS_RECORTES if lema not in LEMA_GRUPO]
fora_dos_recortes = sorted(lema for lema in LEMA_GRUPO if lema not in set(LEMAS_RECORTES))
sem_ocorrencia = sorted(lema for lema in LEMA_GRUPO if not any(lema in freq[s].index for s in DOCUMENTOS))

# Polos: a letra e a posição de cada grupo ativo na figura (como na figura original)
POLOS_QUADRANTE = {"topo": ("A", EIXO_Y[0]), "esquerda": ("B", EIXO_X[1]),
                   "base": ("C", EIXO_Y[1]), "direita": ("D", EIXO_X[0])}
POLO_DO_GRUPO = {g: (letra, lado) for lado, (letra, g) in POLOS_QUADRANTE.items()}
SENTIDO_POLO = {"topo": "topo, Y > 0", "base": "base, Y < 0", "direita": "direita, X > 0",
                "esquerda": "esquerda, X < 0"}


def descrever_polo(grupo):
    if grupo == GRUPO_DESCARTADO:
        return "fora dos eixos"
    letra, lado = POLO_DO_GRUPO[grupo]
    return f"Vocabulário {letra} ({SENTIDO_POLO[lado]})"


# (a) Lemas de cada grupo
linhas_md = []
for grupo, cfg in GRUPOS_QUADRANTE.items():
    lemas = sorted({l.strip().lower() for l in cfg["lemas"]})
    marcados = [f"{l}†" if l in fora_dos_recortes else l for l in lemas]
    linhas_md.append(f"- **{grupo} — {cfg['nome']}** · {descrever_polo(grupo)} · {len(lemas)} lemas: "
                     + ", ".join(marcados))
display(Markdown("**(a) Lemas de cada grupo** († = lema de fora dos recortes de 75, incluído no codebook)\n\n"
                 + "\n".join(linhas_md)))

# (b) Resumo do codebook
composicao_quadrante = pd.DataFrame([{
    "grupo": grupo,
    "nome": cfg["nome"],
    "posicao_na_figura": descrever_polo(grupo),
    "lemas": len(cfg["lemas"]),
    "lemas_dos_recortes_de_75": sum(l.strip().lower() in set(LEMAS_RECORTES) for l in cfg["lemas"]),
    "lemas_de_fora_dos_recortes": sum(l.strip().lower() in fora_dos_recortes for l in cfg["lemas"]),
    "lemas_sensiveis_ao_pre_processamento": sum(l.strip().lower() in LEMAS_SENSIVEIS for l in cfg["lemas"]),
} for grupo, cfg in GRUPOS_QUADRANTE.items()]).set_index("grupo")
display(Markdown("**(b) Resumo do codebook**"))
exibir(composicao_quadrante, 0)
salvar_tabela(pd.DataFrame([{"grupo": g, "nome": c["nome"], "lema": l} for g, c in GRUPOS_QUADRANTE.items()
                            for l in c["lemas"]]), "03_quadrante_codebook", index=False)

if sem_grupo:
    print(f"ATENÇÃO: {len(sem_grupo)} lema(s) dos recortes de 75 sem grupo no codebook (ficam fora de todas "
          f"as somas): {', '.join(sem_grupo)}")
if sem_ocorrencia:
    print(f"ATENÇÃO: lema(s) do codebook sem nenhuma ocorrência nos seis planos (confira a grafia): "
          f"{', '.join(sem_ocorrencia)}")
sensiveis = sorted(l for l in LEMA_GRUPO if l in LEMAS_SENSIVEIS and LEMA_GRUPO[l] != GRUPO_DESCARTADO)
print(f"Síntese factual: o codebook tem {len(LEMA_GRUPO)} lemas, "
      + ", ".join(f"{len(c['lemas'])} em {g}" for g, c in GRUPOS_QUADRANTE.items())
      + f". Os recortes de 75 dos seis planos somam {len(LEMAS_RECORTES)} lemas distintos, "
      + (f"todos classificados" if not sem_grupo else f"{len(sem_grupo)} deles sem grupo")
      + f"; {len(fora_dos_recortes)} lema(s) do codebook vêm de fora dos recortes ({', '.join(fora_dos_recortes) or 'nenhum'}). "
      f"Lemas dos grupos ativos sensíveis às diferenças de pré-processamento da seção 11.1: "
      + (", ".join(f"{l} ({LEMAS_SENSIVEIS[l]})" for l in sensiveis) if sensiveis else "nenhum") + ".")
```

### 15.3. Peso dos grupos e posição de cada plano

Peso de cada grupo, massa temática, coordenadas X e Y e intervalos de confiança por bootstrap; em seguida, a taxa de cada lema em cada plano, que mostra de onde vem o peso de cada grupo.

```python
import zlib


# Funções do bootstrap por unidades, copiadas sem alteração dos notebooks individuais desta pasta
# (seção 9 desses notebooks), para que os intervalos sigam o mesmo procedimento e a mesma semente.
def semente_documento(sigla):
    """Semente do bootstrap própria de cada plano, derivada de SEMENTE e da sigla.
    Garante (a) os mesmos intervalos no notebook individual e no comparativo e
    (b) sequências aleatórias independentes entre planos diferentes."""
    return np.random.SeedSequence([SEMENTE, zlib.crc32(sigla.encode("utf-8"))])


def taxas_bootstrap(contagens, tamanhos, semente):
    """Bootstrap por unidades: reamostra unidades com reposição e recalcula a taxa por
    1.000 tokens. Respeita a dependência entre palavras de uma mesma unidade.
    contagens: matriz (rótulos x unidades). Devolve (rótulos x reamostras)."""
    rng = np.random.default_rng(semente)
    n_u = len(tamanhos)
    idx = rng.integers(0, n_u, size=(N_BOOTSTRAP, n_u))
    return contagens[:, idx].sum(axis=2) / tamanhos[idx].sum(axis=1) * 1000


def indice_balanco(a, b):
    """Índice de balanço entre dois polos: (a − b)/(a + b), no intervalo [−1, 1]; 0 = equilíbrio."""
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    soma = a + b
    return np.divide(a - b, soma, out=np.zeros_like(soma), where=soma > 0)


def quadrante_de(x, y):
    return ("superior" if y >= 0 else "inferior") + "-" + ("direito" if x >= 0 else "esquerdo")


PLANOS = list(DOCUMENTOS)
ALFA_IC = (1 - NIVEL_CONFIANCA) / 2

# Taxa por 1.000 tokens de cada lema do codebook em cada plano, no documento inteiro (tabelas de
# frequência da seção 11, com as exclusões do pesquisador; lema ausente do plano = 0)
taxas_lemas_quadrante = pd.DataFrame({s: freq[s]["por_1000_tokens"].reindex(list(LEMA_GRUPO)).fillna(0.0)
                                      for s in PLANOS})
grupo_do_lema = taxas_lemas_quadrante.index.map(LEMA_GRUPO)
PESO_GRUPOS = taxas_lemas_quadrante.groupby(grupo_do_lema).sum().reindex(list(GRUPOS_QUADRANTE))

# Bootstrap por unidades: em cada plano, as unidades são reamostradas com reposição e os pesos dos
# grupos, X, Y e a massa são recalculados em cada reamostra
reamostras_quadrante = {}
for sigla in PLANOS:
    t = processados[sigla]["tokens"]
    unidades = sorted(t["unidade"].unique())
    tamanhos = t.groupby("unidade").size().reindex(unidades).to_numpy()
    conteudo = t[t["conteudo"]]
    conteudo = conteudo.assign(grupo=conteudo["lema"].map(LEMA_GRUPO))
    conteudo = conteudo[conteudo["grupo"].isin(GRUPOS_ATIVOS)]
    matriz = pd.crosstab(conteudo["grupo"], conteudo["unidade"]).reindex(
        index=GRUPOS_ATIVOS, columns=unidades, fill_value=0)
    # Conferência: as contagens por unidade reproduzem os pesos calculados com as tabelas da seção 11
    assert np.allclose(matriz.sum(axis=1).to_numpy() / len(t) * 1000,
                       PESO_GRUPOS.loc[GRUPOS_ATIVOS, sigla].to_numpy())
    r = dict(zip(GRUPOS_ATIVOS, taxas_bootstrap(matriz.to_numpy(), tamanhos, semente_documento(sigla))))
    reamostras_quadrante[sigla] = {"X": indice_balanco(r[EIXO_X[0]], r[EIXO_X[1]]),
                                   "Y": indice_balanco(r[EIXO_Y[0]], r[EIXO_Y[1]]),
                                   "massa": sum(r[g] for g in GRUPOS_ATIVOS)}

# (a) Peso dos grupos, massa temática, coordenadas e intervalos de confiança
coordenadas_quadrante = pd.DataFrame(index=pd.Index(PLANOS, name="plano"))
for grupo in GRUPOS_QUADRANTE:
    nome = f"{grupo}_descarte" if grupo == GRUPO_DESCARTADO else grupo
    coordenadas_quadrante[nome] = PESO_GRUPOS.loc[grupo, PLANOS].to_numpy()
coordenadas_quadrante["massa"] = PESO_GRUPOS.loc[GRUPOS_ATIVOS, PLANOS].sum().to_numpy()
coordenadas_quadrante["X"] = indice_balanco(PESO_GRUPOS.loc[EIXO_X[0], PLANOS], PESO_GRUPOS.loc[EIXO_X[1], PLANOS])
coordenadas_quadrante["Y"] = indice_balanco(PESO_GRUPOS.loc[EIXO_Y[0], PLANOS], PESO_GRUPOS.loc[EIXO_Y[1], PLANOS])
for medida in ("X", "Y", "massa"):
    coordenadas_quadrante[f"{medida}_ic_inferior"] = [np.quantile(reamostras_quadrante[s][medida], ALFA_IC)
                                                      for s in PLANOS]
    coordenadas_quadrante[f"{medida}_ic_superior"] = [np.quantile(reamostras_quadrante[s][medida], 1 - ALFA_IC)
                                                      for s in PLANOS]
coordenadas_quadrante["quadrante"] = [quadrante_de(x, y) for x, y in
                                      zip(coordenadas_quadrante["X"], coordenadas_quadrante["Y"])]
colunas_coord = ([c for c in coordenadas_quadrante.columns if c.startswith("G")]
                 + ["massa", "massa_ic_inferior", "massa_ic_superior", "X", "X_ic_inferior", "X_ic_superior",
                    "Y", "Y_ic_inferior", "Y_ic_superior", "quadrante"])
coordenadas_quadrante = coordenadas_quadrante[colunas_coord]
display(Markdown(f"**(a) Peso de cada grupo (por 1.000 tokens), massa temática e coordenadas, com intervalos de "
                 f"confiança de {num(NIVEL_CONFIANCA * 100, 0)}% (bootstrap por unidades, {num(N_BOOTSTRAP, 0)} "
                 f"reamostras)**"))
exibir(coordenadas_quadrante)
salvar_tabela(coordenadas_quadrante, "03_quadrante_coordenadas")


# (b) e (c) Taxa de cada lema do codebook em cada plano, por grupo
def tabela_lemas_do_grupo(grupos):
    tab = taxas_lemas_quadrante[grupo_do_lema.isin(grupos)].copy()
    tab.insert(0, "grupo", tab.index.map(LEMA_GRUPO))
    tab["media_dos_planos"] = tab[PLANOS].mean(axis=1)
    tab["sensivel_ao_pre_processamento"] = [LEMAS_SENSIVEIS.get(l, "") for l in tab.index]
    tab["ordem"] = tab["grupo"].map({g: i for i, g in enumerate(GRUPOS_QUADRANTE)})
    tab = tab.sort_values(["ordem", "media_dos_planos"], ascending=[True, False]).drop(columns="ordem")
    tab.index.name = "lema"
    return tab


lemas_ativos_quadrante = tabela_lemas_do_grupo(GRUPOS_ATIVOS)
display(Markdown("**(b) Taxa por 1.000 tokens de cada lema dos grupos G2 a G5, por plano** (documento inteiro; "
                 "a soma das taxas de um grupo é o peso do grupo na tabela a)"))
exibir(lemas_ativos_quadrante)
salvar_tabela(lemas_ativos_quadrante, "03_quadrante_taxas_por_lema_grupos_ativos")
lemas_descarte_quadrante = tabela_lemas_do_grupo([GRUPO_DESCARTADO])
display(Markdown(f"**(c) Taxa por 1.000 tokens de cada lema do descarte ({GRUPO_DESCARTADO}), por plano** "
                 "(referência para eventuais redirecionamentos de lemas)"))
exibir(lemas_descarte_quadrante)
salvar_tabela(lemas_descarte_quadrante, "03_quadrante_taxas_por_lema_descarte")

maior = lambda col: coordenadas_quadrante[col].idxmax()
menor = lambda col: coordenadas_quadrante[col].idxmin()
print(f"Síntese factual: X vai de {num(coordenadas_quadrante['X'].min())} ({menor('X')}) a "
      f"{num(coordenadas_quadrante['X'].max())} ({maior('X')}); Y, de {num(coordenadas_quadrante['Y'].min())} "
      f"({menor('Y')}) a {num(coordenadas_quadrante['Y'].max())} ({maior('Y')}); a massa temática, de "
      f"{num(coordenadas_quadrante['massa'].min(), 1)} ({menor('massa')}) a {num(coordenadas_quadrante['massa'].max(), 1)} "
      f"({maior('massa')}) por 1.000 tokens. O descarte ({GRUPO_DESCARTADO}) soma de "
      f"{num(PESO_GRUPOS.loc[GRUPO_DESCARTADO].min(), 1)} a {num(PESO_GRUPOS.loc[GRUPO_DESCARTADO].max(), 1)} "
      f"por 1.000 tokens. Quadrantes: "
      + "; ".join(f"{q}: {', '.join(s)}" for q, s in
                  coordenadas_quadrante.groupby("quadrante").groups.items()) + ".")
```

### 15.4. Gráfico de quadrantes

```python
import textwrap

from matplotlib import font_manager
from matplotlib.patches import Circle, FancyBboxPatch
from matplotlib.transforms import Affine2D, Bbox

# ---------------------------------------------------------------------------
# Desenho do gráfico de quadrantes. Reproduz a figura original (pasta notebooks/ Novo/quadrantes,
# módulos plotar.py e plotar_v2.py, arranjo "quadrante"): mesmas dimensões, eixos, linhas de
# referência, caixas dos quatro vocabulários, bolhas (área ∝ massa, halo branco e anel tracejado no
# destaque), rótulos, painel de legenda e nota de rodapé. O código foi trazido para dentro deste
# notebook, que não lê arquivos de fora desta pasta. Textos, cores e rótulos: seção 15.1.
# ---------------------------------------------------------------------------
INK_Q, CINZA_EIXO_Q, CINZA_LINHA_Q, CINZA_TEXTO_Q = "#1a1a1a", "#9a9a9a", "#e3e3e3", "#4d4d4d"
COR_BORDA_PAINEL_Q, COR_SEPARADOR_Q = "#c4c4c4", "#dedede"
# Fontes da figura original, na ordem de preferência; ficam só as instaladas nesta máquina
_fontes_instaladas = {f.name for f in font_manager.fontManager.ttflist}
FONTES_QUADRANTE = [f for f in ("Arial", "Liberation Sans", "Helvetica", "DejaVu Sans")
                    if f in _fontes_instaladas] or ["DejaVu Sans"]
# Estilo próprio da figura (o da seção 9 tem grade e outra fonte); vale só dentro desta célula
ESTILO_QUADRANTE = {
    "font.family": FONTES_QUADRANTE,
    "font.size": 10, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.unicode_minus": True,
    "figure.facecolor": "white", "savefig.facecolor": "white",
    "axes.grid": False, "axes.axisbelow": "line",
}
# Dimensões, em polegadas, as da figura original
LADO_PLOT_Q, MARGEM_LATERAL_Q = 7.4, 1.95          # lado do gráfico; espaço para as caixas B e D
TOPO_TITULO_Q, TOPO_SUBTITULO_Q, TOPO_PLOT_Q = 0.15, 0.53, 1.65   # distâncias a partir do topo
ALTURA_EXTRA_Q = 1.35                               # espaço abaixo do gráfico
PAD_PAINEL_Q, SEPARACAO_SECOES_Q, FOLGA_NOTA_Q = 0.14, 0.24, 0.30
LARGURA_POLO_Q = {"topo": 70, "base": 60, "esquerda": 22, "direita": 24}   # caracteres por linha
MARGEM_BLOCOS_PT_Q = 4                              # folga mínima entre blocos de texto, em pontos

TEXTOS_POLOS_Q = {lado: f"Vocabulário {letra} — {GRUPOS_QUADRANTE[g]['nome']}"
                  for lado, (letra, g) in POLOS_QUADRANTE.items()}
SOMA_ATIVOS_Q = " + ".join(GRUPOS_ATIVOS)
NOTA_QUADRANTE = (
    f"X = ({EIXO_X[0]} − {EIXO_X[1]})/({EIXO_X[0]} + {EIXO_X[1]});  "
    f"Y = ({EIXO_Y[0]} − {EIXO_Y[1]})/({EIXO_Y[0]} + {EIXO_Y[1]});  "
    f"Gk = soma das taxas por 1.000 tokens dos lemas do grupo k, no documento inteiro;  "
    f"área da bolha proporcional a {SOMA_ATIVOS_Q}.\n"
    f"Grupos: os {len(LEMAS_RECORTES)} lemas dos recortes de 75 lemas mais frequentes de cada plano"
    + (f", mais {len(fora_dos_recortes)} acrescentados ao codebook" if fora_dos_recortes else "")
    + f"; {GRUPO_DESCARTADO} (descarte) fica fora dos eixos."
)


def _q_pol(fig):
    """Polegadas (a partir do canto inferior esquerdo da figura) -> tela. Passa por fig.transFigure
    para continuar correta quando a figura é gravada com bbox_inches="tight"."""
    largura, altura = fig.get_size_inches()
    return Affine2D().scale(1 / largura, 1 / altura) + fig.transFigure


def _q_extensao_pol(fig, artista, renderer):
    """Caixa de um artista em polegadas, a partir do canto inferior esquerdo da figura."""
    return artista.get_window_extent(renderer).transformed(_q_pol(fig).inverted())


def _q_escurecer(cor, fator=0.72):
    """Tom mais escuro da cor, usado na borda das bolhas."""
    r, g, b = matplotlib.colors.to_rgb(cor)
    return matplotlib.colors.to_hex((r * fator, g * fator, b * fator))


def _q_raio_pt(massa):
    """Raio da bolha em pontos (área proporcional à massa), usado nos rótulos e nas conferências."""
    return math.sqrt(massa * ESCALA_BOLHA_QUADRANTE / math.pi)


def _q_eixos(ax):
    ax.set_aspect("equal", adjustable="box")
    ax.set_xlim(-1, 1)
    ax.set_ylim(-1, 1)
    ax.grid(False)
    for lado in ("left", "bottom"):                  # eixos cruzando a origem
        ax.spines[lado].set_position("zero")
        ax.spines[lado].set_color(CINZA_EIXO_Q)
        ax.spines[lado].set_linewidth(1.0)
        ax.spines[lado].set_zorder(1)
    for lado in ("top", "right"):
        ax.spines[lado].set_visible(False)
    for v in (-0.5, 0.5):                            # referências discretas em ±0,5
        ax.axhline(v, color=CINZA_LINHA_Q, linewidth=0.6, linestyle=(0, (2, 4)), zorder=0)
        ax.axvline(v, color=CINZA_LINHA_Q, linewidth=0.6, linestyle=(0, (2, 4)), zorder=0)
    ticks = [-1, -0.5, 0.5, 1]
    rot = {-1: "−1", -0.5: "−0,5", 0.5: "0,5", 1: "1"}
    ax.set_xticks(ticks, labels=[rot[t] for t in ticks])
    ax.set_yticks(ticks, labels=[rot[t] for t in ticks])
    ax.tick_params(axis="both", length=3, width=0.8, color=CINZA_EIXO_Q, labelsize=9,
                   labelcolor=CINZA_TEXTO_Q, pad=3)


def _q_polos(ax):
    """Caixas dos quatro vocabulários, nas pontas dos eixos."""
    pos = {"topo": ((0, 1), (0, 12), "center", "bottom"), "base": ((0, -1), (0, -14), "center", "top"),
           "esquerda": ((-1, 0), (-12, 0), "right", "center"), "direita": ((1, 0), (12, 0), "left", "center")}
    caixas = {}
    for lado, texto in TEXTOS_POLOS_Q.items():
        xy, dxy, ha, va = pos[lado]
        caixas[f"polo {lado}"] = ax.annotate(
            textwrap.fill(texto, LARGURA_POLO_Q[lado]), xy=xy, xycoords="data", xytext=dxy,
            textcoords="offset points", ha=ha, va=va, fontsize=9.5, color=INK_Q, linespacing=1.25,
            annotation_clip=False, zorder=5,
            bbox=dict(boxstyle="round,pad=0.45", facecolor="#f4f4f4", edgecolor="#bdbdbd", linewidth=0.6))
    return caixas


def _q_bolhas(ax, coords):
    """Bolhas (as maiores primeiro, para que as menores fiquem por cima) e rótulos. Cada bolha
    recebe um gid, para que a posição desenhada possa ser conferida depois."""
    bolhas, rotulos = [], []
    for sigla, r in coords.sort_values("massa", ascending=False).iterrows():
        foco = sigla == DESTAQUE_QUADRANTE
        massa, cor = float(r["massa"]), CORES_QUADRANTE[sigla]
        rr = _q_raio_pt(massa)
        ax.scatter(r["X"], r["Y"], s=math.pi * (rr + 1.5) ** 2, facecolor="white", edgecolor="none",
                   zorder=2.5)                                          # halo branco
        ax.scatter(r["X"], r["Y"], s=massa * ESCALA_BOLHA_QUADRANTE, facecolor=cor,
                   edgecolor=_q_escurecer(cor), linewidth=0.9, zorder=3, gid=f"bolha:{sigla}")
        if foco:                                                        # anel tracejado do destaque
            ax.scatter(r["X"], r["Y"], s=math.pi * (rr + 3.5) ** 2, facecolor="none", edgecolor=cor,
                       linewidth=1.0, linestyle=(0, (4, 2.5)), zorder=3.1)
            rr += 3.5
        rotulo = DOCUMENTOS_QUADRANTE[sigla]["rotulo"]
        bolhas.append((rotulo, float(r["X"]), float(r["Y"]), rr))
        dx, dy, ha, va, com_guia = ROTULOS_QUADRANTE[sigla]
        dx_total = dx + (rr * np.sign(dx) if dx else 0)
        dy_total = dy + (rr * np.sign(dy) if dy else 0)
        guia = (dict(arrowstyle="-", color="#8c8c8c", linewidth=0.6, shrinkA=2, shrinkB=rr + 1.0)
                if com_guia else None)
        t = ax.annotate(rotulo, xy=(r["X"], r["Y"]), xytext=(dx_total, dy_total), textcoords="offset points",
                        ha=ha, va=va, fontsize=10, fontweight="bold" if foco else "normal", color=INK_Q,
                        zorder=4, arrowprops=guia)
        rotulos.append((rotulo, t))
    return bolhas, rotulos


def _q_legenda_documentos(fig, coords):
    """Legenda de documentos: país ou bloco, título e ano; o destaque em negrito."""
    handles, textos, negrito = [], [], []
    for sigla in [s for s in ORDEM_LEGENDA_QUADRANTE if s in coords.index]:
        doc, cor = DOCUMENTOS_QUADRANTE[sigla], CORES_QUADRANTE[sigla]
        handles.append(Line2D([], [], marker="o", linestyle="none", markersize=9.5, markerfacecolor=cor,
                              markeredgecolor=_q_escurecer(cor), markeredgewidth=0.9))
        textos.append(f"{doc['pais']} — {doc['titulo']}, {doc['ano']}")
        negrito.append(sigla == DESTAQUE_QUADRANTE)
    leg = fig.legend(handles, textos, loc="upper left", bbox_to_anchor=(0, 0), bbox_transform=_q_pol(fig),
                     frameon=False, ncol=1, fontsize=9.5, labelspacing=0.55, handletextpad=0.7,
                     handlelength=1.2, borderpad=0.0, borderaxespad=0.0,
                     title="Documentos analisados (país ou bloco — título, ano)", title_fontsize=9.5,
                     alignment="left")
    leg.set_zorder(5)
    leg.get_title().set_fontweight("bold")
    leg.get_title().set_color(INK_Q)
    for txt, bold in zip(leg.get_texts(), negrito):
        txt.set_color(INK_Q)
        if bold:
            txt.set_fontweight("bold")
    return leg


def _q_legenda_massa(fig, renderer):
    """Legenda de tamanho das bolhas: círculos com o raio real das bolhas do gráfico (num scatter, o
    marcador de área s tem diâmetro √s pontos), lado a lado. Os artistas nascem com o canto superior
    esquerdo em (0, 0) e usam a transformação `deslocamento`, posicionada depois por quem chama."""
    deslocamento = Affine2D()
    t = deslocamento + _q_pol(fig)
    por_pol = _q_pol(fig).inverted()

    def medir(texto):
        return texto.get_window_extent(renderer).transformed(por_pol)

    titulo = fig.text(0, 0, "Massa temática", transform=t, ha="left", va="top", fontsize=9.5,
                      fontweight="bold", color=INK_Q, zorder=5)
    y = -medir(titulo).height - 0.05
    sub = fig.text(0, y, f"{SOMA_ATIVOS_Q}, por 1.000 tokens", transform=t, ha="left", va="top",
                   fontsize=9, color=CINZA_TEXTO_Q, zorder=5)
    y -= medir(sub).height + 0.12
    largura = max(medir(titulo).width, medir(sub).width)
    raios = [math.sqrt(m * ESCALA_BOLHA_QUADRANTE) / 2 / 72 for m in REFERENCIAS_MASSA_QUADRANTE]
    r_max = max(raios)
    x, cy = 0.0, y - r_max
    for r, valor in zip(raios, REFERENCIAS_MASSA_QUADRANTE):
        fig.add_artist(Circle((x + r, cy), r, transform=t, facecolor="white", edgecolor="#6f6f6f",
                              linewidth=0.8, zorder=5))
        rotulo = fig.text(x + 2 * r + 0.07, cy, str(valor), transform=t, ha="left", va="center",
                          fontsize=9.5, color=CINZA_TEXTO_Q, zorder=5)
        x = x + 2 * r + 0.07 + medir(rotulo).width + 0.30
    return {"deslocamento": deslocamento, "largura": max(largura, x - 0.30), "altura": -(cy - r_max)}


def _q_painel(fig, ax, coords, polos, renderer):
    """Painel no canto superior esquerdo: o topo coincide com a borda superior do gráfico e a margem
    esquerda com a da caixa do vocabulário B; documentos em cima, massa temática embaixo."""
    caixa_b = _q_extensao_pol(fig, polos["polo esquerda"], renderer)
    topo = _q_extensao_pol(fig, ax, renderer).y1
    x0 = caixa_b.x0
    leg_doc = _q_legenda_documentos(fig, coords)
    leg_doc.set_bbox_to_anchor((x0 + PAD_PAINEL_Q, topo - PAD_PAINEL_Q), transform=_q_pol(fig))
    fig.canvas.draw()
    ext_doc = _q_extensao_pol(fig, leg_doc, renderer)
    massa = _q_legenda_massa(fig, renderer)
    y_massa = ext_doc.y0 - SEPARACAO_SECOES_Q
    massa["deslocamento"].translate(x0 + PAD_PAINEL_Q, y_massa)
    largura = max(ext_doc.width, massa["largura"]) + 2 * PAD_PAINEL_Q
    base = y_massa - massa["altura"] - PAD_PAINEL_Q
    moldura = FancyBboxPatch((x0, base), largura, topo - base, boxstyle="round,pad=0,rounding_size=0.06",
                             transform=_q_pol(fig), facecolor="white", edgecolor=COR_BORDA_PAINEL_Q,
                             linewidth=0.7, zorder=1.5)
    fig.add_artist(moldura)
    y_sep = ext_doc.y0 - SEPARACAO_SECOES_Q / 2
    fig.add_artist(Line2D([x0 + PAD_PAINEL_Q, x0 + largura - PAD_PAINEL_Q], [y_sep, y_sep],
                          transform=_q_pol(fig), color=COR_SEPARADOR_Q, linewidth=0.7, zorder=4))
    return moldura


def _q_conferir(fig, ax, coords, bolhas, rotulos, moldura, blocos):
    """Conferências automáticas da figura. Devolve a lista de avisos (vazia = tudo certo):
    (1) a posição de cada bolha desenhada é igual à calculada; (2) nenhum rótulo encosta em outro
    rótulo ou numa bolha que não seja a sua; (3) o painel de legenda não encobre bolha, rótulo ou
    valor dos eixos; (4) os blocos de texto (título, subtítulo, caixas, painel, nota) não se sobrepõem."""
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    avisos = []
    desenhadas = {col.get_gid()[len("bolha:"):]: np.asarray(col.get_offsets(), dtype=float)[0]
                  for col in ax.collections if (col.get_gid() or "").startswith("bolha:")}
    for sigla, r in coords.iterrows():
        xy = desenhadas.get(sigla)
        if xy is None or not np.array_equal(xy, np.array([float(r["X"]), float(r["Y"])])):
            avisos.append(f"a bolha de {sigla} não está na posição calculada")
    circulos = {}
    for nome, x, y, r_pt in bolhas:
        px, py = ax.transData.transform((x, y))
        r = r_pt * fig.dpi / 72
        circulos[nome] = Bbox.from_extents(px - r, py - r, px + r, py + r)
    caixas = {nome: t.get_window_extent(renderer) for nome, t in rotulos}
    nomes = list(caixas)
    for i, a in enumerate(nomes):
        for b in nomes[i + 1:]:
            if caixas[a].overlaps(caixas[b]):
                avisos.append(f"o rótulo '{a}' encosta no rótulo '{b}'")
        for b, c in circulos.items():
            if b != a and caixas[a].overlaps(c):
                avisos.append(f"o rótulo '{a}' encosta na bolha '{b}'")
    painel = moldura.get_window_extent(renderer)
    for nome, c in circulos.items():
        if painel.overlaps(c):
            avisos.append(f"o painel de legenda encobre a bolha '{nome}'")
    for nome, caixa in caixas.items():
        if painel.overlaps(caixa):
            avisos.append(f"o painel de legenda encobre o rótulo '{nome}'")
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        if t.get_text() and painel.overlaps(t.get_window_extent(renderer)):
            avisos.append(f"o painel de legenda encobre o valor de eixo '{t.get_text()}'")
    margem = MARGEM_BLOCOS_PT_Q * fig.dpi / 72.0
    ext = {nome: obj.get_window_extent(renderer).padded(margem) for nome, obj in blocos.items()}
    nomes = list(ext)
    for i, a in enumerate(nomes):
        for b in nomes[i + 1:]:
            if ext[a].overlaps(ext[b]):
                avisos.append(f"o bloco '{a}' sobrepõe o bloco '{b}'")
    return avisos


def construir_quadrante(coords):
    """Monta a figura e devolve (fig, ax, avisos)."""
    largura_fig = LADO_PLOT_Q + 2 * MARGEM_LATERAL_Q
    altura_fig = TOPO_PLOT_Q + LADO_PLOT_Q + ALTURA_EXTRA_Q
    fig = plt.figure(figsize=(largura_fig, altura_fig))
    ax = fig.add_axes([MARGEM_LATERAL_Q / largura_fig, (altura_fig - TOPO_PLOT_Q - LADO_PLOT_Q) / altura_fig,
                       LADO_PLOT_Q / largura_fig, LADO_PLOT_Q / altura_fig])
    _q_eixos(ax)
    polos = _q_polos(ax)
    bolhas, rotulos = _q_bolhas(ax, coords)
    blocos = dict(polos)
    blocos["título"] = fig.text(0.5, 1 - TOPO_TITULO_Q / altura_fig, TITULO_QUADRANTE, ha="center", va="top",
                                fontsize=15.5, fontweight="bold", color=INK_Q)
    blocos["subtítulo"] = fig.text(0.5, 1 - TOPO_SUBTITULO_Q / altura_fig, SUBTITULO_QUADRANTE, ha="center",
                                   va="top", fontsize=10.5, color=CINZA_TEXTO_Q, linespacing=1.45)
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    moldura = _q_painel(fig, ax, coords, polos, renderer)
    blocos["painel de legenda"] = moldura
    fig.canvas.draw()
    base = min(_q_extensao_pol(fig, polos["polo base"], renderer).y0, _q_extensao_pol(fig, moldura, renderer).y0)
    blocos["nota de rodapé"] = fig.text(largura_fig / 2, base - FOLGA_NOTA_Q, NOTA_QUADRANTE,
                                        transform=_q_pol(fig), ha="center", va="top", fontsize=9,
                                        color=CINZA_TEXTO_Q, linespacing=1.45)
    return fig, ax, _q_conferir(fig, ax, coords, bolhas, rotulos, moldura, blocos)


with plt.rc_context(ESTILO_QUADRANTE):
    fig_quadrante, ax_quadrante, avisos_quadrante = construir_quadrante(coordenadas_quadrante)
    salvar_figura(fig_quadrante, "03_quadrante_perfis_de_vocabulario")
    plt.show()

for aviso in avisos_quadrante:
    print("AVISO:", aviso)
c = coordenadas_quadrante
print(f"Síntese factual: {len(c)} planos posicionados; X de {num(c['X'].min())} ({c['X'].idxmin()}) a "
      f"{num(c['X'].max())} ({c['X'].idxmax()}); Y de {num(c['Y'].min())} ({c['Y'].idxmin()}) a "
      f"{num(c['Y'].max())} ({c['Y'].idxmax()}); massa temática de {num(c['massa'].min(), 1)} a "
      f"{num(c['massa'].max(), 1)} por 1.000 tokens. "
      + ("Conferências da figura sem avisos: posições desenhadas iguais às calculadas, sem sobreposição de "
         "rótulos, bolhas, painel de legenda e blocos de texto." if not avisos_quadrante
         else f"{len(avisos_quadrante)} aviso(s) nas conferências da figura (acima)."))
```

```python
# =====================================================================
# 15.4.1. CODEBOOK DA VERSÃO 2 — parâmetros editáveis pelo pesquisador
# Por comando do pesquisador (2026-10-06): o codebook da seção 15.1 com 18 lemas redirecionados
# (programme acompanha program, como na seção 15.1) e novos nomes para os vocabulários C (G5) e D (G3).
# As células da versão 2 usam nomes próprios (sufixo _V2 ou _v2) e não alteram nenhuma variável das
# seções 15.1 a 15.6: a figura da versão 1 (acima) e as seções 15.5 e 15.6 (abaixo) continuam iguais
# quando o notebook é executado de novo.
# Para editar: mova ou acrescente lemas nas listas abaixo e execute de novo as três células da versão 2.
# Eixos, textos da figura, cores e destaque: os da seção 15.1.
# =====================================================================
GRUPOS_QUADRANTE_V2 = {
    "G1": {
        "nome": "Descarte",
        "lemas": [
            # Proposta do pesquisador (seção 15.1); saíram construction (para G2) e impact e
            # improvement (para G5)
            "achieve", "action", "actively", "address", "advance", "advanced", "ai", "aim", "approach",
            "area", "artificial", "axis", "base", "brazilian", "bring", "broad", "build", "building", "carry",
            "china", "core", "create", "creation", "develop", "development", "digital", "effort", "enable",
            "encourage", "enhance", "establish", "eu", "europe", "european", "exist", "expand", "explore",
            "facilitate", "field", "focus", "form", "formation", "foster", "fully", "future", "generation",
            "great", "high-quality", "implementation", "improve", "initiative", "integrate", "integration",
            "intelligence", "key", "launch", "layout", "level", "major", "management", "mechanism", "member",
            "method", "need", "new", "opportunity", "optimize", "pbia", "pilot", "planning", "potential",
            "process", "project", "provide", "quality", "refine", "relevant", "require", "resource", "role",
            "sector", "sectoral", "set", "sharing", "significant", "solution", "strengthen", "support",
            "technological", "technology", "tool", "trump", "union", "use", "vigorously", "work",
            # Marcadores de "IA" (seção 15.1); saiu smart (para G2)
            "ai-based", "ai-driven", "ai-native", "intelligent", "intelligentize",
            # Autodesignação, nomes de órgão e nomes dos próprios planos (seção 15.1); saíram country,
            # federal e national (para G4)
            "apply", "commission", "good", "government", "plan", "state", "strategy",
            # Verbos e termos genéricos (seção 15.1); saíram challenge, increase e program (para G5)
            "accelerate", "benefit", "capability", "ensure", "help", "include", "promote", "scenario",
            # Lemas com sentidos de grupos diferentes conforme o plano (seção 15.1)
            "basic", "capacity", "collaboration", "control", "coordination", "environment", "space",
            # Grafia britânica de optimize (programme acompanhou program para G5)
            "optimise",
        ],
    },
    "G2": {
        "nome": "Infraestrutura e tecnologias de IA",
        "lemas": [
            "access", "agent", "autonomous", "center", "centre", "cloud", "compute", "computing", "data",
            "device", "energy", "eurohpc", "facility", "factory", "gigafactory", "grid", "group",
            "human-computer", "information", "infrastructure", "knowledge", "learning", "model", "network",
            "open", "open-source", "perception", "platform", "power", "robot", "robotics", "semiconductor",
            "system", "terminal", "unmanned",
            # Versão 2: smart e construction (de G1); application, innovation e industry (de G3)
            "smart", "construction", "application", "innovation", "industry",
        ],
    },
    "G3": {
        "nome": "Economia, competitividade e inovação",
        "lemas": [
            "adopt", "agricultural", "breakthrough", "chain", "company", "competitiveness", "consumption",
            "deployment", "economic", "economy", "ecosystem", "efficiency", "enterprise", "frontier", "hub",
            "incubate", "investment", "market", "product", "production", "startup", "theory", "uptake",
            "industrial", "manufacturing",
            # Versão 2: saíram application, innovation e industry (para G2), adoption (para G4) e research,
            # science, scientific e talent (para G5)
        ],
    },
    "G4": {
        "nome": "Governança, regulação e segurança",
        "lemas": [
            "act", "adversary", "ally", "assessment", "consultation", "cooperation", "critical", "defence",
            "evaluation", "export", "global", "governance", "international", "lead", "policy", "regulation",
            "risk", "security", "stakeholder", "standard", "strategic", "public", "defense",
            # Versão 2: country, federal e national (de G1); adoption (de G3)
            "country", "federal", "national", "adoption",
        ],
    },
    "G5": {
        "nome": "Sociedade, inclusão, capital humano e ciência",
        "lemas": [
            "content", "cultural", "education", "health", "healthcare", "job", "occupation", "population",
            "professional", "qualification", "reduction", "responsible", "service", "skill", "social",
            "society", "sustainable", "training", "worker", "workforce",
            # Versão 2: challenge, impact, improvement, increase e program (de G1); programme, a grafia
            # britânica de program, acompanha program, como na seção 15.1 (para deixá-lo no descarte,
            # devolva-o à lista de G1); research, science, scientific e talent (de G3)
            "challenge", "impact", "improvement", "increase", "program", "programme",
            "research", "science", "scientific", "talent",
        ],
    },
}

# Rótulo de cada bolha na versão 2: mesmo formato de ROTULOS_QUADRANTE (seção 15.1). Mudaram em relação à
# versão 1 os rótulos das bolhas que ficaram juntas (AI Continent, AI+) ou sobre o eixo vertical (Apply AI,
# PBIA). A seção 15.4.3 avisa quando um rótulo encosta em outro rótulo, numa bolha ou no painel de legenda.
ROTULOS_QUADRANTE_V2 = {
    "AI_Continent_2025": (-6, 12, "right", "bottom", True),  # acima, à esquerda, com linha-guia
    "Apply_AI_2025": (-4, -6, "right", "top", False),        # abaixo, à esquerda
    "China_2025": (6, -6, "left", "top", True),              # abaixo, à direita, com linha-guia
    "New_Gen_2017": (6, 0, "left", "center", False),         # à direita
    "EUA_2025": (6, 0, "left", "center", False),             # à direita
    "PBIA_2025": (0, -8, "center", "top", False),            # abaixo
}
# Canto do painel de legenda: "superior esquerdo" (o da versão 1), "superior direito", "inferior esquerdo"
# ou "inferior direito". Na versão 2, o canto superior esquerdo encobriria a bolha dos EUA (X = −0,49;
# Y = 0,25); no inferior direito não há bolha nem rótulo. A seção 15.4.3 avisa se o painel encobrir algo.
CANTO_PAINEL_QUADRANTE_V2 = "inferior direito"

# ---------------------------------------------------------------------------
# Conferências do codebook da versão 2 (as da seção 15.2) e mudanças em relação à versão 1
# ---------------------------------------------------------------------------
if list(GRUPOS_QUADRANTE_V2) != list(GRUPOS_QUADRANTE):
    raise ValueError(f"A versão 2 usa os eixos da seção 15.1 e deve ter os mesmos grupos ({list(GRUPOS_QUADRANTE)}).")
LEMA_GRUPO_V2, repetidos_v2 = {}, []
for grupo, cfg in GRUPOS_QUADRANTE_V2.items():
    for lema in cfg["lemas"]:
        chave = lema.strip().lower()
        if chave in LEMA_GRUPO_V2:
            repetidos_v2.append(f"'{chave}' em {LEMA_GRUPO_V2[chave]} e em {grupo}")
        else:
            LEMA_GRUPO_V2[chave] = grupo
if repetidos_v2:
    raise ValueError("Cada lema deve estar em um único grupo. Repetições: " + "; ".join(repetidos_v2))
sem_grupo_v2 = [lema for lema in LEMAS_RECORTES if lema not in LEMA_GRUPO_V2]
fora_dos_recortes_v2 = sorted(lema for lema in LEMA_GRUPO_V2 if lema not in set(LEMAS_RECORTES))
sem_ocorrencia_v2 = sorted(lema for lema in LEMA_GRUPO_V2 if not any(lema in freq[s].index for s in DOCUMENTOS))
# Lemas em grupo diferente do da versão 1 ("—" = lema ausente de uma das versões)
mudancas_v2 = sorted(lema for lema in set(LEMA_GRUPO) | set(LEMA_GRUPO_V2)
                     if LEMA_GRUPO.get(lema) != LEMA_GRUPO_V2.get(lema))
# Hash do codebook (mesma regra da seção 15.6): identifica a versão dos grupos usada em cada execução
codebook_sha256_v2 = hashlib.sha256(json.dumps(
    {g: sorted(l.strip().lower() for l in c["lemas"]) for g, c in GRUPOS_QUADRANTE_V2.items()},
    ensure_ascii=False, sort_keys=True).encode("utf-8")).hexdigest()

# (a) Lemas de cada grupo
linhas_md = []
for grupo, cfg in GRUPOS_QUADRANTE_V2.items():
    lemas = sorted({l.strip().lower() for l in cfg["lemas"]})
    marcados = [(f"**{l}**" if l in mudancas_v2 else l) + ("†" if l in fora_dos_recortes_v2 else "") for l in lemas]
    linhas_md.append(f"- **{grupo} — {cfg['nome']}** · {descrever_polo(grupo)} · {len(lemas)} lemas: "
                     + ", ".join(marcados))
display(Markdown("**(a) Lemas de cada grupo na versão 2** (em negrito, os lemas que mudaram de grupo em relação "
                 "à versão 1; † = lema de fora dos recortes de 75, incluído no codebook)\n\n" + "\n".join(linhas_md)))

# (b) Grupos e nomes nas duas versões
composicao_quadrante_v2 = pd.DataFrame([{
    "grupo": grupo,
    "posicao_na_figura": descrever_polo(grupo),
    "nome_versao_1": GRUPOS_QUADRANTE[grupo]["nome"],
    "nome_versao_2": GRUPOS_QUADRANTE_V2[grupo]["nome"],
    "lemas_versao_1": len(GRUPOS_QUADRANTE[grupo]["lemas"]),
    "lemas_versao_2": len(GRUPOS_QUADRANTE_V2[grupo]["lemas"]),
    "entraram": sum(LEMA_GRUPO_V2.get(l) == grupo for l in mudancas_v2),
    "sairam": sum(LEMA_GRUPO.get(l) == grupo for l in mudancas_v2),
} for grupo in GRUPOS_QUADRANTE_V2]).set_index("grupo")
display(Markdown("**(b) Grupos, nomes e número de lemas nas versões 1 e 2**"))
exibir(composicao_quadrante_v2, 0)

# (c) Lemas que mudaram de grupo, com a taxa por 1.000 tokens em cada plano (documento inteiro)
mudancas_quadrante_v2 = pd.DataFrame({s: freq[s]["por_1000_tokens"].reindex(mudancas_v2).fillna(0.0)
                                      for s in DOCUMENTOS})
mudancas_quadrante_v2.insert(0, "grupo_versao_1", [LEMA_GRUPO.get(l, "—") for l in mudancas_v2])
mudancas_quadrante_v2.insert(1, "grupo_versao_2", [LEMA_GRUPO_V2.get(l, "—") for l in mudancas_v2])
ordem_grupos = {g: i for i, g in enumerate(list(GRUPOS_QUADRANTE_V2) + ["—"])}
mudancas_quadrante_v2 = mudancas_quadrante_v2.sort_values(
    ["grupo_versao_2", "grupo_versao_1"], key=lambda col: col.map(ordem_grupos), kind="stable")
mudancas_quadrante_v2.index.name = "lema"
display(Markdown(f"**(c) Os {len(mudancas_v2)} lemas que mudaram de grupo, com a taxa por 1.000 tokens em cada "
                 f"plano** (documento inteiro; é o quanto cada lema soma ao peso do seu grupo)"))
exibir(mudancas_quadrante_v2)
salvar_tabela(pd.DataFrame([{"grupo": g, "nome": c["nome"], "lema": l} for g, c in GRUPOS_QUADRANTE_V2.items()
                            for l in c["lemas"]]), "03_quadrante_v2_codebook", index=False)
salvar_tabela(mudancas_quadrante_v2, "03_quadrante_v2_mudancas_de_grupo")

if sem_grupo_v2:
    print(f"ATENÇÃO: {len(sem_grupo_v2)} lema(s) dos recortes de 75 sem grupo no codebook da versão 2 (ficam fora "
          f"de todas as somas): {', '.join(sem_grupo_v2)}")
if sem_ocorrencia_v2:
    print(f"ATENÇÃO: lema(s) do codebook da versão 2 sem nenhuma ocorrência nos seis planos (confira a grafia): "
          f"{', '.join(sem_ocorrencia_v2)}")
print(f"Síntese factual: o codebook da versão 2 tem {len(LEMA_GRUPO_V2)} lemas, "
      + ", ".join(f"{len(c['lemas'])} em {g}" for g, c in GRUPOS_QUADRANTE_V2.items())
      + f"; {len(mudancas_v2)} mudaram de grupo em relação à versão 1 ("
      + "; ".join(f"{d}: " + ", ".join(l for l in mudancas_v2 if LEMA_GRUPO_V2.get(l, "—") == d)
                  for d in ordem_grupos if any(LEMA_GRUPO_V2.get(l, "—") == d for l in mudancas_v2))
      + "). Os recortes de 75 dos seis planos somam "
      + f"{len(LEMAS_RECORTES)} lemas distintos, "
      + ("todos classificados" if not sem_grupo_v2 else f"{len(sem_grupo_v2)} deles sem grupo")
      + f". Nomes: " + "; ".join(f"{g} ({descrever_polo(g).split(' (')[0]}): {GRUPOS_QUADRANTE_V2[g]['nome']}"
                                 for g in GRUPOS_QUADRANTE_V2 if g != GRUPO_DESCARTADO)
      + f". SHA-256 do codebook da versão 2: {codebook_sha256_v2}.")
```

#### 15.4.2. Peso dos grupos e posição de cada plano (versão 2)

Peso de cada grupo, massa temática, coordenadas e intervalos de confiança com o codebook da versão 2; em seguida, as coordenadas das versões 1 e 2 lado a lado e as diferenças entre os planos, com o critério da seção 15.5a.

```python
# Mesmo cálculo da seção 15.3, com o codebook da versão 2: taxas no documento inteiro, pesos dos grupos,
# massa temática, X, Y e intervalos de confiança por bootstrap por unidades. As funções (semente_documento,
# taxas_bootstrap, indice_balanco, quadrante_de) são as da seção 15.3; com a mesma semente por plano, as
# reamostras sorteiam as mesmas unidades da versão 1.
GRUPOS_ATIVOS_V2 = [g for g in GRUPOS_QUADRANTE_V2 if g != GRUPO_DESCARTADO]
taxas_lemas_quadrante_v2 = pd.DataFrame({s: freq[s]["por_1000_tokens"].reindex(list(LEMA_GRUPO_V2)).fillna(0.0)
                                         for s in PLANOS})
grupo_do_lema_v2 = taxas_lemas_quadrante_v2.index.map(LEMA_GRUPO_V2)
PESO_GRUPOS_V2 = taxas_lemas_quadrante_v2.groupby(grupo_do_lema_v2).sum().reindex(list(GRUPOS_QUADRANTE_V2))

reamostras_quadrante_v2 = {}
for sigla in PLANOS:
    t = processados[sigla]["tokens"]
    unidades = sorted(t["unidade"].unique())
    tamanhos = t.groupby("unidade").size().reindex(unidades).to_numpy()
    conteudo = t[t["conteudo"]]
    conteudo = conteudo.assign(grupo=conteudo["lema"].map(LEMA_GRUPO_V2))
    conteudo = conteudo[conteudo["grupo"].isin(GRUPOS_ATIVOS_V2)]
    matriz = pd.crosstab(conteudo["grupo"], conteudo["unidade"]).reindex(
        index=GRUPOS_ATIVOS_V2, columns=unidades, fill_value=0)
    # Conferência: as contagens por unidade reproduzem os pesos calculados com as tabelas da seção 11
    assert np.allclose(matriz.sum(axis=1).to_numpy() / len(t) * 1000,
                       PESO_GRUPOS_V2.loc[GRUPOS_ATIVOS_V2, sigla].to_numpy())
    r = dict(zip(GRUPOS_ATIVOS_V2, taxas_bootstrap(matriz.to_numpy(), tamanhos, semente_documento(sigla))))
    reamostras_quadrante_v2[sigla] = {"X": indice_balanco(r[EIXO_X[0]], r[EIXO_X[1]]),
                                      "Y": indice_balanco(r[EIXO_Y[0]], r[EIXO_Y[1]]),
                                      "massa": sum(r[g] for g in GRUPOS_ATIVOS_V2)}

# (a) Peso dos grupos, massa temática, coordenadas e intervalos de confiança
coordenadas_quadrante_v2 = pd.DataFrame(index=pd.Index(PLANOS, name="plano"))
for grupo in GRUPOS_QUADRANTE_V2:
    nome = f"{grupo}_descarte" if grupo == GRUPO_DESCARTADO else grupo
    coordenadas_quadrante_v2[nome] = PESO_GRUPOS_V2.loc[grupo, PLANOS].to_numpy()
coordenadas_quadrante_v2["massa"] = PESO_GRUPOS_V2.loc[GRUPOS_ATIVOS_V2, PLANOS].sum().to_numpy()
coordenadas_quadrante_v2["X"] = indice_balanco(PESO_GRUPOS_V2.loc[EIXO_X[0], PLANOS],
                                               PESO_GRUPOS_V2.loc[EIXO_X[1], PLANOS])
coordenadas_quadrante_v2["Y"] = indice_balanco(PESO_GRUPOS_V2.loc[EIXO_Y[0], PLANOS],
                                               PESO_GRUPOS_V2.loc[EIXO_Y[1], PLANOS])
for medida in ("X", "Y", "massa"):
    coordenadas_quadrante_v2[f"{medida}_ic_inferior"] = [np.quantile(reamostras_quadrante_v2[s][medida], ALFA_IC)
                                                         for s in PLANOS]
    coordenadas_quadrante_v2[f"{medida}_ic_superior"] = [np.quantile(reamostras_quadrante_v2[s][medida], 1 - ALFA_IC)
                                                         for s in PLANOS]
coordenadas_quadrante_v2["quadrante"] = [quadrante_de(x, y) for x, y in
                                         zip(coordenadas_quadrante_v2["X"], coordenadas_quadrante_v2["Y"])]
coordenadas_quadrante_v2 = coordenadas_quadrante_v2[colunas_coord]
display(Markdown(f"**(a) Versão 2: peso de cada grupo (por 1.000 tokens), massa temática e coordenadas, com "
                 f"intervalos de confiança de {num(NIVEL_CONFIANCA * 100, 0)}% (bootstrap por unidades, "
                 f"{num(N_BOOTSTRAP, 0)} reamostras)**"))
exibir(coordenadas_quadrante_v2)
salvar_tabela(coordenadas_quadrante_v2, "03_quadrante_v2_coordenadas")

# (b) Versão 1 (seção 15.3) e versão 2, lado a lado
comparacao_versoes_quadrante = pd.concat(
    {"versao_1": coordenadas_quadrante[["X", "Y", "massa", "quadrante"]],
     "versao_2": coordenadas_quadrante_v2[["X", "Y", "massa", "quadrante"]]}, axis=1)
for medida in ("X", "Y", "massa"):
    comparacao_versoes_quadrante[("versao_2_menos_versao_1", medida)] = (
        coordenadas_quadrante_v2[medida] - coordenadas_quadrante[medida])
comparacao_versoes_quadrante[("versao_2_menos_versao_1", "deslocamento")] = np.hypot(
    comparacao_versoes_quadrante[("versao_2_menos_versao_1", "X")],
    comparacao_versoes_quadrante[("versao_2_menos_versao_1", "Y")])
display(Markdown("**(b) Coordenadas e massa nas versões 1 e 2** (deslocamento = distância entre as duas posições "
                 "da bolha, em unidades dos eixos)"))
exibir(comparacao_versoes_quadrante)
salvar_tabela(comparacao_versoes_quadrante, "03_quadrante_v2_comparacao_com_versao_1")

# (c) Diferença entre dois planos em X e em Y, com intervalo de confiança por bootstrap (mesmo critério da
# seção 15.5a: uma diferença é afirmada apenas quando o intervalo não inclui zero; skill 03, seção 3.5)
linhas = []
for i, a in enumerate(PLANOS):
    for b in PLANOS[i + 1:]:
        linha = {"plano_a": a, "plano_b": b}
        for eixo in ("X", "Y"):
            dif = reamostras_quadrante_v2[a][eixo] - reamostras_quadrante_v2[b][eixo]
            inf, sup = np.quantile(dif, [ALFA_IC, 1 - ALFA_IC])
            linha[f"d{eixo}"] = coordenadas_quadrante_v2.loc[a, eixo] - coordenadas_quadrante_v2.loc[b, eixo]
            linha[f"d{eixo}_ic_inferior"], linha[f"d{eixo}_ic_superior"] = inf, sup
            linha[f"d{eixo}_ic_exclui_zero"] = "sim" if inf > 0 or sup < 0 else "não"
        linhas.append(linha)
diferencas_quadrante_v2 = pd.DataFrame(linhas).set_index(["plano_a", "plano_b"])
display(Markdown(f"**(c) Versão 2: diferenças entre os planos (plano a − plano b), com intervalo de confiança de "
                 f"{num(NIVEL_CONFIANCA * 100, 0)}%**"))
exibir(diferencas_quadrante_v2)
salvar_tabela(diferencas_quadrante_v2, "03_quadrante_v2_diferencas_entre_planos")

c2, d2 = coordenadas_quadrante_v2, diferencas_quadrante_v2
deslocamento_v2 = comparacao_versoes_quadrante[("versao_2_menos_versao_1", "deslocamento")]
mudou = [s for s in PLANOS if coordenadas_quadrante.loc[s, "quadrante"] != c2.loc[s, "quadrante"]]
print(f"Síntese factual: na versão 2, X vai de {num(c2['X'].min())} ({c2['X'].idxmin()}) a {num(c2['X'].max())} "
      f"({c2['X'].idxmax()}); Y, de {num(c2['Y'].min())} ({c2['Y'].idxmin()}) a {num(c2['Y'].max())} "
      f"({c2['Y'].idxmax()}); a massa temática, de {num(c2['massa'].min(), 1)} ({c2['massa'].idxmin()}) a "
      f"{num(c2['massa'].max(), 1)} ({c2['massa'].idxmax()}) por 1.000 tokens. Quadrantes: "
      + "; ".join(f"{q}: {', '.join(s)}" for q, s in c2.groupby("quadrante").groups.items())
      + f". Em relação à versão 1, as bolhas se deslocam de {num(deslocamento_v2.min())} ({deslocamento_v2.idxmin()}) a "
      f"{num(deslocamento_v2.max())} ({deslocamento_v2.idxmax()}); "
      + (f"mudam de quadrante: {', '.join(mudou)}" if mudou else "nenhum plano muda de quadrante")
      + f". Das {len(d2)} comparações entre pares de planos, o intervalo da diferença exclui zero em "
      f"{(d2['dX_ic_exclui_zero'] == 'sim').sum()} no eixo X e em {(d2['dY_ic_exclui_zero'] == 'sim').sum()} "
      f"no eixo Y.")
```

#### 15.4.3. Gráfico de quadrantes (versão 2)

```python
from contextlib import contextmanager

# ---------------------------------------------------------------------------
# Desenho da versão 2 com as funções da seção 15.4, sem alteração. Essas funções leem os textos das
# caixas dos vocabulários, a largura dessas caixas, a nota de rodapé, a posição dos rótulos e o painel de
# legenda em variáveis globais da versão 1 (TEXTOS_POLOS_Q, LARGURA_POLO_Q, NOTA_QUADRANTE,
# ROTULOS_QUADRANTE e _q_painel), que recebem os valores da versão 2 só durante o desenho e voltam aos da
# versão 1 logo depois. Rótulos e canto do painel: seção 15.4.1; título, subtítulo, cores, destaque e
# escala: seção 15.1.
# ---------------------------------------------------------------------------
TEXTOS_POLOS_Q_V2 = {lado: f"Vocabulário {letra} — {GRUPOS_QUADRANTE_V2[g]['nome']}"
                     for lado, (letra, g) in POLOS_QUADRANTE.items()}
# Caixa C com até 70 caracteres por linha, como a caixa A: com os 60 da versão 1, o nome novo, mais longo,
# deixaria "ciência" sozinha na segunda linha
LARGURA_POLO_Q_V2 = {**LARGURA_POLO_Q, "base": 70}
NOTA_QUADRANTE_V2 = (
    f"X = ({EIXO_X[0]} − {EIXO_X[1]})/({EIXO_X[0]} + {EIXO_X[1]});  "
    f"Y = ({EIXO_Y[0]} − {EIXO_Y[1]})/({EIXO_Y[0]} + {EIXO_Y[1]});  "
    f"Gk = soma das taxas por 1.000 tokens dos lemas do grupo k, no documento inteiro;  "
    f"área da bolha proporcional a {SOMA_ATIVOS_Q}.\n"
    f"Grupos: os {len(LEMAS_RECORTES)} lemas dos recortes de 75 lemas mais frequentes de cada plano"
    + (f", mais {len(fora_dos_recortes_v2)} acrescentados ao codebook" if fora_dos_recortes_v2 else "")
    + f"; {GRUPO_DESCARTADO} (descarte) fica fora dos eixos."
)


def _q_painel_no_canto(fig, ax, coords, polos, renderer, canto):
    """Painel de legenda da seção 15.4 (_q_painel), com o mesmo conteúdo e o mesmo desenho, num canto
    qualquer do gráfico. Na vertical, o topo (canto superior) ou a base (canto inferior) do painel
    coincide com a borda do gráfico; na horizontal, a margem esquerda coincide com a da caixa do
    vocabulário B (canto esquerdo), ou a margem direita com a da caixa do vocabulário D (canto direito)."""
    vertical, horizontal = canto.split()
    # Tamanho do painel, medido antes de posicioná-lo
    leg_doc = _q_legenda_documentos(fig, coords)
    fig.canvas.draw()
    ext_doc = _q_extensao_pol(fig, leg_doc, renderer)
    massa = _q_legenda_massa(fig, renderer)
    largura = max(ext_doc.width, massa["largura"]) + 2 * PAD_PAINEL_Q
    altura = 2 * PAD_PAINEL_Q + ext_doc.height + SEPARACAO_SECOES_Q + massa["altura"]
    # Posição
    ext_ax = _q_extensao_pol(fig, ax, renderer)
    if horizontal == "esquerdo":
        x0 = _q_extensao_pol(fig, polos["polo esquerda"], renderer).x0
    else:
        x0 = _q_extensao_pol(fig, polos["polo direita"], renderer).x1 - largura
    topo = ext_ax.y1 if vertical == "superior" else ext_ax.y0 + altura
    leg_doc.set_bbox_to_anchor((x0 + PAD_PAINEL_Q, topo - PAD_PAINEL_Q), transform=_q_pol(fig))
    fig.canvas.draw()
    ext_doc = _q_extensao_pol(fig, leg_doc, renderer)
    y_massa = ext_doc.y0 - SEPARACAO_SECOES_Q
    massa["deslocamento"].translate(x0 + PAD_PAINEL_Q, y_massa)
    base = y_massa - massa["altura"] - PAD_PAINEL_Q
    moldura = FancyBboxPatch((x0, base), largura, topo - base, boxstyle="round,pad=0,rounding_size=0.06",
                             transform=_q_pol(fig), facecolor="white", edgecolor=COR_BORDA_PAINEL_Q,
                             linewidth=0.7, zorder=1.5)
    fig.add_artist(moldura)
    y_sep = ext_doc.y0 - SEPARACAO_SECOES_Q / 2
    fig.add_artist(Line2D([x0 + PAD_PAINEL_Q, x0 + largura - PAD_PAINEL_Q], [y_sep, y_sep],
                          transform=_q_pol(fig), color=COR_SEPARADOR_Q, linewidth=0.7, zorder=4))
    return moldura


def _q_painel_v2(fig, ax, coords, polos, renderer):
    return _q_painel_no_canto(fig, ax, coords, polos, renderer, CANTO_PAINEL_QUADRANTE_V2)


@contextmanager
def _q_globais_trocadas(**valores):
    """Dá às variáveis globais indicadas os valores pedidos e restaura os anteriores ao sair, mesmo se
    houver erro no desenho."""
    anteriores = {nome: globals()[nome] for nome in valores}
    globals().update(valores)
    try:
        yield
    finally:
        globals().update(anteriores)


with plt.rc_context(ESTILO_QUADRANTE), _q_globais_trocadas(TEXTOS_POLOS_Q=TEXTOS_POLOS_Q_V2,
                                                           LARGURA_POLO_Q=LARGURA_POLO_Q_V2,
                                                           NOTA_QUADRANTE=NOTA_QUADRANTE_V2,
                                                           ROTULOS_QUADRANTE=ROTULOS_QUADRANTE_V2,
                                                           _q_painel=_q_painel_v2):
    fig_quadrante_v2, ax_quadrante_v2, avisos_quadrante_v2 = construir_quadrante(coordenadas_quadrante_v2)
    salvar_figura(fig_quadrante_v2, "03_quadrante_v2_perfis_de_vocabulario")
    plt.show()

for aviso in avisos_quadrante_v2:
    print("AVISO:", aviso)
c2 = coordenadas_quadrante_v2
print(f"Síntese factual: versão 2 com {len(c2)} planos posicionados; X de {num(c2['X'].min())} ({c2['X'].idxmin()}) "
      f"a {num(c2['X'].max())} ({c2['X'].idxmax()}); Y de {num(c2['Y'].min())} ({c2['Y'].idxmin()}) a "
      f"{num(c2['Y'].max())} ({c2['Y'].idxmax()}); massa temática de {num(c2['massa'].min(), 1)} a "
      f"{num(c2['massa'].max(), 1)} por 1.000 tokens. Caixas dos vocabulários: "
      + "; ".join(f"{TEXTOS_POLOS_Q_V2[lado]} ({lado})" for lado in POLOS_QUADRANTE) + ". "
      + ("Conferências da figura sem avisos: posições desenhadas iguais às calculadas, sem sobreposição de "
         "rótulos, bolhas, painel de legenda e blocos de texto." if not avisos_quadrante_v2
         else f"{len(avisos_quadrante_v2)} aviso(s) nas conferências da figura (acima)."))
```

**Método e escolhas (versão 2).** Os procedimentos são os da versão 1, descritos no texto *Método e escolhas* da seção 15.5 (itens 2 a 4 e 6): o peso de um grupo é a soma das taxas por 1.000 tokens dos seus lemas no documento inteiro; X = (G3 − G4)/(G3 + G4) e Y = (G2 − G5)/(G2 + G5); a área da bolha é proporcional a G2 + G3 + G4 + G5; os intervalos de confiança de 95% vêm do *bootstrap* por unidades (2.000 reamostras, com a semente de cada plano, de modo que as duas versões usam as mesmas reamostras); e uma diferença entre dois planos é afirmada apenas quando o intervalo da diferença não inclui zero (skill 03, seção 3.5). Só o codebook mudou: o descarte passou de 135 para 124 lemas, G2 de 35 para 40, G3 de 33 para 25, G4 de 23 para 27 e G5 de 20 para 30. As alternativas de robustez e a influência de cada lema (tabelas 15.5b e 15.5c) não foram recalculadas para a versão 2.

**Interpretação.** *Observação:*

- **Deslocamento em relação à versão 1 (tabela 15.4.2b).** Todas as bolhas vão para a esquerda: X diminui de 0,29 (Apply AI) a 0,52 (PBIA), porque G3 perdeu oito lemas, que eram de 37% (Apply AI) a 53% (*AI+*, New Generation e EUA) do seu peso, e G4 ganhou quatro, que somam de 3,04 (Apply AI) a 11,47 (PBIA) por 1.000 tokens. Y diminui de 0,16 a 0,24 em cinco planos e 0,01 no *AI+*: os dez lemas que entraram em G5 aumentaram o grupo de 64% (*AI+*) a 108% (New Generation e EUA), e os cinco que entraram em G2, de 18% (EUA) a 60% (*AI+*). A massa temática aumentou em todos os planos (de 5,71 por 1.000 tokens no AI Continent a 26,54 no PBIA), porque onze lemas saíram do descarte.
- **Eixo X (G3 × G4).** Os EUA (−0,49 [IC 95%: −0,59; −0,39]) e o PBIA (−0,37 [−0,45; −0,28]) têm X negativo, com intervalos que não incluem zero: nos dois, G4 pesa mais que G3 (39,28 contra 13,42 por 1.000 tokens nos EUA; 32,94 contra 15,31 no PBIA). Nos quatro planos da UE e da China, X fica entre −0,08 e 0,02 (AI Continent −0,08 [−0,22; 0,06]; Apply AI −0,06 [−0,21; 0,09]; *AI+* −0,05 [−0,34; 0,27]; New Generation 0,02 [−0,13; 0,16]), com intervalos que incluem zero: G3 e G4 se equilibram. O X dos EUA e o do PBIA são menores que o de cada um desses quatro planos, com intervalo da diferença que exclui zero nas oito comparações (a menos folgada, entre o *AI+* e o PBIA: 0,32 [0,004; 0,64]). A diferença entre os EUA e o PBIA (−0,13 [−0,26; 0,01]) não atende ao critério, nem as diferenças em X entre os quatro planos da UE e da China.
- **Eixo Y (G2 × G5).** Cinco planos têm Y positivo, com intervalos que não incluem zero: New Generation 0,39 [0,28; 0,49], *AI+* 0,34 [0,16; 0,51], AI Continent 0,33 [0,19; 0,46], EUA 0,25 [0,10; 0,40] e Apply AI 0,20 [0,05; 0,34]. No PBIA, Y = −0,17 [−0,24; −0,10]: G5 (48,20 por 1.000 tokens) supera G2 (34,23), e o intervalo, que na versão 1 incluía zero, fica inteiro abaixo dele. O Y do PBIA é menor que o de cada um dos outros cinco planos, com intervalo da diferença que exclui zero em todos os casos (de 0,37 [0,20; 0,53], em relação ao Apply AI, a 0,56 [0,43; 0,69], em relação ao New Generation). Entre os outros cinco, só uma diferença em Y atende ao critério: o New Generation fica acima do Apply AI (0,20 [0,02; 0,37]).
- **Quadrantes.** Os EUA ficam no quadrante superior esquerdo, e o PBIA, no inferior esquerdo, ambos com intervalos de X e de Y que não incluem zero. Pelas estimativas pontuais, o AI Continent, o Apply AI e o *AI+* ficam no superior esquerdo, e o New Generation, no superior direito; como os intervalos de X desses quatro planos incluem zero, eles estão sobre a fronteira vertical, e o lado em que caem não é determinado pelos dados. As bolhas do AI Continent e do *AI+* quase coincidem (distância de 0,04 entre os centros) e se sobrepõem na figura.
- **Massa temática.** O *AI+* (174,32 [159,41; 190,54]) e o New Generation (171,12 [161,03; 180,96]) continuam com a maior massa; os outros quatro ficam entre 114,64 (Apply AI) e 130,68 (PBIA), com intervalos que não se sobrepõem aos dos dois planos chineses.
- **De onde vêm os pesos (tabelas 15.4.1c e 15.4.2a).** Os lemas redirecionados são parte grande dos grupos que os receberam. Os dez que entraram em G5 somam de 39% (*AI+*) a 52% (New Generation e EUA) do grupo: no PBIA, *increase* (4,39), *impact* (4,03), *challenge* (3,17) e *program* (2,93) são os quatro lemas de maior peso em G5; nos EUA, *program* (3,29) e *research* (2,81) são os dois maiores; no New Generation, *research* (7,85) responde sozinho por 22% de G5. Em G2, os cinco lemas que entraram são 37% do grupo no *AI+*, em que *application* (11,13) é o lema de maior peso, e 26% no New Generation. *Inferência:* com o codebook da versão 2, o eixo X separa dois conjuntos: os EUA e o PBIA, em que o vocabulário de governança, regulação e segurança predomina sobre o de economia, competitividade e inovação, e os quatro planos da UE e da China, em que os dois se equilibram. O eixo Y separa o PBIA, único plano em que o vocabulário de sociedade, inclusão, capital humano e ciência supera o de infraestrutura e tecnologias de IA, dos outros cinco. Em relação à versão 1, o deslocamento para a esquerda é geral e a ordem dos planos em X pouco muda (os EUA continuam com o menor X, e o New Generation, com o maior), mas o PBIA, que na versão 1 ficava mais perto dos planos da UE e da China do que dos EUA, passa a ficar mais perto dos EUA. Isso ocorre sobretudo porque os quatro lemas que entraram em G4 pesam mais no PBIA (11,47 por 1.000 tokens, sobretudo *national* e *country*) do que nos planos da UE e da China (de 3,04 a 4,50). O PBIA continua com o menor Y. As posições medem pesos lexicais, e não prioridades políticas.

**Fragilidades e ajustes possíveis (versão 2).**

- **Termos genéricos em G5.** *increase*, *impact*, *challenge*, *improvement*, *program* e *programme*, que a versão 1 tratava como termos genéricos e deixava no descarte, somam 34% de G5 no PBIA (16,53 por 1.000 tokens) e 26% nos EUA. Sem eles, o Y do PBIA seria 0,04, e não −0,17: a posição do PBIA abaixo do eixo horizontal depende desses seis lemas.
- **Autodesignação em G4.** *country*, *national* e *federal* ficavam no descarte na versão 1, como autodesignação (seção 15.1), e são 30% de G4 no PBIA e 17% nos EUA e no New Generation. Sem eles, o X do PBIA seria −0,20 (e não −0,37), e o dos EUA, −0,41 (e não −0,49); os dois continuariam com X negativo.
- **Lemas próximos em grupos diferentes.** Alguns lemas próximos dos redirecionados ficaram onde estavam: *adopt*, *uptake* e *deployment* continuam em G3, e *adoption* foi para G4; *industrial* e *manufacturing* continuam em G3, e *industry* foi para G2; *breakthrough* e *theory* continuam em G3, e *research*, *science* e *scientific* foram para G5; *improve* continua no descarte, e *improvement* foi para G5; *intelligent* continua no descarte, e *smart* foi para G2; *government* continua no descarte, e *federal* e *national* foram para G4. Para reunir esses lemas, basta movê-los no codebook da célula 15.4.1.
- **O zero não é neutro.** Os grupos dos eixos têm agora 40, 25, 27 e 30 lemas (G2 a G5), e o equilíbrio de cada par depende de quantos e de quais lemas o compõem; as posições devem ser lidas umas em relação às outras.
- **Plano curto.** O *AI+* (2.874 tokens, 65 unidades) continua com os intervalos mais largos (X de −0,34 a 0,27; Y de 0,16 a 0,51).
- **Bolhas sobrepostas.** A bolha do *AI+*, a maior, fica parcialmente encoberta pelas do AI Continent e do New Generation; as linhas-guia ligam os rótulos do AI Continent e do *AI+* às suas bolhas.

```python
# =====================================================================
# 15.4.4. CODEBOOK DA VERSÃO 3 — parâmetros editáveis pelo pesquisador
# Por comando do pesquisador (2026-10-07): o codebook que ele montou no site interativo "Perfis de
# vocabulário dos planos nacionais de IA" (exportado pelo botão "Copiar codebook (Python)" e colado no
# comando), sem alteração de nenhum lema, com o vocabulário D (G3) renomeado para "Economia e mercado".
# As células da versão 3 usam nomes próprios (sufixo _V3 ou _v3) e não alteram nenhuma variável das versões
# 1 e 2 nem das seções 15.5 e 15.6, que continuam iguais quando o notebook é executado de novo.
# Para editar: mova ou acrescente lemas nas listas abaixo e execute de novo as células da versão 3
# (15.4.4 a 15.4.7). Eixos, título, subtítulo, cores, destaque e escala das bolhas: os da seção 15.1.
# =====================================================================
GRUPOS_QUADRANTE_V3 = {
    "G1": {
        "nome": "Vocabulário comum",
        "lemas": [
            "achieve", "action", "actively", "address", "advance", "ai", "aim", "apply", "approach",
            "area", "artificial", "axis", "base", "basic", "benefit", "brazilian", "bring", "broad",
            "capability", "carry", "china", "content", "core", "create", "creation", "develop",
            "development", "digital", "effort", "enable", "encourage", "enhance", "ensure", "establish",
            "eu", "europe", "european", "evaluation", "exist", "expand", "explore", "facilitate",
            "facility", "field", "focus", "form", "formation", "foster", "fully", "future", "generation",
            "great", "group", "help", "improve", "improvement", "include", "incubate", "initiative",
            "integrate", "intelligence", "intelligent", "intelligentize", "key", "launch", "layout",
            "lead", "level", "major", "management", "mechanism", "member", "method", "need", "new",
            "occupation", "opportunity", "optimise", "optimize", "pbia", "pilot", "plan", "planning",
            "potential", "process", "project", "provide", "quality", "refine", "relevant", "require",
            "resource", "role", "scenario", "sectoral", "service", "set", "significant", "space", "state",
            "strategy", "strengthen", "technological", "technology", "tool", "trump", "union", "uptake",
            "use", "vigorously", "work",
        ],
    },
    "G2": {
        "nome": "Infraestrutura e tecnologias de IA",
        "lemas": [
            "accelerate", "advanced", "agent", "ai-based", "ai-driven", "ai-native", "application",
            "autonomous", "breakthrough", "build", "building", "capacity", "center", "centre", "cloud",
            "compute", "computing", "construction", "coordination", "data", "device", "ecosystem",
            "efficiency", "energy", "environment", "eurohpc", "factory", "gigafactory", "grid",
            "high-quality", "human-computer", "implementation", "information", "infrastructure",
            "innovation", "integration", "knowledge", "learning", "model", "network", "open",
            "open-source", "perception", "platform", "power", "robot", "robotics", "semiconductor",
            "skill", "smart", "system", "terminal", "training", "unmanned",
        ],
    },
    "G3": {
        # Nome dado pelo comando de 2026-10-07 (no codebook colado, vindo do site: "Economia, competitividade
        # e inovação")
        "nome": "Economia e mercado",
        "lemas": [
            "adopt", "adoption", "agricultural", "chain", "company", "competitiveness", "consumption",
            "deployment", "economic", "economy", "enterprise", "industrial", "industry", "investment",
            "manufacturing", "market", "product", "production", "startup",
        ],
    },
    "G4": {
        "nome": "Governança, regulação e segurança",
        "lemas": [
            "act", "adversary", "ally", "assessment", "collaboration", "commission", "consultation",
            "control", "cooperation", "country", "critical", "defence", "defense", "export", "federal",
            "frontier", "global", "governance", "government", "hub", "international", "national", "policy",
            "programme", "regulation", "risk", "sector", "security", "sharing", "stakeholder", "standard",
            "strategic",
        ],
    },
    "G5": {
        "nome": "Sociedade, inclusão, capital humano e ciência",
        "lemas": [
            "access", "challenge", "cultural", "education", "good", "health", "healthcare", "impact",
            "increase", "job", "population", "professional", "program", "promote", "public",
            "qualification", "reduction", "research", "responsible", "science", "scientific", "social",
            "society", "solution", "support", "sustainable", "talent", "theory", "worker", "workforce",
        ],
    },
}

# Linha que identifica a versão, abaixo do subtítulo da figura, como na imagem anexada ao comando (deixe ""
# para não desenhá-la)
LINHA_VERSAO_QUADRANTE_V3 = "Versão 3: codebook do pesquisador, montado no site interativo"

# Codebook recomendado do site (visão "Recomendado"), só como referência nas comparações e nas críticas
# (15.4.4c e 15.4.7): o da versão 1 com sete lemas em outro grupo
MUDANCAS_RECOMENDADO_SITE = {
    "research": "G5", "science": "G5", "scientific": "G5", "theory": "G5",   # "ciência" no nome de C
    "talent": "G5",                                                           # "capital humano" no nome de C
    "service": "G1",                                                          # sentido muda conforme o plano
    "reduction": "G1",                                                        # termo genérico
}

# A imagem anexada ao comando é a figura exportada pelo site num estado do codebook diferente do colado no
# comando (a imagem informa 57 lemas fora do grupo recomendado; o codebook colado tem 53). Centro de cada
# bolha, medido nos pixels da imagem e convertido para as coordenadas do gráfico, e os quatro lemas que,
# noutro grupo, reproduzem essas posições (conferência na tabela 15.4.5d).
POSICOES_IMAGEM_COMANDO_V3 = {
    "AI_Continent_2025": (-0.440, 0.322), "Apply_AI_2025": (-0.454, 0.230), "China_2025": (-0.262, 0.418),
    "New_Gen_2017": (-0.241, 0.323), "EUA_2025": (-0.582, 0.344), "PBIA_2025": (-0.513, -0.115),
}
LEMAS_EM_OUTRO_GRUPO_NA_IMAGEM_V3 = {"industry": "G2", "industrial": "G2", "adopt": "G2", "competitiveness": "G4"}

# ---------------------------------------------------------------------------
# Conferências do codebook da versão 3 (as da seção 15.2) e mudanças em relação às versões 1 e 2
# ---------------------------------------------------------------------------
if list(GRUPOS_QUADRANTE_V3) != list(GRUPOS_QUADRANTE):
    raise ValueError(f"A versão 3 usa os eixos da seção 15.1 e deve ter os mesmos grupos ({list(GRUPOS_QUADRANTE)}).")
LEMA_GRUPO_V3, repetidos_v3 = {}, []
for grupo, cfg in GRUPOS_QUADRANTE_V3.items():
    for lema in cfg["lemas"]:
        chave = lema.strip().lower()
        if chave in LEMA_GRUPO_V3:
            repetidos_v3.append(f"'{chave}' em {LEMA_GRUPO_V3[chave]} e em {grupo}")
        else:
            LEMA_GRUPO_V3[chave] = grupo
if repetidos_v3:
    raise ValueError("Cada lema deve estar em um único grupo. Repetições: " + "; ".join(repetidos_v3))
sem_grupo_v3 = [lema for lema in LEMAS_RECORTES if lema not in LEMA_GRUPO_V3]
fora_dos_recortes_v3 = sorted(lema for lema in LEMA_GRUPO_V3 if lema not in set(LEMAS_RECORTES))
sem_ocorrencia_v3 = sorted(lema for lema in LEMA_GRUPO_V3 if not any(lema in freq[s].index for s in DOCUMENTOS))
# Grafias de um mesmo lema, que a seção 15.1 mantém no mesmo grupo
PARES_DE_GRAFIA_Q3 = [("center", "centre"), ("defence", "defense"), ("optimize", "optimise"), ("program", "programme")]
grafias_separadas_v3 = [(a, b) for a, b in PARES_DE_GRAFIA_Q3 if LEMA_GRUPO_V3.get(a) != LEMA_GRUPO_V3.get(b)]
LEMA_GRUPO_REC_SITE = {**LEMA_GRUPO, **MUDANCAS_RECOMENDADO_SITE}
todos_v3 = set(LEMA_GRUPO) | set(LEMA_GRUPO_V2) | set(LEMA_GRUPO_V3)
mudancas_v3 = sorted(l for l in todos_v3 if LEMA_GRUPO_V2.get(l) != LEMA_GRUPO_V3.get(l))   # em relação à versão 2
n_mudancas_v3 = {"versao_1": sum(LEMA_GRUPO.get(l) != LEMA_GRUPO_V3.get(l) for l in todos_v3),
                 "versao_2": len(mudancas_v3),
                 "recomendado_do_site": sum(LEMA_GRUPO_REC_SITE.get(l) != LEMA_GRUPO_V3.get(l) for l in todos_v3)}
codebook_sha256_v3 = hashlib.sha256(json.dumps(
    {g: sorted(l.strip().lower() for l in c["lemas"]) for g, c in GRUPOS_QUADRANTE_V3.items()},
    ensure_ascii=False, sort_keys=True).encode("utf-8")).hexdigest()

# (a) Lemas de cada grupo
linhas_md = []
for grupo, cfg in GRUPOS_QUADRANTE_V3.items():
    lemas = sorted({l.strip().lower() for l in cfg["lemas"]})
    marcados = [(f"**{l}**" if l in mudancas_v3 else l) + ("†" if l in fora_dos_recortes_v3 else "") for l in lemas]
    linhas_md.append(f"- **{grupo} — {cfg['nome']}** · {descrever_polo(grupo)} · {len(lemas)} lemas: "
                     + ", ".join(marcados))
display(Markdown("**(a) Lemas de cada grupo na versão 3** (em negrito, os lemas em grupo diferente do da "
                 "versão 2; † = lema de fora dos recortes de 75, incluído no codebook)\n\n" + "\n".join(linhas_md)))

# (b) Grupos e nomes nas três versões
composicao_quadrante_v3 = pd.DataFrame([{
    "grupo": grupo,
    "posicao_na_figura": descrever_polo(grupo),
    "nome_versao_1": GRUPOS_QUADRANTE[grupo]["nome"],
    "nome_versao_2": GRUPOS_QUADRANTE_V2[grupo]["nome"],
    "nome_versao_3": GRUPOS_QUADRANTE_V3[grupo]["nome"],
    "lemas_versao_1": len(GRUPOS_QUADRANTE[grupo]["lemas"]),
    "lemas_versao_2": len(GRUPOS_QUADRANTE_V2[grupo]["lemas"]),
    "lemas_versao_3": len(GRUPOS_QUADRANTE_V3[grupo]["lemas"]),
    "entraram_em_relacao_a_v2": sum(LEMA_GRUPO_V3.get(l) == grupo for l in mudancas_v3),
    "sairam_em_relacao_a_v2": sum(LEMA_GRUPO_V2.get(l) == grupo for l in mudancas_v3),
} for grupo in GRUPOS_QUADRANTE_V3]).set_index("grupo")
display(Markdown("**(b) Grupos, nomes e número de lemas nas versões 1, 2 e 3**"))
exibir(composicao_quadrante_v3, 0)

# (c) Lemas em grupo diferente do da versão 2, com o grupo nas outras versões e a taxa por 1.000 tokens em cada
# plano (documento inteiro; é o quanto cada lema soma ao peso do seu grupo)
mudancas_quadrante_v3 = pd.DataFrame({s: freq[s]["por_1000_tokens"].reindex(mudancas_v3).fillna(0.0)
                                      for s in DOCUMENTOS})
mudancas_quadrante_v3.insert(0, "grupo_versao_1", [LEMA_GRUPO.get(l, "—") for l in mudancas_v3])
mudancas_quadrante_v3.insert(1, "grupo_recomendado_site", [LEMA_GRUPO_REC_SITE.get(l, "—") for l in mudancas_v3])
mudancas_quadrante_v3.insert(2, "grupo_versao_2", [LEMA_GRUPO_V2.get(l, "—") for l in mudancas_v3])
mudancas_quadrante_v3.insert(3, "grupo_versao_3", [LEMA_GRUPO_V3.get(l, "—") for l in mudancas_v3])
ordem_grupos_v3 = {g: i for i, g in enumerate(list(GRUPOS_QUADRANTE_V3) + ["—"])}
mudancas_quadrante_v3 = mudancas_quadrante_v3.sort_values(
    ["grupo_versao_3", "grupo_versao_2"], key=lambda col: col.map(ordem_grupos_v3), kind="stable")
mudancas_quadrante_v3.index.name = "lema"
display(Markdown(f"**(c) Os {len(mudancas_v3)} lemas em grupo diferente do da versão 2, com o grupo nas versões "
                 f"1 e 2 e no recomendado do site, e a taxa por 1.000 tokens em cada plano**"))
exibir(mudancas_quadrante_v3)
salvar_tabela(pd.DataFrame([{"grupo": g, "nome": c["nome"], "lema": l} for g, c in GRUPOS_QUADRANTE_V3.items()
                            for l in c["lemas"]]), "03_quadrante_v3_codebook", index=False)
salvar_tabela(mudancas_quadrante_v3, "03_quadrante_v3_mudancas_de_grupo")

if sem_grupo_v3:
    print(f"ATENÇÃO: {len(sem_grupo_v3)} lema(s) dos recortes de 75 sem grupo no codebook da versão 3 (ficam fora "
          f"de todas as somas): {', '.join(sem_grupo_v3)}")
if sem_ocorrencia_v3:
    print(f"ATENÇÃO: lema(s) do codebook da versão 3 sem nenhuma ocorrência nos seis planos (confira a grafia): "
          f"{', '.join(sem_ocorrencia_v3)}")
for a, b in grafias_separadas_v3:
    print(f"ATENÇÃO: as grafias '{a}' ({LEMA_GRUPO_V3.get(a)}) e '{b}' ({LEMA_GRUPO_V3.get(b)}) estão em grupos "
          f"diferentes; a seção 15.1 mantém as duas grafias de um lema no mesmo grupo (efeito na tabela 15.4.7b).")
print(f"Síntese factual: o codebook da versão 3 tem {len(LEMA_GRUPO_V3)} lemas, "
      + ", ".join(f"{len(c['lemas'])} em {g}" for g, c in GRUPOS_QUADRANTE_V3.items())
      + f"; em grupo diferente: {n_mudancas_v3['versao_2']} em relação à versão 2, {n_mudancas_v3['versao_1']} em "
      f"relação à versão 1 e {n_mudancas_v3['recomendado_do_site']} em relação ao recomendado do site. Os recortes "
      f"de 75 dos seis planos somam {len(LEMAS_RECORTES)} lemas distintos, "
      + ("todos classificados" if not sem_grupo_v3 else f"{len(sem_grupo_v3)} deles sem grupo")
      + ". Nomes: " + "; ".join(f"{g} ({descrever_polo(g).split(' (')[0]}): {GRUPOS_QUADRANTE_V3[g]['nome']}"
                                 for g in GRUPOS_QUADRANTE_V3 if g != GRUPO_DESCARTADO)
      + f". SHA-256 do codebook da versão 3: {codebook_sha256_v3}.")
```

#### 15.4.5. Peso dos grupos e posição de cada plano (versão 3)

Peso de cada grupo, massa temática, coordenadas e intervalos de confiança com o codebook da versão 3; as coordenadas das versões 1, 2 e 3 lado a lado; as diferenças entre os planos, com o critério da seção 15.5a; e a conferência da imagem anexada ao comando.

```python
# Mesmo cálculo das seções 15.3 e 15.4.2, com o codebook da versão 3: taxas no documento inteiro, pesos dos
# grupos, massa temática, X, Y e intervalos de confiança por bootstrap por unidades. As funções são as da
# seção 15.3; com a mesma semente por plano, as reamostras sorteiam as mesmas unidades das versões 1 e 2.
GRUPOS_ATIVOS_V3 = [g for g in GRUPOS_QUADRANTE_V3 if g != GRUPO_DESCARTADO]
taxas_lemas_quadrante_v3 = pd.DataFrame({s: freq[s]["por_1000_tokens"].reindex(list(LEMA_GRUPO_V3)).fillna(0.0)
                                         for s in PLANOS})
grupo_do_lema_v3 = taxas_lemas_quadrante_v3.index.map(LEMA_GRUPO_V3)
PESO_GRUPOS_V3 = taxas_lemas_quadrante_v3.groupby(grupo_do_lema_v3).sum().reindex(list(GRUPOS_QUADRANTE_V3))

reamostras_quadrante_v3 = {}
for sigla in PLANOS:
    t = processados[sigla]["tokens"]
    unidades = sorted(t["unidade"].unique())
    tamanhos = t.groupby("unidade").size().reindex(unidades).to_numpy()
    conteudo = t[t["conteudo"]]
    conteudo = conteudo.assign(grupo=conteudo["lema"].map(LEMA_GRUPO_V3))
    conteudo = conteudo[conteudo["grupo"].isin(GRUPOS_ATIVOS_V3)]
    matriz = pd.crosstab(conteudo["grupo"], conteudo["unidade"]).reindex(
        index=GRUPOS_ATIVOS_V3, columns=unidades, fill_value=0)
    # Conferência: as contagens por unidade reproduzem os pesos calculados com as tabelas da seção 11
    assert np.allclose(matriz.sum(axis=1).to_numpy() / len(t) * 1000,
                       PESO_GRUPOS_V3.loc[GRUPOS_ATIVOS_V3, sigla].to_numpy())
    r = dict(zip(GRUPOS_ATIVOS_V3, taxas_bootstrap(matriz.to_numpy(), tamanhos, semente_documento(sigla))))
    reamostras_quadrante_v3[sigla] = {"X": indice_balanco(r[EIXO_X[0]], r[EIXO_X[1]]),
                                      "Y": indice_balanco(r[EIXO_Y[0]], r[EIXO_Y[1]]),
                                      "massa": sum(r[g] for g in GRUPOS_ATIVOS_V3)}

# (a) Peso dos grupos, massa temática, coordenadas e intervalos de confiança
coordenadas_quadrante_v3 = pd.DataFrame(index=pd.Index(PLANOS, name="plano"))
for grupo in GRUPOS_QUADRANTE_V3:
    nome = f"{grupo}_descarte" if grupo == GRUPO_DESCARTADO else grupo
    coordenadas_quadrante_v3[nome] = PESO_GRUPOS_V3.loc[grupo, PLANOS].to_numpy()
coordenadas_quadrante_v3["massa"] = PESO_GRUPOS_V3.loc[GRUPOS_ATIVOS_V3, PLANOS].sum().to_numpy()
coordenadas_quadrante_v3["X"] = indice_balanco(PESO_GRUPOS_V3.loc[EIXO_X[0], PLANOS],
                                               PESO_GRUPOS_V3.loc[EIXO_X[1], PLANOS])
coordenadas_quadrante_v3["Y"] = indice_balanco(PESO_GRUPOS_V3.loc[EIXO_Y[0], PLANOS],
                                               PESO_GRUPOS_V3.loc[EIXO_Y[1], PLANOS])
for medida in ("X", "Y", "massa"):
    coordenadas_quadrante_v3[f"{medida}_ic_inferior"] = [np.quantile(reamostras_quadrante_v3[s][medida], ALFA_IC)
                                                         for s in PLANOS]
    coordenadas_quadrante_v3[f"{medida}_ic_superior"] = [np.quantile(reamostras_quadrante_v3[s][medida], 1 - ALFA_IC)
                                                         for s in PLANOS]
coordenadas_quadrante_v3["quadrante"] = [quadrante_de(x, y) for x, y in
                                         zip(coordenadas_quadrante_v3["X"], coordenadas_quadrante_v3["Y"])]
coordenadas_quadrante_v3 = coordenadas_quadrante_v3[colunas_coord]
display(Markdown(f"**(a) Versão 3: peso de cada grupo (por 1.000 tokens), massa temática e coordenadas, com "
                 f"intervalos de confiança de {num(NIVEL_CONFIANCA * 100, 0)}% (bootstrap por unidades, "
                 f"{num(N_BOOTSTRAP, 0)} reamostras)**"))
exibir(coordenadas_quadrante_v3)
salvar_tabela(coordenadas_quadrante_v3, "03_quadrante_v3_coordenadas")

# (b) Versões 1, 2 e 3 lado a lado
comparacao_versoes_quadrante_v3 = pd.concat(
    {"versao_1": coordenadas_quadrante[["X", "Y", "massa", "quadrante"]],
     "versao_2": coordenadas_quadrante_v2[["X", "Y", "massa", "quadrante"]],
     "versao_3": coordenadas_quadrante_v3[["X", "Y", "massa", "quadrante"]]}, axis=1)
for medida in ("X", "Y", "massa"):
    comparacao_versoes_quadrante_v3[("versao_3_menos_versao_2", medida)] = (
        coordenadas_quadrante_v3[medida] - coordenadas_quadrante_v2[medida])
comparacao_versoes_quadrante_v3[("versao_3_menos_versao_2", "deslocamento")] = np.hypot(
    comparacao_versoes_quadrante_v3[("versao_3_menos_versao_2", "X")],
    comparacao_versoes_quadrante_v3[("versao_3_menos_versao_2", "Y")])
display(Markdown("**(b) Coordenadas e massa nas versões 1, 2 e 3** (deslocamento = distância entre as posições "
                 "da bolha nas versões 2 e 3, em unidades dos eixos)"))
exibir(comparacao_versoes_quadrante_v3)
salvar_tabela(comparacao_versoes_quadrante_v3, "03_quadrante_v3_comparacao_com_versoes_1_e_2")

# (c) Diferença entre dois planos em X e em Y, com intervalo de confiança por bootstrap (critério da seção 15.5a:
# uma diferença é afirmada apenas quando o intervalo não inclui zero; skill 03, seção 3.5)
linhas = []
for i, a in enumerate(PLANOS):
    for b in PLANOS[i + 1:]:
        linha = {"plano_a": a, "plano_b": b}
        for eixo in ("X", "Y"):
            dif = reamostras_quadrante_v3[a][eixo] - reamostras_quadrante_v3[b][eixo]
            inf, sup = np.quantile(dif, [ALFA_IC, 1 - ALFA_IC])
            linha[f"d{eixo}"] = coordenadas_quadrante_v3.loc[a, eixo] - coordenadas_quadrante_v3.loc[b, eixo]
            linha[f"d{eixo}_ic_inferior"], linha[f"d{eixo}_ic_superior"] = inf, sup
            linha[f"d{eixo}_ic_exclui_zero"] = "sim" if inf > 0 or sup < 0 else "não"
        linhas.append(linha)
diferencas_quadrante_v3 = pd.DataFrame(linhas).set_index(["plano_a", "plano_b"])
display(Markdown(f"**(c) Versão 3: diferenças entre os planos (plano a − plano b), com intervalo de confiança de "
                 f"{num(NIVEL_CONFIANCA * 100, 0)}%**"))
exibir(diferencas_quadrante_v3)
salvar_tabela(diferencas_quadrante_v3, "03_quadrante_v3_diferencas_entre_planos")


def coordenadas_do_codebook_v3(lema_grupo):
    """X, Y e massa de cada plano para um codebook {lema: grupo}, com as taxas da versão 3 (documento inteiro)."""
    pesos = taxas_lemas_quadrante_v3.groupby(taxas_lemas_quadrante_v3.index.map(lema_grupo)).sum().reindex(
        list(GRUPOS_QUADRANTE_V3)).fillna(0.0)
    return pd.DataFrame({
        "X": indice_balanco(pesos.loc[EIXO_X[0], PLANOS], pesos.loc[EIXO_X[1], PLANOS]),
        "Y": indice_balanco(pesos.loc[EIXO_Y[0], PLANOS], pesos.loc[EIXO_Y[1], PLANOS]),
        "massa": pesos.loc[GRUPOS_ATIVOS_V3, PLANOS].sum().to_numpy(),
    }, index=pd.Index(PLANOS, name="plano"))


# (d) A imagem anexada ao comando: posições medidas na imagem, posições da versão 3 e posições com os quatro
# lemas de LEMAS_EM_OUTRO_GRUPO_NA_IMAGEM_V3 (seção 15.4.4) no grupo em que estavam na imagem
lema_grupo_imagem_v3 = {**LEMA_GRUPO_V3, **LEMAS_EM_OUTRO_GRUPO_NA_IMAGEM_V3}
coords_imagem_v3 = coordenadas_do_codebook_v3(lema_grupo_imagem_v3)
imagem_quadrante_v3 = pd.DataFrame({
    "X_medido_na_imagem": [POSICOES_IMAGEM_COMANDO_V3[s][0] for s in PLANOS],
    "Y_medido_na_imagem": [POSICOES_IMAGEM_COMANDO_V3[s][1] for s in PLANOS],
    "X_versao_3": coordenadas_quadrante_v3["X"].to_numpy(),
    "Y_versao_3": coordenadas_quadrante_v3["Y"].to_numpy(),
    "X_com_os_4_lemas_da_imagem": coords_imagem_v3["X"].to_numpy(),
    "Y_com_os_4_lemas_da_imagem": coords_imagem_v3["Y"].to_numpy(),
}, index=pd.Index(PLANOS, name="plano"))
imagem_quadrante_v3["diferenca_maxima_para_a_imagem"] = np.maximum(
    (imagem_quadrante_v3["X_com_os_4_lemas_da_imagem"] - imagem_quadrante_v3["X_medido_na_imagem"]).abs(),
    (imagem_quadrante_v3["Y_com_os_4_lemas_da_imagem"] - imagem_quadrante_v3["Y_medido_na_imagem"]).abs())
n_fora_rec_imagem = sum(lema_grupo_imagem_v3.get(l) != LEMA_GRUPO_REC_SITE.get(l) for l in todos_v3)
display(Markdown(f"**(d) A imagem anexada ao comando: posições medidas, posições da versão 3 e posições com "
                 f"{', '.join(LEMAS_EM_OUTRO_GRUPO_NA_IMAGEM_V3)} nos grupos em que estavam na imagem** (com esses "
                 f"quatro lemas, o codebook tem {n_fora_rec_imagem} lemas fora do grupo recomendado, como informa a "
                 f"imagem)"))
exibir(imagem_quadrante_v3, 3)

c3, d3 = coordenadas_quadrante_v3, diferencas_quadrante_v3
deslocamento_v3 = comparacao_versoes_quadrante_v3[("versao_3_menos_versao_2", "deslocamento")]
mudou_v3 = [s for s in PLANOS if coordenadas_quadrante_v2.loc[s, "quadrante"] != c3.loc[s, "quadrante"]]
print(f"Síntese factual: na versão 3, X vai de {num(c3['X'].min())} ({c3['X'].idxmin()}) a {num(c3['X'].max())} "
      f"({c3['X'].idxmax()}); Y, de {num(c3['Y'].min())} ({c3['Y'].idxmin()}) a {num(c3['Y'].max())} "
      f"({c3['Y'].idxmax()}); a massa temática, de {num(c3['massa'].min(), 1)} ({c3['massa'].idxmin()}) a "
      f"{num(c3['massa'].max(), 1)} ({c3['massa'].idxmax()}) por 1.000 tokens. Quadrantes: "
      + "; ".join(f"{q}: {', '.join(s)}" for q, s in c3.groupby("quadrante").groups.items())
      + f". Em relação à versão 2, as bolhas se deslocam de {num(deslocamento_v3.min())} ({deslocamento_v3.idxmin()}) a "
      f"{num(deslocamento_v3.max())} ({deslocamento_v3.idxmax()}); "
      + (f"mudam de quadrante: {', '.join(mudou_v3)}" if mudou_v3 else "nenhum plano muda de quadrante")
      + f". Das {len(d3)} comparações entre pares de planos, o intervalo da diferença exclui zero em "
      f"{(d3['dX_ic_exclui_zero'] == 'sim').sum()} no eixo X e em {(d3['dY_ic_exclui_zero'] == 'sim').sum()} "
      f"no eixo Y. Imagem do comando: com os quatro lemas no grupo da imagem, a maior diferença entre as posições "
      f"calculadas e as medidas é {num(imagem_quadrante_v3['diferenca_maxima_para_a_imagem'].max(), 3)}.")
```

#### 15.4.6. Gráfico de quadrantes (versão 3)

```python
import matplotlib.patheffects as path_effects
from matplotlib.text import Text

# ---------------------------------------------------------------------------
# Desenho da versão 3: o da figura exportada pelo site interativo e anexada ao comando de 2026-10-07. Título,
# subtítulo e linha da versão; caixas dos vocabulários com a letra num círculo escuro (A no topo, B à esquerda,
# C na base, D à direita); bolhas com área proporcional à massa, na escala das versões 1 e 2; rótulos colocados
# por busca automática, com linha-guia quando ficam longe da bolha ou quando a bolha se sobrepõe a outra (então
# a linha vai até o centro, marcado com um ponto); legenda de documentos e de massa temática abaixo do gráfico;
# nota com as fórmulas. Sem os círculos tracejados da posição recomendada, retirados do site pelo mesmo comando.
# Da seção 15.4 vêm as transformações, as cores, o raio das bolhas e as duas legendas (_q_pol, _q_escurecer,
# _q_raio_pt, _q_legenda_documentos e _q_legenda_massa), sem alteração; da seção 15.4.3, _q_globais_trocadas.
# ---------------------------------------------------------------------------
LADO_PLOT_Q3 = LADO_PLOT_Q                 # 7,4 polegadas, como nas versões 1 e 2 (mesma escala das bolhas)
MARGEM_LATERAL_Q3 = 2.0                    # espaço para as caixas B e D
LARGURA_FIG_Q3 = LADO_PLOT_Q3 + 2 * MARGEM_LATERAL_Q3
MARGEM_TOPO_Q3, MARGEM_BASE_Q3 = 0.25, 0.25
FOLGA_CAIXA_Q3 = 0.14                      # entre o gráfico e as caixas dos vocabulários
FOLGA_LEGENDA_Q3 = 0.40                    # entre a caixa C e a legenda
TAMANHOS_Q3 = {"titulo": 15.5, "subtitulo": 10.5, "versao": 10, "polo": 10, "rotulo": 10, "nota": 9}
COR_CAIXA_Q3, COR_BORDA_CAIXA_Q3 = "#f4f4f4", "#bdbdbd"
COR_REFERENCIA_Q3 = "#e8e8e8"              # linhas em ±0,5, contínuas como no site
COR_GUIA_Q3 = "#6f6f6f"
# Direções testadas para cada rótulo: ângulo (graus, sentido trigonométrico), alinhamento do texto e
# preferência (menor = preferida); e distâncias além da borda da bolha, em pontos
DIRECOES_Q3 = [(0, "left", "center", 0.0), (180, "right", "center", 0.6), (-90, "center", "top", 1.0),
               (90, "center", "bottom", 1.0), (-40, "left", "top", 1.4), (220, "right", "top", 1.6),
               (40, "left", "bottom", 1.4), (140, "right", "bottom", 1.6)]
DISTANCIAS_Q3 = [4, 12, 22, 34, 48]
# Subtítulo da figura exportada pelo site e anexada ao comando ("quatro vocabulários"; na seção 15.1, "quatro
# campos lexicais")
SUBTITULO_QUADRANTE_V3 = (
    "Seis documentos oficiais de política de IA (2017–2025) posicionados pelo peso relativo de quatro vocabulários.\n"
    "Cada eixo contrapõe dois vocabulários: quanto mais distante do centro, mais desequilibrado é o par naquele documento."
)
NOTA_QUADRANTE_V3 = (
    "X = (D − B)/(D + B);  Y = (A − C)/(A + C);  A, B, C, D = soma das taxas por 1.000 tokens dos lemas de cada "
    "vocabulário, no documento inteiro;\nárea da bolha proporcional a A + B + C + D. "
    f"Lemas: os {len(LEMAS_RECORTES)} dos recortes de 75 lemas mais frequentes de cada plano, mais "
    f"{len(fora_dos_recortes_v3)} acrescentados ao codebook;\no vocabulário comum fica fora dos eixos. "
    f"No notebook, A = {POLOS_QUADRANTE['topo'][1]}, B = {POLOS_QUADRANTE['esquerda'][1]}, "
    f"C = {POLOS_QUADRANTE['base'][1]} e D = {POLOS_QUADRANTE['direita'][1]}."
)


def _q3_topo(fig):
    """(x, y) em polegadas, com y medido a partir do TOPO da figura -> tela."""
    largura, altura = fig.get_size_inches()
    return Affine2D().scale(1 / largura, -1 / altura).translate(0, 1) + fig.transFigure


def _q3_largura(fig, renderer, texto, tamanho, peso="normal"):
    """Largura de um texto, em polegadas."""
    t = fig.text(0, 0, texto, fontsize=tamanho, fontweight=peso)
    largura = t.get_window_extent(renderer).width / fig.dpi
    t.remove()
    return largura


def _q3_quebrar(fig, renderer, texto, largura_max, tamanho):
    """Quebra o texto em linhas de até largura_max polegadas."""
    linhas, atual = [], ""
    for palavra in texto.split():
        teste = f"{atual} {palavra}".strip()
        if atual and _q3_largura(fig, renderer, teste, tamanho) > largura_max:
            linhas.append(atual)
            atual = palavra
        else:
            atual = teste
    return linhas + ([atual] if atual else [])


def _q3_medir_caixa(fig, renderer, nome, modo):
    """Tamanho da caixa de um vocabulário. modo "linha" (A e C): letra e nome na mesma linha; "coluna" (B e
    D): letra acima do nome, quebrado na largura da margem lateral."""
    tam = TAMANHOS_Q3["polo"]
    caixa = {"modo": modo, "pad": 0.09, "r": 0.105, "folga_letra": 0.08, "lh": tam * 1.3 / 72}
    if modo == "linha":
        caixa["linhas"] = [nome]
        caixa["w"] = 2 * caixa["pad"] + 2 * caixa["r"] + caixa["folga_letra"] + _q3_largura(fig, renderer, nome, tam)
        caixa["h"] = 2 * caixa["pad"] + max(2 * caixa["r"], caixa["lh"])
    else:
        caixa["linhas"] = _q3_quebrar(fig, renderer, nome, MARGEM_LATERAL_Q3 - FOLGA_CAIXA_Q3 - 2 * caixa["pad"] - 0.15, tam)
        caixa["w"] = 2 * caixa["pad"] + max([2 * caixa["r"]] + [_q3_largura(fig, renderer, l, tam) for l in caixa["linhas"]])
        caixa["h"] = 2 * caixa["pad"] + 2 * caixa["r"] + 0.05 + len(caixa["linhas"]) * caixa["lh"]
    return caixa


def _q3_desenhar_caixa(fig, T, caixa, x, y, letra, direita=False):
    """Desenha a caixa com o canto superior esquerdo em (x, y), em polegadas a partir do topo. direita=True
    (caixa B) alinha a letra e o texto à direita, junto do eixo."""
    tam, w, h, pad, r = TAMANHOS_Q3["polo"], caixa["w"], caixa["h"], caixa["pad"], caixa["r"]
    moldura = FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0,rounding_size=0.07", transform=T,
                             facecolor=COR_CAIXA_Q3, edgecolor=COR_BORDA_CAIXA_Q3, linewidth=0.7, zorder=5)
    fig.add_artist(moldura)
    if caixa["modo"] == "linha":
        cx, cy = x + pad + r, y + h / 2
    else:
        cx, cy = (x + w - pad - r) if direita else (x + pad + r), y + pad + r
    fig.add_artist(Circle((cx, cy), r, transform=T, facecolor=INK_Q, edgecolor="none", zorder=6))
    fig.text(cx, cy, letra, transform=T, ha="center", va="center", fontsize=tam - 1, fontweight="bold",
             color="white", zorder=7)
    if caixa["modo"] == "linha":
        fig.text(x + pad + 2 * r + caixa["folga_letra"], y + h / 2, caixa["linhas"][0], transform=T, ha="left",
                 va="center", fontsize=tam, color=INK_Q, zorder=7)
    else:
        for i, linha in enumerate(caixa["linhas"]):
            fig.text((x + w - pad) if direita else (x + pad), y + pad + 2 * r + 0.05 + (i + 0.5) * caixa["lh"], linha,
                     transform=T, ha="right" if direita else "left", va="center", fontsize=tam, color=INK_Q, zorder=7)
    return moldura


def _q3_eixos(ax):
    """Eixos de −1 a +1 cruzados na origem e referências em ±0,5 (contínuas, como no site)."""
    ax.set_aspect("equal", adjustable="box")
    ax.set_xlim(-1, 1)
    ax.set_ylim(-1, 1)
    ax.grid(False)
    ax.patch.set_alpha(0)
    for lado in ("left", "bottom"):
        ax.spines[lado].set_position("zero")
        ax.spines[lado].set_color(CINZA_EIXO_Q)
        ax.spines[lado].set_linewidth(1.0)
        ax.spines[lado].set_zorder(1)
    for lado in ("top", "right"):
        ax.spines[lado].set_visible(False)
    for v in (-0.5, 0.5):
        ax.axhline(v, color=COR_REFERENCIA_Q3, linewidth=0.8, zorder=0)
        ax.axvline(v, color=COR_REFERENCIA_Q3, linewidth=0.8, zorder=0)
    ticks = [-1, -0.5, 0.5, 1]
    rot = {-1: "−1", -0.5: "−0,5", 0.5: "0,5", 1: "1"}
    ax.set_xticks(ticks, labels=[rot[t] for t in ticks])
    ax.set_yticks(ticks, labels=[rot[t] for t in ticks])
    ax.tick_params(axis="both", length=3, width=0.8, color=CINZA_EIXO_Q, labelsize=9, labelcolor=CINZA_TEXTO_Q, pad=3)


def _q3_bolhas(ax, coords):
    """Bolhas como na seção 15.4 (as maiores primeiro, halo branco, borda escura, anel tracejado no destaque).
    Devolve, por plano, o raio do corpo e o raio externo (halo ou anel), em pontos (num scatter, o marcador de
    área s tem diâmetro √s pontos)."""
    raios = {}
    for sigla, r in coords.sort_values("massa", ascending=False).iterrows():
        massa, cor = float(r["massa"]), CORES_QUADRANTE[sigla]
        rr = _q_raio_pt(massa)
        s_externo = math.pi * (rr + 1.5) ** 2
        ax.scatter(r["X"], r["Y"], s=s_externo, facecolor="white", edgecolor="none", zorder=2.5)
        ax.scatter(r["X"], r["Y"], s=massa * ESCALA_BOLHA_QUADRANTE, facecolor=cor, edgecolor=_q_escurecer(cor),
                   linewidth=0.9, zorder=3, gid=f"bolha:{sigla}")
        if sigla == DESTAQUE_QUADRANTE:
            s_externo = math.pi * (rr + 3.5) ** 2
            ax.scatter(r["X"], r["Y"], s=s_externo, facecolor="none", edgecolor=cor, linewidth=1.0,
                       linestyle=(0, (4, 2.5)), zorder=3.1)
        raios[sigla] = {"corpo": math.sqrt(massa * ESCALA_BOLHA_QUADRANTE) / 2, "externo": math.sqrt(s_externo) / 2}
    return raios


def _q3_sobreposicao(a, b):
    """Área de interseção de duas caixas (x0, y0, x1, y1)."""
    return max(0.0, min(a[2], b[2]) - max(a[0], b[0])) * max(0.0, min(a[3], b[3]) - max(a[1], b[1]))


def _q3_circulo_na_caixa(cx, cy, r, caixa):
    """Quanto um círculo entra numa caixa (0 = não encosta)."""
    px, py = min(max(cx, caixa[0]), caixa[2]), min(max(cy, caixa[1]), caixa[3])
    return max(0.0, r - math.hypot(px - cx, py - cy))


def _q3_rotulos(fig, ax, coords, raios, caixas_fixas, renderer):
    """Rótulos das bolhas: para cada uma (as maiores primeiro), a posição em volta da bolha que menos encosta em
    rótulos já colocados, em outras bolhas, nos eixos, nos valores dos eixos e nas caixas dos vocabulários (a
    mesma busca do site). Tudo em pixels da tela."""
    px = fig.dpi / 72
    centros = {s: ax.transData.transform((coords.loc[s, "X"], coords.loc[s, "Y"])) for s in coords.index}
    corpo = {s: raios[s]["corpo"] * px for s in coords.index}
    externo = {s: raios[s]["externo"] * px for s in coords.index}
    sobrepostas = {a for a in coords.index for b in coords.index
                   if a != b and math.hypot(*(centros[a] - centros[b])) < 0.85 * (corpo[a] + corpo[b])}
    (ex0, ey0), (ex1, ey1) = ax.transData.transform((-1, -1)), ax.transData.transform((1, 1))
    (ox, oy) = ax.transData.transform((0, 0))
    eixo_h, eixo_v = (ex0, oy - 2 * px, ex1, oy + 2 * px), (ox - 2 * px, ey0, ox + 2 * px, ey1)
    valores = [t.get_window_extent(renderer).extents for t in ax.get_xticklabels() + ax.get_yticklabels() if t.get_text()]
    largura_fig, altura_fig = fig.get_size_inches() * fig.dpi
    colocados, rotulos = [], []
    for sigla in coords.sort_values("massa", ascending=False).index:
        foco = sigla == DESTAQUE_QUADRANTE
        texto = DOCUMENTOS_QUADRANTE[sigla]["rotulo"]
        t = ax.text(0, 0, texto, fontsize=TAMANHOS_Q3["rotulo"], fontweight="bold" if foco else "normal")
        bb = t.get_window_extent(renderer)
        w, h = bb.width, bb.height
        t.remove()
        cx, cy = centros[sigla]
        rr = externo[sigla] + px
        melhor = None
        for dist in DISTANCIAS_Q3:
            for ang, ha, va, pref in DIRECOES_Q3:
                a = math.radians(ang)
                axp, ayp = cx + (rr + dist * px) * math.cos(a), cy + (rr + dist * px) * math.sin(a)
                x0 = axp if ha == "left" else axp - w if ha == "right" else axp - w / 2
                y0 = ayp - h if va == "top" else ayp if va == "bottom" else ayp - h / 2
                caixa = (x0, y0, x0 + w, y0 + h)
                frac = lambda o: _q3_sobreposicao(caixa, o) / (w * h)
                s = pref * 2 + dist * 0.5
                for o in colocados:
                    if frac(o) > 0:
                        s += 60 + 120 * frac(o)
                for o in coords.index:
                    if o != sigla:
                        s += 3 * _q3_circulo_na_caixa(*centros[o], externo[o] + 3 * px, caixa) / px
                for o in caixas_fixas:
                    if frac(o) > 0:
                        s += 60 + 100 * frac(o)
                for o in valores:
                    if frac(o) > 0:
                        s += 22 + 60 * frac(o)
                if frac(eixo_h) > 0:
                    s += 18 + 120 * frac(eixo_h)
                if frac(eixo_v) > 0:
                    s += 6 + 120 * frac(eixo_v)
                fora_plot = max(0, ex0 - x0) + max(0, x0 + w - ex1) + max(0, ey0 - y0) + max(0, y0 + h - ey1)
                fora_fig = max(0, -x0) + max(0, x0 + w - largura_fig) + max(0, -y0) + max(0, y0 + h - altura_fig)
                s += 0.15 * fora_plot / px + 4 * fora_fig / px
                if melhor is None or s < melhor[0]:
                    melhor = (s, caixa, axp, ayp, ha, va, dist)
        _, caixa, axp, ayp, ha, va, dist = melhor
        colocados.append(caixa)
        centro = sigla in sobrepostas
        guia = None
        if centro or dist > 6:
            guia = dict(arrowstyle="-", color=COR_GUIA_Q3, linewidth=0.6, shrinkA=1.5,
                        shrinkB=2.6 if centro else raios[sigla]["externo"] + 1)
        anot = ax.annotate(texto, xy=(coords.loc[sigla, "X"], coords.loc[sigla, "Y"]),
                           xytext=((axp - cx) / px, (ayp - cy) / px), textcoords="offset points", ha=ha, va=va,
                           fontsize=TAMANHOS_Q3["rotulo"], fontweight="bold" if foco else "normal", color=INK_Q,
                           zorder=6, arrowprops=guia, annotation_clip=False)
        anot.set_path_effects([path_effects.withStroke(linewidth=3, foreground="white")])
        if centro:      # ponto no centro da bolha sobreposta, onde termina a linha-guia
            ax.scatter(coords.loc[sigla, "X"], coords.loc[sigla, "Y"], s=4.6 ** 2, facecolor=INK_Q,
                       edgecolor="white", linewidth=0.8, zorder=5.5)
        rotulos.append((texto, anot))
    return rotulos


def _q3_conferir(fig, ax, coords, raios, rotulos, blocos):
    """Conferências da figura (lista vazia = tudo certo): (1) cada bolha está na posição calculada; (2) nenhum
    rótulo encosta em outro rótulo ou no corpo de outra bolha; (3) os blocos de texto (título, subtítulo, linha
    da versão, caixas, legendas e nota) não se sobrepõem."""
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    avisos = []
    desenhadas = {col.get_gid()[len("bolha:"):]: np.asarray(col.get_offsets(), dtype=float)[0]
                  for col in ax.collections if (col.get_gid() or "").startswith("bolha:")}
    for sigla, r in coords.iterrows():
        xy = desenhadas.get(sigla)
        if xy is None or not np.array_equal(xy, np.array([float(r["X"]), float(r["Y"])])):
            avisos.append(f"a bolha de {sigla} não está na posição calculada")
    px = fig.dpi / 72
    # Só o texto de cada rótulo (sem a linha-guia, que Annotation.get_window_extent incluiria)
    caixas = {nome: Text.get_window_extent(t, renderer).extents for nome, t in rotulos}
    nomes = list(caixas)
    for i, a in enumerate(nomes):
        for b in nomes[i + 1:]:
            if _q3_sobreposicao(caixas[a], caixas[b]) > 0:
                avisos.append(f"o rótulo '{a}' encosta no rótulo '{b}'")
    rotulo_de = {DOCUMENTOS_QUADRANTE[s]["rotulo"]: s for s in coords.index}
    for nome, caixa in caixas.items():
        for s in coords.index:
            if s != rotulo_de[nome]:
                cx, cy = ax.transData.transform((coords.loc[s, "X"], coords.loc[s, "Y"]))
                if _q3_circulo_na_caixa(cx, cy, raios[s]["corpo"] * px, caixa) > 0:
                    avisos.append(f"o rótulo '{nome}' encosta na bolha de {s}")
    margem = MARGEM_BLOCOS_PT_Q * px
    ext = {nome: Bbox.from_extents(*e).padded(margem) for nome, e in blocos.items()}
    nomes = list(ext)
    for i, a in enumerate(nomes):
        for b in nomes[i + 1:]:
            if ext[a].overlaps(ext[b]):
                avisos.append(f"o bloco '{a}' sobrepõe o bloco '{b}'")
    return avisos


def construir_quadrante_v3(coords, altura):
    """Monta a figura com a altura dada (polegadas). Devolve (fig, ax, avisos, altura necessária)."""
    largura = LARGURA_FIG_Q3
    fig = plt.figure(figsize=(largura, altura))
    T = _q3_topo(fig)
    renderer = fig.canvas.get_renderer()
    blocos = {}
    ext_pol = lambda artista: artista.get_window_extent(renderer).height / fig.dpi

    # Título, subtítulo e linha da versão
    y = MARGEM_TOPO_Q3
    titulo = fig.text(largura / 2, y, TITULO_QUADRANTE, transform=T, ha="center", va="top",
                      fontsize=TAMANHOS_Q3["titulo"], fontweight="bold", color=INK_Q)
    y += ext_pol(titulo) + 0.13
    subtitulo = fig.text(largura / 2, y, SUBTITULO_QUADRANTE_V3, transform=T, ha="center", va="top",
                         fontsize=TAMANHOS_Q3["subtitulo"], color=CINZA_TEXTO_Q, linespacing=1.45)
    y += ext_pol(subtitulo) + 0.15
    versao = None
    if LINHA_VERSAO_QUADRANTE_V3:
        versao = fig.text(largura / 2, y, LINHA_VERSAO_QUADRANTE_V3, transform=T, ha="center", va="top",
                          fontsize=TAMANHOS_Q3["versao"], fontweight="bold", color=CINZA_TEXTO_Q)
        y += ext_pol(versao)
    y += 0.24

    # Caixas dos vocabulários e gráfico
    caixas = {lado: _q3_medir_caixa(fig, renderer, GRUPOS_QUADRANTE_V3[g]["nome"],
                                    "linha" if lado in ("topo", "base") else "coluna")
              for lado, (letra, g) in POLOS_QUADRANTE.items()}
    y_plot = y + caixas["topo"]["h"] + FOLGA_CAIXA_Q3
    x_plot = MARGEM_LATERAL_Q3
    ax = fig.add_axes([x_plot / largura, 1 - (y_plot + LADO_PLOT_Q3) / altura,
                       LADO_PLOT_Q3 / largura, LADO_PLOT_Q3 / altura])
    _q3_eixos(ax)
    xc, yc = x_plot + LADO_PLOT_Q3 / 2, y_plot + LADO_PLOT_Q3 / 2
    posicao = {
        "topo": (xc - caixas["topo"]["w"] / 2, y_plot - FOLGA_CAIXA_Q3 - caixas["topo"]["h"]),
        "base": (xc - caixas["base"]["w"] / 2, y_plot + LADO_PLOT_Q3 + FOLGA_CAIXA_Q3),
        "esquerda": (x_plot - FOLGA_CAIXA_Q3 - caixas["esquerda"]["w"], yc - caixas["esquerda"]["h"] / 2),
        "direita": (x_plot + LADO_PLOT_Q3 + FOLGA_CAIXA_Q3, yc - caixas["direita"]["h"] / 2),
    }
    molduras = {lado: _q3_desenhar_caixa(fig, T, caixas[lado], *posicao[lado], POLOS_QUADRANTE[lado][0],
                                         direita=(lado == "esquerda"))
                for lado in POLOS_QUADRANTE}

    # Bolhas e rótulos
    raios = _q3_bolhas(ax, coords)
    fig.canvas.draw()
    caixas_fixas = [m.get_window_extent(renderer).extents for m in molduras.values()]
    rotulos = _q3_rotulos(fig, ax, coords, raios, caixas_fixas, renderer)

    # Legenda: documentos (à esquerda) e massa temática (à direita), abaixo da caixa C
    y_leg = posicao["base"][1] + caixas["base"]["h"] + FOLGA_LEGENDA_Q3
    x_doc, x_massa = 0.55, largura * 0.63
    leg_doc = _q_legenda_documentos(fig, coords)
    leg_doc.set_bbox_to_anchor((x_doc, altura - y_leg), transform=_q_pol(fig))
    with _q_globais_trocadas(SOMA_ATIVOS_Q="A + B + C + D"):
        massa = _q_legenda_massa(fig, renderer)
    massa["deslocamento"].translate(x_massa, altura - y_leg)
    y_anel = y_leg + massa["altura"] + 0.16
    anel = fig.text(x_massa, y_anel, "\n".join(_q3_quebrar(
        fig, renderer, f"Anel tracejado: destaque visual do {DOCUMENTOS_QUADRANTE[DESTAQUE_QUADRANTE]['rotulo'].split(' — ')[-1]} "
        "(cálculo igual ao dos demais).", largura - x_massa - 0.4, 9)), transform=T, ha="left", va="top",
        fontsize=9, color=CINZA_TEXTO_Q, linespacing=1.4)
    fig.canvas.draw()
    fim_doc = altura - leg_doc.get_window_extent(renderer).y0 / fig.dpi
    fim_anel = y_anel + ext_pol(anel)
    nota = fig.text(largura / 2, max(fim_doc, fim_anel) + 0.32, NOTA_QUADRANTE_V3, transform=T, ha="center",
                    va="top", fontsize=TAMANHOS_Q3["nota"], color=CINZA_TEXTO_Q, linespacing=1.45)
    fig.canvas.draw()
    altura_necessaria = altura - nota.get_window_extent(renderer).y0 / fig.dpi + MARGEM_BASE_Q3

    para_tela = lambda x0, y0, w, h: (x0 * fig.dpi, (altura - y0 - h) * fig.dpi, (x0 + w) * fig.dpi, (altura - y0) * fig.dpi)
    blocos = {"título": titulo.get_window_extent(renderer).extents,
              "subtítulo": subtitulo.get_window_extent(renderer).extents,
              **({"linha da versão": versao.get_window_extent(renderer).extents} if versao is not None else {}),
              **{f"caixa {lado}": m.get_window_extent(renderer).extents for lado, m in molduras.items()},
              "legenda de documentos": leg_doc.get_window_extent(renderer).extents,
              "legenda de massa": para_tela(x_massa, y_leg, massa["largura"], massa["altura"]),
              "nota do anel": anel.get_window_extent(renderer).extents,
              "nota de rodapé": nota.get_window_extent(renderer).extents}
    return fig, ax, _q3_conferir(fig, ax, coords, raios, rotulos, blocos), altura_necessaria


# Duas passagens: a primeira mede a altura necessária; a segunda desenha a figura com essa altura
with plt.rc_context(ESTILO_QUADRANTE):
    fig_rascunho, _, _, altura_quadrante_v3 = construir_quadrante_v3(coordenadas_quadrante_v3, 16.0)
    plt.close(fig_rascunho)
    fig_quadrante_v3, ax_quadrante_v3, avisos_quadrante_v3, _ = construir_quadrante_v3(coordenadas_quadrante_v3,
                                                                                      altura_quadrante_v3)
    salvar_figura(fig_quadrante_v3, "03_quadrante_v3_perfis_de_vocabulario")
    plt.show()

for aviso in avisos_quadrante_v3:
    print("AVISO:", aviso)
c3 = coordenadas_quadrante_v3
print(f"Síntese factual: versão 3 com {len(c3)} planos posicionados; X de {num(c3['X'].min())} ({c3['X'].idxmin()}) "
      f"a {num(c3['X'].max())} ({c3['X'].idxmax()}); Y de {num(c3['Y'].min())} ({c3['Y'].idxmin()}) a "
      f"{num(c3['Y'].max())} ({c3['Y'].idxmax()}); massa temática de {num(c3['massa'].min(), 1)} a "
      f"{num(c3['massa'].max(), 1)} por 1.000 tokens. Caixas dos vocabulários: "
      + "; ".join(f"{letra} — {GRUPOS_QUADRANTE_V3[g]['nome']} ({lado})" for lado, (letra, g) in POLOS_QUADRANTE.items())
      + f". Figura de {num(LARGURA_FIG_Q3, 1)} × {num(altura_quadrante_v3, 1)} polegadas. "
      + ("Conferências da figura sem avisos: posições desenhadas iguais às calculadas, sem sobreposição de "
         "rótulos, bolhas e blocos de texto." if not avisos_quadrante_v3
         else f"{len(avisos_quadrante_v3)} aviso(s) nas conferências da figura (acima)."))
```

**Método e escolhas (versão 3).** Os procedimentos são os das versões 1 e 2 (texto *Método e escolhas* da seção 15.5, itens 2 a 4 e 6): o peso de um grupo é a soma das taxas por 1.000 tokens dos seus lemas no documento inteiro; X = (G3 − G4)/(G3 + G4) e Y = (G2 − G5)/(G2 + G5), isto é, X = (D − B)/(D + B) e Y = (A − C)/(A + C); a área da bolha é proporcional a A + B + C + D, na escala das versões 1 e 2; os intervalos de confiança de 95% vêm do *bootstrap* por unidades, com as mesmas reamostras das versões 1 e 2; e uma diferença entre dois planos é afirmada apenas quando o intervalo da diferença não inclui zero (skill 03, seção 3.5). Mudam o codebook e o desenho (seção 15.4.4). O desenho foi refeito em matplotlib a partir da figura do site; os rótulos são posicionados pela mesma busca automática do site, e a célula confere se cada bolha está na posição calculada e se rótulos, bolhas e blocos de texto se sobrepõem.

**Interpretação.** *Observação:*

- **Posições (tabela 15.4.5a).** Todos os planos têm X negativo. Os EUA (−0,48 [IC 95%: −0,58; −0,37]), o PBIA (−0,37 [−0,46; −0,28]), o Apply AI (−0,33 [−0,45; −0,20]) e o AI Continent (−0,26 [−0,38; −0,14]) têm intervalos que não incluem zero; nos dois planos chineses, X é −0,07, com intervalos que incluem zero (New Generation [−0,21; 0,07]; *AI+* [−0,37; 0,21]). Em Y, cinco planos ficam acima do eixo horizontal, de 0,20 (Apply AI) a 0,38 (*AI+*), e o PBIA fica abaixo (−0,15 [−0,20; −0,09]); nenhum desses seis intervalos inclui zero.
- **Diferenças entre os planos (tabela 15.4.5c).** O Y do PBIA é menor que o de cada um dos outros cinco planos, com intervalo da diferença que exclui zero em todos os casos (de 0,35 [0,22; 0,48], em relação ao Apply AI, a 0,53 [0,37; 0,70], em relação ao *AI+*); entre os outros cinco, nenhuma diferença em Y atende ao critério. Em X, seis das quinze comparações atendem ao critério: o AI Continent (diferença de 0,22 [0,07; 0,38]), o New Generation (0,41 [0,23; 0,58]) e o *AI+* (0,40 [0,08; 0,71]) ficam à direita dos EUA; o New Generation fica à direita do Apply AI (0,26 [0,07; 0,44]), do PBIA (0,30 [0,13; 0,46]) e do AI Continent (0,19 [0,003; 0,38], no limite do critério). A diferença entre o PBIA e os EUA em X (0,11 [−0,03; 0,25]) não atende ao critério, nem as diferenças entre o PBIA e os planos da UE.
- **Em relação à versão 2 (tabela 15.4.5b).** As bolhas se deslocam de 0,02 (PBIA) a 0,27 (Apply AI). Os planos da UE vão para a esquerda (X do Apply AI de −0,06 para −0,33; do AI Continent, de −0,08 para −0,26), e o New Generation passa do quadrante superior direito para o superior esquerdo (X de 0,02 para −0,07). A massa temática aumenta em todos os planos, de 14,15 por 1.000 tokens (EUA) a 37,93 (*AI+*), e vai agora de 133,2 (EUA) a 212,2 (*AI+*), porque 33 lemas que a versão 1 deixava no descarte entram nos eixos.

*Inferência:* com o codebook da versão 3, o eixo X põe os EUA, o PBIA e os dois planos da UE do lado do vocabulário de governança, regulação e segurança (intervalos de X abaixo de zero) e os dois planos chineses sobre a fronteira com o de economia e mercado; entre esses dois conjuntos, as diferenças do New Generation atendem ao critério, e as do *AI+*, só em relação aos EUA, porque o intervalo do *AI+*, o plano mais curto, é largo. O eixo Y separa o PBIA, único plano em que o vocabulário de sociedade, inclusão, capital humano e ciência supera o de infraestrutura e tecnologias de IA, dos outros cinco. As posições medem pesos lexicais, e não prioridades políticas, e dependem muito das escolhas do codebook, como mostram as críticas da seção 15.4.7.

#### 15.4.7. Críticas ao gráfico (versão 3)

Por comando do pesquisador, críticas ao gráfico da versão 3. As tabelas abaixo medem o quanto as posições dependem de escolhas do codebook: (a) os lemas que a versão 1 deixava no descarte e que a versão 3 leva aos eixos, e o peso deles em cada grupo; (b) as coordenadas com uma escolha desfeita de cada vez; (c) a influência de cada lema; (d) pares de lemas de sentido próximo em grupos diferentes. Nenhuma tabela altera a versão 3.

```python
# Críticas ao gráfico da versão 3, por comando do pesquisador (2026-10-07). As tabelas medem o quanto as
# posições dependem de escolhas do codebook; nenhuma delas altera a versão 3.

# (a) Lemas que a versão 3 leva aos eixos e que a versão 1 deixava no descarte, com o motivo registrado na
# seção 15.1 (listas copiadas dos comentários de GRUPOS_QUADRANTE)
MOTIVOS_DESCARTE_V1 = {
    "genérico (proposta inicial do pesquisador)": [
        "achieve", "action", "actively", "address", "advance", "advanced", "ai", "aim", "approach", "area",
        "artificial", "axis", "base", "brazilian", "bring", "broad", "build", "building", "carry", "china",
        "construction", "core", "create", "creation", "develop", "development", "digital", "effort", "enable",
        "encourage", "enhance", "establish", "eu", "europe", "european", "exist", "expand", "explore",
        "facilitate", "field", "focus", "form", "formation", "foster", "fully", "future", "generation", "great",
        "high-quality", "impact", "implementation", "improve", "improvement", "initiative", "integrate",
        "integration", "intelligence", "key", "launch", "layout", "level", "major", "management", "mechanism",
        "member", "method", "need", "new", "opportunity", "optimize", "pbia", "pilot", "planning", "potential",
        "process", "project", "provide", "quality", "refine", "relevant", "require", "resource", "role", "sector",
        "sectoral", "set", "sharing", "significant", "solution", "strengthen", "support", "technological",
        "technology", "tool", "trump", "union", "use", "vigorously", "work"],
    "marcador de “IA”": ["ai-based", "ai-driven", "ai-native", "intelligent", "intelligentize", "smart"],
    "autodesignação ou nome de órgão ou de plano": ["apply", "commission", "country", "federal", "good",
                                                    "government", "national", "plan", "state", "strategy"],
    "verbo ou termo genérico (revisão)": ["accelerate", "benefit", "capability", "challenge", "ensure", "help",
                                          "include", "increase", "program", "promote", "scenario"],
    "sentido muda conforme o plano": ["basic", "capacity", "collaboration", "control", "coordination",
                                      "environment", "space"],
    "grafia britânica": ["optimise", "programme"],
}
MOTIVO_DESCARTE_V1 = {l: motivo for motivo, lemas in MOTIVOS_DESCARTE_V1.items() for l in lemas}
assert set(MOTIVO_DESCARTE_V1) == {l for l, g in LEMA_GRUPO.items() if g == GRUPO_DESCARTADO}
vindos_do_descarte_v3 = sorted(l for l, g in LEMA_GRUPO_V3.items()
                               if g != GRUPO_DESCARTADO and LEMA_GRUPO.get(l) == GRUPO_DESCARTADO)
lemas_descarte_v3 = taxas_lemas_quadrante_v3.loc[vindos_do_descarte_v3].copy()
lemas_descarte_v3.insert(0, "grupo_versao_3", [LEMA_GRUPO_V3[l] for l in vindos_do_descarte_v3])
lemas_descarte_v3.insert(1, "motivo_do_descarte_na_versao_1", [MOTIVO_DESCARTE_V1[l] for l in vindos_do_descarte_v3])
lemas_descarte_v3["media_dos_planos"] = lemas_descarte_v3[PLANOS].mean(axis=1)
lemas_descarte_v3 = lemas_descarte_v3.sort_values(["grupo_versao_3", "media_dos_planos"], ascending=[True, False])
lemas_descarte_v3.index.name = "lema"
display(Markdown(f"**(a1) Os {len(vindos_do_descarte_v3)} lemas que a versão 3 leva aos eixos e que a versão 1 "
                 f"deixava no descarte, com o motivo da seção 15.1 e a taxa por 1.000 tokens em cada plano**"))
exibir(lemas_descarte_v3)
salvar_tabela(lemas_descarte_v3, "03_quadrante_v3_criticas_lemas_do_descarte")

peso_descarte_v3 = (lemas_descarte_v3.groupby("grupo_versao_3")[PLANOS].sum()
                    .reindex(GRUPOS_ATIVOS_V3).fillna(0.0))
parte_descarte_v3 = peso_descarte_v3 / PESO_GRUPOS_V3.loc[GRUPOS_ATIVOS_V3, PLANOS] * 100
tab_a2 = pd.concat({"peso_por_1000_tokens": peso_descarte_v3, "percentual_do_grupo": parte_descarte_v3}, axis=1)
tab_a2.index = [f"{g} ({descrever_polo(g).split(' (')[0]})" for g in tab_a2.index]
tab_a2.index.name = "grupo"
display(Markdown("**(a2) Peso desses lemas em cada grupo e plano, e a parte do grupo que eles representam (%)**"))
exibir(tab_a2, 1)
salvar_tabela(tab_a2, "03_quadrante_v3_criticas_peso_do_descarte")

# (b) Alternativas: a versão 3 com uma escolha do codebook desfeita de cada vez
n_lemas_grupo_v3 = pd.Series({g: len(c["lemas"]) for g, c in GRUPOS_QUADRANTE_V3.items()})
pesos_media_v3 = PESO_GRUPOS_V3.div(n_lemas_grupo_v3, axis=0)
ALTERNATIVAS_V3 = {
    "versao_3": coordenadas_do_codebook_v3(LEMA_GRUPO_V3),
    "sem_os_lemas_do_descarte_da_v1": coordenadas_do_codebook_v3(
        {**LEMA_GRUPO_V3, **{l: GRUPO_DESCARTADO for l in vindos_do_descarte_v3}}),
    "programme_com_program": coordenadas_do_codebook_v3({**LEMA_GRUPO_V3, "programme": LEMA_GRUPO_V3["program"]}),
    "media_por_lema": pd.DataFrame({
        "X": indice_balanco(pesos_media_v3.loc[EIXO_X[0], PLANOS], pesos_media_v3.loc[EIXO_X[1], PLANOS]),
        "Y": indice_balanco(pesos_media_v3.loc[EIXO_Y[0], PLANOS], pesos_media_v3.loc[EIXO_Y[1], PLANOS]),
        "massa": np.nan}, index=pd.Index(PLANOS, name="plano")),
    "como_na_imagem_do_comando": coordenadas_do_codebook_v3(lema_grupo_imagem_v3),
}
partes_v3 = {}
for nome, c in ALTERNATIVAS_V3.items():
    partes_v3[nome] = c[["X", "Y"]].assign(quadrante=[quadrante_de(a, b) for a, b in zip(c["X"], c["Y"])])
alternativas_quadrante_v3 = pd.concat(partes_v3, axis=1)
alternativas_quadrante_v3[("todas", "muda_de_quadrante")] = [
    "sim" if alternativas_quadrante_v3.loc[s, (slice(None), "quadrante")].nunique() > 1 else "não" for s in PLANOS]
display(Markdown("**(b) Alternativas: coordenadas da versão 3 e com uma escolha desfeita de cada vez** "
                 "(sem os lemas da tabela a1; *programme* no grupo de *program*; peso do grupo dividido pelo número "
                 "de lemas do grupo; os quatro lemas no grupo em que estavam na imagem do comando)"))
exibir(alternativas_quadrante_v3)
salvar_tabela(alternativas_quadrante_v3, "03_quadrante_v3_criticas_alternativas")

# (c) Influência de cada lema: deslocamento de cada bolha se o lema fosse para o vocabulário comum
N_INFLUENCIA_EXIBIDOS_V3 = 15
linhas = []
for lema in taxas_lemas_quadrante_v3.index[grupo_do_lema_v3.isin(GRUPOS_ATIVOS_V3)]:
    pesos = PESO_GRUPOS_V3.copy()
    pesos.loc[LEMA_GRUPO_V3[lema], PLANOS] -= taxas_lemas_quadrante_v3.loc[lema, PLANOS]
    dx = indice_balanco(pesos.loc[EIXO_X[0], PLANOS], pesos.loc[EIXO_X[1], PLANOS]) - coordenadas_quadrante_v3["X"].to_numpy()
    dy = indice_balanco(pesos.loc[EIXO_Y[0], PLANOS], pesos.loc[EIXO_Y[1], PLANOS]) - coordenadas_quadrante_v3["Y"].to_numpy()
    linhas.append({"lema": lema, "grupo": LEMA_GRUPO_V3[lema],
                   "no_descarte_da_v1": "sim" if lema in vindos_do_descarte_v3 else "não",
                   **dict(zip(PLANOS, np.hypot(dx, dy)))})
influencia_quadrante_v3 = pd.DataFrame(linhas).set_index("lema")
influencia_quadrante_v3["maior_deslocamento"] = influencia_quadrante_v3[PLANOS].max(axis=1)
influencia_quadrante_v3 = influencia_quadrante_v3.sort_values("maior_deslocamento", ascending=False)
display(Markdown(f"**(c) Influência de cada lema na versão 3: deslocamento de cada bolha se o lema fosse para o "
                 f"vocabulário comum** (os {N_INFLUENCIA_EXIBIDOS_V3} de maior efeito, entre "
                 f"{len(influencia_quadrante_v3)} lemas dos grupos {', '.join(GRUPOS_ATIVOS_V3)})"))
exibir(influencia_quadrante_v3.head(N_INFLUENCIA_EXIBIDOS_V3))
salvar_tabela(influencia_quadrante_v3, "03_quadrante_v3_criticas_influencia_por_lema")

# (d) Pares de lemas de sentido próximo (a lista abaixo) que a versão 3 põe em grupos diferentes
PARES_PROXIMOS_Q3 = [
    ("program", "programme", "grafias do mesmo lema"),
    ("adopt", "uptake", "adoção e difusão da IA"), ("adoption", "uptake", "adoção e difusão da IA"),
    ("deployment", "uptake", "adoção e difusão da IA"), ("application", "deployment", "adoção e difusão da IA"),
    ("assessment", "evaluation", "avaliação"),
    ("job", "occupation", "trabalho"), ("worker", "occupation", "trabalho"),
    ("education", "training", "formação de pessoas"), ("education", "skill", "formação de pessoas"),
    ("qualification", "skill", "formação de pessoas"),
    ("factory", "facility", "instalações"), ("center", "facility", "instalações"),
    ("startup", "incubate", "empreendedorismo"), ("startup", "hub", "empreendedorismo"),
    ("research", "breakthrough", "pesquisa e descobertas"),
    ("intelligent", "smart", "marcadores de “IA”"), ("ai", "ai-based", "marcadores de “IA”"),
    ("government", "public", "Estado e setor público"), ("government", "state", "Estado e setor público"),
    ("national", "brazilian", "autodesignação"), ("commission", "eu", "autodesignação"),
]
pares_v3 = pd.DataFrame([{"lema_1": a, "grupo_1": LEMA_GRUPO_V3[a], "lema_2": b, "grupo_2": LEMA_GRUPO_V3[b],
                          "tema": tema}
                         for a, b, tema in PARES_PROXIMOS_Q3 if LEMA_GRUPO_V3[a] != LEMA_GRUPO_V3[b]])
display(Markdown(f"**(d) Pares de lemas de sentido próximo que a versão 3 põe em grupos diferentes** "
                 f"({len(pares_v3)} pares)"))
exibir(pares_v3)

desloc_sem = np.hypot(partes_v3["sem_os_lemas_do_descarte_da_v1"]["X"] - partes_v3["versao_3"]["X"],
                      partes_v3["sem_os_lemas_do_descarte_da_v1"]["Y"] - partes_v3["versao_3"]["Y"])
print(f"Síntese factual: {len(vindos_do_descarte_v3)} lemas que a versão 1 deixava no descarte entram nos eixos da "
      f"versão 3; eles respondem por "
      + "; ".join(f"{g}: {num(parte_descarte_v3.loc[g].min(), 0)}% a {num(parte_descarte_v3.loc[g].max(), 0)}%"
                  for g in GRUPOS_ATIVOS_V3)
      + " do peso do grupo, conforme o plano. Sem eles, as bolhas se deslocam de "
      f"{num(desloc_sem.min())} ({desloc_sem.idxmin()}) a {num(desloc_sem.max())} ({desloc_sem.idxmax()}). Número de "
      f"lemas por grupo: " + ", ".join(f"{g} {n_lemas_grupo_v3[g]}" for g in GRUPOS_ATIVOS_V3)
      + f". Planos que mudam de quadrante em alguma alternativa: "
      + (", ".join(alternativas_quadrante_v3.index[alternativas_quadrante_v3[("todas", "muda_de_quadrante")] == "sim"])
         or "nenhum")
      + ". Lemas de maior influência: "
      + "; ".join(f"{l} ({r['grupo']}; {num(r['maior_deslocamento'])} em {r[PLANOS].astype(float).idxmax()})"
                  for l, r in influencia_quadrante_v3.head(5).iterrows())
      + f". Pares de lemas próximos em grupos diferentes: {len(pares_v3)}.")
```

**Críticas ao gráfico (versão 3).**

1. **Termos genéricos e autodesignações entram nos eixos.** Trinta e três lemas que a versão 1 deixava no descarte, pelos motivos registrados na seção 15.1, passam a contar (tabela 15.4.7a1): verbos e termos genéricos (*support*, *promote*, *increase*, *impact*, *challenge*, *solution*, *accelerate*, *build*, *integration*, *implementation*), autodesignações e nomes de órgãos (*commission*, *national*, *country*, *government*, *federal*, e *good*, que a versão 1 tratava como autodesignação por causa do título do PBIA, “AI for the Good of All”), marcadores de “IA” (*smart*, *ai-based*, *ai-driven*, *ai-native*) e termos de sentido variável (*sector*, *control*, *capacity*, *environment*, *coordination*, *collaboration*). Eles respondem, conforme o plano, por 29% a 53% do peso de B, 31% a 52% do de C e 16% a 34% do de A (tabela 15.4.7a2). Sem eles, as bolhas se deslocariam de 0,19 a 0,41: o AI Continent, o New Generation e o *AI+* passariam para o lado de D (X de 0,05 a 0,16), e o PBIA ficaria perto do centro (−0,01; 0,04) (tabela 15.4.7b). O deslocamento geral para a esquerda vem, portanto, sobretudo desses lemas, e não do vocabulário temático de governança.
2. **No eixo X, o peso de B nos planos da UE e no PBIA vem de autodesignação e de um termo genérico.** *sector* (6,16 e 8,34 por 1.000 tokens no AI Continent e no Apply AI; sobretudo “public sector”, “private sector” e “strategic sectors”) e *commission* (4,25 e 6,09; a Comissão Europeia, autora dos dois planos) estão entre os lemas de maior influência (tabela 15.4.7c); no PBIA, *national* e *country* (4,64 e 3,97) fazem o mesmo papel. Esses lemas registram quem escreve o plano e de que setor ele fala, mais do que o tema de governança.
3. **No eixo Y, o peso de C nos planos chineses vem de verbos de ação.** *promote* (12,18 por 1.000 tokens no *AI+* e 7,15 no New Generation) e *support* (5,57 e 6,18) estão entre os lemas que mais movem as bolhas (tabela 15.4.7c): o vocabulário de sociedade, inclusão, capital humano e ciência cresce com verbos que aparecem em qualquer diretriz.
4. **Os grupos têm tamanhos muito diferentes.** A tem 54 lemas, B 32, C 30 e D 19. Com 19 lemas, D tende a pesar menos que B em todos os planos, e o eixo X pende para a esquerda por construção. Com o peso de cada grupo dividido pelo número de lemas, as posições mudam muito (tabela 15.4.7b): o AI Continent iria para (−0,00; 0,01), o Apply AI passaria para baixo do eixo horizontal, e os dois planos chineses, para o lado de D. O zero de cada eixo não é neutro, e a leitura deve ser relativa.
5. **D, “Economia e mercado”, ficou estreito.** Saíram de G3 *innovation*, *application*, *efficiency*, *ecosystem*, *breakthrough*, *frontier*, *hub*, *incubate*, *uptake* e os lemas de ciência e talento (*research*, *science*, *scientific*, *talent*, *theory*); ficaram 19 lemas de mercado e produção. *industry* é, com *promote*, o lema de maior influência do gráfico (0,12 no *AI+*). O nome novo descreve melhor o grupo, mas o eixo X passa a opor um vocabulário amplo, que inclui autodesignações (B, 32 lemas), a um vocabulário restrito (D).
6. **Grafias e termos próximos em grupos diferentes (tabela 15.4.7d).** *program* está em C, e *programme*, a grafia britânica, em B, contra a regra da seção 15.1 (o efeito é pequeno: até 0,02 no AI Continent). Outros pares de sentido próximo ficam separados: *adopt*, *adoption* e *deployment* em D, e *uptake* no vocabulário comum; *assessment* em B, e *evaluation* no comum; *job* e *worker* em C, e *occupation* no comum; *education* e *qualification* em C, e *skill* e *training* em A; *factory* e *center* em A, e *facility* no comum; *startup* em D, *hub* em B (no AI Continent, sobretudo “innovation hubs”) e *incubate* no comum; *research* em C, e *breakthrough* em A; *smart* e *ai-based* em A, e *intelligent* e *ai* no comum; *national* e *commission* em B, e *brazilian* e *eu* no comum. *frontier*, em B, é sobretudo “frontier AI” e “frontier models”, isto é, tecnologia.
7. **A imagem e o codebook não coincidem.** Como registrado na seção 15.4.4, a imagem anexada corresponde a outro estado do site, com *industry*, *industrial*, *adopt* e *competitiveness* fora de D; nesse estado, D teria 15 lemas, e todas as bolhas iriam ainda mais para a esquerda (tabela 15.4.5d).

*Ajustes possíveis:* devolver ao vocabulário comum, no todo ou em parte, os lemas genéricos e as autodesignações da tabela 15.4.7a1; juntar *programme* a *program*; reunir os pares da tabela 15.4.7d; aproximar o número de lemas dos quatro grupos, ou comparar as posições com o peso médio por lema; e, antes de interpretar as posições, conferir na tabela 15.4.7c o efeito de cada lema que mudar de grupo.

### 15.5. Diferenças entre os planos, robustez e influência de cada lema

Diferenças entre pares de planos com intervalo de confiança (critério da skill 03, seção 3.5), coordenadas recalculadas com três alternativas de método e efeito de cada lema sobre as posições.

```python
# (a) Diferença entre dois planos em X e em Y, com intervalo de confiança por bootstrap. As reamostras
# de cada plano são independentes (seção 15.3); a diferença é calculada reamostra a reamostra. Uma
# diferença é afirmada apenas quando o intervalo não inclui zero (skill 03, seção 3.5).
linhas = []
for i, a in enumerate(PLANOS):
    for b in PLANOS[i + 1:]:
        linha = {"plano_a": a, "plano_b": b}
        for eixo in ("X", "Y"):
            dif = reamostras_quadrante[a][eixo] - reamostras_quadrante[b][eixo]
            inf, sup = np.quantile(dif, [ALFA_IC, 1 - ALFA_IC])
            linha[f"d{eixo}"] = coordenadas_quadrante.loc[a, eixo] - coordenadas_quadrante.loc[b, eixo]
            linha[f"d{eixo}_ic_inferior"], linha[f"d{eixo}_ic_superior"] = inf, sup
            linha[f"d{eixo}_ic_exclui_zero"] = "sim" if inf > 0 or sup < 0 else "não"
        linhas.append(linha)
diferencas_quadrante = pd.DataFrame(linhas).set_index(["plano_a", "plano_b"])
display(Markdown(f"**(a) Diferenças entre os planos (plano a − plano b), com intervalo de confiança de "
                 f"{num(NIVEL_CONFIANCA * 100, 0)}%**"))
exibir(diferencas_quadrante)
salvar_tabela(diferencas_quadrante, "03_quadrante_diferencas_entre_planos")

# (b) Robustez: as coordenadas recalculadas com três alternativas de método
fora_do_recorte = pd.DataFrame({s: ~taxas_lemas_quadrante.index.isin(recortes[s].index) for s in PLANOS},
                               index=taxas_lemas_quadrante.index)
formas_sem_correcao = pd.DataFrame(0.0, index=taxas_lemas_quadrante.index, columns=PLANOS)
for _, l in sem_correcao.iterrows():     # formas que a correção de outro notebook uniria a um lema (seção 11.1)
    if l["lema_de_destino"] in formas_sem_correcao.index and l["lema_sem_correcao"] not in LEMA_GRUPO:
        formas_sem_correcao.loc[l["lema_de_destino"], l["plano"]] += l["frequencia"] / N_TOKENS[l["plano"]] * 1000
n_lemas_grupo = pd.Series({g: len(c["lemas"]) for g, c in GRUPOS_QUADRANTE.items()})
VERSOES_QUADRANTE = {
    "principal": PESO_GRUPOS,
    "media_por_lema": PESO_GRUPOS.div(n_lemas_grupo, axis=0),
    "so_recorte_de_75": taxas_lemas_quadrante.mask(fora_do_recorte, 0.0).groupby(grupo_do_lema).sum(),
    "com_formas_sem_correcao": (taxas_lemas_quadrante + formas_sem_correcao).groupby(grupo_do_lema).sum(),
}
partes = {}
for versao, pesos in VERSOES_QUADRANTE.items():
    x = indice_balanco(pesos.loc[EIXO_X[0], PLANOS], pesos.loc[EIXO_X[1], PLANOS])
    y = indice_balanco(pesos.loc[EIXO_Y[0], PLANOS], pesos.loc[EIXO_Y[1], PLANOS])
    partes[versao] = pd.DataFrame({"X": x, "Y": y, "quadrante": [quadrante_de(a, b) for a, b in zip(x, y)]},
                                  index=PLANOS)
robustez_quadrante = pd.concat(partes, axis=1)
robustez_quadrante[("todas", "muda_de_quadrante")] = [
    "sim" if robustez_quadrante.loc[s, (slice(None), "quadrante")].nunique() > 1 else "não" for s in PLANOS]
robustez_quadrante.index.name = "plano"
display(Markdown("**(b) Robustez: coordenadas com a regra principal e com três alternativas**"))
exibir(robustez_quadrante)
salvar_tabela(robustez_quadrante, "03_quadrante_robustez")

# (c) Influência de cada lema: quanto cada bolha se deslocaria se o lema fosse para o descarte
# (distância entre a posição atual e a nova, em unidades dos eixos). Serve de apoio às edições do codebook.
N_INFLUENCIA_EXIBIDOS = 15
linhas = []
for lema in taxas_lemas_quadrante.index[grupo_do_lema.isin(GRUPOS_ATIVOS)]:
    pesos = PESO_GRUPOS.copy()
    pesos.loc[LEMA_GRUPO[lema], PLANOS] -= taxas_lemas_quadrante.loc[lema, PLANOS]
    dx = indice_balanco(pesos.loc[EIXO_X[0], PLANOS], pesos.loc[EIXO_X[1], PLANOS]) - coordenadas_quadrante["X"].to_numpy()
    dy = indice_balanco(pesos.loc[EIXO_Y[0], PLANOS], pesos.loc[EIXO_Y[1], PLANOS]) - coordenadas_quadrante["Y"].to_numpy()
    linhas.append({"lema": lema, "grupo": LEMA_GRUPO[lema], **dict(zip(PLANOS, np.hypot(dx, dy)))})
influencia_quadrante = pd.DataFrame(linhas).set_index("lema")
influencia_quadrante["maior_deslocamento"] = influencia_quadrante[PLANOS].max(axis=1)
influencia_quadrante = influencia_quadrante.sort_values("maior_deslocamento", ascending=False)
display(Markdown(f"**(c) Influência de cada lema: deslocamento de cada bolha se o lema fosse para o descarte** "
                 f"(os {N_INFLUENCIA_EXIBIDOS} lemas de maior efeito, entre {len(influencia_quadrante)} dos grupos "
                 f"{', '.join(GRUPOS_ATIVOS)})"))
exibir(influencia_quadrante.head(N_INFLUENCIA_EXIBIDOS))
salvar_tabela(influencia_quadrante, "03_quadrante_influencia_por_lema")

d = diferencas_quadrante
n_x, n_y = (d["dX_ic_exclui_zero"] == "sim").sum(), (d["dY_ic_exclui_zero"] == "sim").sum()
desvios = {v: (partes[v][["X", "Y"]] - partes["principal"][["X", "Y"]]).abs().to_numpy().max()
           for v in partes if v != "principal"}
print(f"Síntese factual: das {len(d)} comparações entre pares de planos, o intervalo da diferença exclui zero "
      f"em {n_x} no eixo X e em {n_y} no eixo Y. Maior mudança de coordenada em relação à regra principal: "
      + "; ".join(f"{v}: {num(m)}" for v, m in desvios.items())
      + f". Planos que mudam de quadrante em alguma alternativa: "
      + (", ".join(robustez_quadrante.index[robustez_quadrante[("todas", "muda_de_quadrante")] == "sim"]) or "nenhum")
      + ". Lemas de maior influência: "
      + "; ".join(f"{l} ({r['grupo']}; {num(r['maior_deslocamento'])} em {r[PLANOS].astype(float).idxmax()})"
                  for l, r in influencia_quadrante.head(5).iterrows()) + ".")
```

**Método e escolhas.**

1. **Codebook.** Os 243 lemas distintos dos recortes de 75 dos seis planos (seção 12, com os empates incluídos) foram classificados em cinco grupos, conforme a proposta do pesquisador revista com ele em 2026-10-05. A revisão levou ao descarte (G1) os marcadores de "IA" (*intelligent*, *smart*, *intelligentize*, *ai-based*, *ai-driven*, *ai-native*), a autodesignação e os nomes de órgãos e dos próprios planos (*apply*, *commission*, *country*, *federal*, *good*, *government*, *national*, *plan*, *state*, *strategy*), verbos e termos genéricos (*accelerate*, *benefit*, *capability*, *challenge*, *ensure*, *help*, *include*, *increase*, *program*, *promote*, *scenario*) e lemas com sentidos de grupos diferentes conforme o plano (*basic*, *capacity*, *collaboration*, *control*, *coordination*, *environment*, *space*); levou *application*, *industrial* e *manufacturing* para G3 e *public* para G4. Entraram ainda três lemas de fora dos recortes, para tratar igualmente as grafias britânica e americana, como já ocorre com *center* e *centre*: *defense* em G4, ao lado de *defence* (11 ocorrências no New Generation, 6 nos EUA e 4 no PBIA), e *optimise* e *programme* em G1, ao lado de *optimize* e *program*. A comparação é sempre por lema exato, depois do pré-processamento da seção 11.
2. **Peso de cada grupo.** É a soma das taxas por 1.000 tokens dos lemas do grupo no plano, com as tabelas de frequência completas da seção 11 (com as exclusões do pesquisador nos EUA e no PBIA). A contagem é feita **no documento inteiro**: um lema do codebook conta em todos os planos, inclusive naqueles em que não está entre os 75 mais frequentes. O denominador é o total de tokens do plano, *stopwords* incluídas, como em toda a comparação (skill 03, seção 3.3).
3. **Posição das bolhas.** A mesma regra da figura original: X = (G3 − G4)/(G3 + G4) e Y = (G2 − G5)/(G2 + G5), índices de balanço entre dois polos, de −1 a +1, iguais a zero no equilíbrio do par; a área da bolha é proporcional à massa temática (G2 + G3 + G4 + G5), na escala da figura original (6 pontos² por unidade). Como o numerador e o denominador de X e de Y vêm dos mesmos tokens, os índices não dependem da extensão do plano; a massa, que é uma taxa, também é comparável entre planos.
4. **Incerteza.** Intervalos de confiança de 95% por *bootstrap* por unidades (2.000 reamostras por plano), com as funções e a semente por plano dos notebooks individuais (`taxas_bootstrap` e `semente_documento`, a partir de `SEMENTE = 2026`). A diferença entre dois planos, em X ou em Y, é afirmada apenas quando o intervalo de confiança da diferença não inclui zero (skill 03, seção 3.5).
5. **Robustez.** As coordenadas foram recalculadas com três alternativas: (i) o peso do grupo dividido pelo número de lemas do grupo (`media_por_lema`), como na verificação de robustez da figura original; (ii) só os lemas do recorte de 75 do próprio plano (`so_recorte_de_75`), regra de contagem da figura original, que usava os 50 termos mais frequentes de cada documento; (iii) com as formas que a correção de lema de outro notebook uniria a um lema do codebook (seção 11.1) somadas a esse lema (`com_formas_sem_correcao`). A tabela 15.5c mostra quanto cada bolha se deslocaria se cada lema fosse para o descarte.
6. **Figura.** Reproduz a figura original: título, subtítulo, eixos de −1 a +1 cruzados na origem, referências em ±0,5, as quatro caixas dos vocabulários (A no topo, B à esquerda, C na base, D à direita), bolhas com halo branco e borda escura, painel com a legenda de documentos e a de massa temática, e nota com as fórmulas. Por comando do pesquisador, as cores são as da figura original, uma por país ou bloco (os dois planos da UE em amarelo e os dois da China em vermelho), e não os tons de `COR_DOCUMENTO`, que distinguem o segundo plano de cada bloco. O PBIA tem anel tracejado e rótulo em negrito, como na figura original; o destaque é apenas visual, e o procedimento é o mesmo para todos os planos. Os anos da legenda seguem a data registrada em cada JSON: o PBIA aparece como 2025 (a figura original trazia 2024). As caixas dos polos trazem os nomes dos grupos definidos na seção 15.1. A célula confere, na própria figura, se cada bolha está na posição calculada e se rótulos, bolhas, painel e blocos de texto não se sobrepõem.
7. **Saídas.** Com `SALVAR_SAIDAS = False`, nada é gravado em `resultados/`; a figura (PNG em 300 dpi e PDF) e as tabelas são gravadas apenas por comando do pesquisador.

| Aspecto | Figura original (`notebooks/ Novo/quadrantes`) | Esta versão |
|---|---|---|
| Vocabulário de partida | 50 termos mais frequentes de cada documento, com outros JSONs e outro pré-processamento | Recortes de 75 lemas da seção 12 (só unigramas), com o pipeline e as exclusões desta pasta |
| Grupos | A: infraestrutura, computação e base material; B: Estado, governança e regulação; C: ciência, dimensão social, capital humano, inclusão e sustentabilidade; D: economia, indústria, difusão de mercado e competição geopolítica | A: infraestrutura e tecnologias de IA; B: governança, regulação e segurança; C: sociedade e capital humano; D: economia, ciência e inovação |
| Base da contagem | Só os termos presentes no top-50 do documento | Documento inteiro |
| Fórmulas, área da bolha e desenho | X, Y, massa e escala acima | As mesmas |
| Incerteza | Não havia | Intervalos de confiança por *bootstrap* por unidades |

**Como ler.** Cada bolha é um plano; a cor indica o país ou bloco, e a área, a massa temática do texto. Na horizontal, à direita, o vocabulário do plano pende para G3 (*Economia, ciência e inovação*) em relação a G4 (*Governança, regulação e segurança*); à esquerda, para G4. Na vertical, acima, pende para G2 (*Infraestrutura e tecnologias de IA*) em relação a G5 (*Sociedade e capital humano*); abaixo, para G5. O cruzamento dos eixos é o equilíbrio de cada par. Os índices comparam dois grupos **dentro** do mesmo plano: um plano mais à direita usa G3 mais em relação a G4, e não necessariamente mais G3 que outro plano (os pesos estão na tabela 15.3a). A distância entre duas bolhas só indica diferença quando o intervalo de confiança da diferença não inclui zero (tabela 15.5a). O zero dos eixos depende de quantos e de quais lemas compõem cada grupo, por isso as posições devem ser lidas umas em relação às outras.

**Interpretação.** *Observação:*

- **Eixo X (G3 × G4).** Cinco planos têm X positivo: New Generation 0,45 [IC 95%: 0,35; 0,56], *AI+* 0,39 [0,14; 0,60], AI Continent 0,31 [0,18; 0,43], Apply AI 0,23 [0,08; 0,37] e PBIA 0,16 [0,06; 0,25]. Nos EUA, X = −0,02 [−0,13; 0,10]: G3 e G4 pesam quase o mesmo (28,42 e 29,76 por 1.000 tokens), e o intervalo inclui zero. O X dos EUA é menor que o de cada um dos outros cinco planos, com intervalo da diferença que exclui zero em todos os casos (de −0,18 [−0,32; −0,03], contra o PBIA, a −0,48 [−0,63; −0,32], contra o New Generation). Entre os outros cinco, só duas diferenças em X atendem ao critério: o New Generation fica à direita do PBIA (0,30 [0,16; 0,44]) e do Apply AI (0,23 [0,05; 0,40]).
- **Eixo Y (G2 × G5).** Cinco planos têm Y entre 0,35 e 0,56: New Generation 0,56 [0,43; 0,66], AI Continent 0,53 [0,38; 0,65], EUA 0,49 [0,33; 0,66], Apply AI 0,36 [0,19; 0,52] e *AI+* 0,35 [0,10; 0,59]. No PBIA, Y = −0,01 [−0,10; 0,09]: G2 e G5 pesam quase o mesmo (26,29 e 26,78), e o intervalo inclui zero. O Y do PBIA é menor que o de cada um dos outros cinco, com intervalo da diferença que exclui zero em todos os casos (de 0,36 [0,10; 0,61], com o *AI+*, a 0,57 [0,41; 0,71], com o New Generation). Entre os outros cinco, só uma diferença em Y atende ao critério: o New Generation fica acima do Apply AI (0,20 [0,01; 0,40]).
- **Quadrantes.** Quatro planos ficam no quadrante superior direito (os dois da UE e os dois da China); os EUA, no superior esquerdo; o PBIA, no inferior direito. Como os intervalos do X dos EUA e do Y do PBIA incluem zero, esses dois planos estão na fronteira entre quadrantes, e o lado em que caem não é determinado pelos dados.
- **Massa temática.** O *AI+* (160,06 [145,62; 176,03]) e o New Generation (157,27 [147,10; 167,21]) têm a maior massa; os outros quatro ficam entre 104,14 (PBIA) e 119,47 (AI Continent), com intervalos que não se sobrepõem aos dos dois planos chineses.
- **De onde vêm os pesos (tabela 15.3b).** Nos EUA, G4 reúne sobretudo *security* (3,66), *critical* (2,07), *risk* e *standard* (1,95 cada) e *export* (1,83), e G5 é o menor dos seis planos (11,95), com *workforce* e *worker* à frente. No PBIA, G5 é o maior dos seis (26,78) e o mais distribuído: o lema de maior peso, *service* (2,75), responde por 10% do grupo, seguido de *health* (2,68), *education* (2,38), *reduction* (2,26) e *training* (2,14); em G2, *data* responde por 22% (5,67). No AI Continent, *data* é 23% de G2 (10,08), e *public* e *act* (sobretudo *AI Act*) somam 46% de G4. Nos planos chineses, *application* lidera G3 no *AI+* (11,13; 18%) e, no New Generation, vem depois de *research* (7,85); em G2, pesam *model* no *AI+* (9,05) e *system* (10,33) e *platform* (6,18) no New Generation; em G5, *service* é 39% do grupo no *AI+* e 31% no New Generation.
- **Robustez (tabela 15.5b).** Nenhum plano muda de quadrante quando a contagem se restringe ao recorte de 75 do próprio plano (mudança máxima de 0,20) ou quando se somam as formas sem correção de lema (0,01). Com o peso dividido pelo número de lemas do grupo, o PBIA passa ao quadrante inferior esquerdo (X = −0,02; Y = −0,28), e todos os Y diminuem. Nas quatro versões, os EUA têm o menor X, e o PBIA, o menor Y.
- **Influência (tabela 15.5c).** Os lemas cujo descarte mais deslocaria uma bolha são *service* (0,20 no *AI+*), *public* (0,16 no PBIA), *data* (0,12 no PBIA), *act* (0,10 no AI Continent) e *model* (0,10 no *AI+*).

*Inferência:* com este codebook, o PBIA é o único dos seis planos em que o vocabulário de infraestrutura e tecnologias de IA e o de sociedade e capital humano têm pesos equivalentes; nos outros cinco, o primeiro predomina. Os EUA são o único plano em que o vocabulário de economia, ciência e inovação e o de governança, regulação e segurança se equilibram, com G4 apoiado sobretudo em segurança, risco, padrões e controle de exportações; nos outros cinco, predomina o primeiro. Os planos da UE e da China não se separam uns dos outros nesses índices, salvo o New Generation em relação ao Apply AI. As posições medem pesos lexicais, e não prioridades políticas: os planos diferem em gênero (plano de ação, estratégia, diretrizes do Conselho de Estado, plano de desenvolvimento), em extensão (de 2.874 a 16.391 tokens) e em data (2017 e 2024–2025), e essas diferenças afetam o vocabulário.

**Fragilidades e ajustes possíveis.**

- **O codebook é uma classificação interpretativa.** As posições dependem da atribuição de cada lema a um grupo, e vários lemas mudam de sentido conforme o plano: *service* é sobretudo *public services* no PBIA e *service platforms* e *service robots* no New Generation; *system* é sobretudo *AI systems* nos EUA e no PBIA e *unmanned systems* no New Generation. Poucos lemas carregam parcelas grandes de um grupo, como mostra a tabela 15.5c. *public*, levado a G4 na revisão, é 29% de G4 no PBIA e 25% no AI Continent: com *public* em G5, como na proposta inicial, o PBIA iria para (0,32; −0,11), e o AI Continent, para (0,43; 0,39).
- **O zero não é neutro.** Os grupos têm tamanhos diferentes (35, 33, 23 e 20 lemas de G2 a G5), e o equilíbrio de um par depende de quantos e de quais lemas o compõem: com o peso dividido pelo número de lemas, todos os Y diminuem de 0,22 a 0,27, e o PBIA muda de quadrante. Por isso, a leitura deve ser relativa. Centrar os eixos na média dos seis planos é um ajuste possível; não foi aplicado, para manter a regra de posicionamento da figura original.
- **Base da contagem.** Restringir a contagem ao recorte de 75 de cada plano, como fazia a figura original com os 50 termos mais frequentes, desloca as posições em até 0,20, porque um lema do codebook fora do recorte de um plano passa a valer zero nesse plano.
- **Plano curto.** O *AI+* (2.874 tokens, 65 unidades) tem os intervalos mais largos (X de 0,14 a 0,60; Y de 0,10 a 0,59), e a sua posição é a menos precisa.
- **Nomes e exclusões.** Os nomes de órgãos e do país foram retirados da contagem nos EUA e no PBIA, e não na UE e na China (seção 11.1); no quadrante, os nomes desse tipo que estão nos recortes ficam no descarte e não entram nos eixos. Permanecem nos grupos ativos nomes de leis e iniciativas que designam objetos dos próprios grupos, como *AI Act* (em *act*, G4) e *AI Factories*, *AI Gigafactories* e EuroHPC (em G2).
- **Pré-processamento.** Seis lemas dos grupos ativos são sensíveis às diferenças de correção de lema entre os notebooks (seção 15.2: *center*, *gigafactory*, *network*, *robotics*, *service* e *skill*). Somar as formas não unidas a eles muda as coordenadas em no máximo 0,01.
- **Tradução.** Os dois planos chineses são traduções para o inglês, e a escolha de palavras da tradução afeta as contagens (por exemplo, *intelligent* e *smart*, que estão no descarte).
- **Cores.** Por comando do pesquisador, os dois planos de um mesmo bloco têm a mesma cor e são distinguidos pelos rótulos. Para usar as cores da pasta, basta trocar `CORES_QUADRANTE` por `dict(COR_DOCUMENTO)` na seção 15.1.

### 15.6. Registro desta seção

Parâmetros da seção 15, *hash* do codebook (identifica a versão dos grupos usada em cada execução) e saídas gravadas. O registro geral do notebook está na seção 14.

```python
codebook_ordenado = {g: sorted(l.strip().lower() for l in c["lemas"]) for g, c in GRUPOS_QUADRANTE.items()}
registro_quadrante = {
    "secao": "15. Perfis de vocabulário: gráfico de quadrantes",
    "executado_em": datetime.now().astimezone().isoformat(timespec="seconds"),
    "codebook_sha256": hashlib.sha256(json.dumps(codebook_ordenado, ensure_ascii=False, sort_keys=True)
                                      .encode("utf-8")).hexdigest(),
    "grupos": {g: {"nome": c["nome"], "lemas": len(c["lemas"])} for g, c in GRUPOS_QUADRANTE.items()},
    "grupo_descartado": GRUPO_DESCARTADO,
    "eixos": {"X": f"({EIXO_X[0]} − {EIXO_X[1]})/({EIXO_X[0]} + {EIXO_X[1]})",
              "Y": f"({EIXO_Y[0]} − {EIXO_Y[1]})/({EIXO_Y[0]} + {EIXO_Y[1]})"},
    "base_da_contagem": "documento inteiro: tabelas de frequência da seção 11, com as exclusões do pesquisador",
    "lemas_dos_recortes_de_75": len(LEMAS_RECORTES),
    "lemas_de_fora_dos_recortes": fora_dos_recortes,
    "bootstrap": {"N_BOOTSTRAP": N_BOOTSTRAP, "NIVEL_CONFIANCA": NIVEL_CONFIANCA, "SEMENTE": SEMENTE,
                  "semente_por_plano": "SeedSequence([SEMENTE, crc32(sigla)]), como nos notebooks individuais"},
    "figura": {"cores": CORES_QUADRANTE, "destaque": DESTAQUE_QUADRANTE,
               "escala_bolha": ESCALA_BOLHA_QUADRANTE, "referencias_massa": REFERENCIAS_MASSA_QUADRANTE,
               "avisos_das_conferencias": avisos_quadrante},
}
if SALVAR_SAIDAS:
    (DIR_SAIDA / "registro_quadrante.json").write_text(
        json.dumps(registro_quadrante | {"codebook": codebook_ordenado}, ensure_ascii=False, indent=2),
        encoding="utf-8")
    SAIDAS_GRAVADAS.append("registro_quadrante.json")
    registro_quadrante["saidas_gravadas"] = sorted(s for s in SAIDAS_GRAVADAS if s.startswith(("03_", "registro_q")))
else:
    registro_quadrante["saidas_gravadas"] = "nenhuma (SALVAR_SAIDAS = False)"
print(json.dumps(registro_quadrante, ensure_ascii=False, indent=2, default=str))
```
