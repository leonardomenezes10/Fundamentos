## 4. Por que a comparação é relativa

Os planos têm **tamanhos muito diferentes** (seção 11). Comparar frequências absolutas favoreceria o texto mais longo. Por isso, toda comparação usa **ocorrências por 1.000 tokens**:

> **taxa por 1.000 tokens = (ocorrências do termo no plano ÷ total de tokens do plano) × 1.000**

Nos blocos e no conjunto, os planos são juntados como **um único texto**, como pede o comando ("os documentos juntos"):

> **taxa do grupo = (ocorrências somadas nos planos do grupo ÷ tokens somados dos planos) × 1.000**

O denominador é o total de tokens (palavras), incluindo *stopwords* e as palavras retiradas pelas exclusões do pesquisador, como nos notebooks individuais. Com essa normalização, uma taxa de 5 por 1.000 tem o mesmo significado em um plano de 3.000 tokens e em um grupo de 39.000. Valores absolutos aparecem apenas nas tabelas, como informação auxiliar.

```python
# ===========================================================================
# PIPELINE DE PRÉ-PROCESSAMENTO E MEDIDAS
# Mesmo pipeline dos notebooks individuais da pasta. As poucas escolhas que os notebooks
# individuais fizeram para o seu plano (ajustes de tokenização e correções de lema) entram
# como parâmetros por plano, definidos na configuração (seção 8). As funções são as do notebook
# comparativo TESTE ANÁLISE PLANOS.ipynb, cujo pipeline reproduz, plano a plano, a tabela de
# tokens e a tabela de frequências de cada notebook individual; aqui, a seção 11.1 confere os
# recortes ampliados de cada plano.
# A função tabela_bigramas é a dos notebooks individuais (seção 14), sem alteração.
# ===========================================================================

HIFENS = {"-", "‐", "‑"}

# Valores admitidos nos campos usados como filtro (skill 01, seção 4.3).
TIPOS_VALIDOS = {"heading", "paragraph", "list_item"}
CATEGORIAS_VALIDAS = {"Foreword", "Executive summary", "Body", "Box", "Annex"}

COLUNAS_TOKENS = ["unidade", "frase", "i", "posicao", "forma", "lema_spacy", "lema", "classe", "etiqueta"]


def compilar_re_palavra(siglas_com_e_comercial):
    """Token (palavra) = sequência de letras, admitindo hífen interno (multi-stakeholder), sigla
    com pontos (U.S.) e, se siglas_com_e_comercial = True (notebooks dos planos chineses), sigla
    com "&" (R&D). Números, pontuação e símbolos NÃO são tokens."""
    return re.compile(
        r"(?:[^\W\d_]\.){2,}"
        + (r"|[^\W\d_]+(?:&[^\W\d_]+)+" if siglas_com_e_comercial else "")
        + r"|[^\W\d_]+(?:[-‐‑][^\W\d_]+)*"
    )


def carregar_modelo(separar_simbolo_final):
    """Carrega o modelo do spaCy. Se separar_simbolo_final = True (notebooks dos planos chineses),
    separa do fim da palavra o "+" (AI+) e o hífen suspenso ("intelligentization- and"), que o
    tokenizador mantém colados: a palavra é contada e o símbolo não."""
    nlp = spacy.load(MODELO_SPACY)
    if separar_simbolo_final:
        sufixos = list(nlp.Defaults.suffixes) + [r"(?<=[^\W\d_])\+", r"(?<=[^\W\d_])[-‐‑]"]
        nlp.tokenizer.suffix_search = compile_suffix_regex(sufixos).search
    return nlp


def normalizar_hifen(texto):
    """Uniformiza variantes tipográficas do hífen (apenas nas chaves de contagem;
    o texto do JSON não é alterado)."""
    return texto.replace("‐", "-").replace("‑", "-")


def carregar_extracao(caminho):
    """Lê o JSON produzido pela skill 01 e devolve SOMENTE o título, o subtítulo e as
    unidades (id, tipo, categoria e texto integral). Todo o conteúdo periférico (number,
    level, page, path e demais metadados) é descartado aqui.
    Interrompe a execução se o arquivo não passar nas verificações de integridade."""
    if not caminho.exists():
        raise FileNotFoundError(
            f"Arquivo '{caminho.name}' não encontrado na pasta. "
            "Execute antes a skill 01 (skill_extração.md) para gerar o JSON do plano."
        )
    bruto = caminho.read_bytes()
    dados = json.loads(bruto.decode("utf-8"))

    problemas = [f"chave '{c}' ausente" for c in ("document", "extraction", "sections") if c not in dados]
    if problemas:
        raise ValueError(f"Falha na verificação de '{caminho.name}':\n- " + "\n- ".join(problemas))

    extracao, documento, secoes = dados["extraction"], dados["document"], dados["sections"]
    if extracao.get("skill") != "skill_extração":
        problemas.append("o JSON não registra ter sido produzido pela skill 01 (extraction.skill)")
    if extracao.get("corpus") != "Planos Nacionais de IA":
        problemas.append("o JSON não registra o corpus 'Planos Nacionais de IA' (extraction.corpus)")
    if [s.get("id") for s in secoes] != list(range(1, len(secoes) + 1)):
        problemas.append("os ids das unidades não são sequenciais a partir de 1")
    vazias = [s.get("id") for s in secoes if not isinstance(s.get("text"), str) or not s["text"].strip()]
    if vazias:
        problemas.append(f"unidades com texto vazio: {vazias}")
    if extracao.get("units_extracted") != len(secoes):
        problemas.append(
            f"extraction.units_extracted = {extracao.get('units_extracted')}, mas há {len(secoes)} unidades"
        )
    tipos = {s.get("type") for s in secoes} - TIPOS_VALIDOS
    if tipos:
        problemas.append(f"tipos de unidade não previstos pela skill 01: {sorted(map(str, tipos))}")
    categorias = {s.get("category") for s in secoes} - CATEGORIAS_VALIDAS
    if categorias:
        problemas.append(f"categorias não previstas pela skill 01: {sorted(map(str, categorias))}")
    if not documento.get("title"):
        problemas.append("título (document.title) ausente")
    if documento.get("language") != IDIOMA_ESPERADO:
        problemas.append(
            f"idioma declarado no JSON: {documento.get('language')!r}; esperado: {IDIOMA_ESPERADO!r}"
        )
    if problemas:
        raise ValueError(f"Falha na verificação de '{caminho.name}':\n- " + "\n- ".join(problemas))

    return {
        "arquivo": caminho.name,
        "sha256": hashlib.sha256(bruto).hexdigest(),
        "titulo": documento["title"],
        "subtitulo": documento.get("subtitle"),
        "idioma": documento.get("language"),
        # SOMENTE id (posição), tipo e categoria (filtros) e texto (conteúdo analisado)
        "unidades": [
            {"id": s["id"], "tipo": s["type"], "categoria": s["category"], "texto": s["text"]}
            for s in secoes
        ],
    }


def selecionar_unidades(documento, unidades_excluidas):
    """Monta as unidades analisadas a partir dos parâmetros de recorte da configuração. O título
    e o subtítulo formam a unidade 0, separados por uma quebra de linha, que impede a junção da
    última palavra do título com a primeira do subtítulo sem acrescentar pontuação ao texto.
    Cada unidade retirada é atribuída ao primeiro filtro que a retira. Nenhum texto é alterado."""
    inexistentes = sorted(set(unidades_excluidas) - {u["id"] for u in documento["unidades"]})
    if inexistentes:
        raise ValueError(f"UNIDADES_EXCLUIDAS contém ids que não existem no JSON: {inexistentes}")
    sem_motivo = [i for i, motivo in unidades_excluidas.items() if not str(motivo).strip()]
    if sem_motivo:
        raise ValueError(f"Registre o motivo da exclusão das unidades {sem_motivo} em UNIDADES_EXCLUIDAS.")
    categorias_invalidas = sorted(set(CATEGORIAS_EXCLUIDAS) - CATEGORIAS_VALIDAS)
    if categorias_invalidas:
        raise ValueError(f"CATEGORIAS_EXCLUIDAS contém categorias não previstas: {categorias_invalidas}")

    texto_0 = "\n".join(t for t in (documento["titulo"], documento["subtitulo"]) if t)
    candidatas = [{"id": 0, "tipo": "title", "categoria": None, "texto": texto_0}] + documento["unidades"]
    analisadas, retiradas = [], []
    for u in candidatas:
        if u["tipo"] == "title" and not INCLUIR_TITULO_NA_CONTAGEM:
            filtro = "INCLUIR_TITULO_NA_CONTAGEM"
        elif u["tipo"] == "heading" and not INCLUIR_TITULOS_INTERNOS:
            filtro = "INCLUIR_TITULOS_INTERNOS"
        elif u["categoria"] in CATEGORIAS_EXCLUIDAS:
            filtro = "CATEGORIAS_EXCLUIDAS"
        elif u["id"] in unidades_excluidas:
            filtro = "UNIDADES_EXCLUIDAS"
        else:
            analisadas.append(u)
            continue
        retiradas.append({**u, "filtro": filtro})
    return analisadas, retiradas


def montar_stopwords(nlp):
    """Lista de stopwords do spaCy, ajustada pelas listas editáveis da configuração."""
    base = {w.lower() for w in nlp.Defaults.stop_words}
    return (base | {w.lower() for w in STOPWORDS_ADICIONAIS}) - {w.lower() for w in STOPWORDS_MANTIDAS}


def unir_compostos_hifenizados(doc):
    """Reúne em um único token as palavras compostas com hífen (ex.: open-source, multi-stakeholder,
    human-centric, state-of-the-art), que o spaCy separa em vários tokens. Só une quando não há
    espaço em volta do hífen. Por isso, cada composto conta como um único unigrama."""
    trechos, i, n = [], 0, len(doc)
    while i < n:
        j = i
        while (j + 2 < n and doc[j].is_alpha and doc[j].whitespace_ == ""
               and doc[j + 1].text in HIFENS and doc[j + 1].whitespace_ == ""
               and doc[j + 2].is_alpha):
            j += 2
        if j > i:
            trechos.append(doc[i:j + 1])
            i = j + 1
        else:
            i += 1
    with doc.retokenize() as rt:
        for trecho in trechos:
            ultimo = trecho[-1]
            # Lema do composto = forma escrita; se o último elemento estiver no plural,
            # usa-se o lema dele (multi-stakeholders -> multi-stakeholder).
            if ultimo.tag_ in {"NNS", "NNPS"}:
                lema = trecho.text[: len(trecho.text) - len(ultimo.text)] + ultimo.lemma_
            else:
                lema = trecho.text
            rt.merge(trecho, attrs={"LEMMA": lema, "POS": ultimo.pos_, "TAG": ultimo.tag_})
    return doc


def processar_documento(unidades, nlp, stopwords, re_palavra, correcoes_de_lema):
    """Aplica o pré-processamento a cada unidade (título, parágrafo ou item) e devolve uma tabela
    com uma linha por token (palavra), na ordem original do documento. 'lema_spacy' é o lema dado
    pelo spaCy; 'lema' é o lema depois das correções explícitas do plano."""
    linhas, posicao, frases_por_unidade = [], 0, {}
    for unidade, doc in zip(unidades, nlp.pipe([u["texto"] for u in unidades])):
        doc = unir_compostos_hifenizados(doc)
        uid = unidade["id"]
        for frase in doc.sents:
            k = frases_por_unidade.get(uid, 0)
            frases_por_unidade[uid] = k + 1
            for tok in frase:
                if not re_palavra.fullmatch(tok.text):
                    continue
                lema_spacy = normalizar_hifen(tok.lemma_.lower())
                linhas.append({
                    "unidade": uid,
                    "frase": f"{uid}.{k}",
                    "i": tok.i,                      # índice na unidade (detecta adjacência)
                    "posicao": posicao,              # posição no documento inteiro
                    "forma": normalizar_hifen(tok.text.lower()),
                    "lema_spacy": lema_spacy,
                    "lema": correcoes_de_lema.get(lema_spacy, lema_spacy),
                    "classe": tok.pos_,
                    "etiqueta": tok.tag_,
                })
                posicao += 1
    tokens = pd.DataFrame(linhas, columns=COLUNAS_TOKENS)
    tokens["conteudo"] = (
        tokens["classe"].isin(CLASSES_DE_CONTEUDO)
        & ~tokens["lema"].isin(stopwords)
        & (tokens["lema"].str.len() >= TAMANHO_MINIMO_LEMA)
        & tokens["lema"].map(lambda lema: bool(re_palavra.fullmatch(lema)))
    ).astype(bool)
    return tokens


# ---------------------------------------------------------------------------
# Exclusões por comando do pesquisador (funções da seção 15 do notebook dos EUA, com o padrão
# de palavra do plano como parâmetro). Aplicadas à lista do PBIA, reproduzem exatamente o
# resultado da função da seção 16 do notebook do PBIA (verificado na construção deste notebook).
# ---------------------------------------------------------------------------
CATEGORIA_PROTEGIDA = "Expressão protegida (mantida na contagem)"


def formas_da_expressao(texto, nlp, re_palavra):
    """Sequência de formas (minúsculas) de uma expressão, com o mesmo pré-processamento
    do texto: tokenização do spaCy, reunião de compostos com hífen e padrão de palavra."""
    doc = unir_compostos_hifenizados(nlp(texto))
    return [normalizar_hifen(t.text.lower()) for t in doc if re_palavra.fullmatch(t.text)]


def preparar_exclusoes(exclusoes, nlp, re_palavra, protegidas=()):
    """Converte as expressões em padrões de formas, da mais longa para a mais curta.
    Palavras entre colchetes fazem parte do padrão, mas não são retiradas; as expressões
    protegidas são tratadas como inteiramente entre colchetes. Interrompe a execução se uma
    expressão ficar vazia ou repetir outra (as categorias são exclusivas)."""
    entradas, vistos = [], {}
    grupos = dict(exclusoes)
    grupos[CATEGORIA_PROTEGIDA] = [f"[{x}]" for x in protegidas]
    for categoria, expressoes in grupos.items():
        for expressao in expressoes:
            formas, retirar = [], []
            for parte in re.split(r"(\[[^\]]*\])", expressao):
                if not parte.strip():
                    continue
                f = formas_da_expressao(parte.strip("[]"), nlp, re_palavra)
                formas += f
                retirar += [not parte.startswith("[")] * len(f)
            chave = tuple(formas)
            if not chave:
                raise ValueError(f"A expressão '{expressao}' não contém nenhuma palavra.")
            if chave in vistos:
                raise ValueError(f"A expressão '{expressao}' repete '{vistos[chave]}'.")
            vistos[chave] = expressao
            entradas.append({"categoria": categoria, "expressao": expressao,
                             "formas": chave, "retirar": tuple(retirar)})
    return sorted(entradas, key=lambda e: -len(e["formas"]))


def aplicar_exclusoes(tokens, entradas):
    """Marca os tokens retirados da contagem de conteúdo. A busca percorre cada frase da
    esquerda para a direita; em cada posição, vale a expressão mais longa que coincide, e
    cada token é retirado no máximo uma vez. O total de tokens não muda.
    Devolve os tokens (com 'exclusao', 'expressao_excluida', 'conteudo_antes_das_exclusoes' e
    'conteudo' recalculado) e o número de ocorrências de cada expressão."""
    categoria = pd.Series(None, index=tokens.index, dtype=object)
    expressao = pd.Series(None, index=tokens.index, dtype=object)
    ocorrencias = {e["expressao"]: 0 for e in entradas}
    if entradas:
        for _, frase in tokens.groupby("frase", sort=False):
            indices, formas = frase.index.to_list(), frase["forma"].to_list()
            j = 0
            while j < len(formas):
                for e in entradas:
                    n = len(e["formas"])
                    if tuple(formas[j:j + n]) == e["formas"]:
                        for k in range(n):
                            if e["retirar"][k]:
                                categoria[indices[j + k]] = e["categoria"]
                                expressao[indices[j + k]] = e["expressao"]
                        ocorrencias[e["expressao"]] += 1
                        j += n
                        break
                else:
                    j += 1
    resultado = tokens.copy()
    resultado["exclusao"] = categoria
    resultado["expressao_excluida"] = expressao
    resultado["conteudo_antes_das_exclusoes"] = tokens["conteudo"]
    resultado["conteudo"] = tokens["conteudo"] & categoria.isna()
    return resultado, ocorrencias


# ---------------------------------------------------------------------------
# Frequências e recortes
# ---------------------------------------------------------------------------
def tabela_frequencias(tokens):
    """Frequência, taxa por 1.000 tokens, alcance e dispersão (DP normalizado) de
    TODOS os lemas de conteúdo. Ordenação: frequência decrescente; empate -> ordem alfabética."""
    n_tokens = len(tokens)
    unidades = sorted(tokens["unidade"].unique())
    proporcao_unidade = (tokens.groupby("unidade").size().reindex(unidades) / n_tokens).to_numpy()
    conteudo = tokens[tokens["conteudo"]]
    cruz = pd.crosstab(conteudo["lema"], conteudo["unidade"]).reindex(columns=unidades, fill_value=0)
    freq = cruz.sum(axis=1)
    # DP de Gries (2008), normalizado por Lijffijt & Gries (2012):
    # 0 = distribuição proporcional ao tamanho das unidades; 1 = concentração máxima.
    v = cruz.div(freq, axis=0).to_numpy()
    dp = 0.5 * np.abs(v - proporcao_unidade).sum(axis=1)
    dp_norm = dp / (1 - proporcao_unidade.min()) if len(unidades) > 1 else np.full(len(freq), np.nan)
    formas = conteudo.groupby("lema")["forma"].agg(
        lambda s: "; ".join(f"{forma} ({n})" for forma, n in s.value_counts().items())
    )
    tab = pd.DataFrame({
        "frequencia": freq,
        "por_1000_tokens": freq / n_tokens * 1000,
        "unidades_com_ocorrencia": (cruz > 0).sum(axis=1),
        "pct_unidades": (cruz > 0).sum(axis=1) / len(unidades) * 100,
        "DP_norm": dp_norm,
        "formas_observadas": formas,
    })
    tab.index.name = "lema"
    tab = tab.reset_index().sort_values(["frequencia", "lema"], ascending=[False, True])
    return tab.set_index("lema")


def tabela_bigramas(tokens):
    """Pares de tokens de conteúdo ADJACENTES no texto original (mesma frase, sem
    pontuação entre eles). Não se formam pares artificiais pela remoção de stopwords."""
    a = tokens.iloc[:-1].reset_index(drop=True)
    b = tokens.iloc[1:].reset_index(drop=True)
    ok = (
        (a["frase"].to_numpy() == b["frase"].to_numpy())
        & (b["i"].to_numpy() - a["i"].to_numpy() == 1)
        & a["conteudo"].to_numpy() & b["conteudo"].to_numpy()
    )
    pares = pd.DataFrame({
        "bigrama": (a["lema"] + " " + b["lema"])[ok],
        "forma": (a["forma"] + " " + b["forma"])[ok],
        "unidade": a["unidade"][ok],
    })
    if pares.empty:
        return pd.DataFrame(columns=["frequencia", "por_1000_tokens", "unidades_com_ocorrencia",
                                     "forma_principal", "formas_observadas"]).rename_axis("bigrama")
    grupos = pares.groupby("bigrama")
    tab = pd.DataFrame({
        "frequencia": grupos.size(),
        "por_1000_tokens": grupos.size() / len(tokens) * 1000,
        "unidades_com_ocorrencia": grupos["unidade"].nunique(),
        "forma_principal": grupos["forma"].agg(lambda s: s.value_counts().index[0]),
        "formas_observadas": grupos["forma"].agg(
            lambda s: "; ".join(f"{f} ({n})" for f, n in s.value_counts().items())
        ),
    })
    tab = tab.reset_index().sort_values(["frequencia", "bigrama"], ascending=[False, True])
    return tab.set_index("bigrama")


def recorte_com_empates(tab, n, coluna="frequencia"):
    """Primeiras n linhas de uma tabela ordenada por `coluna` (decrescente), ampliadas até incluir
    todas as linhas com o mesmo valor da n-ésima, para que o desempate alfabético não decida quem
    aparece (regra definida pelo pesquisador). Devolve o recorte e um resumo do ponto de corte."""
    n_base = min(n, len(tab))
    if n_base == 0:
        return tab.iloc[:0], {"pedido": n, "valor_do_corte": None, "n_recorte": 0,
                              "empatados_acrescentados": [], "primeiro_fora": None}
    valor_corte = tab[coluna].iloc[n_base - 1]
    n_recorte = int((tab[coluna] >= valor_corte).sum())
    recorte = tab.iloc[:n_recorte]
    seguinte = tab.iloc[n_recorte] if len(tab) > n_recorte else None
    # Verificação: o primeiro item fora do recorte tem valor menor que o do último incluído.
    assert seguinte is None or seguinte[coluna] < valor_corte
    return recorte, {
        "pedido": n,
        "valor_do_corte": valor_corte,
        "n_recorte": n_recorte,
        "empatados_acrescentados": list(recorte.index[n_base:]),
        "primeiro_fora": (seguinte.name, seguinte[coluna]) if seguinte is not None else None,
    }


# ---------------------------------------------------------------------------
# Equivalência de tratamento entre os planos (skill 03, seção 3.6)
# ---------------------------------------------------------------------------
# Regras e expressões regulares dos ajustes de tokenização dos notebooks dos planos chineses.
REGRAS_DE_TOKENIZACAO = [
    ("'+' colado ao fim da palavra", "separar_simbolo_final", r"[^\W\d_]+\+(?![^\W\d_])"),
    ("hífen suspenso", "separar_simbolo_final", r"[^\W\d_]+[-‐‑](?=\s)"),
    ("sigla com '&'", "siglas_com_e_comercial", r"(?<![^\W\d_])[^\W\d_]+(?:&[^\W\d_]+)+(?![^\W\d_])"),
]


def formas_dos_ajustes_de_tokenizacao(sigla, unidades):
    """Formas do texto do plano que os ajustes de tokenização afetam, com o número de ocorrências e o
    tratamento que recebem NESTE plano (a contagem é feita diretamente no texto das unidades)."""
    linhas = []
    for ajuste, parametro, padrao in REGRAS_DE_TOKENIZACAO:
        ativo = DOCUMENTOS[sigla][parametro]
        contagem = pd.Series([m.group(0) for u in unidades for m in re.finditer(padrao, u["texto"])],
                             dtype=object).value_counts()
        for forma, n in contagem.items():
            linhas.append({"plano": sigla, "ajuste": ajuste, "forma_no_texto": forma, "ocorrencias": int(n),
                           "tratamento": ("contada (ajuste ativo no plano)" if ativo
                                          else "fora da contagem (ajuste inativo no plano)")})
    return pd.DataFrame(linhas, columns=["plano", "ajuste", "forma_no_texto", "ocorrencias", "tratamento"])


def lemas_sem_correcao_de_outro_plano(freq_por_plano):
    """Para cada plano, os lemas que o notebook individual de OUTRO plano corrige (CORRECOES_DE_LEMA)
    e que, neste plano, ficam sem a correção, com a frequência e o lema de destino."""
    linhas = []
    for sigla, tab in freq_por_plano.items():
        proprias = CORRECOES_DE_LEMA.get(sigla, {})
        origens = {}
        for outra, correcoes in CORRECOES_DE_LEMA.items():
            for origem, destino in correcoes.items():
                if outra != sigla and origem not in proprias and origem in tab.index:
                    origens.setdefault((origem, destino), []).append(outra)
        for (origem, destino), outras in sorted(origens.items()):
            linhas.append({"plano": sigla, "lema_sem_correcao": origem,
                           "frequencia": int(tab.loc[origem, "frequencia"]), "lema_de_destino": destino,
                           "correcao_adotada_em": ", ".join(outras)})
    return pd.DataFrame(linhas, columns=["plano", "lema_sem_correcao", "frequencia", "lema_de_destino",
                                         "correcao_adotada_em"])


# ---------------------------------------------------------------------------
# Conferência de continuidade (seção 11.2)
# ---------------------------------------------------------------------------
def ler_recorte_individual(caminho, id_celula):
    """Lê, no notebook individual (somente leitura), a tabela gravada na saída da célula do recorte
    de cerca de 30 lemas. Devolve a tabela (lema, frequência, taxa por 1.000 tokens) ou None se a
    célula ou a tabela não forem encontradas. Nenhum número é usado nos cálculos deste notebook."""
    if not caminho.exists():
        return None
    nb = json.loads(caminho.read_text(encoding="utf-8"))
    celula = next((c for c in nb.get("cells", []) if c.get("id") == id_celula), None)
    if celula is None:
        return None
    for saida in celula.get("outputs", []):
        html = saida.get("data", {}).get("text/html")
        if not html:
            continue
        tab = pd.read_html(io.StringIO("".join(html)), decimal=",", thousands=".", index_col=0)[0]
        if isinstance(tab.columns, pd.MultiIndex):
            tab.columns = [c[0] for c in tab.columns]
        if {"frequencia", "por_1000_tokens"} <= set(tab.columns):
            tab = tab[tab.index.notna()]
            tab.index = tab.index.astype(str)
            return tab[["frequencia", "por_1000_tokens"]].astype({"frequencia": int})
    return None


def versoes_ambiente(nlp):
    return {
        "python": platform.python_version(),
        "spacy": spacy.__version__,
        "modelo_spacy": f"{MODELO_SPACY} {nlp.meta.get('version', '?')}",
        "pandas": pd.__version__,
        "numpy": np.__version__,
        "matplotlib": matplotlib.__version__,
    }
```

```python
# ===========================================================================
# FUNÇÕES COMPARATIVAS (skill 03): grupos de planos, recortes, keyness em pares e gráfico de
# convergência e divergência. O desenho do gráfico e o posicionamento dos rótulos reproduzem os do
# notebook comparativo da pasta Declarações IA (Análise_Comparativa_BRICS_OCDE.ipynb, seções 16
# e 17), por comando do pesquisador; as adaptações estão comentadas no código.
# ===========================================================================
import textwrap

import matplotlib.patheffects as path_effects
from matplotlib.colors import to_rgb
from scipy.stats import chi2

TIPOS = {
    "unigramas": {"titulo": "Unigramas", "singular": "lema", "plural": "lemas",
                  "top_n": TOP_N_LEMAS_CONVERGENCIA, "freq_minima": 1,
                  "descricao_ponto": "lema de conteúdo (unigrama)",
                  "recorte": "os {n} lemas de conteúdo mais frequentes {de}, com os empatados no {n}º"},
    "bigramas": {"titulo": "Bigramas", "singular": "bigrama", "plural": "bigramas",
                 "top_n": TOP_N_BIGRAMAS_CONVERGENCIA, "freq_minima": FREQ_MINIMA_BIGRAMA,
                 "descricao_ponto": "bigrama (expressão de duas palavras de conteúdo)",
                 "recorte": "as {n} expressões de duas palavras mais frequentes {de}, com as empatadas na {n}ª"},
}


# ---------------------------------------------------------------------------
# Grupos de planos (blocos e conjunto dos cinco planos)
# ---------------------------------------------------------------------------
def contar_formas(*textos):
    """Soma as contagens de formas escritas no formato 'forma (n); forma (n)' de uma ou mais tabelas."""
    contagem = {}
    for texto in textos:
        if not isinstance(texto, str) or not texto:
            continue
        for item in texto.split("; "):
            forma, n = re.fullmatch(r"(.*) \((\d+)\)", item).groups()
            contagem[forma] = contagem.get(forma, 0) + int(n)
    return contagem


def formatar_formas(contagem):
    """Formas escritas no formato 'forma (n); forma (n)', da mais frequente para a menos frequente."""
    return "; ".join(f"{f} ({n})" for f, n in sorted(contagem.items(), key=lambda kv: (-kv[1], kv[0])))


def tabela_do_grupo(tabelas, n_tokens):
    """Junta os planos de um grupo (bloco ou conjunto dos cinco planos) como um único texto: soma as
    ocorrências de cada termo e as formas escritas observadas em cada plano e divide pela soma dos
    tokens dos planos. Taxa do grupo = (ocorrências somadas ÷ tokens somados) × 1.000. Ordenação:
    frequência decrescente; empate -> ordem alfabética (a mesma das tabelas de cada plano)."""
    nome_indice = tabelas[0].index.name
    termos = sorted(set().union(*(t.index for t in tabelas)))
    freq = sum(t["frequencia"].reindex(termos, fill_value=0).astype(int) for t in tabelas)
    formas = [formatar_formas(contar_formas(*(t["formas_observadas"].get(termo) for t in tabelas))) for termo in termos]
    tab = pd.DataFrame({"frequencia": freq.astype(int), "por_1000_tokens": freq / n_tokens * 1000,
                        "formas_observadas": formas}, index=pd.Index(termos, name=nome_indice))
    tab = tab.reset_index().sort_values(["frequencia", nome_indice], ascending=[False, True])
    return tab.set_index(nome_indice)


def montar_lado(chave, nome, no, do, planos, cor, agrupado):
    """Um lado da comparação: um plano (agrupado = False) ou um grupo de planos tratado como um único
    texto (agrupado = True). Guarda as tabelas completas de unigramas e bigramas, com e sem as exclusões
    do pesquisador, e os recortes ampliados (TOP_N, com os empatados no N-ésimo)."""
    n_tokens = int(sum(N_TOKENS[p] for p in planos))
    lado = {"chave": chave, "nome": nome, "no": no, "do": do, "planos": list(planos), "cor": cor,
            "agrupado": agrupado, "n_tokens": n_tokens, "tabelas": {}, "tabelas_sem_exclusoes": {},
            "recortes": {}, "info_recortes": {}}
    for tipo, t in TIPOS.items():
        if agrupado:
            tab = tabela_do_grupo([TABELAS[tipo][p] for p in planos], n_tokens)
            sem = tabela_do_grupo([TABELAS_SEM_EXCLUSOES[tipo][p] for p in planos], n_tokens)
        else:
            tab, sem = TABELAS[tipo][planos[0]], TABELAS_SEM_EXCLUSOES[tipo][planos[0]]
        elegiveis = tab[tab["frequencia"] >= t["freq_minima"]]
        recorte, info = recorte_com_empates(elegiveis, t["top_n"])
        info["elegiveis"] = len(elegiveis)
        lado["tabelas"][tipo], lado["tabelas_sem_exclusoes"][tipo] = tab, sem
        lado["recortes"][tipo], lado["info_recortes"][tipo] = recorte, info
    return lado


# ---------------------------------------------------------------------------
# Medidas termo a termo (keyness em pares; skill 03, seção 3.5)
# ---------------------------------------------------------------------------
def tabela_keyness(freq_a, freq_b, n_a, n_b):
    """Compara as frequências de dois lados (A = PBIA, eixo X; B = outro lado, eixo Y).
    G² (log-verossimilhança; Dunning, 1993; Rayson & Garside, 2000) mede a significância;
    Log Ratio (Hardie, 2014) mede o tamanho do efeito: log2 da razão entre as taxas relativas.
    Convenção: Log Ratio > 0 -> mais característico de A; < 0 -> de B.
    Frequência zero recebe correção de 0,5 apenas no cálculo do Log Ratio."""
    termos = freq_a.index.union(freq_b.index)
    a = freq_a.reindex(termos, fill_value=0).astype(float)
    b = freq_b.reindex(termos, fill_value=0).astype(float)
    esp_a = n_a * (a + b) / (n_a + n_b)
    esp_b = n_b * (a + b) / (n_a + n_b)
    with np.errstate(divide="ignore", invalid="ignore"):
        g2 = 2 * (np.where(a > 0, a * np.log(a / esp_a), 0.0) + np.where(b > 0, b * np.log(b / esp_b), 0.0))
    log_ratio = np.log2((a.where(a > 0, 0.5) / n_a) / (b.where(b > 0, 0.5) / n_b))
    return pd.DataFrame({
        "freq_a": a.astype(int), "freq_b": b.astype(int),
        "x": a / n_a * 1000, "y": b / n_b * 1000,
        "freq_combinada": (a + b).astype(int),
        "G2": g2, "p_valor": chi2.sf(g2, 1), "log_ratio": log_ratio,
    })


def zona_do_termo(freq_a, freq_b, taxa_a, taxa_b):
    """Zona do termo pela razão entre as taxas: 'A' (PBIA) ou 'B' (outro lado) se a taxa de um lado for
    ao menos RAZAO_ZONA_NEUTRA vezes a do outro (ou se o termo só ocorrer nele); 'neutra' nos demais casos."""
    if freq_a > 0 and freq_b == 0:
        return "A"
    if freq_b > 0 and freq_a == 0:
        return "B"
    if taxa_a >= RAZAO_ZONA_NEUTRA * taxa_b:
        return "A"
    if taxa_b >= RAZAO_ZONA_NEUTRA * taxa_a:
        return "B"
    return "neutra"


def normalizar_termo(texto):
    """Minúsculas, sem hífen e com espaços simples (ex.: 'Multi-Stakeholder' -> 'multistakeholder')."""
    return " ".join(str(texto).lower().replace("-", "").split())


def _primeiros_com_empate(serie, n):
    """Os n primeiros itens de uma série já ordenada (valor decrescente); se o n-ésimo estiver empatado,
    entram também os empatados com ele, desde que o total não passe de n + 3 (sem isso, os empatados
    ficam todos de fora, e não apenas parte deles). Usada só para escolher os rótulos."""
    if len(serie) <= n:
        return list(serie.index)
    corte = serie.iloc[n - 1]
    com_empate = list(serie.index[serie >= corte])
    if len(com_empate) <= n + 3:
        return com_empate
    return list(serie.index[serie > corte])


def nome_da_zona(zona, A, B):
    """Nome da zona para os textos: o nome do lado ou 'neutra'."""
    return {"A": A["nome"], "B": B["nome"]}.get(zona, "neutra")


def preparar_convergencia(A, B, tipo):
    """Une os recortes ampliados dos dois lados e calcula, para cada termo: ocorrências, taxas por 1.000
    tokens, G² e Log Ratio, zona, zona sem as exclusões do pesquisador, zona com pesos iguais (grupos),
    taxa em cada plano do grupo, posição na escala e rótulo. Termos pedidos pelo pesquisador
    (DESTAQUES) que estejam fora dos recortes são acrescentados e sinalizados."""
    tab_a, tab_b = A["tabelas"][tipo], B["tabelas"][tipo]
    rec_a, rec_b = A["recortes"][tipo], B["recortes"][tipo]
    termos = list(dict.fromkeys(list(rec_a.index) + list(rec_b.index)))
    origem = {}
    for t in termos:
        dentro = [lado["nome"] for lado in (A, B) if t in lado["recortes"][tipo].index]
        origem[t] = "os dois recortes" if len(dentro) == 2 else f"recorte {dentro[0]}"

    # Formas escritas observadas nos dois lados (rótulo dos bigramas e busca dos pedidos).
    todos = tab_a.index.union(tab_b.index)
    formas = {t: contar_formas(tab_a["formas_observadas"].get(t), tab_b["formas_observadas"].get(t)) for t in todos}

    # Termos pedidos pelo pesquisador: procurados pelo termo (lema) ou por qualquer forma escrita observada.
    pedidos, nao_encontrados = [], []
    for pedido in DESTAQUES.get((B["chave"], tipo), []):
        alvo = normalizar_termo(pedido)
        achados = [t for t in todos if alvo == normalizar_termo(t) or alvo in {normalizar_termo(f) for f in formas[t]}]
        if not achados:
            nao_encontrados.append(pedido)
            continue
        plotados = [t for t in achados if t in termos]
        escolhido = plotados[0] if plotados else max(
            achados, key=lambda t: (tab_a["frequencia"].get(t, 0) + tab_b["frequencia"].get(t, 0), t))
        if escolhido not in termos:   # fora dos recortes: entra no gráfico por pedido do pesquisador
            termos.append(escolhido)
            origem[escolhido] = "pedido do pesquisador"
        pedidos.append(escolhido)

    tab = tabela_keyness(tab_a["frequencia"], tab_b["frequencia"], A["n_tokens"], B["n_tokens"]).reindex(termos)
    tab.index.name = "bigrama" if tipo == "bigramas" else "lema"
    tab["zona"] = [zona_do_termo(*v) for v in zip(tab["freq_a"], tab["freq_b"], tab["x"], tab["y"])]

    # Robustez 1: zona se as exclusões do pesquisador (EUA e PBIA) fossem desfeitas, para os mesmos termos.
    sem = tabela_keyness(A["tabelas_sem_exclusoes"][tipo]["frequencia"], B["tabelas_sem_exclusoes"][tipo]["frequencia"],
                         A["n_tokens"], B["n_tokens"]).reindex(termos).fillna(0)
    tab["freq_a_sem_exclusoes"] = sem["freq_a"].astype(int)
    tab["freq_b_sem_exclusoes"] = sem["freq_b"].astype(int)
    tab["zona_sem_exclusoes"] = [zona_do_termo(*v) for v in zip(sem["freq_a"], sem["freq_b"], sem["x"], sem["y"])]
    tab["depende_das_exclusoes"] = tab["zona"] != tab["zona_sem_exclusoes"]

    # Robustez 2 (grupos com mais de um plano): taxa de cada plano do grupo e zona com pesos iguais
    # (média simples das taxas dos planos, em vez de ocorrências somadas ÷ tokens somados).
    if len(B["planos"]) > 1:
        taxas_planos = pd.DataFrame({p: TABELAS[tipo][p]["frequencia"].reindex(termos, fill_value=0) / N_TOKENS[p] * 1000
                                     for p in B["planos"]})
        for p in B["planos"]:
            tab[f"y_{p}"] = taxas_planos[p]
        tab["y_pesos_iguais"] = taxas_planos.mean(axis=1)
        tab["zona_pesos_iguais"] = [zona_do_termo(*v) for v in
                                    zip(tab["freq_a"], tab["freq_b"], tab["x"], tab["y_pesos_iguais"])]

    if tipo == "bigramas":   # rótulo: forma escrita mais frequente nos dois lados (desempate alfabético)
        tab["rotulo"] = [sorted(formas[t].items(), key=lambda kv: (-kv[1], kv[0]))[0][0] for t in tab.index]
    else:                    # lemas: o próprio lema
        tab["rotulo"] = tab.index
    tab["origem"] = [origem[t] for t in tab.index]
    tab["pedido"] = tab.index.isin(pedidos)

    # Fora da escala: termo cuja maior taxa supera FATOR_FORA_DA_ESCALA vezes a maior taxa dos demais.
    maiores = tab[["x", "y"]].max(axis=1).sort_values(ascending=False)
    fora = []
    while len(maiores) > 1 and maiores.iloc[0] > FATOR_FORA_DA_ESCALA * maiores.iloc[1]:
        fora.append(maiores.index[0])
        maiores = maiores.iloc[1:]
    tab["fora_da_escala"] = tab.index.isin(fora)

    # Rótulos (regra fixa, sem olhar para os termos): os pedidos; os termos cuja zona depende das
    # exclusões (sempre, com †); na zona neutra, os N_ROTULOS_NEUTROS de maior taxa somada; em cada zona
    # colorida, os N_ROTULOS_POR_ZONA de maior taxa no lado em que são mais frequentes. Ordenação por
    # taxa e, no empate, alfabética; empatados no último lugar seguem _primeiros_com_empate.
    base = tab[~tab["fora_da_escala"]].copy()
    base["_soma"] = base["x"] + base["y"]
    escolhidos = {}
    for zona, coluna, n in (("neutra", "_soma", N_ROTULOS_NEUTROS), ("A", "x", N_ROTULOS_POR_ZONA),
                            ("B", "y", N_ROTULOS_POR_ZONA)):
        sub = base[base["zona"] == zona].reset_index()
        sub = sub.sort_values([coluna, tab.index.name], ascending=[False, True]).set_index(tab.index.name)[coluna]
        for t in _primeiros_com_empate(sub, n):
            escolhidos[t] = "regra: zona neutra" if zona == "neutra" else "regra: zona colorida"
    for t in base.index[base["depende_das_exclusoes"]]:
        escolhidos[t] = "zona depende das exclusões"
    for t in base.index[base["pedido"]]:
        escolhidos[t] = "pedido do pesquisador"
    tab["destaque"] = [escolhidos.get(t, "") for t in tab.index]
    tab["rotulo_grafico"] = [r + ("†" if d else "") for r, d in zip(tab["rotulo"], tab["depende_das_exclusoes"])]

    tab["distintivo"] = ((tab["G2"] >= LIMIAR_G2) & (tab["log_ratio"].abs() >= LIMIAR_LOG_RATIO)
                         & (tab["freq_combinada"] >= FREQ_MINIMA_COMPARACAO))
    tab.attrs.update({"nao_encontrados": nao_encontrados, "A": A, "B": B, "tipo": tipo})
    return tab


# ---------------------------------------------------------------------------
# Posicionamento de rótulos sem sobreposição (copiado do notebook BRICS × OCDE). Únicas alterações: o número
# de posições candidatas por rótulo e o custo de um rótulo sobre outro viraram os atributos N_CANDIDATOS e
# CUSTO_SOBREPOSICAO (40 e 1.000 no original, mantidos como padrão da classe). Os gráficos deste notebook,
# mais densos, usam os valores da configuração (ROTULADOR_*), aplicados em grafico_convergencia.
# ---------------------------------------------------------------------------
def _sobrepoe(a, b):
    """As caixas (x0, y0, x1, y1) se sobrepõem?"""
    return a[0] < b[2] and b[0] < a[2] and a[1] < b[3] and b[1] < a[3]


def _cruza(p, q, caixa):
    """O segmento p-q atravessa a caixa? (recorte de Liang-Barsky)"""
    (x0, y0), (x1, y1) = p, q
    dx, dy = x1 - x0, y1 - y0
    t0, t1 = 0.0, 1.0
    for pk, qk in ((-dx, x0 - caixa[0]), (dx, caixa[2] - x0), (-dy, y0 - caixa[1]), (dy, caixa[3] - y0)):
        if pk == 0:
            if qk < 0:
                return False
            continue
        t = qk / pk
        if pk < 0:
            t0 = max(t0, t)
        else:
            t1 = min(t1, t)
        if t0 > t1:
            return False
    return True


def _distancia_caixa(p, caixa):
    """Distância de um ponto a uma caixa (0 se estiver dentro)."""
    dx = max(caixa[0] - p[0], 0, p[0] - caixa[2])
    dy = max(caixa[1] - p[1], 0, p[1] - caixa[3])
    return float(np.hypot(dx, dy))


def _segmentos_cruzam(p1, p2, q1, q2):
    """Os segmentos p1-p2 e q1-q2 se cruzam? (pontas comuns não contam)"""
    def orient(a, b, c):
        return (b[0] - a[0]) * (c[1] - a[1]) - (b[1] - a[1]) * (c[0] - a[0])
    return (orient(q1, q2, p1) * orient(q1, q2, p2) < 0) and (orient(p1, p2, q1) * orient(p1, p2, q2) < 0)


class Rotulador:
    """Posiciona rótulos sem sobreposição. Cada rótulo tem posições candidatas ao redor do seu ponto, da mais
    próxima para a mais distante, e cada posição tem um custo: distância ao ponto e sobreposição a pontos,
    a outros rótulos e a linhas de ligação (proibitivas), além de passagem sobre a diagonal ou os limites da
    faixa (evitada). Um rótulo sem linha de ligação não pode ficar mais perto de outro ponto do que do seu.
    Os rótulos são distribuídos em várias ordens (a primeira por densidade; as demais sorteadas com semente
    fixa) e fica a distribuição de menor custo total. O resultado é sempre o mesmo para os mesmos dados."""

    ANGULOS = (0, 35, -35, 70, -70, 110, -110, 145, -145, 180, 90, -90)
    DISTANCIAS_PT = (3, 6, 10, 15, 21, 28, 37, 48, 61, 76, 94, 115, 140)
    N_CANDIDATOS = 40   # posições candidatas mantidas por rótulo (as de menor custo)
    CUSTO_SOBREPOSICAO = 1000   # custo de um rótulo sobre outro rótulo (texto sobre texto)

    def __init__(self, ax, pontos_dados, raios_pt, linhas_suaves_dados):
        self.ax = ax
        self.renderer = ax.figure.canvas.get_renderer()
        self.px = ax.figure.dpi / 72
        self.pontos = ax.transData.transform(pontos_dados)
        self.raios = np.asarray(raios_pt, dtype=float) * self.px
        self.caixas_pontos = [(x - r - 1, y - r - 1, x + r + 1, y + r + 1) for (x, y), r in zip(self.pontos, self.raios)]
        e = ax.get_window_extent(self.renderer)
        self.limite = (e.x0 + 2, e.y0 + 2, e.x1 - 2, e.y1 - 2)
        self.linhas_suaves = [tuple(map(tuple, ax.transData.transform(seg))) for seg in linhas_suaves_dados]
        self.ocupadas, self.segmentos = [], []

    def caixa(self, artista, folga=2):
        bb = artista.get_window_extent(self.renderer)
        return (bb.x0 - folga, bb.y0 - folga, bb.x1 + folga, bb.y1 + folga)

    def ocupar(self, artista):
        self.ocupadas.append(self.caixa(artista))

    def ocupar_segmento(self, p_dados, q_dados):
        p, q = self.ax.transData.transform([p_dados, q_dados])
        self.segmentos.append((tuple(p), tuple(q)))

    def _candidatos(self, alvo, raio, largura, altura, proprio, distancias):
        """Posições candidatas de um rótulo, com o custo que não depende dos outros rótulos a posicionar."""
        candidatos = []
        for d_pt in distancias:
            for k, ang in enumerate(self.ANGULOS):
                u = np.array([np.cos(np.radians(ang)), np.sin(np.radians(ang))])
                ancora = alvo + (raio + d_pt * self.px) * u
                ha = "left" if u[0] > 0.3 else "right" if u[0] < -0.3 else "center"
                va = "bottom" if u[1] > 0.3 else "top" if u[1] < -0.3 else "center"
                x0 = ancora[0] - {"left": 0, "right": largura, "center": largura / 2}[ha]
                y0 = ancora[1] - {"bottom": 0, "top": altura, "center": altura / 2}[va]
                caixa = (x0 - 2, y0 - 2, x0 + largura + 2, y0 + altura + 2)
                ligacao = None
                if d_pt >= 6 and proprio >= 0:
                    ligacao = (tuple(alvo), (float(np.clip(alvo[0], caixa[0] + 2, caixa[2] - 2)),
                                             float(np.clip(alvo[1], caixa[1] + 2, caixa[3] - 2))))
                custo = d_pt + 2.0 * k
                if not (caixa[0] >= self.limite[0] and caixa[1] >= self.limite[1]
                        and caixa[2] <= self.limite[2] and caixa[3] <= self.limite[3]):
                    custo += 5000
                custo += self.CUSTO_SOBREPOSICAO * sum(_sobrepoe(caixa, c) for c in self.ocupadas)
                custo += 400 * sum(_sobrepoe(caixa, c) for c in self.caixas_pontos)
                custo += 600 * sum(_cruza(p, q, caixa) for p, q in self.segmentos)
                custo += 25 * sum(_cruza(p, q, caixa) for p, q in self.linhas_suaves)
                outros = [_distancia_caixa(p, caixa) - r for i, (p, r) in enumerate(zip(self.pontos, self.raios)) if i != proprio]
                if ligacao is None:
                    # Sem ligação, o rótulo precisa estar claramente mais perto do seu ponto.
                    if proprio >= 0 and outros and min(outros) < _distancia_caixa(alvo, caixa) + 4:
                        custo += 60
                else:
                    # Com ligação, o rótulo também não deve encostar em outro ponto (leitura ambígua).
                    if outros and min(outros) < 7:
                        custo += 80
                    custo += 600 * sum(_cruza(ligacao[0], ligacao[1], c) for c in self.ocupadas)
                    custo += 300 * sum(_cruza(ligacao[0], ligacao[1], c) for i, c in enumerate(self.caixas_pontos) if i != proprio)
                candidatos.append((custo, ancora, ha, va, caixa, ligacao))
        candidatos.sort(key=lambda c: c[0])
        return candidatos[:self.N_CANDIDATOS]

    def _custo_mutuo(self, candidato, colocados):
        """Custo de um candidato diante dos rótulos já colocados nesta distribuição."""
        caixa, ligacao = candidato[4], candidato[5]
        custo = 0.0
        for caixa_2, ligacao_2 in colocados:
            if _sobrepoe(caixa, caixa_2):
                custo += self.CUSTO_SOBREPOSICAO
            if ligacao_2 is not None and _cruza(ligacao_2[0], ligacao_2[1], caixa):
                custo += 600
            if ligacao is not None:
                if _cruza(ligacao[0], ligacao[1], caixa_2):
                    custo += 600
                if ligacao_2 is not None and _segmentos_cruzam(ligacao[0], ligacao[1], ligacao_2[0], ligacao_2[1]):
                    custo += 200
        return custo

    def posicionar(self, pedidos, tentativas=150, semente=0, distancias=None):
        """Cria os rótulos de 'pedidos' — lista de (xy, texto, cor_da_ligacao, indice_do_ponto, estilo);
        índice -1 indica uma âncora que não é ponto (ex.: a seta de um termo fora da escala)."""
        if not pedidos:
            return []
        artistas, candidatos = [], []
        for xy, texto, _, proprio, estilo in pedidos:
            artista = self.ax.text(0, 0, texto, zorder=6, clip_on=False, ha="left", va="bottom", **estilo)
            bb = artista.get_window_extent(self.renderer)
            alvo = self.ax.transData.transform([xy])[0]
            raio = self.raios[proprio] if proprio >= 0 else 0.0
            candidatos.append(self._candidatos(alvo, raio, bb.width, bb.height, proprio, distancias or self.DISTANCIAS_PT))
            artistas.append(artista)

        alvos = self.ax.transData.transform([p[0] for p in pedidos])
        densidade = [int((np.hypot(*(self.pontos - a).T) < 45 * self.px).sum()) for a in alvos]
        ordens = [sorted(range(len(pedidos)), key=lambda i: -densidade[i])]
        rng = np.random.default_rng(semente)
        ordens += [list(rng.permutation(len(pedidos))) for _ in range(tentativas)]
        melhor = None
        for ordem in ordens:
            escolha, colocados, total = {}, [], 0.0
            for i in ordem:
                custo, candidato = min(((c[0] + self._custo_mutuo(c, colocados), c) for c in candidatos[i]),
                                       key=lambda par: par[0])
                escolha[i] = candidato
                colocados.append((candidato[4], candidato[5]))
                total += custo
                if melhor is not None and total >= melhor[0]:
                    break
            else:
                if melhor is None or total < melhor[0]:
                    melhor = (total, escolha)

        inv = self.ax.transData.inverted()
        for i, (artista, (xy, _, cor, proprio, _)) in enumerate(zip(artistas, pedidos)):
            _, ancora, ha, va, caixa, ligacao = melhor[1][i]
            artista.set_position(inv.transform(ancora))
            artista.set_ha(ha)
            artista.set_va(va)
            self.ocupadas.append(caixa)
            if ligacao is not None:
                self.segmentos.append(ligacao)
                raio_pt = self.raios[proprio] / self.px if proprio >= 0 else 0
                self.ax.annotate("", xy=xy, xytext=inv.transform(ligacao[1]), zorder=5, annotation_clip=False,
                                 arrowprops=dict(arrowstyle="-", color=cor, lw=0.7, shrinkA=0, shrinkB=raio_pt + 1))
        return artistas


# ---------------------------------------------------------------------------
# Gráfico
# ---------------------------------------------------------------------------
def _luminancia(cor):
    """Luminância relativa (WCAG 2.1) de uma cor."""
    canais = [c / 12.92 if c <= 0.03928 else ((c + 0.055) / 1.055) ** 2.4 for c in to_rgb(cor)]
    return 0.2126 * canais[0] + 0.7152 * canais[1] + 0.0722 * canais[2]


def cor_de_texto(cor, contraste_minimo=3.0):
    """Tom da mesma cor, escurecido só o necessário para que o TEXTO escrito nela tenha contraste
    mínimo com o fundo branco (ex.: o amarelo da UE). Os pontos mantêm a cor original."""
    rgb = np.array(to_rgb(cor))
    while (1.05 / (_luminancia(rgb) + 0.05)) < contraste_minimo:
        rgb = rgb * 0.95
    return tuple(rgb)


def grafico_convergencia(tab, titulo_b, subtitulos, nota_grupo=None):
    """Gráfico de convergência e divergência: X = taxa no PBIA (lado A), Y = taxa no outro lado (B), por
    1.000 tokens, mesma escala nos dois eixos, diagonal de frequência igual e faixa da zona neutra."""
    A, B, tipo = tab.attrs["A"], tab.attrs["B"], tab.attrs["tipo"]
    cor_zona = {"A": A["cor"], "B": B["cor"], "neutra": COR_NEUTRA}
    cor_txt = {z: cor_de_texto(c) for z, c in cor_zona.items()}
    r_txt = f"{RAZAO_ZONA_NEUTRA:g}".replace(".", ",")
    dentro = tab[~tab["fora_da_escala"]]
    lim = MaxNLocator(nbins=7, steps=[1, 2, 2.5, 5, 10]).tick_values(0, dentro[["x", "y"]].to_numpy().max() * 1.1)[-1]

    # Mesmas medidas do gráfico de referência (polegadas); a altura cresce com as linhas do subtítulo.
    # Linhas longas são quebradas em partes de tamanho parecido, para caber na largura da figura.
    subtitulos = [parte for linha in subtitulos
                  for parte in textwrap.wrap(linha, math.ceil(len(linha) / math.ceil(len(linha) / 96)) + 8)]
    extra = 0.2 * (len(subtitulos) - 1)
    largura, altura, lado, esq, base = 7.6, 9.4 + extra, 6.0, 1.0, 2.75
    fig = plt.figure(figsize=(largura, altura))
    ax = fig.add_axes([esq / largura, base / altura, lado / largura, lado / altura])
    ax.set_xlim(0, lim)
    ax.set_ylim(0, lim)
    ax.set_aspect("equal", adjustable="box")

    # Zona neutra (faixa entre y = x / R e y = R·x) e diagonal de frequência igual.
    xs = np.linspace(0, lim, 400)
    ax.fill_between(xs, xs / RAZAO_ZONA_NEUTRA, np.minimum(xs * RAZAO_ZONA_NEUTRA, lim),
                    color=COR_FAIXA, alpha=0.75, lw=0, zorder=0)
    for fator in (RAZAO_ZONA_NEUTRA, 1 / RAZAO_ZONA_NEUTRA):
        ax.plot(xs, xs * fator, color=COR_LIGACAO, lw=0.7, ls=(0, (1.5, 2)), zorder=1)
    ax.plot([0, lim], [0, lim], color=COR_TINTA_2, lw=1.0, ls=(0, (5, 4)), zorder=2)

    # Pontos. Termos com os mesmos valores formam um único ponto, maior, com o número de termos dentro
    # (ou ao lado, quando o ponto maior se sobrepõe a outro ponto; ver adiante).
    grupos = []
    for (x, y), g in dentro.groupby([dentro["x"].round(9), dentro["y"].round(9)], sort=False):
        grupos.append({"x": g["x"].iloc[0], "y": g["y"].iloc[0], "zona": g["zona"].iloc[0], "n": len(g),
                       "destacados": sorted(g.loc[g["destaque"] != "", "rotulo_grafico"])})
    tam_simples, tam_multiplo = 46, 150
    for zona in ("neutra", "A", "B"):
        for multiplo in (False, True):
            sel = [p for p in grupos if p["zona"] == zona and (p["n"] > 1) == multiplo]
            if sel:
                ax.scatter([p["x"] for p in sel], [p["y"] for p in sel], s=tam_multiplo if multiplo else tam_simples,
                           color=cor_zona[zona], alpha=0.92, edgecolors="white", linewidths=0.8, zorder=3, clip_on=False)

    ax.xaxis.set_major_locator(MaxNLocator(nbins=7, steps=[1, 2, 2.5, 5, 10]))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=7, steps=[1, 2, 2.5, 5, 10]))
    ax.xaxis.set_major_formatter(FMT_BR)
    ax.yaxis.set_major_formatter(FMT_BR)
    ax.set_xlabel(f"Frequência relativa {A['no']} (ocorrências por 1.000 tokens)", labelpad=8)
    ax.set_ylabel(f"Frequência relativa {B['no']} (ocorrências por 1.000 tokens)", labelpad=8)
    centro = (esq + lado / 2) / largura   # título centrado sobre a área do gráfico
    fig.text(centro, 1 - 0.18 / altura, f"Convergência e divergência de termos — {A['nome']} × {titulo_b}",
             ha="center", va="top", fontsize=13, fontweight="bold")
    fig.text(centro, 1 - 0.48 / altura, "\n".join(subtitulos), ha="center", va="top", fontsize=10,
             color=COR_TINTA_2, linespacing=1.4)
    fig.canvas.draw()

    contorno = [path_effects.withStroke(linewidth=3, foreground="white")]
    raios = [np.sqrt(tam_multiplo if p["n"] > 1 else tam_simples) / 2 for p in grupos]
    rot = Rotulador(ax, np.array([(p["x"], p["y"]) for p in grupos]), raios,
                    [[(0, 0), (lim, lim)], [(0, 0), (lim / RAZAO_ZONA_NEUTRA, lim)], [(0, 0), (lim, lim / RAZAO_ZONA_NEUTRA)]])
    rot.N_CANDIDATOS = ROTULADOR_CANDIDATOS
    rot.CUSTO_SOBREPOSICAO = ROTULADOR_CUSTO_SOBREPOSICAO

    # Número de termos de cada ponto maior: dentro do ponto (branco sobre cores escuras; tinta escura sobre
    # cores claras, como o amarelo) quando ele não se sobrepõe a nenhum outro ponto. Se houver sobreposição,
    # números vizinhos se juntariam (ex.: "7" e "5" lidos como "75"); por isso, o número vai para o lado do
    # ponto, em negrito e na cor da zona, posicionado como os rótulos.
    centros = rot.pontos
    for i, p in enumerate(grupos):
        p["numero_ao_lado"] = p["n"] > 1 and any(
            np.hypot(*(centros[i] - centros[j])) < rot.raios[i] + rot.raios[j] - 1 for j in range(len(grupos)) if j != i)
        if p["n"] > 1 and not p["numero_ao_lado"]:
            claro = _luminancia(cor_zona[p["zona"]]) > 0.4
            numero = ax.text(p["x"], p["y"], str(p["n"]), ha="center", va="center", fontsize=7.5, fontweight="bold",
                             color=COR_TINTA if claro else "white", zorder=4, clip_on=False)
            rot.ocupar(numero)   # nenhum rótulo cobre o número de um ponto

    # Nome das regiões: no primeiro canto livre de pontos, inteiramente dentro da zona de cada lado.
    inv = ax.transData.inverted()
    for zona, texto, candidatos in (
        ("B", f"↑ mais frequente {B['no']}",
         [(0.03, f, "left", "top") for f in (0.97, 0.91, 0.85, 0.79, 0.73, 0.67)]),
        ("A", f"mais frequente {A['no']} →",
         [(0.97, f, "right", "bottom") for f in (0.03, 0.09, 0.15, 0.21, 0.27, 0.33)]),
    ):
        artista = ax.text(0, 0, texto, transform=ax.transAxes, fontsize=9.5, style="italic",
                          color=cor_txt[zona], zorder=6, path_effects=contorno)
        escolhido = None
        for fx, fy, ha, va in candidatos:
            artista.set_position((fx, fy))
            artista.set_ha(ha)
            artista.set_va(va)
            caixa = rot.caixa(artista, folga=6)   # folga maior: o nome da região não encosta em pontos
            (x0, y0), (x1, y1) = inv.transform([(caixa[0], caixa[1]), (caixa[2], caixa[3])])
            na_zona = y0 >= RAZAO_ZONA_NEUTRA * x1 if zona == "B" else y1 <= x0 / RAZAO_ZONA_NEUTRA
            livre = not any(_sobrepoe(caixa, c) for c in rot.caixas_pontos)
            if na_zona and livre:
                escolhido = (fx, fy, ha, va)
                break
        fx, fy, ha, va = escolhido or candidatos[0]
        artista.set_position((fx, fy))
        artista.set_ha(ha)
        artista.set_va(va)
        rot.ocupar(artista)

    # Termos fora da escala: seta na borda, na direção do valor real, com os valores exatos.
    for termo, linha in tab[tab["fora_da_escala"]].iterrows():
        ponta = np.array([min(linha["x"], lim), min(linha["y"], lim)])
        direcao = np.array([linha["x"] > lim, linha["y"] > lim], dtype=float)
        direcao /= np.linalg.norm(direcao)
        cauda = ponta - direcao * lim * 0.075
        ax.annotate("", xy=ponta, xytext=cauda, zorder=6, annotation_clip=False,
                    arrowprops=dict(arrowstyle="-|>", color=cor_zona[linha["zona"]], lw=1.6, mutation_scale=12))
        rot.ocupar_segmento(cauda, ponta)
        rot.posicionar([(cauda, f"{linha['rotulo_grafico']} (fora da escala)\n"
                                f"{A['nome']}: {num(linha['x'])} · {B['nome']}: {num(linha['y'])}",
                         cor_zona[linha["zona"]], -1,
                         dict(fontsize=9, color=COR_TINTA, linespacing=1.3, path_effects=contorno))],
                       tentativas=0, distancias=(2, 4, 7, 11))

    # Rótulos dos termos destacados: um por ponto (termos com os mesmos valores dividem o rótulo).
    estilo = dict(fontsize=9, color=COR_TINTA, linespacing=1.15, path_effects=contorno)
    rot.posicionar([((p["x"], p["y"]), "\n".join(p["destacados"]), cor_zona[p["zona"]], i, estilo)
                    for i, p in enumerate(grupos) if p["destacados"]], tentativas=ROTULADOR_TENTATIVAS, semente=SEMENTE,
                   distancias=Rotulador.DISTANCIAS_PT + (170, 200))

    # Números dos pontos maiores sobrepostos, ao lado do ponto (depois dos rótulos dos termos, que têm prioridade).
    rot.posicionar([((p["x"], p["y"]), str(p["n"]), cor_zona[p["zona"]], i,
                     dict(fontsize=8, fontweight="bold", color=cor_txt[p["zona"]], path_effects=contorno))
                    for i, p in enumerate(grupos) if p["numero_ao_lado"]],
                   tentativas=100, semente=SEMENTE, distancias=(1, 3, 6, 10, 15, 21, 28, 37, 48))

    # Legenda e instrução de leitura, abaixo do gráfico.
    marcador = dict(ls="", marker="o", ms=7.5, mec="white", mew=0.8)
    itens = [
        Line2D([], [], mfc=A["cor"], label=f"Mais frequente {A['no']}: taxa {A['no']} ≥ {r_txt}× a taxa {B['no']}", **marcador),
        Line2D([], [], mfc=B["cor"], label=f"Mais frequente {B['no']}: taxa {B['no']} ≥ {r_txt}× a taxa {A['no']}", **marcador),
        Line2D([], [], mfc=COR_NEUTRA, label=f"Zona neutra: usado por ambos, com diferença menor que {r_txt}×", **marcador),
        Line2D([], [], color=COR_TINTA_2, lw=1.0, ls=(0, (5, 4)),
               label=f"Linha tracejada: frequência igual {A['no']} e {B['no']} (x = y)"),
        Patch(facecolor=COR_FAIXA, edgecolor=COR_LIGACAO, lw=0.7, ls=":",
              label=f"Faixa cinza: zona neutra, entre y = x/{r_txt} e y = {r_txt}·x"),
    ]
    fig.legend(handles=itens, loc="upper left", bbox_to_anchor=(esq / largura, (base - 0.62) / altura),
               ncol=1, fontsize=9, handlelength=2.2, handletextpad=0.8, labelspacing=0.45, borderaxespad=0)
    leitura = (f"Como ler: cada ponto é um {TIPOS[tipo]['descricao_ponto']}. O eixo X mostra quantas vezes o termo "
               f"aparece a cada 1.000 tokens (palavras) {A['no']}; o eixo Y, o mesmo {B['no']}"
               + (f" ({nota_grupo})" if nota_grupo else "") + ". "
               f"Ex.: um ponto em x = 4 e y = 2 indica 4 ocorrências a cada 1.000 palavras {A['no']} "
               f"e 2 {B['no']}. Acima da diagonal, o termo é proporcionalmente mais frequente {B['no']}; "
               f"abaixo, {A['no']}. Pontos sobre um eixo (x = 0 ou y = 0) aparecem em um só lado. "
               "Um ponto maior reúne termos com exatamente os mesmos valores; o número de termos aparece dentro dele "
               "ou, quando ele se sobrepõe a outro ponto, ao lado, em negrito.")
    if tab["depende_das_exclusoes"].any():
        leitura += (" † A zona do termo depende das exclusões do pesquisador (nome do país e de órgãos, no PBIA e nos "
                    "EUA): sem elas, ele ficaria em outra zona (ver tabela).")
    fig.text(esq / largura, (base - 1.75) / altura, textwrap.fill(leitura, 100), ha="left", va="top",
             fontsize=9, color=COR_TINTA_2, linespacing=1.45)
    return fig


def tabela_convergencia_exibicao(tab):
    """Tabela com os valores exatos do gráfico (uma linha por termo), ordenada por zona e taxa."""
    A, B = tab.attrs["A"], tab.attrs["B"]
    nome_zona = {"A": f"mais frequente {A['no']}", "neutra": "neutra", "B": f"mais frequente {B['no']}"}
    ordem_zona = {"A": 0, "neutra": 1, "B": 2}
    t = tab.copy()
    t["_ordem"] = t["zona"].map(ordem_zona)
    t["_taxa"] = np.where(t["zona"] == "B", t["y"], np.where(t["zona"] == "A", t["x"], t["x"] + t["y"]))
    t = t.sort_values(["_ordem", "_taxa"], ascending=[True, False])
    saida = pd.DataFrame({
        "rótulo": t["rotulo"],
        f"{A['nome']}: ocorrências": t["freq_a"],
        f"{A['nome']}: por 1.000 tokens": t["x"],
        f"{B['nome']}: ocorrências": t["freq_b"],
        f"{B['nome']}: por 1.000 tokens": t["y"],
    })
    if len(B["planos"]) > 1:
        for p in B["planos"]:
            saida[f"{DOCUMENTOS[p]['nome']}: por 1.000 tokens"] = t[f"y_{p}"]
    saida["Log Ratio"] = t["log_ratio"]
    saida["G²"] = t["G2"]
    saida["zona"] = t["zona"].map(nome_zona)
    saida["distintivo (G², Log Ratio)"] = np.where(t["distintivo"], "sim", "não")
    saida["resiste a Bonferroni"] = np.where(t["resiste_bonferroni"], "sim", "não")
    saida["zona sem as exclusões"] = np.where(t["depende_das_exclusoes"], t["zona_sem_exclusoes"].map(nome_zona), "igual")
    if len(B["planos"]) > 1:
        saida["zona com pesos iguais"] = np.where(t["zona_pesos_iguais"] != t["zona"],
                                                  t["zona_pesos_iguais"].map(nome_zona), "igual")
    saida["origem"] = t["origem"]
    saida["rotulado"] = np.where(t["destaque"] != "", t["destaque"], "—")
    saida["fora da escala"] = np.where(t["fora_da_escala"], "sim", "—")
    saida.index.name = tab.index.name
    return saida


def sintese_convergencia(tab):
    """Síntese factual do gráfico: contagens por zona, vocabulário em comum (zona neutra), termos fora
    da escala, termos cuja zona depende das exclusões ou do peso dos planos do grupo."""
    A, B, tipo = tab.attrs["A"], tab.attrs["B"], tab.attrs["tipo"]
    t = TIPOS[tipo]
    contagem = tab["zona"].value_counts()
    ambos = int((tab["origem"] == "os dois recortes").sum())
    print(f"Síntese factual: {len(tab)} {t['plural']} no gráfico (união dos recortes: {A['nome']} "
          f"{len(A['recortes'][tipo])}, {B['nome']} {len(B['recortes'][tipo])}; {ambos} estão nos dois recortes). "
          f"Zona {A['nome']}: {contagem.get('A', 0)}; zona neutra: {contagem.get('neutra', 0)}; "
          f"zona {B['nome']}: {contagem.get('B', 0)}.")
    neutros = tab[tab["zona"] == "neutra"].assign(_s=lambda d: d["x"] + d["y"]).sort_values("_s", ascending=False)
    print(f"Vocabulário em comum (zona neutra, da maior para a menor taxa somada; {A['nome']}; {B['nome']}): "
          + ("; ".join(f"{l['rotulo']} ({num(l['x'])}; {num(l['y'])})" for _, l in neutros.iterrows()) or "(nenhum)") + ".")
    for _, l in tab[tab["fora_da_escala"]].iterrows():
        print(f"Fora da escala (indicado por seta): {l['rotulo']} — {A['nome']} {num(l['x'])}; {B['nome']} {num(l['y'])} "
              f"por 1.000 tokens; zona {nome_da_zona(l['zona'], A, B)}.")
    for _, l in tab[tab["depende_das_exclusoes"]].iterrows():
        print(f"Atenção (†): '{l['rotulo']}' está na zona {nome_da_zona(l['zona'], A, B)} "
              f"com as exclusões do pesquisador ({A['nome']} {l['freq_a']} ocorrências; {B['nome']} {l['freq_b']}); "
              f"sem elas, ficaria na zona {nome_da_zona(l['zona_sem_exclusoes'], A, B)} "
              f"({A['nome']} {l['freq_a_sem_exclusoes']}; {B['nome']} {l['freq_b_sem_exclusoes']}).")
    if "zona_pesos_iguais" in tab:
        muda = tab[tab["zona_pesos_iguais"] != tab["zona"]]
        print(f"Com pesos iguais para os planos do grupo (média simples das taxas), {len(muda)} termo(s) mudariam de zona"
              + (": " + "; ".join(f"{l['rotulo']} ({nome_da_zona(l['zona'], A, B)} → "
                                  f"{nome_da_zona(l['zona_pesos_iguais'], A, B)})" for _, l in muda.iterrows())
                 if len(muda) else "") + ".")
    acrescentados = tab[tab["origem"] == "pedido do pesquisador"]
    if len(acrescentados):
        print("Acrescentados por pedido do pesquisador (fora dos recortes): " + ", ".join(acrescentados["rotulo"]) + ".")
    for item in tab.attrs.get("nao_encontrados", []):
        print(f"Pedido não encontrado em nenhum dos dois lados: {item}.")
    print(f"Passam nos critérios de keyness (G² ≥ {num(LIMIAR_G2)}, |Log Ratio| ≥ {num(LIMIAR_LOG_RATIO, 1)} e frequência "
          f"combinada ≥ {FREQ_MINIMA_COMPARACAO}): {int(tab['distintivo'].sum())} de {len(tab)}; resistem à correção de "
          f"Bonferroni: {int(tab['resiste_bonferroni'].sum())}.")
```

```python
modelos = {}        # um modelo do spaCy por configuração de tokenização
PROCESSADOS = {}    # por plano: unidades, tokens e exclusões
for sigla, cfg in DOCUMENTOS.items():
    if cfg["separar_simbolo_final"] not in modelos:
        modelos[cfg["separar_simbolo_final"]] = carregar_modelo(cfg["separar_simbolo_final"])
    nlp = modelos[cfg["separar_simbolo_final"]]
    re_palavra = compilar_re_palavra(cfg["siglas_com_e_comercial"])
    stopwords = montar_stopwords(nlp)

    analisadas, retiradas = selecionar_unidades(documentos[sigla], UNIDADES_EXCLUIDAS.get(sigla, {}))
    tokens = processar_documento(analisadas, nlp, stopwords, re_palavra, CORRECOES_DE_LEMA.get(sigla, {}))
    entradas = preparar_exclusoes(EXCLUSOES.get(sigla, {}), nlp, re_palavra, EXPRESSOES_PROTEGIDAS.get(sigla, []))
    tokens, ocorrencias = aplicar_exclusoes(tokens, entradas)
    PROCESSADOS[sigla] = {"analisadas": analisadas, "retiradas": retiradas, "tokens": tokens,
                          "n_tokens": len(tokens), "ocorrencias_exclusoes": ocorrencias}
N_TOKENS = pd.Series({sigla: p["n_tokens"] for sigla, p in PROCESSADOS.items()})
STOPWORDS = montar_stopwords(modelos[False])

# Tabelas completas de cada plano: unigramas (lemas de conteúdo) e bigramas (pares de tokens de conteúdo
# adjacentes no texto original). Com as exclusões do pesquisador: base dos recortes e dos gráficos.
# Sem elas: usadas apenas para verificar se a zona de um termo depende das exclusões.
TABELAS = {tipo: {} for tipo in TIPOS}
TABELAS_SEM_EXCLUSOES = {tipo: {} for tipo in TIPOS}
for sigla, p in PROCESSADOS.items():
    tokens_sem = p["tokens"].assign(conteudo=p["tokens"]["conteudo_antes_das_exclusoes"])
    TABELAS["unigramas"][sigla] = tabela_frequencias(p["tokens"])
    TABELAS["bigramas"][sigla] = tabela_bigramas(p["tokens"])
    TABELAS_SEM_EXCLUSOES["unigramas"][sigla] = tabela_frequencias(tokens_sem)
    TABELAS_SEM_EXCLUSOES["bigramas"][sigla] = tabela_bigramas(tokens_sem)
    salvar_tabela(TABELAS["unigramas"][sigla], f"00_frequencias_lemas_completa_{sigla}")
    salvar_tabela(TABELAS["bigramas"][sigla], f"00_frequencias_bigramas_completa_{sigla}")

# (a) Perfil comparativo: extensão e vocabulário de cada plano
perfil = pd.DataFrame({sigla: {
    "Unidades analisadas (com a unidade 0)": len(p["analisadas"]),
    "Tokens (palavras): denominador das taxas": p["n_tokens"],
    "Tamanho em relação ao menor plano (vezes)": p["n_tokens"] / N_TOKENS.min(),
    "Tokens de conteúdo (pipeline)": int(p["tokens"]["conteudo_antes_das_exclusoes"].sum()),
    "Tokens de conteúdo retirados por exclusões do pesquisador": int(
        (p["tokens"]["conteudo_antes_das_exclusoes"] & ~p["tokens"]["conteudo"]).sum()),
    "Tokens de conteúdo considerados": int(p["tokens"]["conteudo"].sum()),
    "Lemas de conteúdo distintos": len(TABELAS["unigramas"][sigla]),
    "Bigramas distintos": len(TABELAS["bigramas"][sigla]),
    f"Bigramas com frequência >= {FREQ_MINIMA_BIGRAMA}": int(
        (TABELAS["bigramas"][sigla]["frequencia"] >= FREQ_MINIMA_BIGRAMA).sum()),
} for sigla, p in PROCESSADOS.items()})
perfil.columns = [DOCUMENTOS[s]["nome"] for s in perfil.columns]
display(Markdown("**(a) Perfil comparativo dos planos**"))
display(perfil.style.format(lambda v: num(v, 0) if float(v).is_integer() else num(v, 2)))
salvar_tabela(perfil, "00_perfil_comparativo")

# (b) Tratamentos de cada plano, reproduzidos dos notebooks individuais
tratamentos = pd.DataFrame({sigla: {
    "Correções de lema: entradas na configuração": len(CORRECOES_DE_LEMA.get(sigla, {})),
    "Correções de lema: tokens afetados": int((p["tokens"]["lema_spacy"] != p["tokens"]["lema"]).sum()),
    "Separar '+' e hífen suspenso do fim da palavra": "sim" if DOCUMENTOS[sigla]["separar_simbolo_final"] else "não",
    "Siglas com '&' como token": "sim" if DOCUMENTOS[sigla]["siglas_com_e_comercial"] else "não",
    "Exclusões do pesquisador: expressões na lista": sum(len(v) for v in EXCLUSOES.get(sigla, {}).values()),
    "Exclusões do pesquisador: ocorrências localizadas": sum(
        n for e, n in p["ocorrencias_exclusoes"].items() if not e.startswith("[")),
} for sigla, p in PROCESSADOS.items()})
tratamentos.columns = [DOCUMENTOS[s]["nome"] for s in tratamentos.columns]
display(Markdown("**(b) Tratamentos de cada plano, reproduzidos dos notebooks individuais**"))
exibir(tratamentos, 0)
salvar_tabela(tratamentos, "00_tratamentos_por_plano")

# (c) Grupos: planos que formam cada bloco e o conjunto dos cinco planos, com o peso de cada um no total
# de tokens do grupo (num grupo tratado como um único texto, o plano mais longo pesa mais).
composicao = pd.DataFrame([{
    "grupo": nome_grupo,
    "plano": DOCUMENTOS[p]["nome"],
    "tokens": int(N_TOKENS[p]),
    "% dos tokens do grupo": N_TOKENS[p] / N_TOKENS[grupo["planos"]].sum() * 100,
} for nome_grupo, grupo in list(BLOCOS.items()) + [("Todos", TODOS)] for p in grupo["planos"]]).set_index(["grupo", "plano"])
display(Markdown("**(c) Composição dos grupos (blocos e conjunto dos cinco planos)**"))
exibir(composicao, 1)
salvar_tabela(composicao, "00_composicao_dos_grupos")

print(f"Stopwords aplicadas: {len(STOPWORDS)} (lista do spaCy, {len(STOPWORDS_ADICIONAIS)} adicionada(s), "
      f"{len(STOPWORDS_MANTIDAS)} mantida(s)), as mesmas para todos os planos.")
print(f"Síntese factual: os planos têm de {num(N_TOKENS.min(), 0)} ({DOCUMENTOS[N_TOKENS.idxmin()]['nome']}) a "
      f"{num(N_TOKENS.max(), 0)} ({DOCUMENTOS[N_TOKENS.idxmax()]['nome']}) tokens, uma razão de "
      f"{num(N_TOKENS.max() / N_TOKENS.min(), 1)} vezes. Os grupos têm "
      + "; ".join(f"{g}: {num(N_TOKENS[c['planos']].sum(), 0)}" for g, c in list(BLOCOS.items()) + [("Todos", TODOS)])
      + " tokens. As exclusões do pesquisador retiram "
      + "; ".join(f"{DOCUMENTOS[s]['nome']}: {num(perfil.loc['Tokens de conteúdo retirados por exclusões do pesquisador', DOCUMENTOS[s]['nome']], 0)}"
                  for s in DOCUMENTOS if perfil.loc["Tokens de conteúdo retirados por exclusões do pesquisador", DOCUMENTOS[s]["nome"]])
      + " tokens de conteúdo.")
```

**Método e escolhas.** Cada plano é processado com o pipeline dos notebooks individuais: modelo `en_core_web_sm` do spaCy; título e subtítulo como unidade 0; palavras compostas com hífen reunidas em um só token (*open-source*, *human-computer* e *AI-native* contam como **um unigrama**); **token** = palavra formada por letras, com hífen interno ou sigla com pontos (números, pontuação e símbolos não são tokens); **tokens de conteúdo** = substantivos, nomes próprios, verbos, adjetivos e advérbios fora da lista de *stopwords* do spaCy, verificada pelo lema, com pelo menos 2 caracteres. Cada plano recebe, além disso, o que o seu notebook individual fixou (tabela b): as correções de lema; nos dois planos chineses, a separação do "+" e do hífen suspenso no fim da palavra e a contagem das siglas com "&" como um token; no PBIA e no Winning the Race, as exclusões definidas pelo pesquisador. As palavras excluídas deixam de ser tokens de conteúdo (não formam lemas nem bigramas), mas continuam no total de tokens, que é o denominador de todas as taxas. A tabela (c) mostra a composição dos grupos usados nas seções 14 e 15 e o peso de cada plano na soma dos tokens do grupo.

**Como ler.** A tabela (a) expõe a disparidade de tamanho entre os planos, que justifica a comparação por 1.000 tokens, e o efeito das exclusões sobre os tokens de conteúdo. A tabela (b) resume os tratamentos herdados dos notebooks individuais. Na tabela (c), a última coluna indica quanto cada plano pesa na taxa do grupo: num grupo tratado como um único texto, o plano mais longo pesa mais.

**Interpretação.** *Observação:* o PBIA é o plano mais longo (16.391 tokens, 586 unidades) e o AI+, o mais curto (2.874 tokens, 65 unidades): uma razão de 5,7 vezes. Os demais têm 11.331 (New Gen), 8.931 (AI Continent), 8.198 (Winning the Race) e 7.554 tokens (Apply AI). Os bigramas com frequência mínima de 2 vão de 87 (AI+) a 493 (New Gen); o PBIA tem 479. ## 12. Recortes ampliados, grupos e preparação das 18 comparações

*Construída por comando do pesquisador (2026-10-05).*

```python
# Lados das comparações: o PBIA (eixo X de todos os gráficos), cada um dos outros cinco planos, os três
# blocos e o conjunto dos cinco planos (eixo Y). Os grupos são tratados como um único texto.
cfg_pbia = DOCUMENTOS[REFERENCIA]
LADO_PBIA = montar_lado(REFERENCIA, cfg_pbia["nome"], cfg_pbia["no"], cfg_pbia["do"], [REFERENCIA],
                        COR_DOCUMENTO[REFERENCIA], agrupado=False)
LADOS = {}
for sigla in INDIVIDUAIS:
    cfg = DOCUMENTOS[sigla]
    LADOS[sigla] = montar_lado(sigla, cfg["nome"], cfg["no"], cfg["do"], [sigla], COR_DOCUMENTO[sigla], agrupado=False)
for chave, grupo in list(BLOCOS.items()) + [("Todos", TODOS)]:
    LADOS[chave] = montar_lado(chave, grupo["nome"], grupo["no"], grupo["do"], grupo["planos"], COR_GRUPO[chave],
                               agrupado=True)

# Conferência: o bloco EUA tem um só plano; tratado como grupo, deve reproduzir exatamente o plano.
for tipo in TIPOS:
    g, p = LADOS["EUA"]["tabelas"][tipo], LADOS["EUA_2025"]["tabelas"][tipo]
    assert g["frequencia"].equals(p["frequencia"].reindex(g.index)) and np.allclose(
        g["por_1000_tokens"], p["por_1000_tokens"].reindex(g.index)), "a junção de planos não reproduz o plano único"

# Tabela dos recortes: tamanho pedido, valor do corte, tamanho final e o primeiro termo que fica de fora.
linhas = []
for lado in [LADO_PBIA] + list(LADOS.values()):
    for tipo, t in TIPOS.items():
        info = lado["info_recortes"][tipo]
        fora = info["primeiro_fora"]
        linhas.append({
            "lado": lado["nome"],
            "planos": " + ".join(DOCUMENTOS[p]["nome"] for p in lado["planos"]),
            "tipo": tipo,
            "pedido": info["pedido"],
            "ocorrências do N-ésimo": info["valor_do_corte"],
            "termos no recorte": info["n_recorte"],
            "empatados acrescentados": len(info["empatados_acrescentados"]),
            "primeiro fora do recorte": (f"{fora[0]} ({fora[1]})" if fora else
                                         f"nenhum com frequência ≥ {t['freq_minima']}"),
        })
tabela_recortes = pd.DataFrame(linhas).set_index(["lado", "tipo"])
exibir(tabela_recortes, 0)
salvar_tabela(tabela_recortes, "01_recortes")

# Lemas de cada grupo que não estão no recorte ampliado de nenhum dos seus planos (só entram pela soma).
for chave in list(BLOCOS) + ["Todos"]:
    lado = LADOS[chave]
    for tipo in TIPOS:
        individuais = set().union(*(LADOS[p]["recortes"][tipo].index for p in lado["planos"]))
        so_no_grupo = [t for t in lado["recortes"][tipo].index if t not in individuais]
        if so_no_grupo and len(lado["planos"]) > 1:
            print(f"{lado['nome']} ({tipo}): no recorte do grupo, mas em nenhum recorte dos seus planos: "
                  + ", ".join(f"{t} ({int(lado['tabelas'][tipo].loc[t, 'frequencia'])})" for t in so_no_grupo))

# Todas as comparações pedidas: 5 individuais, 3 blocos e o conjunto, em unigramas e bigramas.
COMPARACOES = {(chave, tipo): preparar_convergencia(LADO_PBIA, LADOS[chave], tipo)
               for chave in INDIVIDUAIS + list(BLOCOS) + ["Todos"] for tipo in TIPOS}

# Correção de Bonferroni sobre o total de testes de todas as comparações (skill 03, seção 3.4):
# um teste por termo com frequência combinada mínima, em cada gráfico.
N_TESTES = int(sum((tab["freq_combinada"] >= FREQ_MINIMA_COMPARACAO).sum() for tab in COMPARACOES.values()))
G2_BONFERRONI = float(chi2.isf(ALFA_BONFERRONI / N_TESTES, 1))
for tab in COMPARACOES.values():
    tab["resiste_bonferroni"] = tab["distintivo"] & (tab["G2"] >= G2_BONFERRONI)

resumo = pd.DataFrame([{
    "comparação": f"{LADO_PBIA['nome']} × {LADOS[chave]['nome']}",
    "tipo": tipo,
    "termos no gráfico": len(tab),
    "nos dois recortes": int((tab["origem"] == "os dois recortes").sum()),
    f"zona {LADO_PBIA['nome']}": int((tab["zona"] == "A").sum()),
    "zona neutra": int((tab["zona"] == "neutra").sum()),
    "zona do outro lado": int((tab["zona"] == "B").sum()),
    "fora da escala": ", ".join(tab.loc[tab["fora_da_escala"], "rotulo"]) or "—",
    "distintivos": int(tab["distintivo"].sum()),
    "zona depende das exclusões": int(tab["depende_das_exclusoes"].sum()),
} for (chave, tipo), tab in COMPARACOES.items()]).set_index(["comparação", "tipo"])
exibir(resumo, 0)
print(f"Síntese factual: {len(COMPARACOES)} gráficos; {N_TESTES} testes de keyness ao todo (termos com frequência "
      f"combinada >= {FREQ_MINIMA_COMPARACAO}); com a correção de Bonferroni (alfa = {num(ALFA_BONFERRONI)}), "
      f"o G² crítico é {num(G2_BONFERRONI)}.")


def mostrar_convergencia(chave, tipo, numero):
    """Desenha o gráfico de uma comparação, exibe a tabela com os valores exatos e imprime a síntese factual."""
    tab = COMPARACOES[(chave, tipo)]
    A, B, t = LADO_PBIA, LADOS[chave], TIPOS[tipo]
    if chave in BLOCOS or chave == "Todos":
        juntos = [DOCUMENTOS[p]["nome"] for p in B["planos"]]
        if len(juntos) > 1:
            composicao_b = f"{', '.join(juntos[:-1])} e {juntos[-1]} juntos"
            nota = f"{composicao_b}: ocorrências somadas ÷ tokens somados"
        else:
            composicao_b = nota = juntos[0]
        titulo_b = "todos os planos" if chave == "Todos" else B["nome"]
        recorte_b = (f"{B['nome']} ({composicao_b})" if chave != "Todos" else f"os cinco planos juntos")
        linha_1 = f"{t['titulo']}: " + t["recorte"].format(n=t["top_n"], de=f"{A['do']} e {B['do']}")
    else:
        nota, titulo_b, recorte_b = None, B["nome"], B["nome"]
        linha_1 = f"{t['titulo']}: " + t["recorte"].format(n=t["top_n"], de="de cada documento")
    linha_2 = (f"Recortes: {A['nome']} {len(A['recortes'][tipo])} · {recorte_b} {len(B['recortes'][tipo])} · "
               f"união: {len(tab)} {t['plural']}")
    fig = grafico_convergencia(tab, titulo_b, [linha_1, linha_2], nota_grupo=nota)
    salvar_figura(fig, f"{numero}_convergencia_{tipo}_PBIA_x_{chave}")
    plt.show()
    tabela = tabela_convergencia_exibicao(tab)
    salvar_tabela(tabela, f"{numero}_convergencia_{tipo}_PBIA_x_{chave}")
    exibir(tabela)
    sintese_convergencia(tab)
```

**Método e escolhas.** Este é o método comum aos 18 gráficos das seções 13 a 15; o texto de cada gráfico traz apenas o que é próprio dele.

1. **Lados.** O eixo X é sempre o PBIA; o eixo Y é o outro lado: um plano (seção 13), um bloco (seção 14) ou os cinco planos (seção 15). Os blocos seguem o comando: UE = AI Continent + Apply AI; China = New Gen + AI+; EUA = Winning the Race, o único plano dos EUA na pasta. Cada grupo é tratado como **um único texto**: para cada termo, as ocorrências nos planos do grupo são somadas e divididas pela soma dos tokens (função `tabela_do_grupo`). Com isso, cada grupo recebe uma **nova lista dos termos mais frequentes**, calculada a partir da soma, e não copiada das listas dos planos. A célula confere que o grupo EUA, com um só plano, reproduz exatamente o plano.
2. **Recortes.** Para cada lado, os 30 termos mais frequentes (`TOP_N_LEMAS_CONVERGENCIA` e `TOP_N_BIGRAMAS_CONVERGENCIA`), ampliados até incluir todos os empatados com o 30º (regra do pesquisador; função `recorte_com_empates`). A faixa cinza é a zona neutra, entre y = x/2 e y = 2x (`RAZAO_ZONA_NEUTRA` = 2, o equivalente a \|*Log Ratio*\| < 1). A cor do ponto indica a zona: verde (cor do PBIA) quando a taxa no PBIA é ao menos o dobro da taxa do outro lado (ou quando o termo só ocorre no PBIA); a cor do outro lado, no caso inverso; cinza na zona neutra, que é o **vocabulário em comum**. As cores dos planos são as do notebook comparativo já existente na pasta; cada bloco recebe a cor-base do seu país ou bloco (UE amarelo, China vermelho, EUA azul-escuro), e o conjunto dos cinco planos, roxo, cor que não pertence a nenhum deles.
4. **Medidas por termo.** A tabela abaixo de cada gráfico traz as ocorrências e as taxas, o *Log Ratio* (Hardie, 2014) e o G² de log-verossimilhança (Dunning, 1993; Rayson & Garside, 2000), calculados pela mesma função do notebook BRICS × OCDE (`tabela_keyness`, *keyness* em pares). Um termo é **distintivo** quando G² ≥ 6,63, \|*Log Ratio*\| ≥ 1 e frequência combinada ≥ 5 (skill 03, seção 3.5). A correção de Bonferroni considera os 992 testes dos 18 gráficos (um por termo com frequência combinada ≥ 5), com G² crítico de 19,50. As zonas são descritivas. Por construção, nenhum termo da zona neutra é distintivo; um G² alto na zona neutra indica uma diferença estatisticamente detectável, mas menor que o dobro.
5. **Robustez.** (a) **Zona sem as exclusões:** os mesmos termos são recontados sem as exclusões do pesquisador no PBIA e no Winning the Race. (b) **Pesos iguais** (grupos com mais de um plano): a tabela traz a taxa de cada plano do grupo e a zona que o termo teria se a taxa do grupo fosse a média simples das taxas dos planos, e não a soma.
6. **Rótulos.** Seguem uma regra fixa, sem escolha termo a termo. Na zona neutra, são rotulados até 10 termos, os de maior taxa somada (`N_ROTULOS_NEUTROS`). Em cada zona colorida, até 7, os de maior taxa no lado em que o termo é mais frequente (`N_ROTULOS_POR_ZONA`). Os empatados no último lugar entram se não passarem de 3 a mais; se passarem, nenhum deles entra. Os termos com † são sempre rotulados. O pesquisador pode acrescentar termos em `DESTAQUES`. A tabela traz todos os termos do gráfico.
7. **Alterações em relação ao gráfico de referência** (notebook BRICS × OCDE, seções 16 e 17):
   - o PBIA fica sempre no eixo X; título, eixos, legenda e instrução de leitura trazem o nome de cada lado, e, nos grupos, a instrução de leitura informa os planos somados;

## 16. Síntese: o vocabulário que o PBIA tem em comum com cada plano, cada bloco e o conjunto

*Construída por comando do pesquisador (2026-10-05): identificar em que o PBIA tem vocabulário em comum com cada plano, cada bloco e todos eles, em unigramas e bigramas.*

```python
# Síntese das 18 comparações: em que zona cada termo ficou em cada comparação. "em comum" = zona neutra
# (os dois lados usam o termo com taxas que diferem menos que RAZAO_ZONA_NEUTRA vezes).
CHAVES_SINTESE = INDIVIDUAIS + list(BLOCOS) + ["Todos"]
NOME_COLUNA = {chave: ("todos (5 planos)" if chave == "Todos" else
                       f"{LADOS[chave]['nome']} (bloco)" if chave in BLOCOS else LADOS[chave]["nome"])
               for chave in CHAVES_SINTESE}
SIMBOLO_ZONA = {"neutra": "em comum", "A": "mais no PBIA", "B": "mais no outro"}

SINTESE = {}
for tipo, t in TIPOS.items():
    # Uma linha por termo (lema ou par de lemas). Todo termo em comum ocorre no PBIA; o rótulo dos bigramas
    # é a forma escrita mais frequente no PBIA (as tabelas de cada gráfico usam a dos dois lados).
    zonas = {}
    for chave in CHAVES_SINTESE:
        for termo, l in COMPARACOES[(chave, tipo)].iterrows():
            zonas.setdefault(termo, {})[NOME_COLUNA[chave]] = SIMBOLO_ZONA[l["zona"]]
    matriz = pd.DataFrame(zonas).T.reindex(columns=[NOME_COLUNA[c] for c in CHAVES_SINTESE]).fillna("—")
    tab_pbia = LADO_PBIA["tabelas"][tipo]
    matriz.insert(0, "PBIA: por 1.000 tokens", tab_pbia["por_1000_tokens"].reindex(matriz.index).fillna(0.0))
    if tipo == "bigramas":
        matriz.insert(0, "forma escrita no PBIA", tab_pbia["forma_principal"].reindex(matriz.index).fillna("—"))
    individuais = [NOME_COLUNA[c] for c in INDIVIDUAIS]
    matriz["em comum: planos (de 5)"] = (matriz[individuais] == "em comum").sum(axis=1)
    matriz["em comum: comparações (de 9)"] = (matriz[[NOME_COLUNA[c] for c in CHAVES_SINTESE]] == "em comum").sum(axis=1)
    em_comum = matriz[matriz["em comum: comparações (de 9)"] > 0].sort_values(
        ["em comum: comparações (de 9)", "PBIA: por 1.000 tokens"], ascending=[False, False])
    em_comum.index.name = "bigrama (lemas)" if tipo == "bigramas" else "lema"
    SINTESE[tipo] = em_comum
    display(Markdown(f"**{t['titulo']}: termos em comum (zona neutra) em pelo menos uma das 9 comparações** "
                     f"({len(em_comum)} {t['plural']}; ordenados pelo número de comparações e pela taxa no PBIA)"))
    exibir(em_comum)
    salvar_tabela(em_comum, f"20_sintese_vocabulario_em_comum_{tipo}")

    em_todos = em_comum[em_comum["em comum: planos (de 5)"] == len(INDIVIDUAIS)]
    print(f"Síntese factual ({tipo}): {len(em_comum)} {t['plural']} ficam na zona neutra em ao menos uma comparação. "
          f"Em comum com os {len(INDIVIDUAIS)} planos, um a um: "
          + (", ".join(em_todos["forma escrita no PBIA"] if tipo == "bigramas" else em_todos.index)
             if len(em_todos) else "nenhum") + ". "
          + " · ".join(f"{NOME_COLUNA[c]}: {int((matriz[NOME_COLUNA[c]] == 'em comum').sum())} em comum"
                       for c in CHAVES_SINTESE) + ".")
```

**Método e escolhas.** Para cada termo de algum dos 18 gráficos, a tabela mostra a zona em que ele ficou em cada uma das 9 comparações: "em comum" (zona neutra), "mais no PBIA" ou "mais no outro" (lado). O travessão indica que o termo não está entre os mais frequentes de nenhum dos dois lados naquela comparação, e não que esteja ausente do plano. Entram na tabela os termos que ficam em comum em ao menos uma comparação; todos ocorrem no PBIA, e a coluna da taxa no PBIA ajuda a ordená-los. A coluna "em comum: planos (de 5)" conta só as comparações individuais (seção 13); a coluna "comparações (de 9)" conta também os blocos e o conjunto. Nos bigramas, a linha é o par de lemas, e o rótulo é a forma escrita mais frequente no PBIA.

**Como ler.** Uma linha com "em comum" nas cinco primeiras colunas de comparação indica um termo que o PBIA usa com peso parecido ao de cada um dos cinco planos. A linha de síntese abaixo de cada tabela conta os termos em comum por comparação.

**Interpretação.** *Observação:* nos unigramas, 44 lemas ficam em comum em ao menos uma comparação.
