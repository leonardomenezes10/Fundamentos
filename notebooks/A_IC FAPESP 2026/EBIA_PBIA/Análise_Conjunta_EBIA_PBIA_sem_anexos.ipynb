```python
# =============================================================================
# AMBIENTE, ESTILO E FUNÇÕES DO PIPELINE
# Os trechos entre as marcas "início/fim da cópia literal" são cópia letra por letra da célula de funções dos
# notebooks individuais (skill 03, seção 4.2, item 3). A seção de carregamento confere essa identidade.
# =============================================================================
import ast
import hashlib
import json
import math
import platform
import re
import sys
import textwrap
import zlib
from collections import Counter, defaultdict
from datetime import datetime
from pathlib import Path

import matplotlib
import matplotlib.patheffects as path_effects
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy
import spacy
from IPython.display import Markdown, display
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
from matplotlib.ticker import FuncFormatter, LogLocator, MaxNLocator, NullLocator
from scipy.stats import chi2
from spacy.lang.pt.stop_words import STOP_WORDS as STOPWORDS_SPACY

# --- Restrição de pasta (skill 03, seção 1.3) -------------------------------
PASTA = Path.cwd()
if PASTA.name != "EBIA_PBIA":
    raise RuntimeError(f"Execute este notebook de dentro da pasta EBIA_PBIA (pasta atual: {PASTA}).")
PASTA_SAIDA = PASTA / "resultados" / "Conjunta_EBIA_PBIA" / VERSAO_SAIDA   # usada só com SALVAR_SAIDAS = True (seção 4.4)
SAIDAS_GRAVADAS = []

SIGLAS = list(DOCUMENTOS)
SIG_A, SIG_B = COMPARACOES[0]   # EBIA: eixo X e Log Ratio positivo; PBIA: eixo Y e Log Ratio negativo (convenção fixa)
ROTULO = {"EBIA": "EBIA", "PBIA_sem_anexos": "PBIA (sem anexos)", "PBIA_com_anexos": "PBIA (com anexos)"}
CURTO = {"EBIA": "EBIA", "PBIA_sem_anexos": "PBIA", "PBIA_com_anexos": "PBIA"}
NOME_NO = {"EBIA": "na EBIA", "PBIA_sem_anexos": "no PBIA", "PBIA_com_anexos": "no PBIA"}   # "taxa na EBIA"
NOME_DO = {"EBIA": "da EBIA", "PBIA_sem_anexos": "do PBIA", "PBIA_com_anexos": "do PBIA"}
CUT_POR_SIGLA = {"EBIA": None, "PBIA_sem_anexos": "sem_anexos", "PBIA_com_anexos": "com_anexos"}
NOMES_TIPO = {"heading": "Títulos internos", "paragraph": "Parágrafos", "list_item": "Itens de lista"}
MARCADOR_DOC = {"EBIA": "o", "PBIA_sem_anexos": "o"}   # comando do pesquisador (2026-10-14): bolinhas para os dois documentos

# --- Estilo das figuras: o dos notebooks individuais (fonte ≥ 9 pt; PNG 300 dpi e PDF só com comando de gravação)
plt.rcParams.update({
    "font.family": "DejaVu Sans", "font.size": 10, "axes.titlesize": 11, "axes.labelsize": 10,
    "xtick.labelsize": 9, "ytick.labelsize": 9, "legend.fontsize": 9, "figure.dpi": 110,
    "savefig.dpi": 300, "axes.spines.top": False, "axes.spines.right": False,
})
COR_TINTA = "#0b0b0b"     # texto principal
COR_TINTA_2 = "#52514e"   # texto secundário (rótulos, anotações)
COR_GRADE = "#e4e3df"     # grade e linhas de apoio
COR_LIGACAO = "#b9b8b2"   # linhas de ligação e limites da faixa neutra
# Estilo do gráfico de convergência e divergência: o do notebook BRICS × OCDE, por comando do pesquisador.
ESTILO_CONVERGENCIA = {
    "figure.dpi": 110, "savefig.dpi": 300, "savefig.bbox": "tight", "figure.facecolor": "white",
    "axes.facecolor": "white", "font.size": 10, "axes.titlesize": 11, "axes.labelsize": 10, "xtick.labelsize": 9,
    "ytick.labelsize": 9, "legend.fontsize": 9, "axes.spines.top": False, "axes.spines.right": False,
    "axes.edgecolor": COR_TINTA_2, "axes.linewidth": 0.6, "axes.grid": True, "axes.axisbelow": True,
    "grid.color": COR_GRADE, "grid.linewidth": 0.6, "grid.linestyle": "-", "text.color": COR_TINTA,
    "axes.labelcolor": COR_TINTA, "xtick.color": COR_TINTA_2, "ytick.color": COR_TINTA_2, "legend.frameon": False,
    "pdf.fonttype": 42,
}

# >>> início da cópia literal (notebooks individuais: números no padrão brasileiro)
# --- Números no padrão brasileiro (o valor nunca é alterado, só a escrita) ----
def br(valor, casas=0):
    if valor is None or (isinstance(valor, float) and math.isnan(valor)):
        return "–"
    texto = f"{valor:,.{casas}f}"
    return texto.replace(",", " ").replace(".", ",").replace(" ", ".")


def br_pct(fracao, casas=1):
    return f"{br(100 * fracao, casas)}%"


EIXO_BR = FuncFormatter(lambda v, _: br(v, 0) if abs(v - round(v)) < 1e-9 else br(v, 1))


def tabela_br(df, casas=None):
    """Cópia da tabela com os números escritos no padrão brasileiro, só para exibição."""
    casas = casas or {}
    out = df.copy()
    for col in out.columns:
        serie = out[col]
        if pd.api.types.is_bool_dtype(serie) or not pd.api.types.is_numeric_dtype(serie):
            continue
        c = casas.get(col, 0 if pd.api.types.is_integer_dtype(serie) else 2)
        out[col] = serie.map(lambda v: br(v, c))
    return out
# <<< fim da cópia literal


def salvar_figura(fig, nome):
    """Grava PNG (300 dpi) e PDF em resultados/Conjunta_EBIA_PBIA/<VERSAO_SAIDA>/ — só com SALVAR_SAIDAS = True."""
    if not SALVAR_SAIDAS:
        return
    PASTA_SAIDA.mkdir(parents=True, exist_ok=True)
    for ext in ("png", "pdf"):
        alvo = PASTA_SAIDA / f"{NOME_SAIDA}_{nome}.{ext}"
        if alvo.exists() and not SOBRESCREVER:
            raise FileExistsError(f"{alvo} já existe: use uma nova VERSAO_SAIDA.")
        fig.savefig(alvo, dpi=300, bbox_inches="tight")
        SAIDAS_GRAVADAS.append(str(alvo.relative_to(PASTA)))


def salvar_tabela(df, nome):
    """Grava a tabela completa (CSV, UTF-8, separador ';', vírgula decimal) — só com SALVAR_SAIDAS = True."""
    if not SALVAR_SAIDAS:
        return
    PASTA_SAIDA.mkdir(parents=True, exist_ok=True)
    alvo = PASTA_SAIDA / f"{NOME_SAIDA}_{nome}.csv"
    if alvo.exists() and not SOBRESCREVER:
        raise FileExistsError(f"{alvo} já existe: use uma nova VERSAO_SAIDA.")
    df.to_csv(alvo, sep=";", decimal=",", encoding="utf-8", index=True)
    SAIDAS_GRAVADAS.append(str(alvo.relative_to(PASTA)))


# >>> início da cópia literal (notebooks individuais: marcações, pipeline do spaCy, medidas e recorte "top N")
# --- Passo 0: leitura das marcações (skill 02, seção 3.2) ---------------------
RX_MARCADA = re.compile(r"([^\W\d_]+(?:-[^\W\d_]+)*)" + re.escape(MARCADOR_REPETICAO))


def ler_marcacoes(texto):
    """Tira os asteriscos e devolve o texto literal e (início, fim, palavra) de cada palavra marcada."""
    partes, marcadas, ultimo, tamanho = [], [], 0, 0
    for m in RX_MARCADA.finditer(texto):
        trecho = texto[ultimo:m.start()]
        partes.append(trecho)
        tamanho += len(trecho)
        marcadas.append((tamanho, tamanho + len(m.group(1)), m.group(1)))
        partes.append(m.group(1))
        tamanho += len(m.group(1))
        ultimo = m.end()
    partes.append(texto[ultimo:])
    limpo = "".join(partes)
    if MARCADOR_REPETICAO in limpo:
        raise ValueError(f"Marcador de repetição fora de palavra: {texto[:80]}")
    return limpo, marcadas


def agrupar_ocorrencias(limpo, marcadas):
    """Uma ocorrência = palavras marcadas seguidas, separadas apenas por um espaço (como na skill 01)."""
    grupos = []
    for ini, fim, palavra in marcadas:
        if grupos and limpo[grupos[-1][-1][1]:ini] == " ":
            grupos[-1].append((ini, fim, palavra))
        else:
            grupos.append([(ini, fim, palavra)])
    return grupos


def sha256_arquivo(caminho):
    return hashlib.sha256(Path(caminho).read_bytes()).hexdigest()


def carregar_e_verificar(caminho):
    """Lê o JSON da skill 01 e confere integridade e marcações (skill 02, seção 4.2, item 4)."""
    if not caminho.exists():
        raise FileNotFoundError(f"{caminho.name} não existe: a skill 01 (extração) precisa ser executada antes.")
    js = json.loads(caminho.read_text(encoding="utf-8"))
    doc, ext, secoes = js["document"], js["extraction"], js["sections"]
    catalogo = {p["id"]: p for p in ext.get("repetition_patterns", [])}
    linhas = []

    def checar(nome, ok, detalhe=""):
        linhas.append((nome, "ok" if ok else "FALHOU", detalhe))

    checar("produzido pela skill 01 deste corpus",
           ext.get("skill") == "skill_extração" and ext.get("corpus") == "EBIA_PBIA",
           f"skill = {ext.get('skill')}; corpus = {ext.get('corpus')}")
    checar("recorte corresponde à SIGLA", ext.get("cut") == CUT_ESPERADO, f"cut = {ext.get('cut')}")
    checar("ids sequenciais", [s["id"] for s in secoes] == list(range(1, len(secoes) + 1)), f"1 a {len(secoes)}")
    checar("nenhuma unidade vazia", all(s["text"].strip() for s in secoes))
    checar("número de unidades = registrado", len(secoes) == ext.get("units_extracted"),
           f"{len(secoes)} unidades; registrado {ext.get('units_extracted')}")
    checar("título presente", bool(doc.get("title")), doc.get("title") or "")
    checar("idioma = " + IDIOMA_ESPERADO, doc.get("language") == IDIOMA_ESPERADO, f"language = {doc.get('language')}")
    total_marcadas, grupos_ok, ids_ok, por_padrao = 0, True, True, Counter()
    for s in secoes:
        limpo, marcadas = ler_marcacoes(s["text"])
        total_marcadas += len(marcadas)
        grupos = agrupar_ocorrencias(limpo, marcadas)
        grupos_ok &= len(grupos) == len(s["marked_patterns"])
        for p in s["marked_patterns"]:
            ids_ok &= p in catalogo
            por_padrao[p] += 1
    checar("palavras marcadas = extraction.marked_words", total_marcadas == ext.get("marked_words"),
           f"{total_marcadas} no texto; registrado {ext.get('marked_words')}")
    checar("ocorrências marcadas de cada unidade = marked_patterns", grupos_ok)
    checar("todo padrão usado existe no catálogo", ids_ok, f"{len(catalogo)} padrões no catálogo")
    checar("ocorrências por padrão = catálogo",
           all(por_padrao[i] == p["occurrences"] for i, p in catalogo.items()) and set(por_padrao) <= set(catalogo))
    return js, pd.DataFrame(linhas, columns=["verificação", "resultado", "detalhe"])


# --- Passos 1 a 6: spaCy, ênclises, tokens, lemas e classes (skill 02, seção 3.4)
CLITICOS = sorted(set(PRONOMES_CLITICOS), key=len, reverse=True)
RX_ENCLISE = re.compile(r"([^\W\d_]+)-(" + "|".join(CLITICOS) + r")")
RX_MESOCLISE = re.compile(r"([^\W\d_]+)-(" + "|".join(CLITICOS) + r")-([^\W\d_]+)")
TRUNCADAS = {"á": "ar", "ê": "er", "í": "ir", "ô": "or"}   # alcançá-los → alcançar; fazê-lo → fazer
RX_PALAVRA = re.compile(r"[^\W\d_]+(?:[-&][^\W\d_]+)*")      # letras, com hífen ou & internos (P&D)
RX_ENUMERACAO = re.compile(r"(?i)(?=[a-z])(?:x{0,3}(?:ix|iv|v?i{0,3})|[a-z])")   # i, ii, iv, x, a, b...
EXPRESSOES = [tuple(e.lower().split()) for e in EXPRESSOES_RETIRADAS]
STOPWORDS = (set(STOPWORDS_SPACY) | set(STOPWORDS_ADICIONAIS)) - set(STOPWORDS_MANTIDAS)


def tokens_especiais(doc):
    """Ênclises e mesóclises (verbo separado do pronome) e hífens usados como travessão."""
    especiais, trocas = {}, []
    for t in doc:
        m = RX_MESOCLISE.fullmatch(t.text)
        if m:
            raiz, cl, fim = m.groups()
            especiais[t.i] = ("mesóclise", raiz + fim, cl)
            trocas.append((t.idx, t.idx + len(t.text), f"{cl} {(raiz + fim).lower()}".ljust(len(t.text))))
            continue
        m = RX_ENCLISE.fullmatch(t.text)
        if m:
            verbo, cl = m.groups()
            if cl in ("lo", "la", "los", "las") and verbo[-1].lower() in TRUNCADAS:
                especiais[t.i] = ("ênclise (forma truncada)", verbo, cl)
            else:
                especiais[t.i] = ("ênclise", verbo, cl)
                trocas.append((t.idx, t.idx + len(t.text), f"{cl} {verbo.lower()}"))
            continue
        if t.text in HIFEN_COMO_TRAVESSAO:
            especiais[t.i] = ("hífen como travessão", t.text.split("-"), None)
            trocas.append((t.idx, t.idx + len(t.text), t.text.replace("-", " ")))
    return especiais, trocas


def texto_auxiliar(texto, trocas):
    """Texto da passagem auxiliar: mesmo comprimento, só os trechos trocados mudam."""
    pedacos, ultimo = [], 0
    for ini, fim, novo in sorted(trocas):
        assert len(novo) == fim - ini
        pedacos += [texto[ultimo:ini], novo]
        ultimo = fim
    pedacos.append(texto[ultimo:])
    return "".join(pedacos)


def novo_item(unidade, texto, lema, upos, sent, ini=None, origem=None):
    return {"unidade": unidade, "texto": texto, "lema_modelo": lema, "upos": upos,
            "sent": sent, "ini": ini, "origem": origem}


def itens_da_unidade(uid, doc, nlp, registro):
    """Sequência de itens (palavras e não palavras) de uma unidade, na ordem do texto."""
    especiais, trocas = tokens_especiais(doc)
    aux = {}
    if trocas:
        aux = {x.idx: x for x in nlp(texto_auxiliar(doc.text, trocas))}
    itens = []
    for t in doc:
        if t.is_space:
            continue
        sent = t.sent.start
        if t.i not in especiais:
            itens.append(novo_item(uid, t.text, (t.lemma_ or t.text).lower(), t.pos_, sent, ini=t.idx))
            continue
        tipo, a, b = especiais[t.i]
        if tipo in ("ênclise", "mesóclise"):
            v = aux.get(t.idx + len(b) + 1)
            if v is None:
                raise RuntimeError(f"Ênclise não localizada na passagem auxiliar: {t.text} (unidade {uid})")
            upos = v.pos_ if v.pos_ in ("VERB", "AUX") else "VERB"
            itens.append(novo_item(uid, a, v.lemma_.lower(), upos, sent, origem=tipo))
            itens.append(novo_item(uid, b, b.lower(), "PRON", sent, origem=tipo))
            registro.append((t.text, tipo, a, v.lemma_.lower(), v.pos_, b))
        elif tipo == "ênclise (forma truncada)":
            lema = a[:-1].lower() + TRUNCADAS[a[-1].lower()]
            itens.append(novo_item(uid, a, lema, "VERB", sent, origem=tipo))
            itens.append(novo_item(uid, b, b.lower(), "PRON", sent, origem=tipo))
            registro.append((t.text, tipo, a, lema, "VERB (regra)", b))
        else:  # hífen como travessão: cada parte vira palavra; o hífen fica como pontuação
            pos = t.idx
            for k, parte in enumerate(a):
                x = aux.get(pos)
                if x is None or x.text != parte:
                    raise RuntimeError(f"Parte não localizada na passagem auxiliar: {t.text} (unidade {uid})")
                if k:
                    itens.append(novo_item(uid, "-", "-", "PUNCT", sent, origem=tipo))
                itens.append(novo_item(uid, parte, x.lemma_.lower(), x.pos_, sent, origem=tipo))
                pos += len(parte) + 1
            registro.append((t.text, tipo, " + ".join(a), "", "", ""))
    return itens


def processar(js, nlp):
    """Aplica os filtros, o spaCy, a retirada do excesso marcado e as classes de conteúdo."""
    secoes = js["sections"]
    unidades = []
    if INCLUIR_TITULO_NA_CONTAGEM:
        titulo = " ".join(x for x in (js["document"]["title"], js["document"].get("subtitle")) if x)
        unidades.append({"id": 0, "type": "title", "category": "Title", "path": [], "text": titulo, "marked_patterns": []})
    unidades += secoes
    filtros = []
    incluidas = []
    for s in unidades:
        motivo = None
        if s["type"] == "heading" and not INCLUIR_TITULOS_INTERNOS:
            motivo = "títulos internos excluídos (INCLUIR_TITULOS_INTERNOS = False)"
        elif s["category"] in CATEGORIAS_EXCLUIDAS:
            motivo = f"categoria {s['category']} excluída (CATEGORIAS_EXCLUIDAS)"
        elif s["id"] in UNIDADES_EXCLUIDAS:
            motivo = f"unidade excluída: {UNIDADES_EXCLUIDAS[s['id']]}"
        (filtros if motivo else incluidas).append((s, motivo))
    incluidas = [s for s, _ in incluidas]
    limpos, grupos = {}, {}
    for s in incluidas:
        limpo, marcadas = ler_marcacoes(s["text"])
        limpos[s["id"]] = limpo
        grupos[s["id"]] = agrupar_ocorrencias(limpo, marcadas)
    # retirada do excesso: a 1ª ocorrência de cada padrão (na ordem dos id) fica; as demais saem
    ocorrencias, vistos = [], {}
    for s in incluidas:
        for k, (padrao, grupo) in enumerate(zip(s["marked_patterns"], grupos[s["id"]])):
            excesso = RETIRAR_EXCESSO_MARCADO and padrao in vistos
            vistos.setdefault(padrao, s["id"])
            ocorrencias.append({"padrao": padrao, "unidade": s["id"], "ordem": k, "excesso": excesso,
                                "palavras": [p for _, _, p in grupo], "spans": [(i, f) for i, f, _ in grupo]})
    marcada_em = {}
    for o in ocorrencias:
        for ini, fim in o["spans"]:
            marcada_em[(o["unidade"], ini)] = o
    registro_especiais, itens, enumeracoes, expressoes = [], [], [], []
    docs = nlp.pipe([limpos[s["id"]] for s in incluidas], batch_size=32)
    for s, doc in zip(incluidas, docs):
        its = itens_da_unidade(s["id"], doc, nlp, registro_especiais)
        # comando do pesquisador: marcadores de enumeração entre parênteses — (i), (ii), (a) — não são palavras
        for k in range(1, len(its) - 1):
            if (RETIRAR_MARCADORES_DE_ENUMERACAO and its[k - 1]["texto"] == "(" and its[k + 1]["texto"] == ")"
                    and RX_ENUMERACAO.fullmatch(its[k]["texto"])):
                its[k]["enumeracao"] = True
                enumeracoes.append((s["id"], its[k]["texto"]))
        # comando do pesquisador: expressões retiradas do texto analisado (todas as ocorrências)
        textos = [it["texto"].lower() for it in its]
        for expr in EXPRESSOES:
            for k in range(len(its) - len(expr) + 1):
                if tuple(textos[k:k + len(expr)]) == expr:
                    for it in its[k:k + len(expr)]:
                        it["retirada_comando"] = " ".join(expr)
                    expressoes.append((s["id"], " ".join(expr)))
        localizadas = 0
        for it in its:
            o = marcada_em.get((s["id"], it["ini"])) if it["ini"] is not None else None
            it["padrao"] = o["padrao"] if o else None
            it["excesso"] = bool(o and o["excesso"])
            if o:
                if it["texto"] not in o["palavras"]:
                    raise RuntimeError(f"Palavra marcada não corresponde ao token: unidade {s['id']}, {it['texto']}")
                localizadas += 1
        esperadas = sum(len(g) for g in grupos[s["id"]])
        if localizadas != esperadas:
            raise RuntimeError(f"Unidade {s['id']}: {esperadas} palavras marcadas, {localizadas} localizadas no spaCy.")
        itens += its
    for it in itens:
        it["enumeracao"] = bool(it.get("enumeracao"))
        it["retirada_comando"] = it.get("retirada_comando")
        it["palavra"] = (bool(RX_PALAVRA.fullmatch(it["texto"])) and not it["enumeracao"]
                         and it["retirada_comando"] is None)
        it["lema"] = CORRECOES_DE_LEMA.get((it["lema_modelo"], it["upos"]), CORRECOES_DE_LEMA.get(it["lema_modelo"], it["lema_modelo"]))
        it["corrigido"] = it["lema"] != it["lema_modelo"]
        it["stopword"] = it["lema"] in STOPWORDS
        it["modal"] = it["lema"] in LEMAS_MODAIS and it["upos"] in ("VERB", "AUX")
        it["classe_conteudo"] = it["upos"] in CLASSES_DE_CONTEUDO
        it["conteudo_sem_retirada"] = (it["palavra"] and it["classe_conteudo"] and not it["stopword"]
                                       and not it["modal"] and len(it["lema"]) >= TAMANHO_MINIMO_LEMA)
        it["conta"] = it["palavra"] and not it["excesso"]
        it["conteudo"] = it["conteudo_sem_retirada"] and not it["excesso"]
    tab = pd.DataFrame(itens)
    tab["posicao"] = tab["conta"].cumsum().where(tab["conta"])   # posição do token no documento (1..N)
    return {"incluidas": incluidas, "filtros": filtros, "limpos": limpos, "ocorrencias": ocorrencias,
            "itens": tab, "especiais": registro_especiais, "enumeracoes": enumeracoes, "expressoes": expressoes}


# --- Medidas (skill 02, seção 3.5) ------------------------------------------
def tabela_lemas(R):
    """Frequência, taxa por 1.000 tokens, alcance e DP normalizado de todos os lemas de conteúdo."""
    t = R["itens"]
    n_tokens = int(t["conta"].sum())
    n_antes = int(t["palavra"].sum())
    por_unidade = t[t["conta"]].groupby("unidade").size()
    partes = por_unidade[por_unidade > 0]
    s = partes / partes.sum()
    s_min = float(s.min())
    cont = t[t["conteudo"]]
    linhas = []
    freq_antes = t[t["conteudo_sem_retirada"]].groupby("lema").size()
    for lema, g in cont.groupby("lema"):
        f_u = g.groupby("unidade").size()
        v = f_u / f_u.sum()
        diff = (v - s.reindex(v.index)).abs().sum() + (1 - s.reindex(v.index).sum())
        dp = 0.5 * diff
        linhas.append({"lema": lema, "frequência": len(g), "por_mil": len(g) / n_tokens * 1000,
                       "alcance_unidades": len(f_u), "alcance_pct": len(f_u) / len(partes),
                       "DP_norm": dp / (1 - s_min),
                       "frequência_antes": int(freq_antes.get(lema, 0)),
                       "por_mil_antes": freq_antes.get(lema, 0) / n_antes * 1000})
    df = pd.DataFrame(linhas)
    df["retiradas_excesso"] = df["frequência_antes"] - df["frequência"]
    return df.sort_values(["frequência", "lema"], ascending=[False, True]).reset_index(drop=True)


def bigramas(R, sem_retirada=False):
    """Pares de tokens de conteúdo adjacentes no texto, na mesma frase, sem pontuação nem palavra retirada entre eles.
    Ocorrências com a mesma forma escrita ficam no mesmo bigrama: quando o modelo dá lemas diferentes à mesma forma
    (políticas → política ou político), elas vão para o par de lemas mais frequente entre os dessa forma."""
    t = R["itens"]
    chave = "conteudo_sem_retirada" if sem_retirada else "conteudo"
    ocorrencias = []
    registros = t.to_dict("records")
    for a, b in zip(registros, registros[1:]):
        if a["unidade"] != b["unidade"] or a["sent"] != b["sent"] or not (a[chave] and b[chave]):
            continue
        ocorrencias.append(((a["lema"], b["lema"]), f"{a['texto'].lower()} {b['texto'].lower()}", a["unidade"]))
    total_par = Counter(k for k, _, _ in ocorrencias)
    por_forma = defaultdict(Counter)
    for k, forma, _ in ocorrencias:
        por_forma[forma][k] += 1
    canonico = {forma: sorted(c, key=lambda k: (-total_par[k], k))[0] for forma, c in por_forma.items()}
    pares, formas, unidades, reunidos = Counter(), defaultdict(Counter), defaultdict(set), defaultdict(Counter)
    for k, forma, u in ocorrencias:
        kk = canonico[forma]
        pares[kk] += 1
        formas[kk][forma] += 1
        unidades[kk].add(u)
        reunidos[kk][k] += 1
    return pares, formas, unidades, reunidos


def tabela_bigramas(R):
    n_tokens = int(R["itens"]["conta"].sum())
    pares, formas, unidades, reunidos = bigramas(R)
    pares_antes, formas_antes, _, _ = bigramas(R, sem_retirada=True)
    antes_por_forma = Counter()
    for k, c in formas_antes.items():
        antes_por_forma.update(c)
    linhas = []
    for k, f in pares.items():
        variantes = sorted(formas[k].items(), key=lambda kv: (-kv[1], kv[0]))
        outros = [f"{a} + {b} ({br(n)})" for (a, b), n in reunidos[k].items() if (a, b) != k]
        linhas.append({"bigrama": variantes[0][0], "lemas": f"{k[0]} + {k[1]}", "frequência": f,
                       "por_mil": f / n_tokens * 1000, "alcance_unidades": len(unidades[k]),
                       "frequência_antes": sum(antes_por_forma[x] for x in formas[k]),
                       "formas_no_texto": "; ".join(f"{x} ({br(n)})" for x, n in variantes),
                       "lemas_reunidos": "; ".join(outros) if outros else "–"})
    df = pd.DataFrame(linhas)
    df["retiradas_excesso"] = df["frequência_antes"] - df["frequência"]
    return df.sort_values(["frequência", "bigrama"], ascending=[False, True]).reset_index(drop=True)


def recorte(df, n, coluna="frequência", ampliar_empates=False, chave="lema"):
    """Recorte 'top N'. Com ampliar_empates, inclui todos os empatados com o N-ésimo (comando do pesquisador);
    sem ele, desempate alfabético, com a lista dos empatados que ficaram de fora."""
    if len(df) <= n:
        return df.copy(), {"pedido": n, "exibidos": len(df), "acrescentados": [], "fora_empatados": [], "primeiro_fora": None}
    limite = df.iloc[n - 1][coluna]
    if ampliar_empates:
        sel = df[df[coluna] >= limite]
        acrescentados = list(sel.iloc[n:][chave])
        fora_empatados = []
    else:
        sel = df.iloc[:n]
        acrescentados = []
        fora_empatados = list(df.iloc[n:][df.iloc[n:][coluna] == limite][chave])
    resto = df.iloc[len(sel):]
    primeiro_fora = (resto.iloc[0][chave], resto.iloc[0][coluna]) if len(resto) else None
    return sel.copy(), {"pedido": n, "exibidos": len(sel), "limite": limite, "acrescentados": acrescentados,
                        "fora_empatados": fora_empatados, "primeiro_fora": primeiro_fora}
# <<< fim da cópia literal
```

```python
# =============================================================================
# FUNÇÕES DE VERIFICAÇÃO: aninhamento dos recortes do PBIA (skill 03, seção 3.3) e identidade com os notebooks
# individuais (comando do pesquisador: usar os unigramas e bigramas formados neles)
# =============================================================================
INICIO_CONFIG_INDIVIDUAL = ("# =============================================================================\n"
                            "# CONFIGURAÇÃO — todos os parâmetros editáveis. Só esta célula muda entre os três notebooks da pasta.")
INICIO_FUNCOES_INDIVIDUAL = ("# =============================================================================\n"
                             "# 3. AMBIENTE, ESTILO E FUNÇÕES — código idêntico nos três notebooks da pasta")
INICIO_AMBIENTE_CONJUNTO = ("# =============================================================================\n"
                            "# AMBIENTE, ESTILO E FUNÇÕES DO PIPELINE")
RX_COPIA_LITERAL = re.compile(r"^# >>> início da cópia literal[^\n]*\n(.*?)\n# <<< fim da cópia literal", re.S | re.M)
RX_LISTA_EXIBIDA = re.compile(r"^- Lista exibida \(posição\. (lema|bigrama): frequência\): (.*)$", re.M)
PARAMETROS_DO_PIPELINE = ["MARCADOR_REPETICAO", "RETIRAR_EXCESSO_MARCADO", "INCLUIR_TITULO_NA_CONTAGEM",
                          "INCLUIR_TITULOS_INTERNOS", "CATEGORIAS_EXCLUIDAS", "UNIDADES_EXCLUIDAS", "IDIOMA_ESPERADO",
                          "MODELO_SPACY", "CLASSES_DE_CONTEUDO", "TAMANHO_MINIMO_LEMA", "STOPWORDS_ADICIONAIS",
                          "STOPWORDS_MANTIDAS", "LEMAS_MODAIS", "PRONOMES_CLITICOS", "HIFEN_COMO_TRAVESSAO",
                          "CORRECOES_DE_LEMA", "FREQ_MINIMA_BIGRAMA", "RETIRAR_MARCADORES_DE_ENUMERACAO",
                          "EXPRESSOES_RETIRADAS", "PLURAIS_REUNIDOS"]


def carregar_documento(sigla):
    """Carrega e verifica um JSON com a função dos notebooks individuais (carregar_e_verificar). Aquela função lê o
    recorte esperado da variável global CUT_ESPERADO, que aqui é definida para cada documento antes da chamada."""
    global CUT_ESPERADO
    CUT_ESPERADO = CUT_POR_SIGLA[sigla]
    return carregar_e_verificar(PASTA / DOCUMENTOS[sigla])


def conferir_aninhamento(js_sem, js_com):
    """As unidades do PBIA sem anexos devem ser idênticas (todos os campos, inclusive o texto com as marcações) às
    primeiras unidades do PBIA com anexos."""
    sem, com = js_sem["sections"], js_com["sections"]
    divergentes = [a["id"] for a, b in zip(sem, com) if a != b]
    return len(com) > len(sem) and not divergentes, divergentes, len(sem), len(com)


def celula_por_inicio(caminho, inicio):
    """Texto da única célula de código do arquivo .ipynb que começa por `inicio`."""
    celulas = json.loads(Path(caminho).read_text(encoding="utf-8"))["cells"]
    achadas = ["".join(c["source"]) for c in celulas
               if c["cell_type"] == "code" and "".join(c["source"]).startswith(inicio)]
    if len(achadas) != 1:
        raise RuntimeError(f"{Path(caminho).name}: {len(achadas)} células começam pelo cabeçalho esperado (deveria haver 1).")
    return achadas[0]


def conferir_identidade(sigla):
    """Compara este notebook com o notebook individual do documento: (1) os parâmetros de pré-processamento, lidos da
    célula de configuração daquele notebook; (2) os trechos de código copiados para cá, que precisam aparecer letra por
    letra na célula de funções daquele notebook (lida do arquivo .ipynb, assim como a célula deste notebook)."""
    caminho = PASTA / NOTEBOOKS_INDIVIDUAIS[sigla]
    ns = {}
    exec(celula_por_inicio(caminho, INICIO_CONFIG_INDIVIDUAL), ns)   # a célula de configuração só tem atribuições
    linhas = [("parâmetro", nome, "idêntico" if nome in ns and ns[nome] == globals()[nome] else "DIFERENTE")
              for nome in PARAMETROS_DO_PIPELINE]
    funcoes_individual = celula_por_inicio(caminho, INICIO_FUNCOES_INDIVIDUAL)
    proprio = PASTA / f"{NOME_SAIDA}.ipynb"
    trechos = RX_COPIA_LITERAL.findall(celula_por_inicio(proprio, INICIO_AMBIENTE_CONJUNTO)) if proprio.exists() else []
    if not trechos:
        linhas.append(("código", "trechos copiados", f"não conferido ({proprio.name} não encontrado ou sem as marcas)"))
    for k, trecho in enumerate(trechos, start=1):
        nomes = [no.name if isinstance(no, ast.FunctionDef) else getattr(no.targets[0], "id", "?")
                 for no in ast.parse(trecho).body if isinstance(no, (ast.FunctionDef, ast.Assign))]
        linhas.append((f"código, trecho {k} ({len(trecho.splitlines())} linhas)", ", ".join(nomes),
                       "idêntico" if trecho in funcoes_individual else "DIFERENTE"))
    return pd.DataFrame(linhas, columns=["item", "nome", "resultado"])


def listas_exibidas(caminho):
    """Listas 'posição. termo: frequência' impressas nas sínteses factuais de um notebook individual (saídas gravadas
    no arquivo), com o título da seção em que aparecem."""
    celulas = json.loads(Path(caminho).read_text(encoding="utf-8"))["cells"]
    listas = []
    for i, c in enumerate(celulas):
        titulo = ""
        for anterior in reversed(celulas[:i]):
            cab = [l for l in "".join(anterior["source"]).splitlines() if l.startswith("### ")]
            if anterior["cell_type"] == "markdown" and cab:
                titulo = cab[0].lstrip("# ").strip()
                break
        for o in c.get("outputs", []):
            if o.get("output_type") != "stream":
                continue
            for m in RX_LISTA_EXIBIDA.finditer("".join(o["text"])):
                itens = []
                for parte in m.group(2).rstrip(".").split("; "):
                    _, resto = parte.split(". ", 1)
                    termo, freq = resto.rsplit(": ", 1)
                    itens.append((termo, int(freq.replace(".", ""))))
                listas.append({"seção": titulo, "tipo": m.group(1), "itens": itens})
    return listas


def conferir_listas(sigla):
    """Compara as listas exibidas no notebook individual (termo e frequência, na mesma ordem) com as tabelas deste
    notebook (LEMAS e BIG_VALIDOS), que vêm do mesmo JSON, com o mesmo código e os mesmos parâmetros."""
    linhas = []
    for lista in listas_exibidas(PASTA / NOTEBOOKS_INDIVIDUAIS[sigla]):
        tab, col = (LEMAS[sigla], "lema") if lista["tipo"] == "lema" else (BIG_VALIDOS[sigla], "bigrama")
        n = len(lista["itens"])
        aqui = [(t, int(f)) for t, f in zip(tab[col].head(n), tab["frequência"].head(n))]
        linhas.append({"documento": ROTULO[sigla], "notebook individual": NOTEBOOKS_INDIVIDUAIS[sigla],
                       "seção": lista["seção"], "tipo": "unigramas" if lista["tipo"] == "lema" else "bigramas",
                       "termos na lista": n, "resultado": "idêntico" if aqui == lista["itens"] else "DIFERENTE"})
    if not linhas:
        linhas.append({"documento": ROTULO[sigla], "notebook individual": NOTEBOOKS_INDIVIDUAIS[sigla], "seção": "–",
                       "tipo": "–", "termos na lista": 0,
                       "resultado": "não conferido: o arquivo não tem as saídas gravadas"})
    return linhas
```

```python
# =============================================================================
# FUNÇÕES COMPARATIVAS (skill 03, seções 3.4 a 3.6) E GRÁFICOS-PADRÃO DA COMPARAÇÃO
# =============================================================================
AREA_MARCADOR = {"o": 1.0, "s": math.pi / 4, "^": math.pi / 2}   # fator de tamanho que iguala a área visual à do círculo
ESTILO_ZONA_SEMELHANCA = {"neutra": (COR_TINTA_2, "o", 0.55),          # cor, marcador e transparência por zona
                          SIG_A: (COR_DOCUMENTO[SIG_A], MARCADOR_DOC[SIG_A], 0.45),
                          SIG_B: (COR_DOCUMENTO[SIG_B], MARCADOR_DOC[SIG_B], 0.45)}
FMT_BR = FuncFormatter(lambda v, _: f"{v:g}".replace(".", ","))


def num(x, casas=2):
    """Número no padrão brasileiro: a mesma br(), com o nome usado no código do gráfico do BRICS × OCDE."""
    return br(x, casas)


def cientifico(x, casas=2):
    """Notação científica no padrão brasileiro (ex.: 1,23 × 10⁻⁵)."""
    mantissa, expoente = f"{x:.{casas}e}".split("e")
    sobrescrito = str.maketrans("-0123456789", "⁻⁰¹²³⁴⁵⁶⁷⁸⁹")
    return f"{mantissa.replace('.', ',')} × 10{str(int(expoente)).translate(sobrescrito)}"


def cor_do_texto(cor_fundo):
    """Texto branco ou escuro sobre uma cor, pela mesma regra de luminância dos notebooks individuais."""
    r, g, b = (int(cor_fundo[i:i + 2], 16) / 255 for i in (1, 3, 5))
    return "white" if 0.2126 * r + 0.7152 * g + 0.0722 * b < 0.45 else COR_TINTA


def tabela_keyness(freq_a, freq_b, n_a, n_b):
    """Todos os termos dos dois documentos, com G² (log-verossimilhança; Dunning, 1993; Rayson & Garside, 2000) e
    Log Ratio (Hardie, 2014): log2 da razão entre as taxas relativas. Convenção fixa: Log Ratio > 0 → mais
    característico de SIG_A (EBIA); < 0 → de SIG_B (PBIA). Frequência zero recebe a correção de 0,5 apenas no Log
    Ratio. Distintivo = frequência combinada ≥ FREQ_MINIMA_COMPARACAO, G² ≥ LIMIAR_G2 e |Log Ratio| ≥ LIMIAR_LOG_RATIO."""
    termos = freq_a.index.union(freq_b.index)
    a = freq_a.reindex(termos, fill_value=0).astype(float)
    b = freq_b.reindex(termos, fill_value=0).astype(float)
    esp_a = n_a * (a + b) / (n_a + n_b)
    esp_b = n_b * (a + b) / (n_a + n_b)
    with np.errstate(divide="ignore", invalid="ignore"):
        g2 = 2 * (np.where(a > 0, a * np.log(a / esp_a), 0.0) + np.where(b > 0, b * np.log(b / esp_b), 0.0))
    log_ratio = np.log2((a.where(a > 0, 0.5) / n_a) / (b.where(b > 0, 0.5) / n_b))
    tab = pd.DataFrame({f"freq_{SIG_A}": a.astype(int), f"freq_{SIG_B}": b.astype(int),
                        f"por_mil_{SIG_A}": a / n_a * 1000, f"por_mil_{SIG_B}": b / n_b * 1000,
                        "freq_combinada": (a + b).astype(int), "G2": g2, "p_valor": chi2.sf(g2, 1),
                        "log_ratio": log_ratio}, index=termos)
    tab.index.name = "termo"
    tab["presença"] = np.select([(a > 0) & (b > 0), a > 0], ["ambos", f"só {CURTO[SIG_A]}"], f"só {CURTO[SIG_B]}")
    tab["elegível"] = tab["freq_combinada"] >= FREQ_MINIMA_COMPARACAO
    tab["distintivo"] = tab["elegível"] & (tab["G2"] >= LIMIAR_G2) & (tab["log_ratio"].abs() >= LIMIAR_LOG_RATIO)
    tab["característico de"] = np.where(~tab["distintivo"], "–",
                                        np.where(tab["log_ratio"] > 0, CURTO[SIG_A], CURTO[SIG_B]))
    return tab


def contar_formas(*textos):
    """Soma as formas escritas no formato 'forma (n); forma (n)' (coluna formas_no_texto) de um ou mais documentos."""
    contagem = Counter()
    for texto in textos:
        if not isinstance(texto, str) or not texto or texto == "–":
            continue
        for item in texto.split("; "):
            forma, n = re.fullmatch(r"(.*) \(([\d.]+)\)", item).groups()
            contagem[forma] += int(n.replace(".", ""))
    return contagem


def rotulos_bigramas(big_a, big_b):
    """Rótulo comum de cada par de lemas: a forma escrita mais frequente nos dois documentos somados (desempate
    alfabético), como no gráfico de convergência do BRICS × OCDE."""
    fa = big_a.set_index("lemas")["formas_no_texto"]
    fb = big_b.set_index("lemas")["formas_no_texto"]
    return pd.Series({p: sorted(contar_formas(fa.get(p), fb.get(p)).items(), key=lambda kv: (-kv[1], kv[0]))[0][0]
                      for p in fa.index.union(fb.index)})


def limite_da_escala(valores):
    """Maior valor dentro da escala: um valor maior que FATOR_FORA_DA_ESCALA vezes o seguinte fica fora, assim como os
    maiores que ele (mesma regra do gráfico de convergência)."""
    ordem = sorted(valores, reverse=True)
    k = 0
    while k + 1 < len(ordem) and ordem[k] > FATOR_FORA_DA_ESCALA * ordem[k + 1]:
        k += 1
    return ordem[k]


def grafico_lado_a_lado(tops, infos, chave, rotulo, taxas, n_pedido, titulo, subtitulo):
    """Termos mais comuns, lado a lado (skill 03, seção 4.2, item 7): um painel por documento, com os N termos mais
    frequentes dele (recorte ampliado até o desempate), em ocorrências por 1.000 tokens e na mesma escala. A barra é a
    taxa no documento do painel; o marcador, a taxa do mesmo termo no outro documento (zero = ausente). Em negrito,
    os termos que estão nos recortes dos dois documentos. Um termo com taxa maior que FATOR_FORA_DA_ESCALA vezes a dos
    demais fica fora da escala, como no gráfico de convergência: a barra é interrompida e os valores são escritos."""
    outro = {SIG_A: SIG_B, SIG_B: SIG_A}
    n_max = max(len(t) for t in tops.values())
    em_ambos = set(tops[SIG_A][chave]) & set(tops[SIG_B][chave])
    maiores = {}
    for s, t in tops.items():
        for k, v in zip(t[chave], t["por_mil"]):
            maiores[k] = max(maiores.get(k, 0.0), float(v), float(taxas[outro[s]].get(k, 0.0)))
    lim = limite_da_escala(list(maiores.values())) * 1.22
    esq, lado, vao, base, topo = 1.9, 4.3, 2.35, 1.7, 1.0
    W, H = esq + 2 * lado + vao + 0.4, max(4.6, 0.25 * n_max + 3.05)
    fig = plt.figure(figsize=(W, H))
    por_pt = lim / (lado * 72)                       # unidades do eixo X por ponto tipográfico
    for i, s in enumerate((SIG_A, SIG_B)):
        ax = fig.add_axes([(esq + i * (lado + vao)) / W, base / H, lado / W, (H - base - topo) / H])
        o, t = outro[s], tops[s]
        y = n_max - 1 - np.arange(len(t))
        v = t["por_mil"].to_numpy(dtype=float)
        w = np.array([float(taxas[o].get(k, 0.0)) for k in t[chave]])
        ax.barh(y, np.minimum(v, lim), height=0.72, color=COR_DOCUMENTO[s], zorder=2)
        visivel = w <= lim
        tam = 40 * AREA_MARCADOR[MARCADOR_DOC[o]]
        ax.scatter(w[visivel], y[visivel], marker=MARCADOR_DOC[o], s=tam, color=COR_DOCUMENTO[o], edgecolors="white",
                   linewidths=0.9, zorder=4, clip_on=False)
        meio_marcador = np.sqrt(tam) / 2 * por_pt
        for yi, vi, wi in zip(y, v, w):
            if vi > lim:   # barra interrompida (fora da escala), com os dois valores escritos
                for fx in (0.905, 0.935):
                    ax.plot([lim * (fx - 0.008), lim * (fx + 0.008)], [yi - 0.42, yi + 0.42], color="white", lw=2.2,
                            zorder=3, solid_capstyle="butt")
                ax.text(lim * 0.88, yi, f"{br(vi, 2)} · {CURTO[o]}: {br(wi, 2)} (fora da escala)", ha="right",
                        va="center", fontsize=9, fontweight="bold", color=cor_do_texto(COR_DOCUMENTO[s]), zorder=5)
                continue
            texto = br(vi, 2) if wi <= lim else f"{br(vi, 2)} · {CURTO[o]}: {br(wi, 2)} (fora da escala)"
            x_txt = vi + 3 * por_pt                    # o valor fica logo depois da barra...
            largura_txt = len(texto) * 5.4 * por_pt
            if wi <= lim and x_txt - meio_marcador <= wi <= x_txt + largura_txt + meio_marcador:
                x_txt = wi + meio_marcador + 3 * por_pt   # ...ou depois do marcador, quando o marcador cairia sobre ele
            ax.text(x_txt, yi, texto, va="center", fontsize=9, color=COR_TINTA_2, zorder=5)
        ax.set_yticks(y)
        ax.set_yticklabels(t[rotulo])
        for lab, k in zip(ax.get_yticklabels(), t[chave]):
            if k in em_ambos:
                lab.set_fontweight("bold")
        ax.tick_params(axis="y", length=0)
        ax.set_ylim(-0.6, n_max - 0.4)
        ax.set_xlim(0, lim)
        ax.xaxis.set_major_formatter(EIXO_BR)
        ax.grid(axis="x", color=COR_GRADE, linewidth=0.6)
        ax.set_axisbelow(True)
        ax.set_xlabel("ocorrências por 1.000 tokens")
        ax.set_title(f"{ROTULO[s]}: {len(t)} termos ({br(N_TOKENS[s])} tokens)", fontsize=10.5, loc="left")
        if infos[s]["exibidos"] > n_pedido:
            corte = y[n_pedido - 1] - 0.5
            ax.axhline(corte, color="0.25", linestyle="--", linewidth=1, zorder=1)
            for dy, va, texto in ((0.1, "bottom", f"fim do recorte pedido (N = {n_pedido})"),
                                  (-0.1, "top", "abaixo, empatados no ponto de corte")):
                ax.text(lim * 0.995, corte + dy, texto, ha="right", va=va, fontsize=9, color="0.25", zorder=6,
                        path_effects=[path_effects.withStroke(linewidth=3, foreground="white")])
    fig.text(esq / W, 1 - 0.18 / H, titulo, ha="left", va="top", fontsize=12.5, fontweight="bold")
    fig.text(esq / W, 1 - 0.5 / H, subtitulo, ha="left", va="top", fontsize=10, color=COR_TINTA_2)
    marcador = dict(ls="", mec="white", mew=0.9)
    itens = [
        Patch(facecolor=COR_DOCUMENTO[SIG_A], label=f"Taxa {NOME_NO[SIG_A]}: barra no painel {NOME_DO[SIG_A]}"),
        Line2D([], [], marker=MARCADOR_DOC[SIG_A], ms=7.5, mfc=COR_DOCUMENTO[SIG_A],
               label=f"Taxa {NOME_NO[SIG_A]}: marcador no painel {NOME_DO[SIG_B]}", **marcador),
        Patch(facecolor=COR_DOCUMENTO[SIG_B], label=f"Taxa {NOME_NO[SIG_B]}: barra no painel {NOME_DO[SIG_B]}"),
        Line2D([], [], marker=MARCADOR_DOC[SIG_B], ms=8.5, mfc=COR_DOCUMENTO[SIG_B],
               label=f"Taxa {NOME_NO[SIG_B]}: marcador no painel {NOME_DO[SIG_A]}", **marcador),
    ]
    fig.legend(handles=itens, loc="upper left", bbox_to_anchor=(esq / W, 0.88 / H), ncol=2, fontsize=9,
               frameon=False, handlelength=1.6, columnspacing=2.5, borderaxespad=0)
    fig.text(esq / W, 0.12 / H, "Em negrito: termo presente nos recortes dos dois documentos. Marcador em zero: termo "
             "ausente do outro documento.", ha="left", va="bottom", fontsize=9, color=COR_TINTA_2)
    return fig, {"limite": lim, "fora": sorted(k for k, v in maiores.items() if v > lim)}


def tabela_lado_a_lado(tops, chave, key, rotulos=None):
    """União dos dois recortes: posição, ocorrências e taxa em cada documento, Log Ratio, G² e critérios de keyness."""
    uniao = list(dict.fromkeys(list(tops[SIG_A][chave]) + list(tops[SIG_B][chave])))
    pos = {s: {k: i for i, k in enumerate(tops[s][chave], start=1)} for s in SIGLAS}
    k = key.reindex(uniao)
    tab = pd.DataFrame(index=pd.Index(uniao, name="lema" if chave == "lema" else "par de lemas"))
    if rotulos is not None:
        tab["bigrama"] = [rotulos.get(x, x) for x in uniao]
    for s in SIGLAS:
        tab[f"{CURTO[s]}: posição"] = [pos[s].get(x, np.nan) for x in uniao]
        tab[f"{CURTO[s]}: ocorrências"] = k[f"freq_{s}"].to_numpy()
        tab[f"{CURTO[s]}: por 1.000 tokens"] = k[f"por_mil_{s}"].to_numpy()
    tab["Log Ratio"] = k["log_ratio"].to_numpy()
    tab["G²"] = k["G2"].to_numpy()
    tab["distintivo"] = np.where(k["distintivo"].to_numpy(), "sim: " + k["característico de"].astype(str), "não")
    tab["nos recortes de"] = [f"{CURTO[SIG_A]} e {CURTO[SIG_B]}" if x in pos[SIG_A] and x in pos[SIG_B]
                              else f"só {CURTO[SIG_A]}" if x in pos[SIG_A] else f"só {CURTO[SIG_B]}" for x in uniao]
    return tab


def casas_tabela(tab):
    """Casas decimais para exibir as tabelas comparativas."""
    casas = {c: 2 for c in tab.columns if "por 1.000" in c or c in ("Log Ratio", "G²")}
    casas.update({c: 0 for c in tab.columns if "posição" in c})
    return casas


def sintese_lado_a_lado(tops, infos, tab, chave, n_pedido, nome, escala, rotulos=None):
    """Síntese factual das Visualizações 1 e 2."""
    rot = (lambda x: rotulos.get(x, x)) if rotulos is not None else (lambda x: x)
    print("Síntese factual:")
    for s in SIGLAS:
        i = infos[s]
        print(f"- {ROTULO[s]}: recorte pedido {n_pedido}; exibidos {i['exibidos']}"
              + (f" (acrescentados por empate na frequência {br(int(i['limite']))}: {', '.join(map(str, i['acrescentados']))})"
                 if i["acrescentados"] else " (sem empate no ponto de corte)")
              + (f"; primeiro fora: {i['primeiro_fora'][0]} ({br(int(i['primeiro_fora'][1]))})." if i["primeiro_fora"] else "."))
    ambos = tab[tab["nos recortes de"] == f"{CURTO[SIG_A]} e {CURTO[SIG_B]}"]
    print(f"- União dos recortes: {len(tab)} {nome}; nos dois recortes: {len(ambos)} ("
          + ", ".join(rot(x) for x in ambos.index) + ").")
    for s in SIGLAS:
        print(f"- Mais frequentes {NOME_NO[s]} (por 1.000 tokens): "
              + "; ".join(f"{rot(k)} {br(v, 2)}" for k, v in zip(tops[s][chave].head(5), tops[s]["por_mil"].head(5))) + ".")
    dif = ambos.assign(_abs=ambos["Log Ratio"].abs()).sort_values("_abs", ascending=False).head(5)
    print("- Nos dois recortes, maiores diferenças de taxa (|Log Ratio|): "
          + "; ".join(f"{rot(x)} ({CURTO[SIG_A]} {br(r[f'{CURTO[SIG_A]}: por 1.000 tokens'], 2)}; {CURTO[SIG_B]} "
                      f"{br(r[f'{CURTO[SIG_B]}: por 1.000 tokens'], 2)}; Log Ratio {br(r['Log Ratio'], 2)})"
                      for x, r in dif.iterrows()) + ".")
    dist = tab[tab["distintivo"] != "não"]
    print(f"- Passam nos critérios de keyness: {len(dist)} de {len(tab)} ("
          + "; ".join(f"{CURTO[s]}: {int((dist['distintivo'] == 'sim: ' + CURTO[s]).sum())}" for s in SIGLAS) + ").")
    if escala["fora"]:
        print("- Fora da escala (barra interrompida, valores escritos): " + ", ".join(rot(x) for x in escala["fora"]) + ".")
    n_exc = sum(N_EXCESSO.values())
    print("- Retirada do excesso marcado: " + ("nenhum token retirado (os dois JSONs não têm padrões marcados)."
                                               if n_exc == 0 else f"{br(n_exc)} tokens retirados."))


def tabela_semelhancas(key, n, rotulos=None):
    """Os n termos compartilhados (frequência combinada ≥ FREQ_MINIMA_COMPARACAO) de peso parecido (|Log Ratio| <
    LIMIAR_LOG_RATIO) com maior taxa somada."""
    comp = key[(key["presença"] == "ambos") & key["elegível"] & (key["log_ratio"].abs() < LIMIAR_LOG_RATIO)].copy()
    comp["soma"] = comp[f"por_mil_{SIG_A}"] + comp[f"por_mil_{SIG_B}"]
    comp = comp.reset_index().sort_values(["soma", "termo"], ascending=[False, True]).head(n).set_index("termo")
    tab = pd.DataFrame(index=comp.index)
    if rotulos is not None:
        tab["bigrama"] = [rotulos.get(x, x) for x in comp.index]
    for s in SIGLAS:
        tab[f"{CURTO[s]}: ocorrências"] = comp[f"freq_{s}"]
        tab[f"{CURTO[s]}: por 1.000 tokens"] = comp[f"por_mil_{s}"]
    tab["razão (maior ÷ menor)"] = comp[[f"por_mil_{SIG_A}", f"por_mil_{SIG_B}"]].max(axis=1) / comp[[f"por_mil_{SIG_A}", f"por_mil_{SIG_B}"]].min(axis=1)
    tab["Log Ratio"] = comp["log_ratio"]
    tab["G²"] = comp["G2"]
    tab["p"] = comp["p_valor"]
    tab.index.name = "lema" if rotulos is None else "par de lemas"
    return tab


def tabela_distintivos(key, rotulos=None):
    """Todos os termos distintivos (critérios da seção 3.6), por documento e por ordem de G², com o teste de
    Bonferroni sobre o total de testes do notebook."""
    d = key[key["distintivo"]].reset_index()
    d["_lado"] = np.where(d["log_ratio"] > 0, 0, 1)
    d = d.sort_values(["_lado", "G2", "freq_combinada", "termo"], ascending=[True, False, False, True]).set_index("termo")
    tab = pd.DataFrame(index=d.index)
    if rotulos is not None:
        tab["bigrama"] = [rotulos.get(x, x) for x in d.index]
    tab["característico de"] = d["característico de"]
    for s in SIGLAS:
        tab[f"{CURTO[s]}: ocorrências"] = d[f"freq_{s}"]
        tab[f"{CURTO[s]}: por 1.000 tokens"] = d[f"por_mil_{s}"]
    tab["Log Ratio"] = d["log_ratio"]
    tab["G²"] = d["G2"]
    tab["resiste a Bonferroni"] = np.where(d["p_valor"] < ALFA_CORRIGIDO, "sim", "não")
    tab["exclusivo"] = np.where(d["presença"] == "ambos", "–", d["presença"])
    tab.index.name = "lema" if rotulos is None else "par de lemas"
    return tab


def sintese_keyness(key, tab, nome, rotulos=None):
    """Síntese factual das Visualizações 4 e 5."""
    rot = (lambda x: rotulos.get(x, x)) if rotulos is not None else (lambda x: x)
    eleg = int(key["elegível"].sum())
    print("Síntese factual:")
    print(f"- {br(len(key))} {nome} distintos nos dois documentos; {br(eleg)} com frequência combinada ≥ "
          f"{FREQ_MINIMA_COMPARACAO} (elegíveis); {br(len(tab))} distintivos ({br_pct(len(tab) / eleg)} dos elegíveis).")
    for s in SIGLAS:
        t = tab[tab["característico de"] == CURTO[s]]
        n_bonf, n_excl = int((t["resiste a Bonferroni"] == "sim").sum()), int((t["exclusivo"] != "–").sum())
        print(f"- Característicos {NOME_DO[s]}: {len(t)}, dos quais {n_bonf} {'resiste' if n_bonf == 1 else 'resistem'} "
              f"à correção de Bonferroni e {n_excl} {'é exclusivo' if n_excl == 1 else 'são exclusivos'}. Maiores G²: "
              + "; ".join(f"{rot(x)} (G² {br(r['G²'], 1)}; Log Ratio {br(r['Log Ratio'], 2)})" for x, r in t.head(5).iterrows()) + ".")


def partes_vocabulario(freq_a, freq_b, minimo=1):
    """Termos distintos só no primeiro documento, nos dois e só no segundo (frequência combinada ≥ minimo)."""
    t = pd.DataFrame({"a": freq_a, "b": freq_b}).fillna(0)
    t = t[t["a"] + t["b"] >= minimo]
    return (int(((t["a"] > 0) & (t["b"] == 0)).sum()), int(((t["a"] > 0) & (t["b"] > 0)).sum()),
            int(((t["a"] == 0) & (t["b"] > 0)).sum()))


def cobertura(freq, outra):
    """Ocorrências de um documento que pertencem a termos presentes no outro, e as demais."""
    comp = int(freq[freq.index.isin(outra.index)].sum())
    return comp, int(freq.sum()) - comp


def agrupar_pontos(df, chaves_rotuladas, coluna_rotulo="rotulo"):
    """Termos com exatamente os mesmos valores (x, y) formam um único ponto (como no gráfico de convergência)."""
    grupos = []
    for _, g in df.groupby([df["x"].round(9), df["y"].round(9)], sort=False):
        grupos.append({"x": float(g["x"].iloc[0]), "y": float(g["y"].iloc[0]), "n": len(g), "linhas": g,
                       "rotulos": sorted(g.loc[g.index.isin(chaves_rotuladas), coluna_rotulo])})
    return grupos


def grafico_semelhancas(paineis, titulo, subtitulo):
    """Semelhanças (skill 03, seção 2.2, item 5): termos presentes nos dois documentos, com frequência combinada ≥
    FREQ_MINIMA_COMPARACAO, em escala logarítmica (X = taxa na EBIA; Y = taxa no PBIA, por 1.000 tokens). Na faixa
    cinza, a diferença é menor que o dobro (|Log Ratio| < LIMIAR_LOG_RATIO): peso relativo parecido (ponto escuro).
    São rotulados os N_ROTULOS_DISPERSAO termos de maior taxa somada entre os de peso parecido."""
    W, H = 13.6, 8.6
    fig = plt.figure(figsize=(W, H))
    lado, base, esq, vao = 5.3, 1.95, 1.05, 1.55
    contorno = [path_effects.withStroke(linewidth=3, foreground="white")]
    for i, p in enumerate(paineis):
        ax = fig.add_axes([(esq + i * (lado + vao)) / W, base / H, lado / W, lado / H])
        comp = p["tab"][(p["tab"]["presença"] == "ambos") & p["tab"]["elegível"]].copy()
        comp["x"], comp["y"] = comp[f"por_mil_{SIG_A}"], comp[f"por_mil_{SIG_B}"]
        comp["parecido"] = comp["log_ratio"].abs() < LIMIAR_LOG_RATIO
        comp["rotulo"] = [p["rotulos"].get(k, k) for k in comp.index] if p.get("rotulos") is not None else list(comp.index)
        comp["soma"] = comp["x"] + comp["y"]
        rotular = set(comp[comp["parecido"]].reset_index().sort_values(["soma", "termo"], ascending=[False, True])
                      .head(N_ROTULOS_DISPERSAO)["termo"])
        lo, hi = comp[["x", "y"]].min().min() / 1.5, comp[["x", "y"]].max().max() * 1.5
        ax.set_xscale("log")
        ax.set_yscale("log")
        ax.set_xlim(lo, hi)
        ax.set_ylim(lo, hi)
        ax.set_aspect("equal", adjustable="box")
        xs = np.geomspace(lo, hi, 300)
        ax.fill_between(xs, xs / RAZAO_ZONA_NEUTRA, xs * RAZAO_ZONA_NEUTRA, color=COR_FAIXA, alpha=0.75, lw=0, zorder=0)
        for fator in (RAZAO_ZONA_NEUTRA, 1 / RAZAO_ZONA_NEUTRA):
            ax.plot(xs, xs * fator, color=COR_LIGACAO, lw=0.7, ls=(0, (1.5, 2)), zorder=1)
        ax.plot([lo, hi], [lo, hi], color=COR_TINTA_2, lw=1.0, ls=(0, (5, 4)), zorder=2)
        # Pontos semitransparentes: termos com os mesmos valores ficam sobrepostos e o ponto fica mais escuro (há
        # centenas de termos de baixa frequência com valores repetidos; números dentro dos pontos ficariam ilegíveis).
        # Mesma codificação do gráfico de convergência: cinza = peso parecido; verdes = taxa ao menos 2× maior em um.
        comp["zona"] = np.where(comp["parecido"], "neutra", np.where(comp["log_ratio"] > 0, SIG_A, SIG_B))
        tam = 30
        for zona in (SIG_A, SIG_B, "neutra"):
            cor, m, alfa = ESTILO_ZONA_SEMELHANCA[zona]
            sel = comp[comp["zona"] == zona]
            ax.scatter(sel["x"], sel["y"], s=tam * AREA_MARCADOR[m], marker=m, color=cor, alpha=alfa, linewidths=0,
                       zorder=3 if zona == "neutra" else 2, clip_on=False)
        grupos = agrupar_pontos(comp, rotular)   # posições distintas, para o rotulador
        for eixo in (ax.xaxis, ax.yaxis):
            eixo.set_major_locator(LogLocator(base=10, subs=(1.0, 2.0, 5.0)))
            eixo.set_minor_locator(NullLocator())
            eixo.set_major_formatter(FMT_BR)
        ax.grid(color=COR_GRADE, linewidth=0.6)
        ax.set_axisbelow(True)
        ax.set_xlabel(f"Taxa {NOME_NO[SIG_A]} (ocorrências por 1.000 tokens; escala logarítmica)")
        ax.set_ylabel(f"Taxa {NOME_NO[SIG_B]} (ocorrências por 1.000 tokens; escala logarítmica)")
        n_par = int(comp["parecido"].sum())
        ax.set_title(f"{p['titulo']}\n{len(comp)} termos compartilhados; {n_par} com peso parecido "
                     f"({br_pct(n_par / len(comp))})", fontsize=10.5)
        fig.canvas.draw()
        rot = Rotulador(ax, np.array([(q["x"], q["y"]) for q in grupos]), [np.sqrt(tam) / 2] * len(grupos),
                        [[(lo, lo), (hi, hi)], [(lo, lo * RAZAO_ZONA_NEUTRA), (hi / RAZAO_ZONA_NEUTRA, hi)],
                         [(lo * RAZAO_ZONA_NEUTRA, lo), (hi, hi / RAZAO_ZONA_NEUTRA)]])
        estilo = dict(fontsize=9, color=COR_TINTA, linespacing=1.15, path_effects=contorno)
        rot.posicionar([((q["x"], q["y"]), "\n".join(q["rotulos"]), COR_TINTA_2, j, estilo)
                        for j, q in enumerate(grupos) if q["rotulos"]], tentativas=150, semente=SEMENTE)
    fig.text(esq / W, 1 - 0.18 / H, titulo, ha="left", va="top", fontsize=12.5, fontweight="bold")
    fig.text(esq / W, 1 - 0.5 / H, subtitulo, ha="left", va="top", fontsize=10, color=COR_TINTA_2)
    r_txt = f"{RAZAO_ZONA_NEUTRA:g}".replace(".", ",")
    def item(zona, rotulo):
        cor, m, alfa = ESTILO_ZONA_SEMELHANCA[zona]
        return Line2D([], [], ls="", marker=m, ms=6.5 * np.sqrt(AREA_MARCADOR[m]), mfc=cor, mec="none", alpha=alfa,
                      label=rotulo)

    itens = [
        item("neutra", f"Peso parecido: diferença menor que {r_txt}× (|Log Ratio| < {num(LIMIAR_LOG_RATIO, 0)})"),
        item(SIG_A, f"Taxa ao menos {r_txt}× maior {NOME_NO[SIG_A]}"),
        item(SIG_B, f"Taxa ao menos {r_txt}× maior {NOME_NO[SIG_B]}"),
        Line2D([], [], color=COR_TINTA_2, lw=1.0, ls=(0, (5, 4)), label="Linha tracejada: frequência igual (x = y)"),
        Patch(facecolor=COR_FAIXA, edgecolor=COR_LIGACAO, lw=0.7, ls=":",
              label=f"Faixa cinza: entre y = x/{r_txt} e y = {r_txt}·x"),
    ]
    fig.legend(handles=itens, loc="upper left", bbox_to_anchor=(esq / W, 1.12 / H), ncol=2, fontsize=9,
               frameon=False, columnspacing=2.5, borderaxespad=0)
    fig.text(esq / W, 0.1 / H, "Pontos semitransparentes: termos com os mesmos valores ficam sobrepostos, e o ponto fica "
             "mais escuro (os valores exatos estão nas tabelas).", ha="left", va="bottom", fontsize=9, color=COR_TINTA_2)
    return fig


def grafico_keyness(tab, rotulos, titulo, subtitulo):
    """Distinções (keyness; skill 03, seção 3.6): termos que passam nos dois critérios (G² ≥ LIMIAR_G2 e |Log Ratio| ≥
    LIMIAR_LOG_RATIO, com frequência combinada ≥ FREQ_MINIMA_COMPARACAO), até TOP_N_KEYNESS por documento, por ordem
    de G². A barra é o Log Ratio (positivo = EBIA; negativo = PBIA); à direita, o G² e as ocorrências em cada
    documento. Barra hachurada: termo ausente do outro documento (Log Ratio calculado com a correção de 0,5)."""
    dist = tab[tab["distintivo"]].reset_index()
    dist["rotulo"] = [rotulos.get(k, k) for k in dist["termo"]] if rotulos is not None else dist["termo"]
    ordem = dict(by=["G2", "freq_combinada", "termo"], ascending=[False, False, True])
    linhas = []
    for s, sinal in ((SIG_A, 1), (SIG_B, -1)):
        todos = dist[np.sign(dist["log_ratio"]) == sinal]
        d = todos.sort_values(**ordem).head(TOP_N_KEYNESS)
        linhas.append(("cab", f"Mais característicos {NOME_DO[s]}: {len(d)} de {len(todos)} termos distintivos", s))
        linhas += [("termo", r, s) for r in d.to_dict("records")]
    n = len(linhas)
    W, H = 12.7, 0.29 * n + 2.6
    fig = plt.figure(figsize=(W, H))
    ax = fig.add_axes([2.45 / W, 1.4 / H, 6.25 / W, (H - 2.3) / H])
    y = np.arange(n)[::-1]
    lim = max(float(dist["log_ratio"].abs().max()) if len(dist) else 1.0, 1.5) * 1.06
    ax.set_xlim(-lim, lim)
    ax.set_ylim(-0.6, n - 0.4)
    trans = ax.get_yaxis_transform()
    rotulos_y = []
    for yi, (tipo, r, s) in zip(y, linhas):
        if tipo == "cab":
            ax.text(-lim + lim * 0.02, yi, r, ha="left", va="center", fontsize=10, fontweight="bold", color=COR_TINTA)
            rotulos_y.append("")
            continue
        exclusivo = r[f"freq_{SIG_B if s == SIG_A else SIG_A}"] == 0
        ax.barh(yi, r["log_ratio"], height=0.7, color=COR_DOCUMENTO[s], zorder=2,
                hatch="////" if exclusivo else None, edgecolor="white", linewidth=0)
        ax.text(1.03, yi, br(r["G2"], 1), transform=trans, ha="left", va="center", fontsize=9, color=COR_TINTA_2)
        ax.text(1.2, yi, f"{br(int(r[f'freq_{SIG_A}']))} · {br(int(r[f'freq_{SIG_B}']))}", transform=trans,
                ha="left", va="center", fontsize=9, color=COR_TINTA_2)
        rotulos_y.append(r["rotulo"])
    ax.text(1.03, n - 0.2, "G²", transform=trans, ha="left", va="bottom", fontsize=9, fontweight="bold")
    ax.text(1.2, n - 0.2, f"{CURTO[SIG_A]} · {CURTO[SIG_B]}", transform=trans, ha="left", va="bottom", fontsize=9,
            fontweight="bold")
    ax.set_yticks(y)
    ax.set_yticklabels(rotulos_y)
    ax.tick_params(axis="y", length=0)
    ax.axvline(0, color=COR_TINTA_2, lw=0.8, zorder=3)
    for x in (-LIMIAR_LOG_RATIO, LIMIAR_LOG_RATIO):
        ax.axvline(x, color=COR_LIGACAO, lw=0.8, ls=(0, (1.5, 2)), zorder=1)
    ax.xaxis.set_major_formatter(FMT_BR)
    ax.grid(axis="x", color=COR_GRADE, linewidth=0.6)
    ax.set_axisbelow(True)
    ax.set_xlabel("Log Ratio (log₂ da razão entre as taxas por 1.000 tokens)\n"
                  f"← mais característico {NOME_DO[SIG_B]}    |    mais característico {NOME_DO[SIG_A]} →", linespacing=1.6)
    fig.text(0.3 / W, 1 - 0.18 / H, titulo, ha="left", va="top", fontsize=12.5, fontweight="bold")
    fig.text(0.3 / W, 1 - 0.5 / H, subtitulo, ha="left", va="top", fontsize=10, color=COR_TINTA_2)
    fig.text(0.3 / W, 0.12 / H, f"Linhas pontilhadas: |Log Ratio| = {num(LIMIAR_LOG_RATIO, 0)} (o dobro da taxa). "
             "Barra hachurada: termo ausente do outro documento (Log Ratio com a correção de 0,5).",
             ha="left", va="bottom", fontsize=9, color=COR_TINTA_2)
    return fig


def grafico_vocabulario(linhas_tipos, linhas_ocorrencias, titulo, subtitulo):
    """Vocabulário compartilhado e exclusivo (skill 03, seção 2.2, item 7). Painel de cima: termos distintos (tipos)
    presentes só na EBIA, nos dois documentos ou só no PBIA. Painel de baixo: parte das ocorrências de cada documento
    que pertence a termos que também ocorrem no outro."""
    W, H = 12.2, 7.4
    fig = plt.figure(figsize=(W, H))
    esq, larg = 3.55, 8.2
    ax1 = fig.add_axes([esq / W, 3.75 / H, larg / W, 2.55 / H])
    ax2 = fig.add_axes([esq / W, 0.95 / H, larg / W, 1.95 / H])

    def segmento(ax, yi, ini, frac, cor, valor):
        ax.barh(yi, frac, left=ini, height=0.62, color=cor, edgecolor="white", linewidth=2)
        texto = f"{br(valor)} ({br_pct(frac)})"
        if frac >= 0.12:
            ax.text(ini + frac / 2, yi, texto, ha="center", va="center", fontsize=9, color=cor_do_texto(cor))
        else:
            ax.text(ini + frac / 2, yi + 0.38, texto, ha="center", va="bottom", fontsize=9, color=COR_TINTA_2)

    for i, (rot, so_a, ambos, so_b) in enumerate(linhas_tipos):
        yi = len(linhas_tipos) - 1 - i
        total, ini = so_a + ambos + so_b, 0.0
        for valor, cor in ((so_a, COR_DOCUMENTO[SIG_A]), (ambos, COR_NEUTRA), (so_b, COR_DOCUMENTO[SIG_B])):
            segmento(ax1, yi, ini, valor / total, cor, valor)
            ini += valor / total
    ax1.set_yticks(range(len(linhas_tipos)))
    ax1.set_yticklabels([r[0] for r in linhas_tipos][::-1])
    ax1.set_title("Termos distintos (tipos)", loc="left", fontsize=10.5)
    for j, (rot, s, comp, excl) in enumerate(linhas_ocorrencias):
        yi = len(linhas_ocorrencias) - 1 - j
        total = comp + excl
        segmento(ax2, yi, 0.0, comp / total, COR_NEUTRA, comp)
        segmento(ax2, yi, comp / total, excl / total, COR_DOCUMENTO[s], excl)
    ax2.set_yticks(range(len(linhas_ocorrencias)))
    ax2.set_yticklabels([r[0] for r in linhas_ocorrencias][::-1])
    ax2.set_title("Ocorrências: parte que pertence a termos presentes nos dois documentos", loc="left", fontsize=10.5)
    for ax in (ax1, ax2):
        ax.set_xlim(0, 1)
        ax.xaxis.set_major_formatter(FuncFormatter(lambda v, _: br_pct(v, 0)))
        ax.tick_params(axis="y", length=0)
        ax.spines["left"].set_visible(False)
    fig.text(esq / W, 1 - 0.18 / H, titulo, ha="left", va="top", fontsize=12.5, fontweight="bold")
    fig.text(esq / W, 1 - 0.5 / H, subtitulo, ha="left", va="top", fontsize=10, color=COR_TINTA_2)
    itens = [Patch(facecolor=COR_DOCUMENTO[SIG_A], label=f"Só {NOME_NO[SIG_A]}"),
             Patch(facecolor=COR_NEUTRA, label="Nos dois documentos"),
             Patch(facecolor=COR_DOCUMENTO[SIG_B], label=f"Só {NOME_NO[SIG_B]}")]
    fig.legend(handles=itens, loc="upper left", bbox_to_anchor=(esq / W, 0.42 / H), ncol=3, fontsize=9,
               frameon=False, borderaxespad=0)
    return fig
```

```python
# =============================================================================
# CONVERGÊNCIA E DIVERGÊNCIA DE TERMOS (Visualizações 7 e 8), por comando do pesquisador: "igual foi feito em BRICS e
# OCDE". Código do gráfico copiado da seção 16 do notebook Declarações IA/Análise_Comparativa_BRICS_OCDE.ipynb, com
# estas adaptações: (a) documentos, artigos e cores desta pasta (EBIA no eixo X; PBIA sem anexos no eixo Y); (b)
# recortes ampliados até o desempate, por decisão já registrada do pesquisador, no lugar do desempate alfabético;
# (c) sem destaques pedidos, os rótulos seguem uma regra fixa e igual para os dois documentos (N_DESTAQUES_POR_ZONA e
# N_DESTAQUES_NEUTROS); (d) cores pedidas pelo pesquisador em 2026-10-14 (EBIA amarelo, PBIA verde, bolinhas para todos),
# com o texto e as linhas de ligação da zona da EBIA num amarelo mais escuro, para o contraste com o fundo branco; (e)
# números dentro dos pontos com fonte de 9 pt (skill 02, seção 3.8, item 1). O rotulador e as regras de zona, de faixa
# neutra, de pontos sobrepostos e de "fora da escala" são os mesmos.
# =============================================================================
MARCADOR_ZONA = {SIG_A: MARCADOR_DOC[SIG_A], SIG_B: MARCADOR_DOC[SIG_B], "neutra": "o"}
NOME_ZONA = {SIG_A: CURTO[SIG_A], SIG_B: CURTO[SIG_B], "neutra": "neutra"}


def normalizar_termo(texto):
    """Minúsculas, sem hífen e com espaços simples (ex.: 'Bem-Estar Social' -> 'bemestar social')."""
    return " ".join(str(texto).lower().replace("-", "").split())


def zona_do_termo(freq_a, freq_b, taxa_a, taxa_b):
    """Zona do termo pela razão entre as taxas: SIG_A ou SIG_B se a taxa de um documento for ao menos
    RAZAO_ZONA_NEUTRA vezes a do outro (ou se o termo só ocorrer nele); 'neutra' nos demais casos."""
    if freq_a > 0 and freq_b == 0:
        return SIG_A
    if freq_b > 0 and freq_a == 0:
        return SIG_B
    if taxa_a >= RAZAO_ZONA_NEUTRA * taxa_b:
        return SIG_A
    if taxa_b >= RAZAO_ZONA_NEUTRA * taxa_a:
        return SIG_B
    return "neutra"


def preparar_convergencia(freq_a, freq_b, recortes, pedidos, rotulos=None):
    """Une os recortes 'top N' dos dois documentos e calcula, para cada termo: ocorrências, taxas por 1.000 tokens,
    G² e Log Ratio (mesma função das Visualizações 4 e 5), zona, destaque e posição na escala. Termos pedidos pelo
    pesquisador que estejam fora dos recortes são acrescentados e sinalizados."""
    termos = list(dict.fromkeys(list(recortes[SIG_A]) + list(recortes[SIG_B])))
    origem = {}
    for t in termos:
        dentro = [CURTO[s] for s in (SIG_A, SIG_B) if t in set(recortes[s])]
        origem[t] = "recortes " + " e ".join(dentro) if len(dentro) == 2 else f"recorte {dentro[0]}"
    rotulo_de = (lambda t: rotulos.get(t, t)) if rotulos is not None else (lambda t: t)

    # Termos pedidos: procurados pelo termo (lema ou par de lemas) ou pelo rótulo (forma escrita).
    chaves = defaultdict(list)
    for t in freq_a.index.union(freq_b.index):
        chaves[normalizar_termo(t)].append(t)
        chaves[normalizar_termo(rotulo_de(t))].append(t)
    pedido_por, nao_encontrados = {}, []
    for grupo, lista in pedidos.items():
        for pedido in lista:
            achados = list(dict.fromkeys(chaves.get(normalizar_termo(pedido), [])))
            if not achados:
                nao_encontrados.append(f"{pedido} ({CURTO.get(grupo, grupo)})")
                continue
            plotados = [t for t in achados if t in termos]
            if plotados:
                escolhido = plotados[0]
            else:   # fora dos recortes: entra no gráfico por pedido do pesquisador
                escolhido = max(achados, key=lambda t: (freq_a.get(t, 0) + freq_b.get(t, 0), t))
                termos.append(escolhido)
                origem[escolhido] = "pedido do pesquisador"
            pedido_por.setdefault(escolhido, []).append(CURTO.get(grupo, grupo))

    tab = tabela_keyness(freq_a, freq_b, N_TOKENS[SIG_A], N_TOKENS[SIG_B]).reindex(termos)
    tab.index.name = "termo"
    tab["x"] = tab[f"por_mil_{SIG_A}"]
    tab["y"] = tab[f"por_mil_{SIG_B}"]
    tab["zona"] = [zona_do_termo(fa, fb, xa, yb) for fa, fb, xa, yb in
                   zip(tab[f"freq_{SIG_A}"], tab[f"freq_{SIG_B}"], tab["x"], tab["y"])]
    tab["rotulo"] = [rotulo_de(t) for t in tab.index]
    tab["origem"] = [origem[t] for t in tab.index]
    tab["pedido"] = [" e ".join(dict.fromkeys(pedido_por.get(t, []))) for t in tab.index]

    # Fora da escala: termo cuja maior taxa supera FATOR_FORA_DA_ESCALA vezes a maior taxa dos demais.
    maiores = tab[["x", "y"]].max(axis=1).sort_values(ascending=False)
    fora = []
    while len(maiores) > 1 and maiores.iloc[0] > FATOR_FORA_DA_ESCALA * maiores.iloc[1]:
        fora.append(maiores.index[0])
        maiores = maiores.iloc[1:]
    tab["fora_da_escala"] = tab.index.isin(fora)

    # Destaques: os pedidos e, por regra fixa e igual para os dois documentos, os N_DESTAQUES_POR_ZONA de maior taxa
    # em cada zona de documento (taxa naquele documento) e os N_DESTAQUES_NEUTROS de maior taxa somada na zona neutra,
    # com desempate alfabético.
    livres = tab[(tab["pedido"] == "") & ~tab["fora_da_escala"]].reset_index()
    livres["soma"] = livres["x"] + livres["y"]
    regra = set()
    for zona, n, ordem in ((SIG_A, N_DESTAQUES_POR_ZONA, "x"), (SIG_B, N_DESTAQUES_POR_ZONA, "y"),
                           ("neutra", N_DESTAQUES_NEUTROS, "soma")):
        z = livres[livres["zona"] == zona].sort_values([ordem, "termo"], ascending=[False, True])
        regra |= set(z.head(n)["termo"])
    tab["destaque"] = np.where(tab["pedido"] != "", "pedido do pesquisador",
                               np.where(tab.index.isin(regra), "regra (maior taxa na zona)", ""))
    tab.attrs["nao_encontrados"] = nao_encontrados
    return tab


# ---------------------------------------------------------------------------
# Posicionamento de rótulos sem sobreposição (cópia do notebook BRICS × OCDE)
# ---------------------------------------------------------------------------
def _sobrepoe(a, b):
    """As caixas (x0, y0, x1, y1) se sobrepõem?"""
    return a[0] < b[2] and b[0] < a[2] and a[1] < b[3] and b[1] < a[3]


def _cruza(p, q, caixa):
    """O segmento p-q atravessa a caixa? (recorte de Liang-Barsky)"""
    (x0, y0), (x1, y1) = p, q
    dx, dy = x1 - x0, y1 - y0
    t0, t1 = 0.0, 1.0
    for pk, qk in ((-dx, x0 - caixa[0]), (dx, caixa[2] - x0), (-dy, y0 - caixa[1]), (dy, caixa[3] - y0)):
        if pk == 0:
            if qk < 0:
                return False
            continue
        t = qk / pk
        if pk < 0:
            t0 = max(t0, t)
        else:
            t1 = min(t1, t)
        if t0 > t1:
            return False
    return True


def _distancia_caixa(p, caixa):
    """Distância de um ponto a uma caixa (0 se estiver dentro)."""
    dx = max(caixa[0] - p[0], 0, p[0] - caixa[2])
    dy = max(caixa[1] - p[1], 0, p[1] - caixa[3])
    return float(np.hypot(dx, dy))


def _segmentos_cruzam(p1, p2, q1, q2):
    """Os segmentos p1-p2 e q1-q2 se cruzam? (pontas comuns não contam)"""
    def orient(a, b, c):
        return (b[0] - a[0]) * (c[1] - a[1]) - (b[1] - a[1]) * (c[0] - a[0])
    return (orient(q1, q2, p1) * orient(q1, q2, p2) < 0) and (orient(p1, p2, q1) * orient(p1, p2, q2) < 0)


class Rotulador:
    """Posiciona rótulos sem sobreposição. Cada rótulo tem posições candidatas ao redor do seu ponto, da mais
    próxima para a mais distante, e cada posição tem um custo: distância ao ponto e sobreposição a pontos,
    a outros rótulos e a linhas de ligação (proibitivas), além de passagem sobre a diagonal ou os limites da
    faixa (evitada). Um rótulo sem linha de ligação não pode ficar mais perto de outro ponto do que do seu.
    Os rótulos são distribuídos em várias ordens (a primeira por densidade; as demais sorteadas com semente
    fixa) e fica a distribuição de menor custo total. O resultado é sempre o mesmo para os mesmos dados."""

    ANGULOS = (0, 35, -35, 70, -70, 110, -110, 145, -145, 180, 90, -90)
    DISTANCIAS_PT = (3, 6, 10, 15, 21, 28, 37, 48, 61, 76, 94, 115, 140)

    def __init__(self, ax, pontos_dados, raios_pt, linhas_suaves_dados):
        self.ax = ax
        self.renderer = ax.figure.canvas.get_renderer()
        self.px = ax.figure.dpi / 72
        self.pontos = ax.transData.transform(pontos_dados)
        self.raios = np.asarray(raios_pt, dtype=float) * self.px
        self.caixas_pontos = [(x - r - 1, y - r - 1, x + r + 1, y + r + 1) for (x, y), r in zip(self.pontos, self.raios)]
        e = ax.get_window_extent(self.renderer)
        self.limite = (e.x0 + 2, e.y0 + 2, e.x1 - 2, e.y1 - 2)
        self.linhas_suaves = [tuple(map(tuple, ax.transData.transform(seg))) for seg in linhas_suaves_dados]
        self.ocupadas, self.segmentos = [], []

    def caixa(self, artista, folga=2):
        bb = artista.get_window_extent(self.renderer)
        return (bb.x0 - folga, bb.y0 - folga, bb.x1 + folga, bb.y1 + folga)

    def ocupar(self, artista):
        self.ocupadas.append(self.caixa(artista))

    def ocupar_segmento(self, p_dados, q_dados):
        p, q = self.ax.transData.transform([p_dados, q_dados])
        self.segmentos.append((tuple(p), tuple(q)))

    def _candidatos(self, alvo, raio, largura, altura, proprio, distancias):
        """Posições candidatas de um rótulo, com o custo que não depende dos outros rótulos a posicionar."""
        candidatos = []
        for d_pt in distancias:
            for k, ang in enumerate(self.ANGULOS):
                u = np.array([np.cos(np.radians(ang)), np.sin(np.radians(ang))])
                ancora = alvo + (raio + d_pt * self.px) * u
                ha = "left" if u[0] > 0.3 else "right" if u[0] < -0.3 else "center"
                va = "bottom" if u[1] > 0.3 else "top" if u[1] < -0.3 else "center"
                x0 = ancora[0] - {"left": 0, "right": largura, "center": largura / 2}[ha]
                y0 = ancora[1] - {"bottom": 0, "top": altura, "center": altura / 2}[va]
                caixa = (x0 - 2, y0 - 2, x0 + largura + 2, y0 + altura + 2)
                ligacao = None
                if d_pt >= 6 and proprio >= 0:
                    ligacao = (tuple(alvo), (float(np.clip(alvo[0], caixa[0] + 2, caixa[2] - 2)),
                                             float(np.clip(alvo[1], caixa[1] + 2, caixa[3] - 2))))
                custo = d_pt + 2.0 * k
                if not (caixa[0] >= self.limite[0] and caixa[1] >= self.limite[1]
                        and caixa[2] <= self.limite[2] and caixa[3] <= self.limite[3]):
                    custo += 5000
                custo += 1000 * sum(_sobrepoe(caixa, c) for c in self.ocupadas)
                custo += 400 * sum(_sobrepoe(caixa, c) for c in self.caixas_pontos)
                custo += 600 * sum(_cruza(p, q, caixa) for p, q in self.segmentos)
                custo += 25 * sum(_cruza(p, q, caixa) for p, q in self.linhas_suaves)
                outros = [_distancia_caixa(p, caixa) - r for i, (p, r) in enumerate(zip(self.pontos, self.raios)) if i != proprio]
                if ligacao is None:
                    # Sem ligação, o rótulo precisa estar claramente mais perto do seu ponto.
                    if proprio >= 0 and outros and min(outros) < _distancia_caixa(alvo, caixa) + 4:
                        custo += 60
                else:
                    # Com ligação, o rótulo também não deve encostar em outro ponto (leitura ambígua).
                    if outros and min(outros) < 7:
                        custo += 80
                    custo += 600 * sum(_cruza(ligacao[0], ligacao[1], c) for c in self.ocupadas)
                    custo += 300 * sum(_cruza(ligacao[0], ligacao[1], c) for i, c in enumerate(self.caixas_pontos) if i != proprio)
                candidatos.append((custo, ancora, ha, va, caixa, ligacao))
        candidatos.sort(key=lambda c: c[0])
        return candidatos[:40]

    @staticmethod
    def _custo_mutuo(candidato, colocados):
        """Custo de um candidato diante dos rótulos já colocados nesta distribuição."""
        caixa, ligacao = candidato[4], candidato[5]
        custo = 0.0
        for caixa_2, ligacao_2 in colocados:
            if _sobrepoe(caixa, caixa_2):
                custo += 1000
            if ligacao_2 is not None and _cruza(ligacao_2[0], ligacao_2[1], caixa):
                custo += 600
            if ligacao is not None:
                if _cruza(ligacao[0], ligacao[1], caixa_2):
                    custo += 600
                if ligacao_2 is not None and _segmentos_cruzam(ligacao[0], ligacao[1], ligacao_2[0], ligacao_2[1]):
                    custo += 200
        return custo

    def posicionar(self, pedidos, tentativas=150, semente=0, distancias=None):
        """Cria os rótulos de 'pedidos' — lista de (xy, texto, cor_da_ligacao, indice_do_ponto, estilo);
        índice -1 indica uma âncora que não é ponto (ex.: a seta de um termo fora da escala)."""
        if not pedidos:
            return []
        artistas, candidatos = [], []
        for xy, texto, _, proprio, estilo in pedidos:
            artista = self.ax.text(0, 0, texto, zorder=6, clip_on=False, ha="left", va="bottom", **estilo)
            bb = artista.get_window_extent(self.renderer)
            alvo = self.ax.transData.transform([xy])[0]
            raio = self.raios[proprio] if proprio >= 0 else 0.0
            candidatos.append(self._candidatos(alvo, raio, bb.width, bb.height, proprio, distancias or self.DISTANCIAS_PT))
            artistas.append(artista)

        alvos = self.ax.transData.transform([p[0] for p in pedidos])
        densidade = [int((np.hypot(*(self.pontos - a).T) < 45 * self.px).sum()) for a in alvos]
        ordens = [sorted(range(len(pedidos)), key=lambda i: -densidade[i])]
        rng = np.random.default_rng(semente)
        ordens += [list(rng.permutation(len(pedidos))) for _ in range(tentativas)]
        melhor = None
        for ordem in ordens:
            escolha, colocados, total = {}, [], 0.0
            for i in ordem:
                custo, candidato = min(((c[0] + self._custo_mutuo(c, colocados), c) for c in candidatos[i]),
                                       key=lambda par: par[0])
                escolha[i] = candidato
                colocados.append((candidato[4], candidato[5]))
                total += custo
                if melhor is not None and total >= melhor[0]:
                    break
            else:
                if melhor is None or total < melhor[0]:
                    melhor = (total, escolha)

        inv = self.ax.transData.inverted()
        for i, (artista, (xy, _, cor, proprio, _)) in enumerate(zip(artistas, pedidos)):
            _, ancora, ha, va, caixa, ligacao = melhor[1][i]
            artista.set_position(inv.transform(ancora))
            artista.set_ha(ha)
            artista.set_va(va)
            self.ocupadas.append(caixa)
            if ligacao is not None:
                self.segmentos.append(ligacao)
                raio_pt = self.raios[proprio] / self.px if proprio >= 0 else 0
                self.ax.annotate("", xy=xy, xytext=inv.transform(ligacao[1]), zorder=5, annotation_clip=False,
                                 arrowprops=dict(arrowstyle="-", color=cor, lw=0.7, shrinkA=0, shrinkB=raio_pt + 1))
        return artistas


# ---------------------------------------------------------------------------
# Gráfico
# ---------------------------------------------------------------------------
def grafico_convergencia(tab, subtitulo, descricao_ponto):
    """Gráfico de convergência e divergência: X = taxa na EBIA, Y = taxa no PBIA (por 1.000 tokens), mesma escala
    nos dois eixos, diagonal de frequência igual e faixa da zona neutra."""
    cor_zona = {SIG_A: COR_DOCUMENTO[SIG_A], SIG_B: COR_DOCUMENTO[SIG_B], "neutra": COR_NEUTRA}
    cor_texto = {SIG_A: COR_TEXTO_DOCUMENTO[SIG_A], SIG_B: COR_TEXTO_DOCUMENTO[SIG_B], "neutra": COR_NEUTRA}
    r_txt = f"{RAZAO_ZONA_NEUTRA:g}".replace(".", ",")
    dentro = tab[~tab["fora_da_escala"]]
    lim = MaxNLocator(nbins=7, steps=[1, 2, 2.5, 5, 10]).tick_values(0, dentro[["x", "y"]].to_numpy().max() * 1.1)[-1]

    largura, altura, lado, esq, base = 7.9, 10.1, 6.0, 1.05, 2.95   # polegadas (subtítulo em duas linhas)
    fig = plt.figure(figsize=(largura, altura))
    ax = fig.add_axes([esq / largura, base / altura, lado / largura, lado / altura])
    ax.set_xlim(0, lim)
    ax.set_ylim(0, lim)
    ax.set_aspect("equal", adjustable="box")

    # Zona neutra (faixa entre y = x / R e y = R·x) e diagonal de frequência igual.
    xs = np.linspace(0, lim, 400)
    ax.fill_between(xs, xs / RAZAO_ZONA_NEUTRA, np.minimum(xs * RAZAO_ZONA_NEUTRA, lim),
                    color=COR_FAIXA, alpha=0.75, lw=0, zorder=0)
    for fator in (RAZAO_ZONA_NEUTRA, 1 / RAZAO_ZONA_NEUTRA):
        ax.plot(xs, xs * fator, color=COR_LIGACAO, lw=0.7, ls=(0, (1.5, 2)), zorder=1)
    ax.plot([0, lim], [0, lim], color=COR_TINTA_2, lw=1.0, ls=(0, (5, 4)), zorder=2)

    # Pontos. Termos com os mesmos valores formam um único ponto, maior, com o número de termos dentro.
    grupos = []
    for (x, y), g in dentro.groupby([dentro["x"].round(9), dentro["y"].round(9)], sort=False):
        grupos.append({"x": g["x"].iloc[0], "y": g["y"].iloc[0], "zona": g["zona"].iloc[0], "n": len(g),
                       "destacados": sorted(g.loc[g["destaque"] != "", "rotulo"])})
    tam_simples, tam_multiplo = 46, 190

    def tamanho(p):   # mesma área visual para círculo, quadrado e triângulo
        return (tam_multiplo if p["n"] > 1 else tam_simples) * AREA_MARCADOR[MARCADOR_ZONA[p["zona"]]]

    for zona in ("neutra", SIG_A, SIG_B):
        for multiplo in (False, True):
            sel = [p for p in grupos if p["zona"] == zona and (p["n"] > 1) == multiplo]
            if sel:
                ax.scatter([p["x"] for p in sel], [p["y"] for p in sel], s=[tamanho(p) for p in sel],
                           marker=MARCADOR_ZONA[zona], color=cor_zona[zona], alpha=0.92, edgecolors="white",
                           linewidths=0.8, zorder=3, clip_on=False)
    for p in grupos:
        if p["n"] > 1:   # no triângulo, o número fica no centroide (1/6 do lado abaixo do centro)
            desloc = -np.sqrt(tamanho(p)) / 6 if MARCADOR_ZONA[p["zona"]] == "^" else 0.0
            ax.annotate(str(p["n"]), (p["x"], p["y"]), xytext=(0, desloc), textcoords="offset points", ha="center",
                        va="center", fontsize=9, fontweight="bold", color=cor_do_texto(cor_zona[p["zona"]]), zorder=4,
                        annotation_clip=False)

    ax.xaxis.set_major_locator(MaxNLocator(nbins=7, steps=[1, 2, 2.5, 5, 10]))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=7, steps=[1, 2, 2.5, 5, 10]))
    ax.xaxis.set_major_formatter(FMT_BR)
    ax.yaxis.set_major_formatter(FMT_BR)
    ax.set_xlabel(f"Frequência relativa {NOME_NO[SIG_A]} (ocorrências por 1.000 tokens)", labelpad=8)
    ax.set_ylabel(f"Frequência relativa {NOME_NO[SIG_B]} (ocorrências por 1.000 tokens)", labelpad=8)
    centro = (esq + lado / 2) / largura   # título centrado sobre a área do gráfico
    fig.text(centro, 1 - 0.18 / altura, f"Convergência e divergência de termos — {ROTULO[SIG_A]} × {ROTULO[SIG_B]}",
             ha="center", va="top", fontsize=13, fontweight="bold")
    fig.text(centro, 1 - 0.5 / altura, subtitulo, ha="center", va="top", fontsize=10, color=COR_TINTA_2,
             linespacing=1.35)
    fig.canvas.draw()

    contorno = [path_effects.withStroke(linewidth=3, foreground="white")]
    raios = [np.sqrt(tamanho(p)) / 2 for p in grupos]
    rot = Rotulador(ax, np.array([(p["x"], p["y"]) for p in grupos]), raios,
                    [[(0, 0), (lim, lim)], [(0, 0), (lim / RAZAO_ZONA_NEUTRA, lim)], [(0, 0), (lim, lim / RAZAO_ZONA_NEUTRA)]])

    # Nome das regiões: no primeiro canto livre de pontos, inteiramente dentro da zona de cada documento.
    inv = ax.transData.inverted()
    for zona, texto, candidatos in (
        (SIG_B, f"↑ mais frequente {NOME_NO[SIG_B]}",
         [(0.03, f, "left", "top") for f in (0.97, 0.91, 0.85, 0.79, 0.73, 0.67)]),
        (SIG_A, f"mais frequente {NOME_NO[SIG_A]} →",
         [(0.97, f, "right", "bottom") for f in (0.03, 0.09, 0.15, 0.21, 0.27, 0.33)]),
    ):
        artista = ax.text(0, 0, texto, transform=ax.transAxes, fontsize=9.5, style="italic",
                          color=cor_texto[zona], zorder=6, path_effects=contorno)
        escolhido = None
        for fx, fy, ha, va in candidatos:
            artista.set_position((fx, fy))
            artista.set_ha(ha)
            artista.set_va(va)
            caixa = rot.caixa(artista, folga=6)   # folga maior: o nome da região não encosta em pontos
            (x0, y0), (x1, y1) = inv.transform([(caixa[0], caixa[1]), (caixa[2], caixa[3])])
            na_zona = y0 >= RAZAO_ZONA_NEUTRA * x1 if zona == SIG_B else y1 <= x0 / RAZAO_ZONA_NEUTRA
            livre = not any(_sobrepoe(caixa, c) for c in rot.caixas_pontos)
            if na_zona and livre:
                escolhido = (fx, fy, ha, va)
                break
        fx, fy, ha, va = escolhido or candidatos[0]
        artista.set_position((fx, fy))
        artista.set_ha(ha)
        artista.set_va(va)
        rot.ocupar(artista)

    # Termos fora da escala: seta na borda, na direção do valor real, com os valores exatos.
    for termo, linha in tab[tab["fora_da_escala"]].iterrows():
        ponta = np.array([min(linha["x"], lim), min(linha["y"], lim)])
        direcao = np.array([linha["x"] > lim, linha["y"] > lim], dtype=float)
        direcao /= np.linalg.norm(direcao)
        cauda = ponta - direcao * lim * 0.075
        ax.annotate("", xy=ponta, xytext=cauda, zorder=6, annotation_clip=False,
                    arrowprops=dict(arrowstyle="-|>", color=cor_zona[linha["zona"]], lw=1.6, mutation_scale=12))
        rot.ocupar_segmento(cauda, ponta)
        rot.posicionar([(cauda, f"{linha['rotulo']} (fora da escala)\n{CURTO[SIG_A]}: {num(linha['x'])} · "
                                f"{CURTO[SIG_B]}: {num(linha['y'])}",
                         cor_zona[linha["zona"]], -1,
                         dict(fontsize=9, color=COR_TINTA, linespacing=1.3, path_effects=contorno))],
                       tentativas=0, distancias=(2, 4, 7, 11))

    # Rótulos dos termos destacados: um por ponto (termos com os mesmos valores dividem o rótulo).
    estilo = dict(fontsize=9, color=COR_TINTA, linespacing=1.15, path_effects=contorno)
    rot.posicionar([((p["x"], p["y"]), "\n".join(p["destacados"]), cor_texto[p["zona"]], i, estilo)
                    for i, p in enumerate(grupos) if p["destacados"]], tentativas=150, semente=SEMENTE)

    # Legenda e instrução de leitura, abaixo do gráfico.
    def item(zona, rotulo):
        m = MARCADOR_ZONA[zona]
        return Line2D([], [], ls="", marker=m, ms=7.5 * np.sqrt(AREA_MARCADOR[m]), mfc=cor_zona[zona], mec="white",
                      mew=0.8, label=rotulo)

    itens = [
        item(SIG_A, f"Mais frequente {NOME_NO[SIG_A]}: taxa {NOME_NO[SIG_A]} ≥ {r_txt}× a taxa {NOME_NO[SIG_B]}"),
        item(SIG_B, f"Mais frequente {NOME_NO[SIG_B]}: taxa {NOME_NO[SIG_B]} ≥ {r_txt}× a taxa {NOME_NO[SIG_A]}"),
        item("neutra", f"Zona neutra: usado por ambos, com diferença menor que {r_txt}×"),
        Line2D([], [], color=COR_TINTA_2, lw=1.0, ls=(0, (5, 4)), label="Linha tracejada: frequência igual nos dois documentos (x = y)"),
        Patch(facecolor=COR_FAIXA, edgecolor=COR_LIGACAO, lw=0.7, ls=":", label=f"Faixa cinza: zona neutra, entre y = x/{r_txt} e y = {r_txt}·x"),
    ]
    fig.legend(handles=itens, loc="upper left", bbox_to_anchor=(esq / largura, (base - 0.62) / altura),
               ncol=1, fontsize=9, handlelength=2.2, handletextpad=0.8, labelspacing=0.45, borderaxespad=0)
    ex_x = float(f"{lim * 0.4:.1g}")   # exemplo na escala do próprio gráfico (4 e 2 nos unigramas)
    ex_x, ex_y = (f"{v:g}".replace(".", ",") for v in (ex_x, ex_x / 2))
    leitura = (f"Como ler: cada ponto é um {descricao_ponto}. O eixo X mostra quantas vezes o termo aparece "
               f"a cada 1.000 tokens (palavras) {NOME_NO[SIG_A]}; o eixo Y, o mesmo {NOME_NO[SIG_B]}. "
               f"Ex.: um ponto em x = {ex_x} e y = {ex_y} indica {ex_x} ocorrências a cada 1.000 palavras "
               f"{NOME_NO[SIG_A]} e {ex_y} {NOME_NO[SIG_B]}. Acima da diagonal, o termo é proporcionalmente mais "
               f"frequente {NOME_NO[SIG_B]}; abaixo, {NOME_NO[SIG_A]}. Pontos sobre um eixo (x = 0 ou y = 0) "
               "aparecem em um só documento. Um ponto maior com um número dentro reúne esse número de termos com "
               "exatamente os mesmos valores.")
    fig.text(esq / largura, (base - 1.75) / altura, textwrap.fill(leitura, 102), ha="left", va="top",
             fontsize=9, color=COR_TINTA_2, linespacing=1.45)
    return fig


def tabela_convergencia_exibicao(tab, nome_indice):
    """Tabela com os valores exatos do gráfico (uma linha por termo), ordenada por zona e taxa."""
    ordem_zona = {SIG_A: 0, "neutra": 1, SIG_B: 2}
    t = tab.copy()
    t["_ordem"] = t["zona"].map(ordem_zona)
    t["_taxa"] = np.where(t["zona"] == SIG_B, t["y"], np.where(t["zona"] == SIG_A, t["x"], t["x"] + t["y"]))
    t = t.sort_values(["_ordem", "_taxa"], ascending=[True, False])
    saida = pd.DataFrame({
        "rótulo": t["rotulo"],
        f"{CURTO[SIG_A]}: ocorrências": t[f"freq_{SIG_A}"],
        f"{CURTO[SIG_A]}: por 1.000 tokens": t["x"],
        f"{CURTO[SIG_B]}: ocorrências": t[f"freq_{SIG_B}"],
        f"{CURTO[SIG_B]}: por 1.000 tokens": t["y"],
        "Log Ratio": t["log_ratio"],
        "G²": t["G2"],
        "zona": t["zona"].map(NOME_ZONA),
        "distintivo (critérios de keyness)": np.where(t["distintivo"], "sim", "não"),
        "origem": t["origem"],
        "pedido como destaque de": t["pedido"].replace("", "–"),
        "rotulado": np.where(t["destaque"] != "", t["destaque"], "–"),
        "fora da escala": np.where(t["fora_da_escala"], "sim", "–"),
    })
    saida.index.name = nome_indice
    return saida


def sintese_convergencia(tab, nome_plural, descricao_recorte):
    """Síntese factual do gráfico: contagens por zona, termos fora da escala e destaques."""
    contagem = tab["zona"].value_counts()
    ambos = int((tab["origem"] == f"recortes {CURTO[SIG_A]} e {CURTO[SIG_B]}").sum())
    print(f"Síntese factual: {len(tab)} {nome_plural} no gráfico ({descricao_recorte}; {ambos} estão nos dois "
          f"recortes). Zona {CURTO[SIG_A]}: {contagem.get(SIG_A, 0)}; zona neutra: {contagem.get('neutra', 0)}; "
          f"zona {CURTO[SIG_B]}: {contagem.get(SIG_B, 0)}.")
    so_a = tab[tab[f"freq_{SIG_B}"] == 0]
    so_b = tab[tab[f"freq_{SIG_A}"] == 0]
    print(f"Só em um documento (sobre um eixo): {len(so_a)} só {NOME_NO[SIG_A]}"
          + (f" ({', '.join(so_a['rotulo'])})" if len(so_a) else "") + f"; {len(so_b)} só {NOME_NO[SIG_B]}"
          + (f" ({', '.join(so_b['rotulo'])})" if len(so_b) else "") + ".")
    for termo, l in tab[tab["fora_da_escala"]].iterrows():
        print(f"Fora da escala (indicado por seta): {l['rotulo']} — {CURTO[SIG_A]} {num(l['x'])}; {CURTO[SIG_B]} "
              f"{num(l['y'])} por 1.000 tokens; zona {NOME_ZONA[l['zona']]}.")
    acrescentados = tab[tab["origem"] == "pedido do pesquisador"]
    if len(acrescentados):
        print("Acrescentados por pedido do pesquisador (fora dos recortes): " + ", ".join(acrescentados["rotulo"]) + ".")
    for item in tab.attrs.get("nao_encontrados", []):
        print(f"Pedido não encontrado em nenhum dos documentos: {item}.")
    for zona in (SIG_A, "neutra", SIG_B):
        r = tab[(tab["destaque"] == "regra (maior taxa na zona)") & (tab["zona"] == zona)]
        print(f"Rotulados pela regra na zona {NOME_ZONA[zona]}: " + (", ".join(r["rotulo"]) or "(nenhum)") + ".")
    print(f"Passam nos critérios de keyness (G² ≥ {num(LIMIAR_G2)}, |Log Ratio| ≥ {num(LIMIAR_LOG_RATIO, 1)} e "
          f"frequência combinada ≥ {FREQ_MINIMA_COMPARACAO}): {int(tab['distintivo'].sum())} de {len(tab)}; "
          f"fora da zona neutra sem passar nesses critérios: "
          f"{int(((tab['zona'] != 'neutra') & ~tab['distintivo']).sum())}.")
```

```python
# =============================================================================
# ENQUADRAMENTOS TEMÁTICOS DOS BIGRAMAS (Visualizações 9 a 12), por comando do pesquisador (2026-10-14), no desenho
# do gráfico "Principais termos" enviado por ele: à esquerda, os bigramas mais frequentes, com a cor do enquadramento;
# à direita, o total de ocorrências de cada enquadramento e a proporção no total dos quatro. Contagens absolutas, por
# comando do pesquisador.
# =============================================================================
GRUPOS = list(CODEBOOK_ENQUADRAMENTOS)


def montar_enquadramentos():
    """Confere o codebook do pesquisador (nenhum termo em dois grupos) e classifica a base de cada documento: os
    N_BIGRAMAS_BASE_ENQUADRAMENTO bigramas mais frequentes (ampliados até o desempate), pela forma escrita exibida
    nos notebooks individuais, sem diferenciar maiúsculas. Todo bigrama da base precisa ter grupo. Termos sem grupo
    na lista do pesquisador só entram pela CLASSIFICACAO_PROPOSTA (proposta do agente, a confirmar), sinalizada."""
    grupo_do_termo, repetidos = {}, []
    for grupo, termos in [(GRUPO_NEUTRO, TERMOS_NEUTROS)] + list(CODEBOOK_ENQUADRAMENTOS.items()):
        for termo in termos:
            chave = termo.lower()
            if chave in grupo_do_termo:
                repetidos.append(f"{termo} ({grupo_do_termo[chave]} e {grupo})")
            grupo_do_termo[chave] = grupo
    if repetidos:
        raise ValueError("Termos em mais de um grupo do codebook: " + "; ".join(repetidos))
    propostos = {}
    for termo, grupo in CLASSIFICACAO_PROPOSTA.items():
        if termo.lower() in grupo_do_termo:
            raise ValueError(f"{termo} já está na lista do pesquisador: retire-o de CLASSIFICACAO_PROPOSTA.")
        grupo_do_termo[termo.lower()] = propostos[termo.lower()] = grupo
    enq = {}
    for s in SIGLAS:
        base, info = recorte(BIG_VALIDOS[s], N_BIGRAMAS_BASE_ENQUADRAMENTO, ampliar_empates=AMPLIAR_EMPATES,
                             chave="bigrama")
        base = base.reset_index(drop=True)
        base.insert(0, "posição", np.arange(1, len(base) + 1))
        base["grupo"] = base["bigrama"].str.lower().map(grupo_do_termo)
        base["classificação"] = np.where(base["bigrama"].str.lower().isin(propostos), "proposta do agente", "pesquisador")
        sem_grupo = base.loc[base["grupo"].isna(), "bigrama"].tolist()
        if sem_grupo:
            raise ValueError(f"{ROTULO[s]}: bigramas da base sem grupo no codebook: {', '.join(sem_grupo)}")
        cls = base[base["grupo"] != GRUPO_NEUTRO].copy()
        cls["rótulo"] = [REUNIR_NA_EXIBICAO.get(b.lower(), b) for b in cls["bigrama"]]
        exib = (cls.groupby("rótulo", sort=False)
                .agg(frequência=("frequência", "sum"), grupo=("grupo", "first"), grupos=("grupo", "nunique"),
                     bigramas_reunidos=("bigrama", lambda x: " + ".join(x)), n_reunidos=("bigrama", "size"))
                .reset_index())
        if (exib["grupos"] > 1).any():
            raise ValueError(f"{ROTULO[s]}: bigramas reunidos na exibição estão em grupos diferentes.")
        exib["por_mil"] = exib["frequência"] / N_TOKENS[s] * 1000
        exib = exib.sort_values(["frequência", "rótulo"], ascending=[False, True]).reset_index(drop=True)
        enq[s] = {"base": base, "info": info, "classificados": cls, "exibicao": exib,
                  "neutros": base[base["grupo"] == GRUPO_NEUTRO],
                  "totais": cls.groupby("grupo")["frequência"].sum().reindex(GRUPOS, fill_value=0)}
    return grupo_do_termo, enq, propostos


def subtitulo_enquadramento(sigla, top):
    """Subtítulo da figura: recorte do documento, o que são os termos, critério do painel da esquerda e base do painel
    da direita (o mesmo texto do site)."""
    d = ENQ[sigla]
    return (f"{RECORTE_ENQUADRAMENTO[sigla]}Bigramas (expressões de duas palavras) mais citados no documento, classificados "
            f"em quatro enquadramentos temáticos. À esquerda, os {len(top)} com {int(top['frequência'].min())} ou mais "
            f"ocorrências; à direita, o total de ocorrências dos {len(d['classificados'])} classificados entre os "
            f"{len(d['base'])} mais citados. Termos neutros excluídos; contagens absolutas.")


def nome_curto(grupo):
    """Primeira palavra do nome do enquadramento, para o eixo do painel da direita (o nome inteiro está na legenda)."""
    return grupo.split(",")[0].split()[0]


def grafico_enquadramento(sigla, n_pedido, limites):
    """Figura acadêmica com dois painéis simétricos (mesma largura e mesma altura), desenhada na largura de impressão
    (10 polegadas, 25,4 cm), com fontes de 9 pt ou mais. À esquerda, os bigramas não neutros mais frequentes da base
    (n_pedido, ampliado até o desempate), em ocorrências absolutas, com a cor do enquadramento; à direita, a soma das
    ocorrências de cada enquadramento na base inteira, sem os neutros, e a proporção no total dos quatro. Título e
    subtítulo informam documento, recorte e base; a legenda, logo abaixo do subtítulo, traz o nome inteiro de cada
    enquadramento e vale para os dois painéis (o eixo da direita usa a primeira palavra do nome). Escalas iguais nos
    gráficos da EBIA e do PBIA."""
    d = ENQ[sigla]
    top, info = recorte(d["exibicao"], n_pedido, ampliar_empates=AMPLIAR_EMPATES, chave="rótulo")
    n = len(top)
    W, esq, vao, dir_ = 10.0, 2.55, 0.75, 0.15                 # polegadas; esq inclui a coluna dos nomes
    larg = (W - esq - vao - dir_) / 2
    linhas_sub = textwrap.wrap(subtitulo_enquadramento(sigla, top), 128)
    y_sub, y_leg = 0.52, 0.52 + 0.195 * len(linhas_sub) + 0.13
    topo, base_in = y_leg + 0.5 + 0.22, 0.72
    alt = max(4.0, 0.28 * n + 0.3)
    H = topo + alt + base_in
    fig = plt.figure(figsize=(W, H))
    ax1 = fig.add_axes([esq / W, base_in / H, larg / W, alt / H])
    ax2 = fig.add_axes([(esq + larg + vao) / W, base_in / H, larg / W, alt / H])

    # Painel da esquerda: bigramas mais frequentes, com a cor do enquadramento
    y = np.arange(n)[::-1]
    ax1.barh(y, top["frequência"], height=0.72, color=[COR_ENQUADRAMENTO[g] for g in top["grupo"]], zorder=2)
    for yi, f in zip(y, top["frequência"]):
        ax1.text(f + limites["x"] * 0.012, yi, br(int(f)), va="center", ha="left", fontsize=9, color=COR_TINTA_2)
    ax1.set_yticks(y)
    ax1.set_yticklabels(top["rótulo"], fontsize=10)
    ax1.tick_params(axis="y", length=0)
    ax1.set_ylim(-0.6, n - 0.4)
    ax1.set_xlim(0, limites["x"])
    ax1.xaxis.set_major_locator(MaxNLocator(integer=True, steps=[1, 2, 5, 10]))
    ax1.xaxis.set_major_formatter(EIXO_BR)
    ax1.set_xlabel("Ocorrências absolutas no documento", fontsize=10)
    if info["exibidos"] > n_pedido:   # fim do N pedido; abaixo, os empatados (decisão do pesquisador)
        corte = y[n_pedido - 1] - 0.5
        ax1.axhline(corte, color="#9a9893", linestyle=(0, (4, 3)), linewidth=0.8, zorder=1)
        ax1.text(limites["x"], corte + 0.1, f"N = {n_pedido}", ha="right", va="bottom", fontsize=9, color="#77756f")

    # Painel da direita: soma das ocorrências de cada enquadramento e proporção no total dos quatro
    tot = d["totais"]
    soma = int(tot.sum())
    x = np.arange(len(GRUPOS))
    ax2.bar(x, tot.to_numpy(), width=0.64, color=[COR_ENQUADRAMENTO[g] for g in GRUPOS], zorder=2)
    for xi, v in zip(x, tot.to_numpy()):
        ax2.text(xi, v + limites["y"] * 0.015, f"{br(int(v))}\n({br_pct(v / soma)})", ha="center", va="bottom",
                 fontsize=10, fontweight="bold", color=COR_TINTA, linespacing=1.3)
    ax2.set_xticks(x)
    ax2.set_xticklabels([nome_curto(g) for g in GRUPOS], fontsize=9, va="baseline")   # mesma linha de base, com acento ou não
    ax2.tick_params(axis="x", length=0, pad=13)
    ax2.set_xlim(-0.5, len(GRUPOS) - 0.5)
    ax2.set_ylim(0, limites["y"])
    ax2.yaxis.set_major_locator(MaxNLocator(nbins=6, integer=True, steps=[1, 2, 5, 10]))
    ax2.yaxis.set_major_formatter(EIXO_BR)
    ax2.set_ylabel("Total de ocorrências dos bigramas", fontsize=10)
    for ax, eixo in ((ax1, "x"), (ax2, "y")):
        ax.grid(axis=eixo, color=COR_GRADE, linewidth=0.7)
        ax.set_axisbelow(True)
        for lado in ("left", "bottom"):
            ax.spines[lado].set_color(COR_TINTA_2)

    # Título, subtítulo e legenda, alinhados à esquerda. A legenda tem duas colunas, lidas por linha na ordem das barras
    # da direita (o matplotlib preenche por coluna, daí a ordem 0, 2, 1, 3).
    x0 = 0.15 / W
    fig.text(x0, 1 - 0.18 / H, TITULO_ENQUADRAMENTO[sigla], ha="left", va="top", fontsize=14, fontweight="bold",
             color=COR_TINTA)
    fig.text(x0, 1 - y_sub / H, "\n".join(linhas_sub), ha="left", va="top", fontsize=10, color=COR_TINTA_2,
             linespacing=1.4)
    ordem = [0, 2, 1, 3] if len(GRUPOS) == 4 else list(range(len(GRUPOS)))
    fig.legend(handles=[Patch(facecolor=COR_ENQUADRAMENTO[GRUPOS[i]], label=GRUPOS[i]) for i in ordem], loc="upper left",
               bbox_to_anchor=(x0, 1 - y_leg / H), ncol=2, frameon=False, fontsize=9.5, handlelength=1.3,
               handleheight=1.0, columnspacing=2.4, labelspacing=0.55, borderaxespad=0)
    return fig, top, info


def exibir_enquadramento(sigla, n_pedido, nome_figura, tabelas_da_base=True):
    """Gráfico, tabela dos bigramas exibidos e, se pedido, os totais da base e os neutros excluídos; síntese factual."""
    fig, top, info = grafico_enquadramento(sigla, n_pedido, LIMITES_ENQ)
    salvar_figura(fig, nome_figura)
    plt.show()
    d, tot = ENQ[sigla], ENQ[sigla]["totais"]
    tab = pd.DataFrame({"bigrama": top["rótulo"].to_numpy(), "enquadramento": top["grupo"].to_numpy(),
                        "ocorrências": top["frequência"].to_numpy(), "por 1.000 tokens": top["por_mil"].to_numpy(),
                        "bigramas reunidos na barra": np.where(top["n_reunidos"].to_numpy() > 1,
                                                               top["bigramas_reunidos"].to_numpy(), "–")},
                       index=pd.RangeIndex(1, len(top) + 1, name="posição"))
    with pd.option_context("display.max_rows", None, "display.max_colwidth", None):
        display(tabela_br(tab, {"por 1.000 tokens": 2}))
    salvar_tabela(tab, nome_figura)
    if tabelas_da_base:
        resumo = pd.DataFrame({"bigramas na base": d["classificados"].groupby("grupo").size().reindex(GRUPOS, fill_value=0),
                               "ocorrências": tot, "% do total dos quatro": 100 * tot / tot.sum()})
        resumo.loc["Total"] = [resumo["bigramas na base"].sum(), tot.sum(), 100.0]
        resumo.index.name = "enquadramento"
        display(tabela_br(resumo, {"% do total dos quatro": 1}))
    neutros = d["neutros"]
    print("Síntese factual:")
    print(f"- Base: {len(d['base'])} bigramas; {len(neutros)} neutros excluídos ({br(int(neutros['frequência'].sum()))} "
          f"ocorrências: " + "; ".join(f"{b} {br(int(f))}" for b, f in zip(neutros["bigrama"], neutros["frequência"]))
          + f"); {len(d['classificados'])} nos quatro enquadramentos, com {br(int(tot.sum()))} ocorrências.")
    print(f"- Painel da esquerda: pedido {n_pedido}; exibidos {info['exibidos']}"
          + (f" (acrescentados por empate na frequência {br(int(info['limite']))}: {', '.join(info['acrescentados'])})"
             if info["acrescentados"] else " (sem empate no ponto de corte)")
          + (f"; primeiro fora: {info['primeiro_fora'][0]} ({br(int(info['primeiro_fora'][1]))})" if info["primeiro_fora"] else "")
          + ". Por enquadramento: " + "; ".join(f"{g} {n}" for g, n in
                                                top["grupo"].value_counts().reindex(GRUPOS, fill_value=0).items()) + ".")
    print("- Painel da direita: " + "; ".join(f"{g} {br(int(v))} ({br_pct(v / tot.sum())})" for g, v in tot.items()) + ".")
    return top


def contagens_por_unidade(R):
    """Ocorrências de cada bigrama (par de lemas, com as formas escritas reunidas pela regra de bigramas()) em cada
    unidade do documento. Confere que os totais coincidem com os de bigramas(), a função dos notebooks individuais."""
    registros = R["itens"].to_dict("records")
    ocorr = [((a["lema"], b["lema"]), f"{a['texto'].lower()} {b['texto'].lower()}", a["unidade"])
             for a, b in zip(registros, registros[1:])
             if a["unidade"] == b["unidade"] and a["sent"] == b["sent"] and a["conteudo"] and b["conteudo"]]
    total_par = Counter(k for k, _, _ in ocorr)
    por_forma = defaultdict(Counter)
    for k, forma, _ in ocorr:
        por_forma[forma][k] += 1
    canonico = {forma: sorted(c, key=lambda k: (-total_par[k], k))[0] for forma, c in por_forma.items()}
    contagem = Counter((f"{canonico[f][0]} + {canonico[f][1]}", u) for _, f, u in ocorr)
    totais = Counter()
    for (lemas, _), n in contagem.items():
        totais[lemas] += n
    if totais != Counter({f"{a} + {b}": n for (a, b), n in bigramas(R)[0].items()}):
        raise RuntimeError("As contagens por unidade não coincidem com as de bigramas().")
    return contagem


def bootstrap_enquadramentos(sigla, pares_por_grupo):
    """Bootstrap por unidades (skill 03, seção 3.6): reamostra, com reposição, as unidades do documento N_BOOTSTRAP
    vezes e recalcula as ocorrências de cada enquadramento, a proporção de cada um no total dos quatro e a taxa por
    1.000 tokens. Semente própria do documento, derivada de SEMENTE e da sigla (reprodutível e independente)."""
    ids = [u["id"] for u in RES[sigla]["incluidas"]]
    pos = {u: i for i, u in enumerate(ids)}
    M = np.zeros((len(ids), len(GRUPOS)))
    for j, g in enumerate(GRUPOS):
        alvo = set(pares_por_grupo[g])
        for (lemas, u), n in CONT_UNIDADE[sigla].items():
            if lemas in alvo:
                M[pos[u], j] += n
    tokens = T[sigla][T[sigla]["conta"]].groupby("unidade").size().reindex(ids, fill_value=0).to_numpy()
    if tokens.sum() != N_TOKENS[sigla]:
        raise RuntimeError("Os tokens por unidade não somam o total do documento.")
    rng = np.random.default_rng(np.random.SeedSequence([SEMENTE, zlib.crc32(sigla.encode("utf-8"))]))
    idx = rng.integers(0, len(ids), size=(N_BOOTSTRAP, len(ids)))
    soma = M[idx].sum(axis=1)
    return {"ocorrencias": M.sum(axis=0), "prop": M.sum(axis=0) / M.sum(), "taxa": M.sum(axis=0) / tokens.sum() * 1000,
            "prop_boot": soma / soma.sum(axis=1, keepdims=True), "taxa_boot": soma / tokens[idx].sum(axis=1, keepdims=True) * 1000}


def intervalo(amostras):
    """Intervalo percentil do bootstrap, no NIVEL_CONFIANCA."""
    a = (1 - NIVEL_CONFIANCA) / 2
    return np.quantile(amostras, [a, 1 - a], axis=0)


def comparar_enquadramentos(pares_por_doc):
    """Por enquadramento: ocorrências, proporção no total dos quatro e taxa por 1.000 tokens em cada documento, com
    intervalo de confiança; e a diferença EBIA − PBIA, afirmada só quando o intervalo da diferença não inclui zero."""
    b = {s: bootstrap_enquadramentos(s, pares_por_doc[s]) for s in SIGLAS}
    linhas, numeros = [], []
    for j, g in enumerate(GRUPOS):
        linha = {"enquadramento": g}
        num_linha = {"enquadramento": g}
        for s in SIGLAS:
            ip, it = intervalo(b[s]["prop_boot"][:, j]), intervalo(b[s]["taxa_boot"][:, j])
            linha[f"{CURTO[s]}: ocorrências"] = br(int(b[s]["ocorrencias"][j]))
            linha[f"{CURTO[s]}: % dos quatro [IC]"] = f"{br(100 * b[s]['prop'][j], 1)} [{br(100 * ip[0], 1)}; {br(100 * ip[1], 1)}]"
            linha[f"{CURTO[s]}: por 1.000 tokens [IC]"] = f"{br(b[s]['taxa'][j], 2)} [{br(it[0], 2)}; {br(it[1], 2)}]"
            num_linha.update({f"{s}_ocorrencias": b[s]["ocorrencias"][j], f"{s}_prop": b[s]["prop"][j],
                              f"{s}_prop_ic": ip, f"{s}_taxa": b[s]["taxa"][j], f"{s}_taxa_ic": it})
        for medida, nome, escala, casas in (("prop", "% dos quatro (p.p.)", 100, 1), ("taxa", "por 1.000 tokens", 1, 2)):
            dif = b[SIG_A][medida][j] - b[SIG_B][medida][j]
            ic = intervalo(b[SIG_A][f"{medida}_boot"][:, j] - b[SIG_B][f"{medida}_boot"][:, j])
            afirmada = ("sim: maior " + NOME_NO[SIG_A]) if ic[0] > 0 else ("sim: maior " + NOME_NO[SIG_B]) if ic[1] < 0 else "não (o IC inclui zero)"
            linha[f"diferença EBIA − PBIA, {nome} [IC]"] = f"{br(escala * dif, casas)} [{br(escala * ic[0], casas)}; {br(escala * ic[1], casas)}]"
            linha[f"diferença afirmada ({nome.split(' (')[0]})"] = afirmada
            num_linha.update({f"dif_{medida}": dif, f"dif_{medida}_ic": ic, f"afirmada_{medida}": afirmada})
        linhas.append(linha)
        numeros.append(num_linha)
    return pd.DataFrame(linhas).set_index("enquadramento"), pd.DataFrame(numeros).set_index("enquadramento")
```

## Pré-processamento e perfil comparativo
Os dois documentos passam pelo mesmo pipeline, com os mesmos parâmetros. O perfil expõe a disparidade de tamanho antes e depois da retirada do excesso marcado, os filtros aplicados e os lemas de classes de conteúdo descartados pela lista de *stopwords* em cada documento.

```python
nlp = spacy.load(MODELO_SPACY)
RES, T, LEMAS, BIG, BIG_VALIDOS, N_TOKENS, N_TOKENS_ANTES, N_EXCESSO = {}, {}, {}, {}, {}, {}, {}, {}
for s in SIGLAS:                     # mesmo pipeline e mesmos parâmetros nos dois documentos (skill 03, seção 3.2)
    RES[s] = processar(DOCS[s], nlp)
    T[s] = RES[s]["itens"]
    N_TOKENS[s] = int(T[s]["conta"].sum())               # denominador das taxas (sem o excesso marcado)
    N_TOKENS_ANTES[s] = int(T[s]["palavra"].sum())       # antes da retirada do excesso marcado
    N_EXCESSO[s] = int((T[s]["palavra"] & T[s]["excesso"]).sum())
    LEMAS[s] = tabela_lemas(RES[s])
    BIG[s] = tabela_bigramas(RES[s])
    BIG_VALIDOS[s] = BIG[s][BIG[s]["frequência"] >= FREQ_MINIMA_BIGRAMA].reset_index(drop=True)
FREQ_LEMAS = {s: LEMAS[s].set_index("lema")["frequência"] for s in SIGLAS}
TAXA_LEMAS = {s: LEMAS[s].set_index("lema")["por_mil"] for s in SIGLAS}
FREQ_BIG = {s: BIG[s].set_index("lemas")["frequência"] for s in SIGLAS}     # bigramas: chave = par de lemas
TAXA_BIG = {s: BIG[s].set_index("lemas")["por_mil"] for s in SIGLAS}
ROTULO_BIGRAMA = rotulos_bigramas(BIG[SIG_A], BIG[SIG_B])                   # forma escrita mais frequente nos dois
CONT_UNIDADE = {s: contagens_por_unidade(RES[s]) for s in SIGLAS}          # bigramas por unidade (bootstrap)

# 1. Perfil comparativo e disparidade de tamanho, antes e depois da retirada do excesso marcado
perfil = {}
for s in SIGLAS:
    t, ext = T[s], DOCS[s]["extraction"]
    tipos = Counter(u["type"] for u in RES[s]["incluidas"])
    cont = t[t["conteudo"]]
    sw = t[t["conta"] & t["classe_conteudo"] & t["stopword"] & ~t["modal"]]
    mod = t[t["conta"] & t["modal"]].groupby("lema").size()
    perfil[ROTULO[s]] = {
        "Recorte (páginas do PDF)": f"{ext['page_range'][0]} a {ext['page_range'][1]}"
                                    + (" (documento inteiro)" if ext["cut"] is None else f" ({ext['cut'].replace('_', ' ')})"),
        "Filtros de unidades": "nenhum" if not RES[s]["filtros"] else f"{len(RES[s]['filtros'])} unidades",
        "Palavras retiradas por comando do pesquisador: marcadores de enumeração": br(int(t["enumeracao"].sum())),
        "Palavras retiradas por comando do pesquisador: fórmulas": br(int(t["retirada_comando"].notna().sum())),
        "Unidades analisadas": br(len(RES[s]["incluidas"])),
        "  títulos internos · parágrafos · itens de lista": " · ".join(br(tipos.get(k, 0)) for k in ("heading", "paragraph", "list_item")),
        "Tokens no texto, antes da retirada do excesso marcado": br(N_TOKENS_ANTES[s]),
        "Tokens retirados como excesso marcado": br(N_EXCESSO[s]),
        "Tokens do documento (denominador das taxas)": br(N_TOKENS[s]),
        "Tokens de conteúdo": f"{br(len(cont))} ({br_pct(len(cont) / N_TOKENS[s])})",
        "Lemas de conteúdo distintos": br(len(LEMAS[s])),
        "Bigramas distintos": br(len(BIG[s])),
        f"Bigramas com frequência ≥ {FREQ_MINIMA_BIGRAMA}": br(len(BIG_VALIDOS[s])),
        "Lemas de conteúdo descartados pela lista de stopwords": f"{br(sw['lema'].nunique())} lemas; {br(len(sw))} ocorrências",
        "Verbos modais (ocorrências por 1.000 tokens)": "; ".join(f"{l} {br(n / N_TOKENS[s] * 1000, 2)}" for l, n in mod.items()),
        "Correções de lema aplicadas (ocorrências)": br(int((t["corrigido"] & t["palavra"]).sum())),
        "Formas com hífen tratadas à parte (ênclises e hífen como travessão)": br(len(RES[s]["especiais"])),
    }
PERFIL = pd.DataFrame(perfil)
with pd.option_context("display.max_colwidth", None):
    display(PERFIL)
maior, menor = (SIG_A, SIG_B) if N_TOKENS[SIG_A] >= N_TOKENS[SIG_B] else (SIG_B, SIG_A)
print(f"Disparidade de tamanho: {ROTULO[SIG_A]} {br(N_TOKENS[SIG_A])} tokens; {ROTULO[SIG_B]} {br(N_TOKENS[SIG_B])} "
      f"tokens (razão {CURTO[SIG_A]}/{CURTO[SIG_B]} = {br(N_TOKENS[SIG_A] / N_TOKENS[SIG_B], 3)}): {ROTULO[maior]} tem "
      f"{br_pct(N_TOKENS[maior] / N_TOKENS[menor] - 1)} mais tokens que {ROTULO[menor]}. Por isso, toda comparação usa "
      f"ocorrências por 1.000 tokens (skill 03, seção 3.4), salvo os gráficos dos enquadramentos, em contagens absolutas "
      f"por comando do pesquisador.")

# 2. Lemas de classes de conteúdo descartados pela lista de stopwords, nos dois documentos (continuam no denominador)
desc = {s: T[s][T[s]["conta"] & T[s]["classe_conteudo"] & T[s]["stopword"] & ~T[s]["modal"]].groupby("lema").size()
        for s in SIGLAS}
DESCARTADOS = pd.DataFrame({f"{CURTO[s]}: ocorrências": desc[s] for s in SIGLAS}).fillna(0).astype(int)
for s in SIGLAS:
    DESCARTADOS[f"{CURTO[s]}: por 1.000 tokens"] = DESCARTADOS[f"{CURTO[s]}: ocorrências"] / N_TOKENS[s] * 1000
limite_top = {s: int(LEMAS[s].iloc[TOP_N_COMPARACAO_LEMAS - 1]["frequência"]) for s in SIGLAS}
DESCARTADOS[f"entraria entre os {TOP_N_COMPARACAO_LEMAS} unigramas de"] = [
    ", ".join(CURTO[s] for s in SIGLAS if r[f"{CURTO[s]}: ocorrências"] >= limite_top[s]) or "–"
    for _, r in DESCARTADOS.iterrows()]
DESCARTADOS["_soma"] = sum(DESCARTADOS[f"{CURTO[s]}: ocorrências"] for s in SIGLAS)
DESCARTADOS = DESCARTADOS.sort_values(["_soma"], ascending=False).drop(columns="_soma")
DESCARTADOS.index.name = "lema"
print(f"\nLemas de classes de conteúdo descartados pela lista de stopwords (os 25 mais frequentes nos dois documentos "
      f"somados, de {br(len(DESCARTADOS))}); a última coluna indica onde o lema entraria entre os "
      f"{TOP_N_COMPARACAO_LEMAS} unigramas se fosse mantido:")
display(tabela_br(DESCARTADOS.head(25), {f"{CURTO[s]}: por 1.000 tokens": 2 for s in SIGLAS}))

# 3. Se fossem mantidos (decisão do pesquisador): taxas, Log Ratio e G² dos descartados que entrariam entre os 30
entrariam = DESCARTADOS[DESCARTADOS[f"entraria entre os {TOP_N_COMPARACAO_LEMAS} unigramas de"] != "–"]
if len(entrariam):
    k = tabela_keyness(entrariam[f"{CURTO[SIG_A]}: ocorrências"], entrariam[f"{CURTO[SIG_B]}: ocorrências"],
                       N_TOKENS[SIG_A], N_TOKENS[SIG_B]).reindex(entrariam.index)
    print("Se fossem mantidos (STOPWORDS_MANTIDAS), estes lemas teriam: "
          + "; ".join(f"{l} ({CURTO[SIG_A]} {br(r[f'por_mil_{SIG_A}'], 2)}; {CURTO[SIG_B]} {br(r[f'por_mil_{SIG_B}'], 2)} "
                      f"por 1.000 tokens; Log Ratio {br(r['log_ratio'], 2)}; G² {br(r['G2'], 1)}"
                      + ("; distintivo" if r["distintivo"] else "") + ")" for l, r in k.iterrows()) + ".")
```

## Possível inflação não marcada (mostrada, não retirada)
Repetições sem asterisco não são retiradas pelo agente (skill 03, seção 3.7). Esta seção mostra, nos dois documentos, as unidades de texto idêntico e as fórmulas do texto (expressões de 4 palavras em 5 ou mais unidades), para que o pesquisador avalie o seu peso nas comparações.

```python
# Possível inflação NÃO marcada, nos dois documentos: o notebook mostra e mede; nada aqui é retirado (skill 03, seção 3.7).
def unidades_repetidas(s):
    """Unidades de texto idêntico (títulos e itens que se repetem literalmente) e os tokens de conteúdo além da 1ª."""
    textos = pd.Series({u["id"]: RES[s]["limpos"][u["id"]] for u in RES[s]["incluidas"]})
    linhas = []
    for texto, n in textos.value_counts().items():
        if n < 2:
            continue
        ids = list(textos[textos == texto].index)
        g = T[s][T[s]["unidade"].isin(ids) & T[s]["conteudo"]]
        linhas.append({"documento": ROTULO[s], "texto (literal)": texto, "unidades": n,
                       "tokens de conteúdo além da 1ª": len(g) - int((g["unidade"] == min(ids)).sum()),
                       "lemas de conteúdo": ", ".join(sorted(set(g["lema"]))) or "–"})
    return linhas


def formulas(s, minimo=5, n=10):
    """Expressões de 4 palavras que se repetem em `minimo` ou mais unidades (fórmulas do texto)."""
    quatro = defaultdict(set)
    for uid, g in T[s][T[s]["conta"]].groupby("unidade"):
        for _, h in g.groupby("sent"):
            palavras = [x.lower() for x in h["texto"]]
            for i in range(len(palavras) - 3):
                quatro[" ".join(palavras[i:i + 4])].add(uid)
    tab = pd.DataFrame([{"documento": ROTULO[s], "expressão de 4 palavras": k, "unidades": len(v)}
                        for k, v in quatro.items() if len(v) >= minimo])
    return tab.sort_values(["unidades", "expressão de 4 palavras"], ascending=[False, True]).head(n) if len(tab) else tab


REPETIDAS = pd.DataFrame([l for s in SIGLAS for l in unidades_repetidas(s)])
print("1. Unidades de texto idêntico, por documento:")
if len(REPETIDAS):
    with pd.option_context("display.max_colwidth", None, "display.max_rows", None):
        display(tabela_br(REPETIDAS.sort_values(["documento", "unidades"], ascending=[True, False])
                          .set_index(["documento", "texto (literal)"])))
FORMULAS = pd.concat([formulas(s) for s in SIGLAS], ignore_index=True)
print("\n2. Expressões de 4 palavras em 5 ou mais unidades (até 10 por documento):")
if len(FORMULAS):
    display(tabela_br(FORMULAS.set_index(["documento", "expressão de 4 palavras"])))
```

- **Frequência zero.** Recebe a correção de 0,5 só no *Log Ratio*; a barra hachurada marca esses termos (exclusivos de um documento).
- **Exibição.** Até 15 termos por documento, por ordem de G². A tabela traz todos os distintivos.
- **Excesso marcado.** Nenhum token foi retirado.

- As colunas da direita trazem o G² e as ocorrências em cada documento. A ordem dentro de cada bloco segue o G² (evidência estatística), não o comprimento da barra.

```python
# Codebook dos enquadramentos (comando do pesquisador, 2026-10-14): conferência e classificação das bases.
GRUPO_DO_TERMO, ENQ, PROPOSTOS = montar_enquadramentos()

# 1. Resumo por grupo: termos da lista e bigramas da base de cada documento
linhas = []
for g in [GRUPO_NEUTRO] + GRUPOS:
    termos = TERMOS_NEUTROS if g == GRUPO_NEUTRO else CODEBOOK_ENQUADRAMENTOS[g]
    linha = {"grupo": g, "termos na lista": len(termos)}
    for s in SIGLAS:
        sel = ENQ[s]["base"][ENQ[s]["base"]["grupo"] == g]
        linha[f"{CURTO[s]}: bigramas da base"] = len(sel)
        linha[f"{CURTO[s]}: ocorrências"] = int(sel["frequência"].sum())
    linhas.append(linha)
CODEBOOK_RESUMO = pd.DataFrame(linhas).set_index("grupo")
CODEBOOK_RESUMO.loc["Total"] = CODEBOOK_RESUMO.sum()
display(tabela_br(CODEBOOK_RESUMO))

# 2. Cada termo da lista: grupo, par de lemas, posição na base de cada documento e ocorrências no documento inteiro
PAR_DO_TERMO = {}
for s in SIGLAS:
    for b, lemas in zip(ENQ[s]["base"]["bigrama"], ENQ[s]["base"]["lemas"]):
        if PAR_DO_TERMO.setdefault(b.lower(), lemas) != lemas:
            raise ValueError(f"O termo {b} corresponde a pares de lemas diferentes nos dois documentos.")
pos = {s: dict(zip(ENQ[s]["base"]["bigrama"].str.lower(), ENQ[s]["base"]["posição"])) for s in SIGLAS}
linhas = []
termos_e_grupos = [(t, g, "pesquisador") for g in [GRUPO_NEUTRO] + GRUPOS
                   for t in (TERMOS_NEUTROS if g == GRUPO_NEUTRO else CODEBOOK_ENQUADRAMENTOS[g])]
termos_e_grupos += [(t, g, "proposta do agente (a confirmar)") for t, g in CLASSIFICACAO_PROPOSTA.items()]
for termo, g, origem in termos_e_grupos:
    par = PAR_DO_TERMO.get(termo.lower())
    linha = {"termo (como na lista)": termo, "grupo": g, "classificação": origem, "par de lemas": par or "–",
             "na base de": " e ".join(CURTO[s] for s in SIGLAS if termo.lower() in pos[s]) or "nenhuma"}
    for s in SIGLAS:
        linha[f"{CURTO[s]}: posição na base"] = pos[s].get(termo.lower(), np.nan)
        linha[f"{CURTO[s]}: ocorrências no documento"] = int(FREQ_BIG[s].get(par, 0)) if par else 0
    linhas.append(linha)
CODEBOOK_TAB = pd.DataFrame(linhas).set_index("termo (como na lista)")
with pd.option_context("display.max_rows", None, "display.max_colwidth", None):
    display(tabela_br(CODEBOOK_TAB, {f"{CURTO[s]}: posição na base": 0 for s in SIGLAS}))
salvar_tabela(CODEBOOK_TAB, "codebook_enquadramentos")

# 3. Conferências
fora = CODEBOOK_TAB[CODEBOOK_TAB["na base de"] == "nenhuma"]
print("Conferências do codebook:")
for s in SIGLAS:
    print(f"- {ROTULO[s]}: base de {len(ENQ[s]['base'])} bigramas (pedido {N_BIGRAMAS_BASE_ENQUADRAMENTO}"
          + (f"; ampliado até o desempate na frequência {br(int(ENQ[s]['info']['limite']))}: {', '.join(ENQ[s]['info']['acrescentados'])}"
             if ENQ[s]["info"]["acrescentados"] else "; sem empate no ponto de corte")
          + f"; primeiro fora: {ENQ[s]['info']['primeiro_fora'][0]} ({br(int(ENQ[s]['info']['primeiro_fora'][1]))})). "
          f"Todos têm grupo: {len(ENQ[s]['neutros'])} neutros e {len(ENQ[s]['classificados'])} nos quatro enquadramentos.")
print(f"- Termos da lista do pesquisador: {len(TERMOS_NEUTROS) + sum(len(v) for v in CODEBOOK_ENQUADRAMENTOS.values())} "
      f"({len(TERMOS_NEUTROS)} neutros e {sum(len(v) for v in CODEBOOK_ENQUADRAMENTOS.values())} nos quatro enquadramentos); "
      "nenhum em dois grupos; "
      + (f"sem correspondência nas bases depois das correções de 2026-10-14: {', '.join(fora.index)}." if len(fora)
         else "todos estão na base de ao menos um documento."))
print("- Classificação proposta pelo agente, a confirmar pelo pesquisador: "
      + ("; ".join(f"{t} → {g}" for t, g in CLASSIFICACAO_PROPOSTA.items()) if CLASSIFICACAO_PROPOSTA else "nenhuma") + ".")
print("- União das bases: " + str(len(set(ENQ[SIG_A]['base']['bigrama'].str.lower()) | set(ENQ[SIG_B]['base']['bigrama'].str.lower())))
      + f" bigramas; nas duas bases: {len(set(ENQ[SIG_A]['base']['bigrama'].str.lower()) & set(ENQ[SIG_B]['base']['bigrama'].str.lower()))}.")
reun = {s: ENQ[s]["exibicao"][ENQ[s]["exibicao"]["n_reunidos"] > 1] for s in SIGLAS}
print("- Reunidos na exibição (decisão do pesquisador): "
      + ("; ".join(f"{ROTULO[s]}: “{r['rótulo']}” = {r['bigramas_reunidos']} ({br(int(r['frequência']))} ocorrências)"
                   for s in SIGLAS for r in reun[s].to_dict("records")) or "nenhum") + ".")

# Escalas comuns aos gráficos da EBIA e do PBIA (comparação visual direta entre as figuras)
LIMITES_ENQ = {"x": max(int(ENQ[s]["exibicao"]["frequência"].max()) for s in SIGLAS) * 1.1,
               "y": max(int(ENQ[s]["totais"].max()) for s in SIGLAS) * 1.2}
```

### Comparação dos enquadramentos: EBIA × PBIA (sem anexos)

Uma diferença só é afirmada quando o intervalo da diferença não inclui zero.

```python
# Comparação dos enquadramentos EBIA × PBIA (skill 03, seções 2.2, item 8, e 3.6): intervalos de confiança por
# bootstrap das unidades, para cada documento e para a diferença EBIA − PBIA. Uma diferença só é afirmada quando o
# intervalo da diferença não inclui zero.
PARES_BASE = {s: {g: ENQ[s]["classificados"].loc[ENQ[s]["classificados"]["grupo"] == g, "lemas"].tolist() for g in GRUPOS}
              for s in SIGLAS}
COMP_BASE_TAB, COMP_BASE = comparar_enquadramentos(PARES_BASE)
for s in SIGLAS:   # as ocorrências por enquadramento precisam ser as mesmas dos gráficos
    if not np.allclose(COMP_BASE[f"{s}_ocorrencias"].to_numpy(dtype=float), ENQ[s]["totais"].to_numpy(dtype=float)):
        raise RuntimeError("As ocorrências do bootstrap não coincidem com as dos gráficos dos enquadramentos.")
print("A. Como nos gráficos: a base própria de cada documento ("
      + "; ".join(f"{ROTULO[s]}: {len(ENQ[s]['classificados'])} bigramas não neutros dos {len(ENQ[s]['base'])}" for s in SIGLAS)
      + f"). Proporção no total dos quatro enquadramentos, com IC de {br_pct(NIVEL_CONFIANCA, 0)}:")
with pd.option_context("display.max_colwidth", None):
    display(COMP_BASE_TAB[[c for c in COMP_BASE_TAB.columns if "por 1.000" not in c]])

PARES_COMUNS = {g: sorted({PAR_DO_TERMO[t.lower()] for t in CODEBOOK_ENQUADRAMENTOS[g] + [k for k, v in CLASSIFICACAO_PROPOSTA.items() if v == g]
                             if t.lower() in PAR_DO_TERMO}) for g in GRUPOS}
COMP_COMUM_TAB, COMP_COMUM = comparar_enquadramentos({s: PARES_COMUNS for s in SIGLAS})
print(f"\nB. Robustez: o mesmo conjunto de termos nos dois documentos (os {sum(len(v) for v in PARES_COMUNS.values())} "
      "pares de lemas não neutros da lista do pesquisador e da classificação proposta, contados no documento inteiro, "
      f"estejam ou não na base de cada um). Proporção e taxa por 1.000 tokens, com IC de {br_pct(NIVEL_CONFIANCA, 0)}:")
with pd.option_context("display.max_colwidth", None):
    display(COMP_COMUM_TAB)
salvar_tabela(COMP_BASE_TAB, "enquadramentos_comparacao_base_propria")
salvar_tabela(COMP_COMUM_TAB, "enquadramentos_comparacao_termos_comuns")

print("Síntese factual (diferença EBIA − PBIA na proporção dos quatro enquadramentos, em pontos percentuais):")
for g in GRUPOS:
    a, b = COMP_BASE.loc[g], COMP_COMUM.loc[g]
    print(f"- {g}: base própria {br(100 * a['dif_prop'], 1)} [{br(100 * a['dif_prop_ic'][0], 1)}; "
          f"{br(100 * a['dif_prop_ic'][1], 1)}] → {a['afirmada_prop']}; termos comuns {br(100 * b['dif_prop'], 1)} "
          f"[{br(100 * b['dif_prop_ic'][0], 1)}; {br(100 * b['dif_prop_ic'][1], 1)}] → {b['afirmada_prop']}; taxa por "
          f"1.000 tokens com termos comuns: {br(b['dif_taxa'], 2)} [{br(b['dif_taxa_ic'][0], 2)}; {br(b['dif_taxa_ic'][1], 2)}] "
          f"→ {b['afirmada_taxa']}.")
```

#### Método e escolhas
- **Intervalos.** *Bootstrap* por unidades: as unidades de cada documento (352 na EBIA, 284 no PBIA) são reamostradas com reposição 2.000 vezes (`N_BOOTSTRAP`). A semente é própria de cada documento, derivada de `SEMENTE` = 2026 e da sigla. Em cada reamostragem, recalculam-se as ocorrências de cada enquadramento, a proporção no total dos quatro e a taxa por 1.000 tokens.
- **Diferença.** A diferença EBIA − PBIA usa reamostragens independentes dos dois documentos. Os intervalos são os percentis de 2,5% e 97,5% (`NIVEL_CONFIANCA` = 0,95).
- **A — como nos gráficos.** A base própria de cada documento: os 60 bigramas não neutros da EBIA e os 67 do PBIA. As ocorrências coincidem com as dos painéis da direita, o que a célula confere.
- **B — robustez.** O mesmo conjunto de termos nos dois documentos: os 111 pares de lemas não neutros do codebook (110 da lista do pesquisador e *big data*, de classificação provisória), contados no documento inteiro, estejam ou não na base de cada um. Assim, os documentos são comparados com os mesmos termos, e a taxa por 1.000 tokens permite comparar também a intensidade, e não só a composição.

#### Como ler
- "% dos quatro" é a parte de cada enquadramento no total dos quatro, em cada documento.
- "Por 1.000 tokens" é quantas ocorrências dos bigramas do enquadramento há a cada 1.000 palavras do documento.
- Entre colchetes, o intervalo de confiança de 95%.
- A última coluna de cada medida diz se a diferença é afirmada, isto é, se o intervalo da diferença não inclui zero.

#### Fragilidades e ajustes possíveis
- **Unidades vizinhas.** O *bootstrap* trata as unidades como independentes, mas unidades vizinhas (itens de uma mesma lista, parágrafos de uma mesma seção) tendem a tratar do mesmo tema. Os intervalos podem, por isso, ser mais estreitos do que deveriam; as diferenças de Estado, com limites inferiores de 0,7 e 0,0, devem ser lidas com cautela.
- **Seleção das bases.** O *bootstrap* não incorpora a escolha das bases, os bigramas mais frequentes de cada documento, feita com os mesmos dados. A tabela B reduz esse efeito ao usar o mesmo conjunto de termos nos dois.
- **Codebook.** Foi montado pelo pesquisador sobre a união das duas bases. Termos frequentes em um só documento entram na lista e pesam a favor dele; termos de um enquadramento que não estão entre os bigramas mais frequentes de nenhum dos dois ficam de fora. *Big data* entra em Infraestrutura por classificação provisória do agente (3 ocorrências na EBIA, nenhuma no PBIA); sem ele, a diferença em Infraestrutura seria um pouco maior.
- **Contagem dos bigramas.** As fragilidades das Visualizações 9 e 10 se aplicam: frequências baixas, sobreposição de bigramas em expressões longas e expressões com preposição não captadas.
