## 1. Metodologia

### 1.1 De onde vêm os dados
`nota_mre.json` (acervo bruto do MRE) → **Passo 1** (`analise_mre.json`: triagem das notas com relevância
diplomática Brasil-China) → **Passo 2** (`blocos_analise_mre.json`: separação em 8 blocos temáticos, um dos
quais é o bloco analisado aqui) → **Passo 3** (`relatorio_analise_conteudo_blocos.csv`: leitura de conteúdo
tema a tema — **é essa a fonte deste notebook**).

### 1.2 Como os temas foram definidos e reconhecidos no texto

Cada uma das 10 notas foi lida à luz de um roteiro fixo com **12 grandes temas** que costumam aparecer na
agenda diplomática internacional — por exemplo, *Tecnologia e Espaço*, *Comércio Exterior e Agronegócio*,
*Meio Ambiente e Mudança do Clima*, *Defesa e Segurança Internacional*, *Governança Global*,
*Multilateralismo* e assim por diante. Essa lista de temas não foi inventada para este bloco especificamente:
é a mesma taxonomia usada em toda a Skill AGY, o que permite comparar este bloco com os demais blocos do
corpus Brasil-China de forma consistente.

Para reconhecer, dentro do texto de cada nota, quando um desses temas está sendo tratado, associamos a cada
um deles um **vocabulário característico** — palavras e expressões que, na prática diplomática, costumam
sinalizar aquele assunto. Por exemplo: falar em *"energia renovável"*, *"hidrogênio verde"* ou *"COP"*
indica o tema Transição Energética/Clima; falar em *"BRICS"*, *"cúpula"* ou *"declaração conjunta"* indica
o tema Multilateralismo; falar em *"Conselho de Segurança"* ou *"reforma da ONU"* indica Governança Global.
Esse vocabulário foi levantado a partir da linguagem recorrente do próprio Itamaraty e da literatura de
relações internacionais, incluindo variações de gênero, número e grau de cada palavra (isto é, tanto
*"tecnológico"* quanto *"tecnológica"* ou *"tecnologias"* contam para o mesmo tema), e também equivalentes em
inglês, já que algumas notas — como as grandes Declarações do BRICS — possuem trechos ou anexos bilíngues.

Nenhum tema é atribuído por impressão geral do texto: a contagem é sempre rastreável a termos específicos
efetivamente encontrados, o que torna a classificação **auditável** — qualquer leitor pode conferir, nota a
nota, quais palavras levaram a cada percentual.

### 1.3 Como o "peso" de cada tema foi calculado

Nem toda menção a um tema tem a mesma importância dentro de uma nota. Um assunto que já aparece no
**título**, ou logo no **primeiro parágrafo** — onde o Itamaraty costuma resumir o propósito do documento —,
tende a ser o verdadeiro foco do texto. Já um assunto citado apenas de passagem, no meio de um parágrafo
longo entre dezenas de outros compromissos, é claramente secundário, ainda que mereça ser registrado.

Para refletir essa hierarquia editorial sem depender de leitura subjetiva, adotamos uma regra simples e
declarada: **menções no título valem o triplo, menções no primeiro parágrafo valem o dobro, e menções no
restante do texto valem o valor normal.** Somando essas contagens "com peso" para cada tema dentro de uma
nota, e comparando com a soma de todos os temas encontrados naquela mesma nota, obtém-se o **percentual de
presença de cada tema na nota** — os números que aparecem no mapa de calor (Gráfico 2) e nas barras
empilhadas (Gráfico 3) mais adiante.

O mesmo raciocínio, somado ao longo das 10 notas do bloco, gera o **percentual consolidado do bloco como um
todo** (Tabela 1 e Gráfico 1). Nessa consolidação, notas mais longas e mais densas — como as grandes
Declarações de Líderes do BRICS, que somam dezenas de parágrafos — naturalmente pesam mais do que um
comunicado curto de uma ou duas páginas. Isso é proposital, não uma distorção: um documento mais extenso
carrega, de fato, um volume maior de conteúdo diplomático, e deve influenciar mais o retrato consolidado do
bloco.

### 1.4 Onde traçamos a linha entre "presente" e "irrelevante"
Para não poluir os gráficos com dezenas de menções isoladas e pouco significativas, adotamos um corte de
materialidade: um tema só é exibido individualmente numa nota se responder por pelo menos **3% do seu peso
total**; no consolidado do bloco, o corte é **1,5%**. Tudo o que fica abaixo disso é somado numa categoria
única, **"Outros"**, para que o leitor saiba que aquele conteúdo existe, sem inflar a leitura com dezenas de
fatias minúsculas.

### 1.5 O que este método consegue — e o que não consegue — captar
- É um método de **leitura por vocabulário**, não uma análise semântica profunda: conta a presença de
  palavras e expressões associadas a cada tema, sem "entender" ironia, negação ou nuance de sentido da forma
  como um leitor humano entenderia.
- Notas curtas de anúncio têm pouco texto para "distribuir" entre temas, de modo que poucas palavras podem
  deslocar bastante o percentual — essas notas foram sinalizadas no relatório de origem com
  `revisão_necessária = Sim`.
- Todo o vocabulário utilizado e toda conta feita são documentados e reprodutíveis: a coluna
  `justificativa_metodologica` do arquivo CSV de origem permite a qualquer pesquisador auditar exatamente
  como cada percentual foi obtido.

**Fonte dos dados:** `relatorio_analise_conteudo_blocos.csv` (Passo 3, Skill AGY) e `blocos_analise_mre.json`
(Passo 2), na pasta `notebooks/Análise Agente AntiGravity/`.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import textwrap

pd.set_option("display.max_colwidth", 120)

# --------------------------------------------------------------------------
# Paleta e chrome (design system de referência da skill dataviz — validado
# para separação CVD; ordem categórica fixa, nunca ciclada)
# --------------------------------------------------------------------------
CATEGORICAL = {
    "blue":    "#2a78d6",
    "orange":  "#eb6834",
    "aqua":    "#1baf7a",
    "yellow":  "#eda100",
    "magenta": "#e87ba4",
    "green":   "#008300",
    "violet":  "#4a3aa7",
    "red":     "#e34948",
}
CAT_ORDER = list(CATEGORICAL.values())  # ordem fixa, 8 slots

SEQ_BLUE = ["#cde2fb", "#9ec5f4", "#6da7ec", "#3987e5", "#256abf", "#184f95", "#0d366b"]

SURFACE      = "#fcfcfb"
INK_PRIMARY  = "#0b0b0b"
INK_SECOND   = "#52514e"
INK_MUTED    = "#898781"
GRID         = "#e1e0d9"
BASELINE     = "#c3c2b7"

plt.rcParams.update({
    "figure.facecolor": SURFACE,
    "axes.facecolor": SURFACE,
    "savefig.facecolor": SURFACE,
    "font.family": "sans-serif",
    "text.color": INK_PRIMARY,
    "axes.edgecolor": BASELINE,
    "axes.labelcolor": INK_SECOND,
    "xtick.color": INK_MUTED,
    "ytick.color": INK_MUTED,
    "axes.titlesize": 13,
    "axes.titleweight": "bold",
    "axes.titlecolor": INK_PRIMARY,
    "font.size": 10.5,
})

NOME_BLOCO = "Cooperação Multilateral e Governança Global"
CSV_PATH = "relatorio_analise_conteudo_blocos.csv"
df_all = pd.read_csv(CSV_PATH)
df_raw = df_all[df_all["bloco_analisado"] == NOME_BLOCO].copy()

df_nota  = df_raw[df_raw["nivel_registro"] == "Nota"].copy()
df_bloco = df_raw[df_raw["nivel_registro"] == "Consolidado Bloco"].copy()

for d in (df_nota, df_bloco):
    d["porcentagem_valor"] = d["porcentagem_presenca"].str.rstrip("%").astype(float)

df_nota["data_dt"] = pd.to_datetime(df_nota["data_nota"], dayfirst=True)
df_nota = df_nota.sort_values("data_dt")

# Metadados unicos por nota (ordem cronologica) + rotulo curto p/ eixos
meta = (df_nota[["id_nota", "num_nota", "data_dt", "titulo_nota"]]
        .drop_duplicates()
        .sort_values("data_dt")
        .reset_index(drop=True))
meta["rotulo"] = meta.apply(
    lambda r: f"{r['data_dt'].strftime('%d/%m/%Y')} — " + textwrap.shorten(r["titulo_nota"], width=48, placeholder="…"),
    axis=1
)

print(f"{len(meta)} notas | período: {meta['data_dt'].min():%d/%m/%Y} a {meta['data_dt'].max():%d/%m/%Y}")
meta[["data_dt", "num_nota", "titulo_nota"]]
```

## 2. Tabela 1 — Distribuição Percentual Consolidada de Temas no Bloco

Percentual do peso lexical-semântico acumulado de cada tema sobre o total do bloco (10 notas, 2023–2025),
conforme metodologia da Seção 1.2–1.4. Barras internas (`Styler.bar`) auxiliam a leitura rápida de magnitude.


```python
tabela1 = (df_bloco[["topico_tematica", "porcentagem_valor", "vocabularios_chave_identificados"]]
           .sort_values("porcentagem_valor", ascending=False)
           .rename(columns={
               "topico_tematica": "Tema",
               "porcentagem_valor": "% do Bloco",
               "vocabularios_chave_identificados": "Principais termos computados (freq.)"
           })
           .reset_index(drop=True))
tabela1.index = tabela1.index + 1

(tabela1.style
    .bar(subset=["% do Bloco"], color=CATEGORICAL["blue"], vmin=0, vmax=tabela1["% do Bloco"].max())
    .format({"% do Bloco": "{:.1f}%"})
    .set_caption("Tabela 1. Composição temática consolidada do bloco 'Cooperação Multilateral e Governança "
                 "Global' (n = 10 notas, 2023–2025). Fonte: relatorio_analise_conteudo_blocos.csv (Passo 3, Skill AGY).")
    .set_table_styles([
        {"selector": "caption", "props": [("caption-side", "bottom"), ("font-size", "10px"),
                                           ("color", "#52514e"), ("text-align", "left"), ("padding-top", "6px")]},
        {"selector": "th", "props": [("text-align", "left"), ("font-size", "11px")]},
        {"selector": "td", "props": [("font-size", "11px")]},
    ])
)
```

## 3. Gráfico 1 — Panorama Temático do Bloco (visão consolidada)

Ranking de magnitude de um único indicador (percentual do peso lexical) por categoria: usa-se **um único
matiz sequencial** (não é identidade categórica — é ranking de uma medida), com rótulos diretos ao final de
cada barra, seguindo a regra "cor segue a função" do método.


```python
fig, ax = plt.subplots(figsize=(9, 6))

d = tabela1.sort_values("% do Bloco")
n = len(d)
shades = [SEQ_BLUE[int(round(i/(n-1)*(len(SEQ_BLUE)-1)))] for i in range(n)]

bars = ax.barh(d["Tema"], d["% do Bloco"], color=shades, height=0.62,
                edgecolor=SURFACE, linewidth=2)

for rect, val in zip(bars, d["% do Bloco"]):
    ax.text(rect.get_width() + 0.5, rect.get_y() + rect.get_height()/2, f"{val:.1f}%",
             va="center", ha="left", fontsize=9.5, color=INK_PRIMARY, fontweight="medium")

ax.set_xlim(0, d["% do Bloco"].max() * 1.18)
ax.set_xlabel("% do peso lexical-semântico consolidado do bloco")

ax.spines[["top", "right", "left"]].set_visible(False)
ax.spines["bottom"].set_color(BASELINE)
ax.xaxis.grid(True, color=GRID, linewidth=0.8, zorder=0)
ax.set_axisbelow(True)
ax.tick_params(left=False)
ax.xaxis.set_major_formatter(mticker.PercentFormatter(decimals=0))

plt.tight_layout(rect=[0, 0.03, 1, 0.88])
fig.suptitle("Composição Temática Consolidada", x=0.015, y=0.975, ha="left",
             fontsize=14.5, fontweight="bold", color=INK_PRIMARY)
fig.text(0.015, 0.905, "Bloco: Cooperação Multilateral e Governança Global  ·  n = 10 notas  ·  2023–2025",
          fontsize=9.5, color=INK_SECOND)
fig.text(0.015, 0.015, "Fonte: relatório Passo 3 (Skill AGY), a partir de relatorio_analise_conteudo_blocos.csv.",
          fontsize=8, color=INK_MUTED)

import os; os.makedirs("imagens", exist_ok=True); plt.savefig("imagens/fig1_panorama_tematico_multilateral.png", dpi=200, bbox_inches="tight")
plt.show()
```

**Leitura.** Diferentemente do bloco de Diplomacia Institucional/Visitas de Estado — onde Economia
liderava —, aqui o próprio **Multilateralismo (31,4%)** domina com folga, quase o dobro do segundo colocado.
Isso é coerente com a natureza do bloco: são declarações *do e sobre* o BRICS enquanto mecanismo, não apenas
documentos que mencionam o fórum de passagem. Logo depois vêm **Economia (11,9%)** e **Defesa/Segurança
Internacional/Geopolítica (11,8%)** — quase empatados —, refletindo o papel do BRICS como palco tanto de
pauta econômica (comércio em moedas locais, Novo Banco de Desenvolvimento) quanto de posicionamento
geopolítico conjunto (Ucrânia, Oriente Médio, reforma da arquitetura de segurança internacional).


## 4. Gráfico 2 — Estrutura Temática por Nota ao Longo do Tempo (2023–2025)

A visualização central deste notebook: uma matriz **nota × tema**, ordenada cronologicamente, em que a
intensidade da cor codifica o percentual de presença de cada tema dentro de cada nota. Um mapa de calor
evita o limite de oito matizes categóricas seguras para daltonismo (aqui seriam necessárias 12) ao usar
**um único matiz sequencial** — a identidade de linha/coluna vem do texto, não da cor.


```python
TEMA_ORDEM = tabela1["Tema"].tolist()  # ja ordenado por magnitude no bloco (desc)
TEMA_ORDEM = [t for t in TEMA_ORDEM if not t.startswith("Outros")] + \
             [t for t in TEMA_ORDEM if t.startswith("Outros")]

pivot = df_nota.pivot_table(index="id_nota", columns="topico_tematica",
                             values="porcentagem_valor", aggfunc="sum", fill_value=0.0)

id_order = meta["id_nota"].tolist()
pivot = pivot.reindex(index=id_order, columns=[t for t in TEMA_ORDEM if t in pivot.columns], fill_value=0.0)

TEMA_LABELS = {
    "Outros / Temas Residuais (abaixo do limiar de 3%)": "Outros (< 3%)",
}
col_labels = [TEMA_LABELS.get(c, c) for c in pivot.columns]
row_labels = meta.set_index("id_nota").loc[id_order, "rotulo"].tolist()

pivot.index = row_labels
pivot.columns = col_labels
pivot
```

```python
from matplotlib.colors import LinearSegmentedColormap

cmap = LinearSegmentedColormap.from_list("seq_blue", SEQ_BLUE, N=256)

fig, ax = plt.subplots(figsize=(12, 6.4))
mat = pivot.values

im = ax.imshow(mat, cmap=cmap, vmin=0, vmax=max(50, mat.max()), aspect="auto")

ax.set_xticks(range(pivot.shape[1]))
ax.set_xticklabels(pivot.columns, rotation=38, ha="right", fontsize=9)
ax.set_yticks(range(pivot.shape[0]))
ax.set_yticklabels(pivot.index, fontsize=9)

for i in range(mat.shape[0]):
    for j in range(mat.shape[1]):
        v = mat[i, j]
        if v <= 0.4:
            continue
        txt_color = INK_PRIMARY if v < (0.55 * max(50, mat.max())) else "#ffffff"
        ax.text(j, i, f"{v:.0f}", ha="center", va="center", fontsize=8, color=txt_color)

ax.set_xticks(np.arange(-0.5, pivot.shape[1], 1), minor=True)
ax.set_yticks(np.arange(-0.5, pivot.shape[0], 1), minor=True)
ax.grid(which="minor", color=SURFACE, linewidth=2)
ax.tick_params(which="minor", bottom=False, left=False)
ax.tick_params(which="major", bottom=False, left=False)
for spine in ax.spines.values():
    spine.set_visible(False)

cbar = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.02)
cbar.set_label("% de presença do tema na nota", color=INK_SECOND, fontsize=9.5)
cbar.ax.yaxis.set_tick_params(color=INK_MUTED, labelcolor=INK_MUTED)
cbar.outline.set_visible(False)

plt.tight_layout(rect=[0, 0.03, 1, 0.90])
fig.suptitle("Estrutura Temática das Notas, em Ordem Cronológica (2023–2025)", x=0.015, y=0.975,
             ha="left", fontsize=14.5, fontweight="bold", color=INK_PRIMARY)
fig.text(0.015, 0.925, "Células = % do peso lexical-semântico do tema dentro de cada nota  ·  linhas ordenadas por data",
          fontsize=9.5, color=INK_SECOND)
fig.text(0.015, 0.015, "Fonte: relatorio_analise_conteudo_blocos.csv (Passo 3, Skill AGY). Células em branco = tema abaixo do limiar de 3% na nota.",
          fontsize=8, color=INK_MUTED)

import os; os.makedirs("imagens", exist_ok=True); plt.savefig("imagens/fig2_heatmap_notas_multilateral.png", dpi=200, bbox_inches="tight")
plt.show()
```

**Leitura.** O mapa de calor mostra uma estrutura bem mais **homogênea** do que a do bloco de Diplomacia
Institucional: quase todas as notas exibem Multilateralismo como tema dominante — esperado, já que são, por
definição, declarações do próprio BRICS. As diferenças relevantes aparecem nos temas *secundários*: as duas
grandes Declarações de Líderes (Joanesburgo 2023, Kazan 2024, Rio 2025) equilibram Economia, Segurança e
Meio Ambiente numa combinação ampla — verdadeiros documentos-síntese da agenda do agrupamento —, enquanto o
**Discurso do Ministro Mauro Vieira** (28/04/2025) e as **Declarações de Chanceleres** concentram-se muito
mais fortemente em Defesa/Segurança/Geopolítica, refletindo o tom da diplomacia de crise (Ucrânia, Oriente
Médio) que costuma marcar encontros ministeriais. Já a nota sobre a **escolha do Rio de Janeiro como sede da
Cúpula** e a **Primeira Reunião de Sherpas** são essencialmente protocolares, quase inteiramente concentradas
em Multilateralismo.


## 5. Gráfico 3 — Composição Percentual por Nota (barras empilhadas a 100%)

Mesma matriz do Gráfico 2, agora como composição — mais direta para comparar o "mix" temático entre notas.
Por segurança de contraste ao daltonismo, mantêm-se os **7 temas de maior peso no bloco** como séries
categóricas próprias (ordem fixa de matizes) e os 5 temas restantes são agregados em **"Outros"** (8º
slot), conforme a regra do método ("a 9ª série nunca é um matiz gerado — ela se funde em 'Outros'").


```python
TOP_N = 7
top_temas = [t for t in tabela1["Tema"] if not t.startswith("Outros")][:TOP_N]
outros_cols = [c for c in pivot.columns if c not in top_temas and c != "Outros (< 3%)"]

comp = pivot[top_temas].copy()
comp["Outros"] = pivot[outros_cols].sum(axis=1) + pivot.get("Outros (< 3%)", 0)

color_map = dict(zip(top_temas + ["Outros"], CAT_ORDER[:TOP_N] + [INK_MUTED]))

fig, ax = plt.subplots(figsize=(11.5, 7.2))
left = np.zeros(len(comp))
y_pos = np.arange(len(comp))

for tema in comp.columns:
    vals = comp[tema].values
    bars = ax.barh(y_pos, vals, left=left, height=0.66, color=color_map[tema],
                    edgecolor=SURFACE, linewidth=1.4, label=tema)
    for yi, (v, l) in enumerate(zip(vals, left)):
        if v >= 9:
            ax.text(l + v/2, yi, f"{v:.0f}%", ha="center", va="center", fontsize=8, color="#ffffff")
    left += vals

ax.set_yticks(y_pos)
ax.set_yticklabels(comp.index, fontsize=9)
ax.invert_yaxis()
ax.set_xlim(0, 100)
ax.xaxis.set_major_formatter(mticker.PercentFormatter(decimals=0))
ax.set_xlabel("% da composição temática da nota")

ax.spines[["top", "right", "left"]].set_visible(False)
ax.spines["bottom"].set_color(BASELINE)
ax.xaxis.grid(True, color=GRID, linewidth=0.8, zorder=0)
ax.set_axisbelow(True)
ax.tick_params(left=False)

ax.legend(loc="upper center", bbox_to_anchor=(0.5, 1.04), ncol=4, frameon=False,
          fontsize=8.3, handlelength=1.1, handleheight=1.1, columnspacing=1.1)

plt.tight_layout(rect=[0, 0.03, 1, 0.80])
fig.suptitle("Composição Temática por Nota — Ordem Cronológica", x=0.015, y=0.975,
             ha="left", fontsize=14.5, fontweight="bold", color=INK_PRIMARY)
fig.text(0.015, 0.92, "Top 7 temas do bloco como séries próprias + 'Outros' (5 temas residuais agregados)",
          fontsize=9.5, color=INK_SECOND)
fig.text(0.015, 0.015, "Fonte: relatorio_analise_conteudo_blocos.csv (Passo 3, Skill AGY).", fontsize=8, color=INK_MUTED)

import os; os.makedirs("imagens", exist_ok=True); plt.savefig("imagens/fig3_composicao_percentual_multilateral.png", dpi=200, bbox_inches="tight")
plt.show()
```

**Leitura.** A composição confirma o padrão: Multilateralismo (azul) nunca desaparece, mas sua fatia
varia de quase 100% (notas de agenda/anúncio) a cerca de um quarto do texto (grandes Declarações de Líderes,
que reservam bastante espaço para Economia, Defesa/Segurança e Meio Ambiente). A faixa cinza ("Outros")
cresce nas notas mais longas — sinal de que, quanto mais extenso o documento, mais ele se ramifica para
temas técnicos específicos (tecnologia, saúde, cultura) que, isoladamente, não atingem os 3% de materialidade,
mas cuja soma é notável.


## 6. Gráfico 4 — Densidade Lexical por Nota (contexto para leitura dos percentuais)

Os gráficos anteriores normalizam cada nota a 100%, o que **oculta o volume absoluto de conteúdo**. Uma
nota de anúncio protocolar e uma Declaração de Líderes do BRICS podem exibir percentuais parecidos num
tema — mas com peso substantivo radicalmente diferente. Este gráfico mostra o **peso lexical-semântico
total** de cada nota, em escala logarítmica dada a amplitude extrema do corpus (de 7 a 1.737 — mais de 200
vezes de variação).


```python
pesos_absolutos = {
    516: 1737, 612: 7, 630: 713, 634: 44, 721: 15,
    741: 17, 939: 1525, 1675: 176, 1719: 594, 1818: 253,
}
peso_s = pd.Series(pesos_absolutos, name="peso_total").reindex(id_order)
peso_s.index = row_labels

fig, ax = plt.subplots(figsize=(10, 6.2))
bars = ax.barh(peso_s.index, peso_s.values, height=0.6, color=CATEGORICAL["blue"], edgecolor=SURFACE, linewidth=1.5)
ax.invert_yaxis()
ax.set_xscale("log")
ax.set_xlabel("Peso lexical-semântico total da nota — escala logarítmica")

for rect, val in zip(bars, peso_s.values):
    ax.text(rect.get_width() * 1.08, rect.get_y() + rect.get_height()/2, f"{val:.0f}",
             va="center", ha="left", fontsize=9, color=INK_PRIMARY)

ax.spines[["top", "right", "left"]].set_visible(False)
ax.spines["bottom"].set_color(BASELINE)
ax.xaxis.grid(True, which="major", color=GRID, linewidth=0.8, zorder=0)
ax.set_axisbelow(True)
ax.tick_params(left=False)

plt.tight_layout(rect=[0, 0.03, 1, 0.88])
fig.suptitle("Densidade Lexical-Semântica por Nota", x=0.015, y=0.975, ha="left",
             fontsize=14.5, fontweight="bold", color=INK_PRIMARY)
fig.text(0.015, 0.905, "Amplitude de 7 (anúncio de sede) a 1.737 (Declaração de Líderes, Rio 2025) — mais de 200x de variação",
          fontsize=9.5, color=INK_SECOND)
fig.text(0.015, 0.015, "Fonte: relatorio_analise_conteudo_blocos.csv (Passo 3, Skill AGY).", fontsize=8, color=INK_MUTED)

import os; os.makedirs("imagens", exist_ok=True); plt.savefig("imagens/fig4_densidade_lexical_multilateral.png", dpi=200, bbox_inches="tight")
plt.show()
```
