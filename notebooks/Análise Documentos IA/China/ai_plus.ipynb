# Análise do AI+ — Ação Nacional "Inteligência Artificial+" da China

Este notebook integra uma análise acadêmica de rigor, desenvolvida no âmbito de uma pesquisa em Relações Internacionais dedicada ao estudo comparado das estratégias nacionais de inteligência artificial. Seu objeto específico é o documento que estabelece as diretrizes sobre o aprofundamento da implementação da ação "IA+" (人工智能+), publicado pelo Conselho de Estado da China, que define a estratégia de integração da inteligência artificial aos diversos setores da economia e da sociedade chinesas.

A análise aqui conduzida busca compreender, de modo simultaneamente **quantitativo e qualitativo**, a linguagem empregada pelo documento — os termos, categorias, ênfases retóricas, valores e prioridades estratégicas que estruturam o texto —, de modo a identificar os marcos conceituais e os campos semânticos por meio dos quais a China formula sua política de inteligência artificial.

A partir dessa leitura, pretende-se **posicionar o documento internacionalmente**, comparando a linguagem e as escolhas discursivas de a China com o vocabulário e as ênfases adotados pelos demais países e blocos contemplados neste projeto (Brasil, China, Estados Unidos, Europa e Índia), de modo a mapear convergências, divergências e posicionamentos estratégicos distintivos no debate internacional sobre desenvolvimento e governança de inteligência artificial.

As análises e visualizações produzidas neste notebook seguem as diretrizes metodológicas da **Skill02** (Análise e Visualização Gráfica de Documentos): baseiam-se exclusivamente no campo `texto_completo` do JSON de extração correspondente a este documento, produzido na etapa anterior (Skill01), adotam rigor acadêmico integral na leitura do texto-fonte, evitam generalizações, simplificações e inferências não fundamentadas no texto original, e cada visualização construída é acompanhada de sua respectiva descrição, leitura, interpretação e eventuais limitações metodológicas.


## Análise de Vocabulário — Termos mais Frequentes

Esta seção aplica a **Skill02** (Análise e Visualização Gráfica de Documentos) em conjunto com seu complemento especializado, a **Skill 02_Análise_Vocab_A** (Análise de Vocabulário e Termos), ao JSON `ai_plus.json`, produzido na etapa anterior (Skill01).

**Protocolo de uso do JSON.** Conforme exigido pela Skill02, apenas os campos `titulo`, `pais_ou_bloco` e `texto_completo` são utilizados nesta análise. Os campos `elementos_descartados`, `data_extracao`, `data_publicacao` e `fonte` não entram, em nenhuma hipótese, na contagem ou na interpretação do vocabulário.

**Idioma identificado.** O `texto_completo` está integralmente redigido em **inglês** — trata-se da tradução oficial do CSET do documento chinês original. Conforme alerta a própria Skill 02_Análise_Vocab_A, o idioma do documento não corresponde necessariamente ao país de origem, e as listas de palavras funcionais e as regras de normalização morfológica aplicadas abaixo correspondem ao inglês, não ao chinês nem ao português.

**Etapas metodológicas aplicadas** (ver célula de código a seguir para a implementação exata):

1. **Extração bruta** de todos os termos de `texto_completo`, sem qualquer corte prévio de conteúdo.
2. **Pré-processamento do texto corrido, antes da tokenização**, fundamentado em identidade de referente e em expressões compostas com significado próprio (não em conveniência estatística):
   - O nome próprio da iniciativa — grafado ora por extenso, `"Artificial Intelligence+"`/`"Artificial intelligence+"` (2 ocorrências: o título do documento e a definição da sigla logo na abertura), ora abreviado, `"AI+"` (9 ocorrências, incluindo a própria definição da sigla e os seis subtítulos estruturais "(1) 'AI+' Science and Technology" a "(6) 'AI+' Global Cooperation") — foi unificado em um único rótulo `AI+` (11 ocorrências no total) e mantido **deliberadamente separado** do termo genérico `AI` (81 ocorrências). Justificativa: `AI+` designa o nome próprio da política/iniciativa em si, enquanto `AI` genérico designa a tecnologia como campo — referentes distintos apesar da sobreposição lexical, na mesma lógica do par "United States"/"state(s)" já tratado como não-fundível na análise análoga dos EUA neste projeto.
   - A expressão composta `"open-source"` (9 ocorrências, sempre grafada com hífen) foi tratada como uma única unidade lexical, conforme a própria Skill 02_Análise_Vocab_A determina para expressões técnicas de mais de uma palavra "quando é isso que o termo designa": fragmentá-la em "open" + "source" diluiria um dos temas centrais do documento — explicitamente destacado no resumo da tradução CSET ("the value of a flourishing open-source AI ecosystem") e correspondente à seção autônoma "(11) Promote a thriving open-source ecosystem" — em dois termos genéricos que, isolados, perdem esse sentido específico.
3. **Tokenização**: sequências de letras (`[A-Za-z]+`) foram extraídas como tokens; qualquer caractere não alfabético (espaço, hífen, apóstrofo, ponto, parênteses, barra, "&", dígitos) foi tratado como separador. Demais compostos com hífen além dos dois casos protegidos no passo 2 (ex.: *AI-driven*, *AI-native*, *human-computer*, *high-quality*, *decision-making*, *scenario-based*) foram, por consequência, fragmentados em suas palavras componentes — regra uniforme, sem exceções adicionais, para não introduzir uma lista extensa e pouco auditável de "quais outros hifens preservar" (discutido como limitação ao final).
4. **Remoção (exclusão)** de:
   - Palavras funcionais do inglês (artigos, preposições — incluindo partículas fraseológicas como *up*/*out* —, conjunções, pronomes, verbos auxiliares/modais e conectores discursivos genéricos como *also*, *including*, *etc.*) — lista declarada integralmente na célula de código e no registro persistente;
   - Tokens de um único caractere, resíduos não semânticos da tokenização de abreviações com "&" (ex.: "R&D" → fragmentos soltos "r"/"d"; "S&T" → "s"/"t"), de compostos letra-dígito (ex.: "6G" → fragmento solto "g"), de possessivos (ex.: "China's", "AI's" → fragmento solto "s") e de numerais isolados entre aspas (ex.: "from 1 to N" → fragmento solto "n");
   - Os numerais romanos estruturais dos quatro cabeçalhos de seção do documento ("I.", "II.", "III.", "IV." → tokens "i"/"ii"/"iii"/"iv") — marcadores editoriais de organização do texto, mesma natureza dos cabeçalhos/numeração de página já descartados na Skill01, não vocabulário empregado organicamente.
5. **Normalização/agrupamento** de variantes que designam exatamente o mesmo referente: 22 pares singular/plural de substantivos (ex.: *application*/*applications*, *model*/*models*) e 15 famílias de conjugações verbais (ex.: *promote*/*promotes*/*promoting*/*promoted*). Pares com grafia ou sentido próximos, porém classe gramatical ou referente distintos, foram deliberadamente **mantidos separados** — por exemplo: *intelligent* (adjetivo) / *intelligentization* (substantivo) / *intelligentized* (particípio/adjetivo) / *smart* (adjetivo, sinônimo não-cognato), quatro itens lexicais distintos do mesmo campo semântico; *driven* e *integrated*, sempre empregados como adjetivos compostos ("AI-driven", "an integrated system") e mantidos fora da fusão com os verbos "drive" e "integrate"; *development*/*develop*, *innovation*/*innovative*/*innovate* e *implementation*/*implement*, pares derivacionais (substantivo/adjetivo formados a partir do verbo, não flexões do mesmo lema); *capacity* mantido separado de *capability*/*capabilities* (quase-sinônimos, mas lemas distintos); *form(s)* mantido separado de *format(s)* (grafia parecida, lemas distintos); e *compute* mantido separado de *computing*, apesar de o próprio texto os aproximar ("computing power ('compute')") — caso de dúvida real, resolvido em favor da não-fusão, conforme o critério conservador exigido pela Skill.
6. **Critério de corte declarado:** os **25 termos de maior frequência absoluta** após as etapas 1–5, em ordem decrescente. Frequência absoluta é adequada aqui porque esta é uma análise de um único documento — a normalização relativa entre documentos de tamanhos distintos, exigida pela Skill02, aplica-se apenas à pasta "Análise Conjunta".

O registro completo, persistente e auditável de todas as remoções e agrupamentos está salvo em [`registro_vocabulario_ai_plus.md`](./registro_vocabulario_ai_plus.md), nesta mesma pasta, e deve ser consultado — e atualizado de forma cumulativa — antes de qualquer nova visualização de vocabulário sobre este mesmo documento.


```python
import json, re
from collections import Counter
import pandas as pd

with open("ai_plus.json", encoding="utf-8") as f:
    doc = json.load(f)

titulo, pais, texto = doc["titulo"], doc["pais_ou_bloco"], doc["texto_completo"]
print(f"Documento: {titulo[:60]}... | País/bloco: {pais} | Caracteres em texto_completo: {len(texto)}")

STOPWORDS_EN = set("""
a an the
and or but nor so yet if because while although that which who whom whose when where how than whether as
this these those it its they their them theirs he she his her him himself herself itself themselves
we our ours us you your yours i my mine
is are was were be been being am
has have had having
do does did doing
will would shall should can could may might must
to of in on at by for with from into through across throughout under over without within among between via per about upon toward towards up out
not no nor
also more most many much very only further therefore thus however moreover
there here
such other others any all both each every some
including include includes included
etc eg ie
""".split())

STRUCTURAL_RESIDUES = {"ii", "iii", "iv"}  # numerais romanos dos cabeçalhos de seção (I., II., III., IV.)

def preprocess(text):
    # nome próprio da iniciativa: unifica forma por extenso e abreviada, protegendo-o da fusão com "AI" genérico
    text = re.sub(r'Artificial [Ii]ntelligence\+', ' AIPLUSTOKEN ', text)
    text = text.replace('AI+', ' AIPLUSTOKEN ')
    # expressão composta com significado próprio (Skill 02_Análise_Vocab_A, item 3)
    text = re.sub(r'\bopen-source\b', ' OPENSOURCETOKEN ', text, flags=re.I)
    return text

texto_tratado = preprocess(texto)
tokens_brutos = re.findall(r"[A-Za-z]+", texto_tratado)
tokens_lower = [t.lower() for t in tokens_brutos]

tokens_filtrados = [
    t for t in tokens_lower
    if len(t) > 1 and t not in STOPWORDS_EN and t not in STRUCTURAL_RESIDUES
]

NOUN_MERGES = {
    "application": ["application", "applications"], "models": ["model", "models"],
    "services": ["service", "services"], "systems": ["system", "systems"],
    "capabilities": ["capability", "capabilities"], "industry": ["industry", "industries"],
    "products": ["product", "products"], "resources": ["resource", "resources"],
    "technology": ["technology", "technologies"], "talent": ["talent", "talents"],
    "risks": ["risk", "risks"], "levels": ["level", "levels"], "agents": ["agent", "agents"],
    "initiatives": ["initiative", "initiatives"], "sectors": ["sector", "sectors"],
    "structures": ["structure", "structures"], "processes": ["process", "processes"],
    "policies": ["policy", "policies"], "ecosystems": ["ecosystem", "ecosystems"],
    "evaluations": ["evaluation", "evaluations"], "platforms": ["platform", "platforms"],
    "approach": ["approach", "approaches"],
}
VERB_MERGES = {
    "promote": ["promote", "promotes", "promoting", "promoted"],
    "strengthen": ["strengthen", "strengthens", "strengthening", "strengthened"],
    "accelerate": ["accelerate", "accelerates", "accelerating", "accelerated"],
    "support": ["support", "supports", "supporting", "supported"],
    "build": ["build", "builds", "building", "built"],
    "develop": ["develop", "develops", "developing", "developed"],
    "create": ["create", "creates", "creating", "created"],
    "establish": ["establish", "establishes", "establishing", "established"],
    "improve": ["improve", "improves", "improving", "improved"],
    "enhance": ["enhance", "enhances", "enhancing", "enhanced"],
    "explore": ["explore", "explores", "exploring", "explored"],
    "encourage": ["encourage", "encourages", "encouraging", "encouraged"],
    "deepen": ["deepen", "deepens", "deepening", "deepened"],
    "optimize": ["optimize", "optimizes", "optimizing", "optimized"],
    "implement": ["implement", "implements", "implementing", "implemented"],
}

canon = {}
for label, variants in NOUN_MERGES.items():
    for v in variants:
        canon[v] = label
for label, variants in VERB_MERGES.items():
    for v in variants:
        canon[v] = label
canon["aiplustoken"] = "AI+"
canon["opensourcetoken"] = "open-source"
canon["ai"] = "AI"

tokens_finais = [canon.get(t, t) for t in tokens_filtrados]
freq = Counter(tokens_finais)

print(f"Tokens brutos: {len(tokens_brutos)} | após remoção de funcionais/resíduos: {len(tokens_filtrados)} | termos distintos no vocabulário final: {len(freq)}")

CORTE = 25
top_termos = freq.most_common(CORTE)

tabela_top = pd.DataFrame(top_termos, columns=["termo", "frequência"])
tabela_top.index = tabela_top.index + 1
tabela_top
```

```python
import matplotlib.pyplot as plt

termos_plot = [t for t, _ in top_termos][::-1]
freqs_plot = [n for _, n in top_termos][::-1]

fig, ax = plt.subplots(figsize=(9, 10))
barras = ax.barh(termos_plot, freqs_plot, color="#a11d21", height=0.68)

ax.set_xlabel("Frequência absoluta (nº de ocorrências em texto_completo)", fontsize=11)
ax.set_title(
    "Termos mais frequentes — Opinions on Deepening the \"AI+\" Initiative (China, 2025)\n"
    f"Top {CORTE} termos após remoção de palavras funcionais e normalização de variantes",
    fontsize=12.5, fontweight="bold", loc="left", pad=14
)
ax.spines[["top", "right"]].set_visible(False)
ax.tick_params(axis="y", labelsize=10.5)
ax.tick_params(axis="x", labelsize=9.5)
ax.set_axisbelow(True)
ax.xaxis.grid(True, color="#d9d9d9", linewidth=0.8)

for bar, val in zip(barras, freqs_plot):
    ax.text(bar.get_width() + 0.7, bar.get_y() + bar.get_height() / 2, str(val),
            va="center", ha="left", fontsize=9.5, color="#333333")

ax.set_xlim(0, max(freqs_plot) * 1.12)
fig.text(0.02, -0.01,
         "Fonte: texto_completo de ai_plus.json (Skill01), tratado segundo a Skill02 / Skill 02_Análise_Vocab_A.",
         fontsize=8, color="#666666")
fig.tight_layout()
plt.show()
```

## Leitura, Interpretação e Limitações do Gráfico

**O que foi construído.** Um gráfico de barras horizontais com os 25 termos de maior frequência absoluta em `texto_completo`, após remoção de palavras funcionais/resíduos e normalização das variantes descritas acima. Cada barra representa a contagem final (já consolidada) de um termo ou família de termos equivalentes.

**Como ler o gráfico.** O eixo vertical lista os termos, em ordem decrescente de frequência (o termo mais frequente no topo); o eixo horizontal mede o número absoluto de ocorrências no documento. O rótulo numérico ao final de cada barra é a contagem exata.

**Interpretação à luz do documento.**
- **"AI" (81 ocorrências)** domina o vocabulário por larga margem — mais do dobro do segundo termo mais comum —, o que é esperado e consistente com o objeto do documento: uma diretriz de Estado inteiramente dedicada à integração da IA em praticamente todos os setores da economia e da sociedade chinesas. Como no caso de outros planos nacionais já analisados neste projeto, esse resultado isolado tem baixo valor discriminante; o valor analítico do gráfico está principalmente nos termos que disputam a segunda posição em diante.
- **"AI+" (11 ocorrências)**, mantido deliberadamente separado de "AI" (ver metodologia), aparece por volta da 22ª posição — uma frequência modesta em termos absolutos, mas analiticamente relevante: suas 11 ocorrências não se distribuem organicamente pelo texto corrido, e sim concentram-se quase inteiramente em usos **estruturais** — a definição da sigla na abertura do documento e, sobretudo, como prefixo fixo dos seis subtítulos da Parte II ("(1) 'AI+' Science and Technology" a "(6) 'AI+' Global Cooperation"). Isso revela que o nome da própria iniciativa funciona, no texto, mais como um rótulo organizacional recorrente do que como vocabulário discursivo livre — achado equivalente, em natureza, ao papel estrutural de "lead/led" identificado na análise já produzida para o documento dos EUA.
- **"new" (41)**, em segundo lugar com folga, reflete uma marca retórica quase onipresente de renovação/transformação: "new infrastructure", "new technological systems", "new industrial ecosystems", "new job openings", "new quality productive forces" (新质生产力), "new generation smart terminal devices", "new pattern", "new business formats", entre muitos outros — o documento se apresenta, lexicalmente, como um projeto de ruptura com o estado de coisas anterior.
- **"development" (37)**, mantido separado do verbo "develop" (variação derivacional, não flexional — ver metodologia), confirma a centralidade da noção de "desenvolvimento" como categoria organizadora do texto (ex.: "new concept of development", "high-quality development", "intelligentization-based development").
- **"promote" (35), "strengthen" (29) e "accelerate" (23)** formam o núcleo da "gramática de ação" do documento: são os três verbos diretivos que mais frequentemente abrem os itens numerados de cada subseção ("Promote a shift...", "Strengthen the coordination...", "Accelerate the exploration..."), reproduzindo a estrutura formulaica típica de um documento normativo de política de Estado — análoga, em função, aos verbos "support/establish" mais recorrentes no plano de ação dos EUA já analisado.
- **"application" (33), "models" (26), "services" (26), "industry" (20), "systems" (18), "technology" (15), "capabilities" (15), "data" (13), "research" (12) e "production" (11)** compõem o vocabulário técnico-operacional esperado de uma política industrial de IA, em linha com o campo semântico já observado em outros documentos deste projeto (EUA, Europa).
- **"governance" (15)**, empatado com "technology"/"smart"/"build"/"capabilities", ocupa uma posição relativamente alta para um termo que, no documento, também nomeia uma única seção dedicada ("(5) 'AI+' Governance Capacity"). Isso mostra que o vocabulário de governança **extravasa** essa seção específica e recorre ao longo do texto — "global AI governance system", "AI-enabled cyberspace governance", "ecological governance", "a pattern of AI governance that is dynamic, agile, diverse, and collaborative" —, um indício, relevante para uma leitura de Relações Internacionais, de que a governança é tratada como preocupação transversal, e não como tópico isolado.
- **"smart" (15)** e **"intelligent" (32)**, mantidos deliberadamente separados por serem itens lexicais distintos (ver metodologia), evidenciam juntos — ainda que não somados no gráfico — o quanto compostos com "smart"/"intelligent" permeiam o texto para descrever o estado-alvo da política ("smart society", "smart economy", "intelligent agents", "intelligent robots", "intelligent compute").
- **"based" (11)**, na 24ª posição, é majoritariamente um artefato de tokenização (ver limitações): resulta da fragmentação de compostos como "AI-based", "platform-based", "scenario-based" e "intelligentization-based" em torno de um hífen não protegido, e não de um uso lexical unificado do termo "based" em si.
- **"open-source" (9 ocorrências)**, embora não alcance o corte dos 25 primeiros termos, é o compound tratado explicitamente como unidade lexical na metodologia (ver seção acima) precisamente porque, fragmentado, se dissolveria em "open" e "source" e obscureceria um tema que o próprio resumo da tradução CSET aponta como central ao documento — vale destacá-lo aqui ainda que fora do recorte visual do gráfico.

**Ajustes possíveis para aprimorar a visualização.**

- Segmentar a contagem pelas quatro partes de nível superior do documento (I. Overall Requirements; II. Accelerating Implementation of Key Initiatives — com suas seis subseções "AI+"; III. Strengthening Basic Support Capabilities; IV. Organizing Implementation), para verificar se vocabulário como "governance", "compute" ou "data" se concentra em partes específicas — não realizado aqui porque o pedido do usuário foi por uma visão geral do documento como um todo.
- Reportar, ao lado da frequência absoluta, a frequência relativa (percentual do total de tokens tratados), especialmente relevante neste caso porque este documento (3.025 tokens tratados) é sensivelmente mais curto que outros já analisados no projeto — suas contagens absolutas não devem ser comparadas diretamente às de outros países fora da pasta "Análise Conjunta".
- Destacar visualmente "open-source" (9 ocorrências, ver acima) em um painel complementar, já que sua relevância temática, sinalizada pela própria fonte, não se reflete plenamente em sua posição no corte de frequência absoluta.

**Limitações e fraquezas metodológicas.**
- **Polissemia não desambiguada:** por exemplo, "production" agrega tanto o registro econômico-teórico de "productive forces" (生产力) quanto usos industriais/agrícolas mais concretos ("design, pilot testing, production, service, and operation") — mesmo token, registros discursivos distintos, fundidos silenciosamente pela contagem lexical.
- **Fragmentação de compostos com hífen além do único caso corrigido ("open-source"):** expressões como "AI-driven" (5), "AI-native" (5), "human-computer" (5), "high-quality" (5) e "decision-making" (3) permanecem fragmentadas em suas palavras componentes pela regra geral de tokenização (passo 3), diluindo-se em termos genéricos já presentes no vocabulário (ex.: "driven" mantido fora do vocabulário final por ser sempre adjetival, "quality" e "computer" parcialmente inflados por essas fragmentações). Estender a esses casos o mesmo tratamento de unidade composta dado a "open-source" foi considerado, mas não aplicado, para não introduzir uma lista extensa e pouco auditável de exceções — uma decisão de escopo conservadora, mas que constitui uma limitação real, documentada aqui.
- **Cobertura da normalização morfológica:** os pares e famílias foram verificados sistematicamente apenas entre os termos com frequência bruta ≥ 2 no vocabulário já filtrado (faixa plausível de disputar o corte de 25) — não houve lematização exaustiva de toda a cauda longa de 782 termos distintos do vocabulário final. Adequado para o gráfico produzido, mas não garantidamente reprodutível se o corte fosse ampliado, por exemplo, para os 100 termos mais frequentes.
- **Distinção "AI"/"AI+" dependente de regra manual:** a separação entre o termo genérico e o nome próprio da iniciativa exigiu uma regra de pré-processamento específica a este documento (proteção de substrings exatas antes da tokenização), e não um princípio geral de desambiguação reutilizável automaticamente em outros documentos sem inspeção manual equivalente.

A metodologia completa — incluindo a lista integral de palavras removidas, os 22 agrupamentos de substantivos e as 15 famílias verbais normalizadas, e os pares deliberadamente não agrupados — está documentada e versionada em [`registro_vocabulario_ai_plus.md`](./registro_vocabulario_ai_plus.md).
