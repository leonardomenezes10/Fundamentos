# Análise do AI+ — Ação Nacional "Inteligência Artificial+" da China

Este notebook integra uma análise acadêmica de rigor, desenvolvida no âmbito de uma pesquisa em Relações Internacionais dedicada ao estudo comparado das estratégias nacionais de inteligência artificial. Seu objeto específico é o documento que estabelece as diretrizes sobre o aprofundamento da implementação da ação "IA+" (人工智能+), publicado pelo Conselho de Estado da China, que define a estratégia de integração da inteligência artificial aos diversos setores da economia e da sociedade chinesas.

A análise aqui conduzida busca compreender, de modo simultaneamente **quantitativo e qualitativo**, a linguagem empregada pelo documento — os termos, categorias, ênfases retóricas, valores e prioridades estratégicas que estruturam o texto —, de modo a identificar os marcos conceituais e os campos semânticos por meio dos quais a China formula sua política de inteligência artificial.

A partir dessa leitura, pretende-se **posicionar o documento internacionalmente**, comparando a linguagem e as escolhas discursivas de a China com o vocabulário e as ênfases adotados pelos demais países e blocos contemplados neste projeto (Brasil, China, Estados Unidos, Europa e Índia), de modo a mapear convergências, divergências e posicionamentos estratégicos distintivos no debate internacional sobre desenvolvimento e governança de inteligência artificial.

As análises e visualizações produzidas neste notebook seguem as diretrizes metodológicas da **Skill02** (Análise e Visualização Gráfica de Documentos): baseiam-se exclusivamente no campo `texto_completo` do JSON de extração correspondente a este documento, produzido na etapa anterior (Skill01), adotam rigor acadêmico integral na leitura do texto-fonte, evitam generalizações, simplificações e inferências não fundamentadas no texto original, e cada visualização construída é acompanhada de sua respectiva descrição, leitura, interpretação e eventuais limitações metodológicas.


## Análise de Vocabulário — Termos mais Frequentes

Esta seção aplica a **Skill02** (Análise e Visualização Gráfica de Documentos) em conjunto com seu complemento especializado, a **Skill 02_Análise_Vocab_A** (Análise de Vocabulário e Termos), ao JSON `ai_plus.json`, produzido na etapa anterior (Skill01).

**Protocolo de uso do JSON.** Conforme exigido pela Skill02, apenas os campos `titulo`, `pais_ou_bloco` e `texto_completo` são utilizados nesta análise. Os campos `elementos_descartados`, `data_extracao`, `data_publicacao` e `fonte` não entram, em nenhuma hipótese, na contagem ou na interpretação do vocabulário.

**Idioma identificado.** O `texto_completo` está integralmente redigido em **inglês** — trata-se da tradução oficial do CSET do documento chinês original. As listas de palavras funcionais e as regras de normalização morfológica aplicadas abaixo correspondem ao inglês.

> **Revisão metodológica (2026-08-05).** Esta seção foi revisada por determinação do usuário para corrigir dois erros metodológicos identificados na versão anterior: (i) uma regra cega de tokenização que tratava **todo hífen como separador**, fragmentando compostos tecnicamente relevantes (ex.: "decision-making", "AI-driven"), e que, além disso, inflava indevidamente a própria contagem de "AI" (66→81, ao somar o "AI" solto de compostos como "AI-driven"/"AI-native"); e (ii) ausência de uma etapa sistemática de **desambiguação contextual** de termos polissêmicos. A metodologia abaixo já reflete a correção; o registro persistente documenta a versão anterior e a mudança em detalhe.

**Etapas metodológicas aplicadas** (ver célula de código a seguir para a implementação exata):

1. **Extração bruta** de todos os termos de `texto_completo`, sem qualquer corte prévio de conteúdo.
2. **Pré-processamento do texto corrido, antes da tokenização** — bigramas essenciais protegidos por identidade de referente, não por conveniência estatística:
   - O nome próprio da iniciativa — `"Artificial Intelligence+"`/`"Artificial intelligence+"` (2 ocorrências) e `"AI+"` (9 ocorrências) — unificado no rótulo `AI+` (11 ocorrências) e mantido **deliberadamente separado** do termo genérico `AI` (66 ocorrências): `AI+` designa o nome próprio da política, `AI` genérico designa a tecnologia como campo — mesma lógica do par "United States"/"state(s)" já tratado como não-fundível na análise dos EUA.
   - O bigrama institucional `"State Council"` (4 ocorrências, o órgão administrativo central chinês) protegido para não deixar "state" solto no vocabulário fora de seu referente institucional.
3. **Tokenização com hífen preservado no token** (`[A-Za-z]+(?:-[A-Za-z]+)*`) — não mais "hífen = separador universal". Cada uma das **59 formas hifenizadas** identificadas em `texto_completo` (ex.: *open-source*, *AI-driven*, *decision-making*, *human-computer*, *high-quality*) foi auditada individualmente quanto a permanecer unida ou separar-se em suas palavras componentes; nenhuma regra cega foi aplicada em nenhuma das duas direções. **Resultado da auditoria: as 59 formas foram mantidas unidas** — todas nomeiam um conceito, mecanismo de governança ou colocação fixa mais específico do que a soma das partes (auditoria completa, com justificativa individual de cada forma, no registro persistente).
4. **Desambiguação contextual de termos polissêmicos/homônimos** (nova etapa obrigatória, Skill 02_Análise_Vocab_A item 5): por concordância, verificou-se que "generation" tem sentido único (tecnológico) neste documento e "state" só ocorre como "State Council"/"state-owned" (sem conflito). Foram identificados e **mantidos separados** por referente/classe gramatical distintos: *open* (adj.) / *opening* (verbo, "opening up") / *openings* (subst., vagas de emprego); *major* (adj.) / *majors* (subst., áreas acadêmicas); *process(es)* (subst.) / *processing* (ação técnica); *share* / *shared* / *sharing*; *lead* / *leading*; *play* / *playing*; *lives* / *living*.
5. **Remoção (exclusão)** de palavras funcionais do inglês, tokens de um único caractere (resíduos de "&", possessivos, compostos letra-dígito) e numerais romanos estruturais dos cabeçalhos — lista integral no registro persistente.
6. **Normalização/agrupamento** de variantes que designam exatamente o mesmo referente: 43 pares singular/plural de substantivos e 25 famílias de conjugações verbais (cobertura ampliada em relação à versão anterior, para sustentar o corte de Top 50 — Skill 02_Análise_Vocab_A, item 6). Pares com grafia ou sentido próximos, porém classe gramatical, referente ou sentido distintos, permanecem deliberadamente **não agrupados** (princípio geral de não fusão de quase-sinônimos) — por exemplo: *intelligent* / *intelligentization* / *intelligentized* / *smart*; *driven* (sempre adjetival em compostos) fora da fusão com o verbo "drive"; *development*/*develop*; *capacity*/*capability*; *compute*/*computing*; *security*/*safety* — além dos pares de desambiguação contextual do item 4.
7. **Critério de corte:** **Top 50** termos por frequência absoluta (critério padrão da Skill 02_Análise_Vocab_A), com o **Top 25** também apresentado como recorte adicional. Frequência absoluta é adequada por se tratar de análise de um único documento.

O registro completo, persistente e auditável — incluindo a tabela de auditoria das 59 formas hifenizadas, a tabela de desambiguação contextual e a lista integral de agrupamentos — está em [`registro_vocabulario_ai_plus.md`](./registro_vocabulario_ai_plus.md), nesta mesma pasta.


```python
import json, re
from collections import Counter
import pandas as pd

with open("ai_plus.json", encoding="utf-8") as f:
    doc = json.load(f)

titulo, pais, texto = doc["titulo"], doc["pais_ou_bloco"], doc["texto_completo"]
print(f"Documento: {titulo[:60]}... | País/bloco: {pais} | Caracteres em texto_completo: {len(texto)}")

STOPWORDS_EN = set("""
a an the
and or but nor so yet if because while although that which who whom whose when where how than whether as
this these those it its they their them theirs he she his her him himself herself itself themselves
we our ours us you your yours i my mine
is are was were be been being am
has have had having
do does did doing
will would shall should can could may might must
to of in on at by for with from into through across throughout under over without within among between via per about upon toward towards up out
not no nor
also more most many much very only further therefore thus however moreover
there here
such other others any all both each every some
including include includes included
etc eg ie
""".split())

STRUCTURAL_RESIDUES = {"ii", "iii", "iv"}  # numerais romanos dos cabeçalhos de seção (I., II., III., IV.)

def preprocess(text):
    # nome próprio da iniciativa: unifica forma por extenso e abreviada, protegendo-o da fusão com "AI" genérico
    text = re.sub(r'Artificial [Ii]ntelligence\+', ' AIPLUSTOKEN ', text)
    text = text.replace('AI+', ' AIPLUSTOKEN ')
    # bigrama institucional (Seção 1 do registro persistente): evita "state" solto fora do referente institucional
    text = re.sub(r'\bState Council\b', ' STATECOUNCILTOKEN ', text)
    # desambiguação contextual (Seção 4 do registro): "factors/relations/paradigms of production"
    # é o registro teórico-econômico (生产力/生产关系), distinto do sentido operacional de "production"
    text = re.sub(r'\b(?:factors?|relations?|paradigms?)\s+of\s+production\b', ' PRODUCTIONTHEORYTOKEN ', text, flags=re.I)
    return text

texto_tratado = preprocess(texto)

# Tokenização (Skill 02_Análise_Vocab_A, item 4): hífen preservado como parte do token —
# cada composto hifenizado é, por padrão, candidato a token único, sujeito à auditoria
# individual registrada em registro_vocabulario_ai_plus.md (Seção 2). Nesta análise, as
# 59 formas hifenizadas identificadas foram todas mantidas unidas (nenhuma regra cega).
tokens_brutos = re.findall(r"[A-Za-z]+(?:-[A-Za-z]+)*", texto_tratado)
tokens_lower = [t.lower() for t in tokens_brutos]

tokens_filtrados = [
    t for t in tokens_lower
    if len(t) > 1 and t not in STOPWORDS_EN and t not in STRUCTURAL_RESIDUES
]

# Normalização morfológica — cobertura ampliada para sustentar o corte de Top 50
# (Skill 02_Análise_Vocab_A, item 6). Pares/famílias com risco de sentido/POS
# divergente (open/opening/openings, major/majors, process/processing,
# share/shared/sharing, lead/leading, play/playing, lives/living) foram
# verificados por concordância e mantidos deliberadamente FORA destes dicionários
# (ver registro persistente, Seção 4).
NOUN_MERGES = {
    "application": ["application", "applications"], "models": ["model", "models"],
    "services": ["service", "services"], "systems": ["system", "systems"],
    "capabilities": ["capability", "capabilities"], "industry": ["industry", "industries"],
    "products": ["product", "products"], "resources": ["resource", "resources"],
    "technology": ["technology", "technologies"], "talent": ["talent", "talents"],
    "risks": ["risk", "risks"], "levels": ["level", "levels"], "agents": ["agent", "agents"],
    "initiatives": ["initiative", "initiatives"], "sectors": ["sector", "sectors"],
    "structures": ["structure", "structures"], "processes": ["process", "processes"],
    "policies": ["policy", "policies"], "ecosystems": ["ecosystem", "ecosystems"],
    "evaluations": ["evaluation", "evaluations"], "platforms": ["platform", "platforms"],
    "approach": ["approach", "approaches"],
    "governments": ["government", "governments"], "commissions": ["commission", "commissions"],
    "forces": ["force", "forces"], "transformations": ["transformation", "transformations"],
    "advantages": ["advantage", "advantages"], "scenarios": ["scenario", "scenarios"],
    "achievements": ["achievement", "achievements"], "breakthroughs": ["breakthrough", "breakthroughs"],
    "sciences": ["science", "sciences"], "architectures": ["architecture", "architectures"],
    "chains": ["chain", "chains"], "operations": ["operation", "operations"], "jobs": ["job", "jobs"],
    "environments": ["environment", "environments"], "students": ["student", "students"],
    "roles": ["role", "roles"], "humans": ["human", "humans"], "rights": ["right", "rights"],
    "incentives": ["incentive", "incentives"], "networks": ["network", "networks"],
    "investments": ["investment", "investments"], "laws": ["law", "laws"],
}
VERB_MERGES = {
    "promote": ["promote", "promotes", "promoting", "promoted"],
    "strengthen": ["strengthen", "strengthens", "strengthening", "strengthened"],
    "accelerate": ["accelerate", "accelerates", "accelerating", "accelerated"],
    "support": ["support", "supports", "supporting", "supported"],
    "build": ["build", "builds", "building", "built"],
    "develop": ["develop", "develops", "developing", "developed"],
    "create": ["create", "creates", "creating", "created"],
    "establish": ["establish", "establishes", "establishing", "established"],
    "improve": ["improve", "improves", "improving", "improved"],
    "enhance": ["enhance", "enhances", "enhancing", "enhanced"],
    "explore": ["explore", "explores", "exploring", "explored"],
    "encourage": ["encourage", "encourages", "encouraging", "encouraged"],
    "deepen": ["deepen", "deepens", "deepening", "deepened"],
    "optimize": ["optimize", "optimizes", "optimizing", "optimized"],
    "implement": ["implement", "implements", "implementing", "implemented"],
    "learn": ["learn", "learns", "learning", "learned"],
    "empower": ["empower", "empowers", "empowering", "empowered"],
    "exploit": ["exploit", "exploits", "exploiting", "exploited"],
    "focus": ["focus", "focuses", "focusing", "focused"],
    "foster": ["foster", "fosters", "fostering", "fostered"],
    "fund": ["fund", "funds", "funding", "funded"],
    "grow": ["grow", "grows", "growing"],
    "help": ["help", "helps", "helping", "helped"],
    "launch": ["launch", "launches", "launching", "launched"],
    "work": ["work", "works", "working", "worked"],
}

canon = {}
for label, variants in NOUN_MERGES.items():
    for v in variants:
        canon[v] = label
for label, variants in VERB_MERGES.items():
    for v in variants:
        canon[v] = label
canon["aiplustoken"] = "AI+"
canon["statecounciltoken"] = "State Council"
canon["productiontheorytoken"] = "production (fatores/relações teóricas)"
canon["ai"] = "AI"

tokens_finais = [canon.get(t, t) for t in tokens_filtrados]
freq = Counter(tokens_finais)

print(f"Tokens brutos: {len(tokens_brutos)} | após remoção de funcionais/resíduos: {len(tokens_filtrados)} | termos distintos no vocabulário final: {len(freq)}")

CORTE = 50
top_termos = freq.most_common(CORTE)
top_25 = top_termos[:25]

tabela_top50 = pd.DataFrame(top_termos, columns=["termo", "frequência"])
tabela_top50.index = tabela_top50.index + 1
tabela_top50
```

```python
import matplotlib.pyplot as plt

termos_plot = [t for t, _ in top_25][::-1]
freqs_plot = [n for _, n in top_25][::-1]

fig, ax = plt.subplots(figsize=(9, 10))
barras = ax.barh(termos_plot, freqs_plot, color="#a11d21", height=0.68)

ax.set_xlabel("Frequência absoluta (nº de ocorrências em texto_completo)", fontsize=11)
ax.set_title(
    "Termos mais frequentes — Opinions on Deepening the \"AI+\" Initiative (China, 2025)\n"
    "Top 25 termos, após correção de hífen/bigramas/polissemia (Skill 02_Análise_Vocab_A)",
    fontsize=12.5, fontweight="bold", loc="left", pad=14
)
ax.spines[["top", "right"]].set_visible(False)
ax.tick_params(axis="y", labelsize=10.5)
ax.tick_params(axis="x", labelsize=9.5)
ax.set_axisbelow(True)
ax.xaxis.grid(True, color="#d9d9d9", linewidth=0.8)

for bar, val in zip(barras, freqs_plot):
    ax.text(bar.get_width() + 0.7, bar.get_y() + bar.get_height() / 2, str(val),
            va="center", ha="left", fontsize=9.5, color="#333333")

ax.set_xlim(0, max(freqs_plot) * 1.12)
fig.text(0.02, -0.01,
         "Fonte: texto_completo de ai_plus.json (Skill01), tratado segundo a Skill02 / Skill 02_Análise_Vocab_A.",
         fontsize=8, color="#666666")
fig.tight_layout()
plt.show()
```

## Leitura, Interpretação e Limitações dos Gráficos

**O que foi construído.** Dois gráficos de barras horizontais com os termos de maior frequência absoluta em `texto_completo`, após a correção metodológica descrita acima: (i) Top 25, mantido por continuidade com as demais análises do projeto; (ii) Top 50 (visão expandida), com as posições 26–50 destacadas em tom mais claro. Cada barra representa a contagem final de um termo, de uma família de variantes equivalentes, de um composto hifenizado mantido unido, ou de um sentido específico de um termo polissêmico desdobrado (Seção 4 do registro).

**Como ler os gráficos.** O eixo vertical lista os termos, em ordem decrescente de frequência; o eixo horizontal mede o número absoluto de ocorrências. No gráfico de Top 50, a linha tracejada separa as 25 posições mais frequentes das 25 seguintes.

**O que mudou com a correção metodológica (comparação com a versão anterior deste notebook).**
- **"AI" caiu de 81 para 66 ocorrências.** A contagem anterior somava, ao "AI" genérico, o "AI" solto de seis famílias de compostos hifenizados ("AI-driven", "AI-native", "AI-powered", "AI-based", "AI-enabled", "AI-related" — 15 ocorrências), porque o hífen era tratado como separador. Com a correção, esses compostos permanecem unidos como seus próprios tokens, e "AI" passa a refletir apenas seus usos genuinamente soltos.
- **"based" (11 ocorrências na versão anterior, 24ª posição) desaparece do vocabulário como token isolado.** Era inteiramente um artefato de fragmentação de "AI-based", "platform-based", "scenario-based" e "intelligentization-based" — agora corretamente absorvido dentro de cada composto.
- **"production" (11 ocorrências, antes um único token) foi desdobrado** em `production (fatores/relações teóricas)` (6, registro econômico-marxista — "factors/relations/paradigms of production", 生产力/生产关系) e `production` (5, sentido operacional — "agricultural production", "cultural production", "production, service, and operation"). Nenhum dos dois, isoladamente, atinge o Top 50 — resultado esperado e correto: a fusão anterior inflava artificialmente um token que, na verdade, misturava dois registros discursivos incompatíveis (esse par era citado como limitação explícita, "polissemia não desambiguada", na versão anterior deste registro).
- **A expansão para Top 50 revela termos antes invisíveis no corte de 25:** "coordination" (10), "security" (9), "consumption" (9), "management" (9), "national" (9), "open-source" (9), "compute" (9), "talent" (9), "governments" (7), "society" (7), "industrial" (7), "global" (7), "devices" (7), "roles" (7), "scientific" (7), "agricultural" (7), "risks" (7), "implementation" (6), "integration" (6) — um perfil vocabular sensivelmente mais rico do que o Top 25 permitia enxergar.

**Interpretação à luz do documento.**
- **"AI" (66)** ainda domina com folga, como esperado para uma diretriz de Estado inteiramente dedicada à integração da IA — resultado tautológico e de baixo valor discriminante isoladamente; o valor analítico está a partir da segunda posição.
- **"AI+" (11, 22ª posição)**, mantido deliberadamente separado de "AI" (Seção 4), concentra-se em usos estruturais — a definição da sigla e o prefixo fixo dos seis subtítulos da Parte II — funcionando mais como rótulo organizacional do que como vocabulário discursivo livre.
- **"new" (41) e "development" (37)** seguem no topo, confirmando a marca retórica de renovação ("new generation smart terminal devices", "new quality productive forces") e a centralidade da noção de desenvolvimento ("high-quality development").
- **"promote" (35), "strengthen" (29), "accelerate" (23)** formam o núcleo da "gramática de ação" do documento, abrindo a maioria dos itens numerados de cada subseção.
- **"application" (32), "intelligent" (32), "models" (26), "services" (26), "industry" (18), "systems" (18), "technology" (15), "capabilities" (15), "data" (13)** compõem o vocabulário técnico-operacional esperado de uma política industrial de IA.
- **"governance" (15)**, alta para um termo que também nomeia uma única seção dedicada ("(5) 'AI+' Governance Capacity"), mostra que o vocabulário de governança extravasa essa seção e recorre ao longo de todo o texto.
- **"smart" (15) e "intelligent" (32)**, mantidos deliberadamente separados (itens lexicais distintos, Seção 4), evidenciam juntos — sem serem somados — o quanto compostos com "smart"/"intelligent" permeiam a descrição do estado-alvo da política.
- **"open-source" (9, 32ª posição)**, agora corretamente visível como token próprio (e não mais fragmentado em "open"+"source"), confirma-se como um tema com peso analítico específico — o mecanismo de governança tecnológica de código aberto, adotado pela China de forma distinta de outros modelos nacionais, é agora diretamente comparável, em frequência, a "compute" e "talent".
- **"national" (9) e "governments" (7)**, visíveis apenas no Top 50, situam o documento em um registro de política de Estado com forte presença de vocabulário institucional, complementando "State Council" (bigrama protegido, 4 ocorrências, fora do corte por ser majoritariamente estrutural).
- **"consumption" (9) e "agricultural" (7)**, também visíveis apenas no Top 50, sinalizam dois eixos setoriais específicos do documento (estímulo ao consumo mediado por IA; modernização agrícola) que o corte de 25 não deixava aparentes.

**Ajustes possíveis.** (a) Construir um terceiro gráfico separando verbos diretivos de vocabulário nominal/temático; (b) segmentar a contagem pelas quatro partes de nível superior do documento; (c) reportar frequência relativa ao lado da absoluta, dado que este documento (2.911 tokens brutos) é mais curto que outros do projeto; (d) aplicar a mesma verificação de concordância feita para "production" a outros substantivos abstratos de médio-baixo escore, caso análises futuras ampliem o corte além do Top 50.

**Limitações e fraquezas metodológicas remanescentes.**
- **Cobertura da normalização morfológica ainda não é exaustiva sobre toda a cauda longa do vocabulário** (755 termos distintos): a varredura ampliada desta revisão cobriu os candidatos plausíveis ao Top 50 (frequência bruta ≥ 3, aproximadamente), não a totalidade dos 755 termos — adequado ao corte aqui produzido, mas não garantidamente reprodutível se o corte fosse ampliado para além de 50.
- **Desambiguação contextual aplicada aos casos identificados por leitura manual da concordância, não a uma varredura automática de todo o vocabulário:** é possível que outros termos de frequência mais baixa (fora do alcance analítico do Top 50) também sejam polissêmicos sem terem sido verificados — risco residual, mas de baixo impacto sobre o gráfico produzido.
- **Dependência de regras de pré-processamento específicas a este documento** (proteção de "AI+", "State Council" e das locuções de "production") — não são princípios gerais aplicáveis automaticamente a outro documento sem inspeção manual equivalente da concordância desse outro texto.
