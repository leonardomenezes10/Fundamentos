# Análise do New Generation AI Development Plan — China

Este notebook integra uma análise acadêmica de rigor, desenvolvida no âmbito de uma pesquisa em Relações Internacionais dedicada ao estudo comparado das estratégias nacionais de inteligência artificial. Seu objeto específico é o Plano de Desenvolvimento de Inteligência Artificial de Nova Geração (新一代人工智能发展规划), publicado pelo Conselho de Estado da China, marco fundador da estratégia chinesa de longo prazo para o desenvolvimento da inteligência artificial.

A análise aqui conduzida busca compreender, de modo simultaneamente **quantitativo e qualitativo**, a linguagem empregada pelo documento — os termos, categorias, ênfases retóricas, valores e prioridades estratégicas que estruturam o texto —, de modo a identificar os marcos conceituais e os campos semânticos por meio dos quais a China formula sua política de inteligência artificial.

A partir dessa leitura, pretende-se **posicionar o documento internacionalmente**, comparando a linguagem e as escolhas discursivas de a China com o vocabulário e as ênfases adotados pelos demais países e blocos contemplados neste projeto (Brasil, China, Estados Unidos, Europa e Índia), de modo a mapear convergências, divergências e posicionamentos estratégicos distintivos no debate internacional sobre desenvolvimento e governança de inteligência artificial.

As análises e visualizações produzidas neste notebook seguem as diretrizes metodológicas da **Skill02** (Análise e Visualização Gráfica de Documentos): baseiam-se exclusivamente no campo `texto_completo` do JSON de extração correspondente a este documento, produzido na etapa anterior (Skill01), adotam rigor acadêmico integral na leitura do texto-fonte, evitam generalizações, simplificações e inferências não fundamentadas no texto original, e cada visualização construída é acompanhada de sua respectiva descrição, leitura, interpretação e eventuais limitações metodológicas.


## Análise de Vocabulário — Termos mais Frequentes

Esta seção aplica a **Skill02** (Análise e Visualização Gráfica de Documentos) em conjunto com seu complemento especializado, a **Skill 02_Análise_Vocab_A** (Análise de Vocabulário e Termos), ao JSON `new_generation_ai_development_plan.json`, produzido na etapa anterior (Skill01).

**Protocolo de uso do JSON.** Conforme exigido pela Skill02, apenas os campos `titulo`, `pais_ou_bloco` e `texto_completo` são utilizados nesta análise. Os campos `elementos_descartados`, `data_extracao`, `data_publicacao` e `fonte` não entram, em nenhuma hipótese, na contagem ou na interpretação do vocabulário.

**Idioma identificado.** O documento está integralmente redigido em **inglês** — trata-se da tradução publicada pela New America/DigiChina do original chinês (新一代人工智能发展规划).

> **Revisão metodológica (2026-08-05).** Por determinação do usuário, esta seção foi revisada para corrigir a regra cega de tokenização que tratava todo hífen como separador — a versão anterior já registrava, como limitação, que essa regra fragmentava "decision-making" (23 ocorrências) e inflava artificialmente os tokens "high" (53, via "high-end"/"high-level") e "brain" (via "brain-inspired"). Foram também adicionadas: desambiguação contextual de "power" e "driving"/"drive"/"driven" (dois referentes genuinamente distintos sob a mesma grafia), proteção do bigrama institucional "State Council", e expansão do corte para **Top 50**. O registro persistente documenta a versão anterior e a mudança em detalhe.

**Etapas metodológicas aplicadas** (ver célula de código a seguir para a implementação exata):

1. **Extração bruta** de todos os termos de `texto_completo`, sem qualquer corte prévio de conteúdo.
2. **Pré-processamento do texto corrido, antes da tokenização:** `"artificial intelligence"` → `"AI"` (mesmo referente); bigrama institucional `"State Council"` protegido (4 ocorrências); e três desambiguações contextuais protegidas por locução — `"nuclear power"` (sentido energético de "power", distinto do sentido geopolítico/nacional que domina as demais ocorrências) e `"driving force"`/`"driving effect"`/`"driving role"`/`"driving networks"`/`"drive"`(verbo)/`"driven"` (sentido figurado de impulso, distinto do sentido literal de condução autônoma que domina as demais ocorrências de "driving"). Duas variantes de grafia/extração também normalizadas nesta camada: `"co-ordination"`→"coordination" (grafia britânica) e `"rain-inspired"`→"brain-inspired" (artefato mecânico de extração, confirmado por concordância — todas as ocorrências descrevem o mesmo conceito técnico no mesmo parágrafo).
3. **Tokenização com hífen preservado no token** (`[A-Za-z]+(?:-[A-Za-z]+)*`) — não mais "hífen = separador universal". Das **136 formas hifenizadas** identificadas em `texto_completo`, 2 foram resolvidas como variantes de grafia (passo 2) e as **134 restantes foram mantidas unidas** (nenhuma qualificou para separação): cada uma nomeia um conceito técnico, categoria de política ou colocação fixa mais específico do que a soma das partes — auditoria completa, organizada tematicamente (neurociência computacional, interação humano-máquina, categorias "high-X", transversalidade "cross-X", pluralidade "multi-X", autonomia "self-/co-", descritores "-driven/-oriented/-based", escala/geração tecnológica, marcadores temporais, termos técnico-econômicos, militar/geográfico), no registro persistente.
4. **Desambiguação contextual de termos polissêmicos/homônimos** (nova etapa obrigatória): "generation" e "state" foram verificados por concordância e têm sentido único neste documento (sem necessidade de desdobramento); "power" e "driving"/"drive"/"driven" apresentaram referentes genuinamente distintos e foram desdobrados (passo 2).
5. **Remoção (exclusão)** de palavras funcionais do inglês e tokens de um único caractere — lista integral no registro persistente.
6. **Normalização/agrupamento** de variantes que designam exatamente o mesmo referente: 59 pares singular/plural de substantivos (30 já existentes + 29 novos nesta revisão, para sustentar o corte de Top 50) e 18 famílias verbais. Pares mantidos deliberadamente **não agrupados** (herdados da versão anterior, todos reconfirmados nesta revisão): *enhanced* fora do grupo verbal "enhance" (termo técnico fixo "hybrid enhanced intelligence"); *smart*/*intelligent*/*intelligence*/*intelligentization* (quatro itens lexicais distintos); *development*/*develop*; *research*/*researching*; *support*/*supporting*; *building* fora de "build"; *process*/*processing*; *share*/*sharing*.
7. **Critério de corte:** **Top 50** termos por frequência absoluta (critério padrão da Skill 02_Análise_Vocab_A), com o **Top 25** também apresentado como recorte adicional.

O registro completo, persistente e auditável — incluindo a auditoria das 136 formas hifenizadas, a tabela de desambiguação contextual e a lista integral de agrupamentos — está em [`registro_vocabulario_new_generation_ai_development_plan.md`](./registro_vocabulario_new_generation_ai_development_plan.md), nesta mesma pasta.


```python
import json, re
from collections import Counter
import pandas as pd

with open("new_generation_ai_development_plan.json", encoding="utf-8") as f:
    doc = json.load(f)

titulo, pais, texto = doc["titulo"], doc["pais_ou_bloco"], doc["texto_completo"]
print(f"Documento: {titulo} | País/bloco: {pais} | Caracteres em texto_completo: {len(texto)}")

STOPWORDS_EN = set("""
a an the
and or but nor so yet if because while although that which who whom whose when where how than whether as
this these those it its they their them theirs he she his her him himself herself itself themselves
we our ours us you your yours i my mine
is are was were be been being am
has have had having
do does did doing
will would shall should can could may might must
to of in on at by for with from into through across throughout under over without within among between via per about upon toward towards
not no nor
also more most many much very only further therefore thus however moreover
there here
such other others any all both each every some
including include includes included
etc eg ie
""".split())

def preprocess(text):
    text = re.sub(r'\bartificial intelligence\b', 'AI', text, flags=re.I)
    # bigrama institucional (Skill 02_Análise_Vocab_A, item 3)
    text = re.sub(r'\bState Council\b', ' STATECOUNCILTOKEN ', text)
    # desambiguação contextual (registro persistente, Seção 4): sentido figurado de
    # impulso/propulsão, distinto do sentido literal de condução autônoma
    text = re.sub(r'\bdriving\s+force\b', ' DRIVEFIGTOKEN ', text, flags=re.I)
    text = re.sub(r'\bdriving\s+effect\b', ' DRIVEFIGTOKEN ', text, flags=re.I)
    text = re.sub(r'\bdriving\s+role\b', ' DRIVEFIGTOKEN ', text, flags=re.I)
    text = re.sub(r'\bdriving\s+networks\b', ' DRIVEFIGTOKEN ', text, flags=re.I)
    text = re.sub(r'\bdrive\b', ' DRIVEFIGTOKEN ', text, flags=re.I)
    text = re.sub(r'(?<!-)\bdriven\b(?!-)', ' DRIVEFIGTOKEN ', text, flags=re.I)
    # desambiguação contextual: sentido energético/nuclear de "power", distinto do
    # sentido geopolítico/nacional que domina as demais ocorrências
    text = re.sub(r'\bnuclear\s+power\b', ' POWERENERGYTOKEN ', text, flags=re.I)
    # variantes de grafia/extração do mesmo termo (registro persistente, Seção 1)
    text = re.sub(r'\bco-ordination\b', 'coordination', text, flags=re.I)
    text = re.sub(r'(?<![A-Za-z-])rain-inspired\b', 'brain-inspired', text, flags=re.I)
    # bigramas institucionais/conceituais (auditoria de consistência entre documentos,
    # Análise Conjunta): mesmo tratamento já aplicado em outros documentos do projeto
    # (pbia.ipynb, ai_plus.ipynb) para as mesmas categorias de política
    text = re.sub(r'\bResearch and Development\b|\bR&D\b', ' RESEARCHDEVTOKEN ', text, flags=re.I)
    text = re.sub(r'\bData Infrastructure\b', ' DATAINFRATOKEN ', text, flags=re.I)
    text = re.sub(r'\bPublic Services?\b', ' PUBSERVICETOKEN ', text, flags=re.I)
    text = re.sub(r'\bValue Chains?\b', ' VALUECHAINTOKEN ', text, flags=re.I)
    text = re.sub(r'\bMachine Learning\b', ' MACHLEARNTOKEN ', text, flags=re.I)
    return text

texto_tratado = preprocess(texto)

# Tokenização (Skill 02_Análise_Vocab_A, item 4): hífen preservado como parte do token —
# cada composto hifenizado é candidato a token único, sujeito à auditoria individual
# registrada no registro persistente (Seção 2). Das 136 formas identificadas neste
# documento, 134 foram mantidas unidas (nenhuma regra cega em nenhuma direção).
tokens_brutos = re.findall(r"[A-Za-z]+(?:-[A-Za-z]+)*", texto_tratado)
tokens_lower = [t.lower() for t in tokens_brutos]
tokens_filtrados = [t for t in tokens_lower if len(t) > 1 and t not in STOPWORDS_EN]

NOUN_MERGES = {
    "technology": ["technology", "technologies"], "systems": ["system", "systems"],
    "applications": ["application", "applications"], "theory": ["theory", "theories"],
    "platforms": ["platform", "platforms"], "enterprises": ["enterprise", "enterprises"],
    "breakthroughs": ["breakthrough", "breakthroughs"], "products": ["product", "products"],
    "robots": ["robot", "robots"], "capabilities": ["capability", "capabilities"],
    "bases": ["base", "bases"], "resources": ["resource", "resources"],
    "mechanisms": ["mechanism", "mechanisms"], "tasks": ["task", "tasks"],
    "projects": ["project", "projects"], "demonstrations": ["demonstration", "demonstrations"],
    "areas": ["area", "areas"], "methods": ["method", "methods"], "models": ["model", "models"],
    "standards": ["standard", "standards"], "fields": ["field", "fields"],
    "policies": ["policy", "policies"], "regulations": ["regulation", "regulations"],
    "laws": ["law", "laws"], "talent": ["talent", "talents"], "levels": ["level", "levels"],
    "results": ["result", "results"], "sectors": ["sector", "sectors"],
    "environments": ["environment", "environments"], "services": ["service", "services"],
    # -- novos pares (cobertura ampliada para Top 50) --
    "networks": ["network", "networks"], "demands": ["demand", "demands"],
    "disciplines": ["discipline", "disciplines"], "domains": ["domain", "domains"],
    "plans": ["plan", "plans"], "changes": ["change", "changes"],
    "structures": ["structure", "structures"], "risks": ["risk", "risks"],
    "clusters": ["cluster", "clusters"], "forces": ["force", "forces"],
    "advantages": ["advantage", "advantages"], "reforms": ["reform", "reforms"],
    "livelihoods": ["livelihood", "livelihoods"], "chains": ["chain", "chains"],
    "teams": ["team", "teams"], "centers": ["center", "centers"],
    "frameworks": ["framework", "frameworks"], "engines": ["engine", "engines"],
    "sensors": ["sensor", "sensors"], "counterparts": ["counterpart", "counterparts"],
    "architectures": ["architecture", "architectures"], "vehicles": ["vehicle", "vehicles"],
    "markets": ["market", "markets"], "terminals": ["terminal", "terminals"],
    "trials": ["trial", "trials"], "solutions": ["solution", "solutions"],
    "organizations": ["organization", "organizations"], "assistants": ["assistant", "assistants"],
    "programs": ["program", "programs"], "incentives": ["incentive", "incentives"],
}

VERB_MERGES = {
    "strengthen": ["strengthen", "strengthens", "strengthening", "strengthened"],
    "promote": ["promote", "promotes", "promoting", "promoted"],
    "establish": ["establish", "establishes", "establishing", "established"],
    "develop": ["develop", "develops", "developing", "developed"],
    "construct": ["construct", "constructs", "constructing", "constructed"],
    "build": ["build", "builds", "built"],
    "accelerate": ["accelerate", "accelerates", "accelerating", "accelerated"],
    "encourage": ["encourage", "encourages", "encouraging", "encouraged"],
    "improve": ["improve", "improves", "improving", "improved"],
    "achieve": ["achieve", "achieves", "achieving", "achieved"],
    "enhance": ["enhance", "enhances", "enhancing"],  # "enhanced" deliberadamente excluído — ver registro, Seção 4
    "form": ["form", "forms", "forming", "formed"],
    "launch": ["launch", "launches", "launching", "launched"],
    "focus": ["focus", "focuses", "focusing", "focused"],
    "integrate": ["integrate", "integrates", "integrating", "integrated"],
    "coordinate": ["coordinate", "coordinates", "coordinating", "coordinated"],
    "lead": ["lead", "leads", "leading", "led"],
    "create": ["create", "creates", "creating", "created"],
}

canon = {}
for label, variants in NOUN_MERGES.items():
    for v in variants:
        canon[v] = label
for label, variants in VERB_MERGES.items():
    for v in variants:
        canon[v] = label
canon["ai"] = "AI"
canon["china"] = "China"
canon["statecounciltoken"] = "State Council"
canon["drivefigtoken"] = "driving/drive (força motriz)"
canon["powerenergytoken"] = "power (nuclear/energia)"
canon["power"] = "power (nacional/geopolítico)"
canon["driving"] = "driving (condução autônoma)"
canon["researchdevtoken"] = "Research & Development (R&D)"
canon["datainfratoken"] = "Data Infrastructure"
canon["pubservicetoken"] = "Public Service(s)"
canon["valuechaintoken"] = "Value Chain"
canon["machlearntoken"] = "Machine Learning"

tokens_finais = [canon.get(t, t) for t in tokens_filtrados]
freq = Counter(tokens_finais)

print(f"Tokens brutos: {len(tokens_brutos)} | após remoção de funcionais/resíduos: {len(tokens_filtrados)} | termos distintos no vocabulário final: {len(freq)}")

CORTE = 50
top_termos = freq.most_common(CORTE)
top_25 = top_termos[:25]

tabela_top50 = pd.DataFrame(top_termos, columns=["termo", "frequência"])
tabela_top50.index = tabela_top50.index + 1
tabela_top50
```

```python
import matplotlib.pyplot as plt

termos_plot = [t for t, _ in top_25][::-1]
freqs_plot = [n for _, n in top_25][::-1]

fig, ax = plt.subplots(figsize=(9, 10.5))
barras = ax.barh(termos_plot, freqs_plot, color="#8c1d1d", height=0.68)

ax.set_xlabel("Frequência absoluta (nº de ocorrências em texto_completo)", fontsize=11)
ax.set_title(
    "Termos mais frequentes — New Generation AI Development Plan (China, 2017)\n"
    "Top 25 termos, após correção de hífen/bigramas/polissemia (Skill 02_Análise_Vocab_A)",
    fontsize=12.5, fontweight="bold", loc="left", pad=14
)
ax.spines[["top", "right"]].set_visible(False)
ax.tick_params(axis="y", labelsize=10.5)
ax.tick_params(axis="x", labelsize=9.5)
ax.set_axisbelow(True)
ax.xaxis.grid(True, color="#d9d9d9", linewidth=0.8)

for bar, val in zip(barras, freqs_plot):
    ax.text(bar.get_width() + 3, bar.get_y() + bar.get_height() / 2, str(val),
            va="center", ha="left", fontsize=9.5, color="#333333")

ax.set_xlim(0, max(freqs_plot) * 1.12)
fig.text(0.02, -0.01,
         "Fonte: texto_completo de new_generation_ai_development_plan.json (Skill01), tratado segundo a Skill02 / Skill 02_Análise_Vocab_A.",
         fontsize=8, color="#666666")
fig.tight_layout()
plt.show()
```

## Leitura, Interpretação e Limitações dos Gráficos

**O que foi construído.** Dois gráficos de barras horizontais: Top 25 (mantido por continuidade) e Top 50 (visão expandida, posições 26–50 em tom mais claro), após a correção metodológica descrita acima.

**O que mudou com a correção (comparação com a versão anterior).**
- **"AI" caiu de 344 para 339 ocorrências** — a diferença de 5 vinha do "AI" solto de "AI-related"(4) e "AI-based"(1), agora corretamente absorvidos em seus próprios compostos.
- **"high" (53 ocorrências, 20º lugar na versão anterior) desaparece inteiramente do vocabulário como token isolado** — era artefato de fragmentação de 12 compostos distintos ("high-end", "high-performance", "high-level" etc.), cada um agora seu próprio token, preservando a distinção entre qualidade, hierarquia, desempenho e precisão que a versão anterior apagava.
- **"open" (30 ocorrências na versão anterior) some do topo do vocabulário** — a maior parte vinha de "open-source" (7) e "open-style" (1), agora tokens próprios; o "open" genuinamente solto (ex. "open cooperation") cai para uma posição de baixa frequência.
- **"power" (6 ocorrências, antes um único token) foi desdobrado** em `power (nacional/geopolítico)` (5 — "global power in science and technology", "an economic power") e `power (nuclear/energia)` (1 — "nuclear power security operations"): dois referentes incompatíveis que a contagem anterior fundia silenciosamente.
- **"driving"/"drive"/"driven" (21 ocorrências, antes um único registro confuso) foi desdobrado** em `driving (condução autônoma)` (13 — veículos autônomos) e `driving/drive (força motriz)` (8 — "driving force", "driving effect", verbo "drive"): a própria versão anterior já apontava esses como "referentes distintos", sem contudo separá-los na contagem — agora corrigido.
- **A expansão para Top 50 revela um perfil vocabular mais completo:** "social" (42), "key" (42), "learning" (41), "construction" (39), "science"/"autonomous"/"breakthroughs"/"models"/"basic" (36), "knowledge" (35), "security" (34), "China" (31), "core" (31), "talent" (31), "sensing" (28), "generation" (27), "plans" (26).

**Interpretação à luz do documento.**
- **"AI" (339)** domina como esperado; valor discriminante baixo isoladamente.
- **"technology" (156), "systems" (130)** seguem confirmando o enquadramento técnico-arquitetural, junto de "development" (139) — mantido separado do verbo "develop" (46) — que confirma a vocação substantiva/programática do plano.
- **"intelligent" (121) e "smart" (82)**, mantidos deliberadamente separados de "intelligence" (57) e entre si (sinônimos de tradução, não fundidos), somados ultrapassariam "technology" — sem que o gráfico permita afirmar isso diretamente, dada a proibição de fusão de itens lexicais distintos.
- **"decision-making" (23)**, agora visível como token próprio (era invisível como tal na versão anterior, dissolvido em "decision"+"making"), confirma a centralidade do processo decisório informado por IA como categoria autônoma do vocabulário do plano — ainda que, com 23 ocorrências, fique abaixo do corte de Top 50 (fronteira em 26): o próprio fato de agora ser contável como unidade única, em vez de inflar dois tokens genéricos, já é o ganho metodológico relevante, independentemente de entrar ou não no gráfico.
- **"open-source" (7) e "AI-related" (4)**, mesma lógica: agora tokens próprios e auditáveis, mesmo sem entrar no corte de 50 — o mecanismo de governança de código aberto identificado como tema central no documento `ai_plus.json` (2025) já aparece, em grau mais modesto, também neste plano de 2017, um indício de continuidade entre os dois planos chineses do projeto.
- **"national" (45) e "China" (31)** seguem como únicos termos de enquadramento nacional/soberano no Top 50 em frequência bruta — o vocabulário do documento permanece predominantemente técnico-teórico.

**Ajustes possíveis.** (a) Gráfico complementar somando lado a lado (sem fundir) "smart"+"intelligent"; (b) segmentação por seção do documento; (c) frequência relativa ao lado da absoluta.

**Limitações remanescentes.**
- **Texto de partida é uma tradução** (New America/DigiChina), não o original em chinês.
- **"smart"/"intelligent" seguem não fundidos**, por exigir julgamento de equivalência de tradução não autorizado pela Skill.
- **Cobertura de normalização e desambiguação concentrada nos candidatos plausíveis ao Top 50** — não há varredura exaustiva de toda a cauda longa de 1.611 termos distintos.
- **Duas normalizações de grafia/extração** ("co-ordination"→"coordination", "rain-inspired"→"brain-inspired") dependem de leitura manual da concordância, não de regra automática generalizável.
