# Análise do New Generation AI Development Plan — China

Este notebook integra uma análise acadêmica de rigor, desenvolvida no âmbito de uma pesquisa em Relações Internacionais dedicada ao estudo comparado das estratégias nacionais de inteligência artificial. Seu objeto específico é o Plano de Desenvolvimento de Inteligência Artificial de Nova Geração (新一代人工智能发展规划), publicado pelo Conselho de Estado da China, marco fundador da estratégia chinesa de longo prazo para o desenvolvimento da inteligência artificial.

A análise aqui conduzida busca compreender, de modo simultaneamente **quantitativo e qualitativo**, a linguagem empregada pelo documento — os termos, categorias, ênfases retóricas, valores e prioridades estratégicas que estruturam o texto —, de modo a identificar os marcos conceituais e os campos semânticos por meio dos quais a China formula sua política de inteligência artificial.

A partir dessa leitura, pretende-se **posicionar o documento internacionalmente**, comparando a linguagem e as escolhas discursivas de a China com o vocabulário e as ênfases adotados pelos demais países e blocos contemplados neste projeto (Brasil, China, Estados Unidos, Europa e Índia), de modo a mapear convergências, divergências e posicionamentos estratégicos distintivos no debate internacional sobre desenvolvimento e governança de inteligência artificial.

As análises e visualizações produzidas neste notebook seguem as diretrizes metodológicas da **Skill02** (Análise e Visualização Gráfica de Documentos): baseiam-se exclusivamente no campo `texto_completo` do JSON de extração correspondente a este documento, produzido na etapa anterior (Skill01), adotam rigor acadêmico integral na leitura do texto-fonte, evitam generalizações, simplificações e inferências não fundamentadas no texto original, e cada visualização construída é acompanhada de sua respectiva descrição, leitura, interpretação e eventuais limitações metodológicas.


## Análise de Vocabulário — Termos mais Frequentes

Esta seção aplica a **Skill02** (Análise e Visualização Gráfica de Documentos) em conjunto com seu complemento especializado, a **Skill 02_Análise_Vocab_A** (Análise de Vocabulário e Termos), ao JSON `new_generation_ai_development_plan.json`, produzido na etapa anterior (Skill01).

**Protocolo de uso do JSON.** Conforme exigido pela Skill02, apenas os campos `titulo`, `pais_ou_bloco` e `texto_completo` são utilizados nesta análise. Os campos `elementos_descartados`, `data_extracao`, `data_publicacao` e `fonte` não entram, em nenhuma hipótese, na contagem ou na interpretação do vocabulário.

**Idioma identificado.** O documento está integralmente redigido em **inglês** — trata-se da tradução publicada pela New America/DigiChina do original chinês (新一代人工智能发展规划). Todas as listas de palavras funcionais e as regras de normalização morfológica aplicadas abaixo correspondem ao inglês, e não ao português, ao chinês ou a qualquer outro idioma.

**Etapas metodológicas aplicadas** (ver célula de código a seguir para a implementação exata):

1. **Extração bruta** de todos os termos de `texto_completo`, sem qualquer corte prévio de conteúdo.
2. **Pré-processamento do texto corrido, antes da tokenização**, fundamentado em identidade de referente: a expressão composta `"artificial intelligence"` (case-insensitive) foi unificada ao termo `"AI"` — sigla e forma por extenso designando o mesmo conceito, mesma lógica do par "IA"/"Inteligência Artificial" citado na própria Skill 02_Análise_Vocab_A. A expressão ocorre apenas 3 vezes no texto corrido (2x no título/cabeçalho do documento, 1x na frase de definição do primeiro parágrafo); a partir daí o documento emprega exclusivamente a sigla "AI". A ocorrência de "artificial data-oriented" (Seção IV) foi verificada e não corresponde a esse padrão, não sendo afetada.
3. **Tokenização**: sequências de letras (`[A-Za-z]+`) foram extraídas como tokens; qualquer caractere não alfabético (espaço, hífen, apóstrofo, ponto, parênteses, barra) foi tratado como separador — mesma regra única e uniforme adotada na análise do documento dos Estados Unidos (`americas_ai_action_plan.json`), para manter consistência metodológica entre os documentos do projeto. Compostos com hífen (ex.: *decision-making*, *data-driven*, *high-end*) são, por consequência, fragmentados em suas palavras componentes — o impacto dessa regra é discutido como limitação ao final, sendo mais acentuado neste documento do que no dos EUA.
4. **Remoção (exclusão)** de: palavras funcionais do inglês (artigos, preposições, conjunções, pronomes, verbos auxiliares/modais e conectores discursivos genéricos como *also*, *including*, *etc.*) — lista declarada integralmente na célula de código e no registro persistente; e tokens de um único caractere (resíduo de possessivos, ex.: "China's" → fragmento solto "s").
5. **Normalização/agrupamento** de variantes que designam exatamente o mesmo referente: 30 pares singular/plural de substantivos (ex.: *system*/*systems*, *service*/*services*) e 18 famílias de conjugações/flexões verbais (ex.: *strengthen*/*strengthens*/*strengthening*/*strengthened*). Pares com grafia ou lema semelhante, mas referente distinto, foram deliberadamente **mantidos separados** — com destaque para dois casos verificados individualmente por leitura de contexto: **"enhanced"**, mantido fora do grupo verbal de "enhance/enhancing" por funcionar, neste documento, como parte do termo técnico fixo do próprio plano "*hybrid (and) enhanced intelligence*" (categoria nomeada na taxonomia de teorias de IA da Seção III), não como flexão livre do verbo genérico; e **"driving"/"driven"/"drive"**, mantidos totalmente separados entre si por reunirem ao menos três referentes distintos (direção autônoma de veículos, compostos adjetivais "data-driven"/"knowledge-driven", e a locução "driving force"). A lista completa de agrupamentos e de pares deliberadamente não agrupados, com a justificativa individual de cada decisão, está no registro persistente.
6. **Critério de corte declarado:** os **25 termos de maior frequência absoluta** após as etapas 1–5, com um ajuste explícito de empate — a 25ª posição ("major", 43 ocorrências) empata exatamente com o termo seguinte ("focus", 43 ocorrências); para evitar um corte arbitrário entre termos de frequência idêntica, o corte foi estendido para incluir todos os termos empatados no valor de fronteira, resultando em **26 termos** no gráfico final. Frequência absoluta é adequada aqui porque esta é uma análise de um único documento — a normalização relativa entre documentos de tamanhos distintos, exigida pela Skill02, aplica-se apenas à pasta "Análise Conjunta".

O registro completo, persistente e auditável de todas as remoções e agrupamentos — incluindo os pares deliberadamente não fundidos e as limitações identificadas — está salvo em [`registro_vocabulario_new_generation_ai_development_plan.md`](./registro_vocabulario_new_generation_ai_development_plan.md), nesta mesma pasta, e deve ser consultado — e atualizado de forma cumulativa — antes de qualquer nova visualização de vocabulário sobre este mesmo documento.


```python
import json, re
from collections import Counter
import pandas as pd

with open("new_generation_ai_development_plan.json", encoding="utf-8") as f:
    doc = json.load(f)

titulo, pais, texto = doc["titulo"], doc["pais_ou_bloco"], doc["texto_completo"]
print(f"Documento: {titulo} | País/bloco: {pais} | Caracteres em texto_completo: {len(texto)}")

STOPWORDS_EN = set("""
a an the
and or but nor so yet if because while although that which who whom whose when where how than whether as
this these those it its they their them theirs he she his her him himself herself itself themselves
we our ours us you your yours i my mine
is are was were be been being am
has have had having
do does did doing
will would shall should can could may might must
to of in on at by for with from into through across throughout under over without within among between via per about upon toward towards
not no nor
also more most many much very only further therefore thus however moreover
there here
such other others any all both each every some
including include includes included
etc eg ie
""".split())

def preprocess(text):
    return re.sub(r'\bartificial intelligence\b', 'AI', text, flags=re.I)

texto_tratado = preprocess(texto)
tokens_brutos = re.findall(r"[A-Za-z]+", texto_tratado)
tokens_lower = [t.lower() for t in tokens_brutos]
tokens_filtrados = [t for t in tokens_lower if len(t) > 1 and t not in STOPWORDS_EN]

NOUN_MERGES = {
    "technology": ["technology", "technologies"], "systems": ["system", "systems"],
    "applications": ["application", "applications"], "theory": ["theory", "theories"],
    "platforms": ["platform", "platforms"], "enterprises": ["enterprise", "enterprises"],
    "breakthroughs": ["breakthrough", "breakthroughs"], "products": ["product", "products"],
    "robots": ["robot", "robots"], "capabilities": ["capability", "capabilities"],
    "bases": ["base", "bases"], "resources": ["resource", "resources"],
    "mechanisms": ["mechanism", "mechanisms"], "tasks": ["task", "tasks"],
    "projects": ["project", "projects"], "demonstrations": ["demonstration", "demonstrations"],
    "areas": ["area", "areas"], "methods": ["method", "methods"], "models": ["model", "models"],
    "standards": ["standard", "standards"], "fields": ["field", "fields"],
    "policies": ["policy", "policies"], "regulations": ["regulation", "regulations"],
    "laws": ["law", "laws"], "talent": ["talent", "talents"], "levels": ["level", "levels"],
    "results": ["result", "results"], "sectors": ["sector", "sectors"],
    "environments": ["environment", "environments"], "services": ["service", "services"],
}

VERB_MERGES = {
    "strengthen": ["strengthen", "strengthens", "strengthening", "strengthened"],
    "promote": ["promote", "promotes", "promoting", "promoted"],
    "establish": ["establish", "establishes", "establishing", "established"],
    "develop": ["develop", "develops", "developing", "developed"],
    "construct": ["construct", "constructs", "constructing", "constructed"],
    "build": ["build", "builds", "built"],
    "accelerate": ["accelerate", "accelerates", "accelerating", "accelerated"],
    "encourage": ["encourage", "encourages", "encouraging", "encouraged"],
    "improve": ["improve", "improves", "improving", "improved"],
    "achieve": ["achieve", "achieves", "achieving", "achieved"],
    "enhance": ["enhance", "enhances", "enhancing"],  # "enhanced" deliberadamente excluído — ver Seção 5 do registro
    "form": ["form", "forms", "forming", "formed"],
    "launch": ["launch", "launches", "launching", "launched"],
    "focus": ["focus", "focuses", "focusing", "focused"],
    "integrate": ["integrate", "integrates", "integrating", "integrated"],
    "coordinate": ["coordinate", "coordinates", "coordinating", "coordinated"],
    "lead": ["lead", "leads", "leading", "led"],
    "create": ["create", "creates", "creating", "created"],
}

canon = {}
for label, variants in NOUN_MERGES.items():
    for v in variants:
        canon[v] = label
for label, variants in VERB_MERGES.items():
    for v in variants:
        canon[v] = label
canon["ai"] = "AI"
canon["china"] = "China"

tokens_finais = [canon.get(t, t) for t in tokens_filtrados]
freq = Counter(tokens_finais)

print(f"Tokens brutos: {len(tokens_brutos)} | após remoção de funcionais/resíduos: {len(tokens_filtrados)} | termos distintos no vocabulário final: {len(freq)}")

ordered = freq.most_common()
CORTE = 25
cutoff_val = ordered[CORTE - 1][1]
top_termos = [item for item in ordered if item[1] >= cutoff_val]
print(f"Valor de frequência na fronteira do corte: {cutoff_val} | termos incluídos após ajuste de empate: {len(top_termos)}")

tabela_top = pd.DataFrame(top_termos, columns=["termo", "frequência"])
tabela_top.index = tabela_top.index + 1
tabela_top
```

```python
import matplotlib.pyplot as plt

termos_plot = [t for t, _ in top_termos][::-1]
freqs_plot = [n for _, n in top_termos][::-1]

fig, ax = plt.subplots(figsize=(9, 10.5))
barras = ax.barh(termos_plot, freqs_plot, color="#8c1d1d", height=0.68)

ax.set_xlabel("Frequência absoluta (nº de ocorrências em texto_completo)", fontsize=11)
ax.set_title(
    "Termos mais frequentes — New Generation AI Development Plan (China, 2017)\n"
    f"Top {len(top_termos)} termos após remoção de palavras funcionais e normalização de variantes",
    fontsize=12.5, fontweight="bold", loc="left", pad=14
)
ax.spines[["top", "right"]].set_visible(False)
ax.tick_params(axis="y", labelsize=10.5)
ax.tick_params(axis="x", labelsize=9.5)
ax.set_axisbelow(True)
ax.xaxis.grid(True, color="#d9d9d9", linewidth=0.8)

for bar, val in zip(barras, freqs_plot):
    ax.text(bar.get_width() + 3, bar.get_y() + bar.get_height() / 2, str(val),
            va="center", ha="left", fontsize=9.5, color="#333333")

ax.set_xlim(0, max(freqs_plot) * 1.12)
fig.text(0.02, -0.01,
         "Fonte: texto_completo de new_generation_ai_development_plan.json (Skill01), tratado segundo a Skill02 / Skill 02_Análise_Vocab_A.",
         fontsize=8, color="#666666")
fig.tight_layout()
plt.show()
```

## Leitura, Interpretação e Limitações do Gráfico

**O que foi construído.** Um gráfico de barras horizontais com os 26 termos de maior frequência absoluta em `texto_completo` (top 25 com ajuste de empate na fronteira do corte, ver metodologia), após remoção de palavras funcionais/resíduos e normalização das variantes descritas acima. Cada barra representa a contagem final (já consolidada) de um termo ou família de termos equivalentes.

**Como ler o gráfico.** O eixo vertical lista os termos, em ordem decrescente de frequência (o termo mais frequente no topo); o eixo horizontal mede o número absoluto de ocorrências no documento. O rótulo numérico ao final de cada barra é a contagem exata.

**Interpretação à luz do documento.**
- **"AI" (344 ocorrências)** domina o vocabulário por ampla margem — mais de duas vezes a frequência do segundo termo —, resultado esperado para um plano de desenvolvimento inteiramente dedicado à IA. Como no caso análogo do documento dos EUA, esse resultado isolado tem baixo valor discriminante; o valor analítico do gráfico está principalmente nos termos que disputam a segunda posição em diante.
- **"technology" (157) e "systems" (130)** ocupam a 2ª e a 4ª posição, confirmando o enquadramento fortemente técnico-arquitetural do documento — a própria Seção III do plano ("Focus Tasks") é organizada em torno de "novos sistemas de teoria e tecnologia de IA" e de plataformas de suporte.
- **"development" (139, 3º lugar)**, mantido deliberadamente separado do verbo "develop" (46, 22º lugar), evidencia que a vocação do documento é primariamente **substantiva/programática** ("um plano de desenvolvimento") e não uma lista de ações verbais — diferença relevante frente ao documento dos EUA, onde o vocabulário verbal de ação ("lead/led", "ensure", "build") tinha peso comparativamente maior nas primeiras posições.
- **"intelligent" (122, 5º lugar)**, mantido separado de "intelligence" (58, 18º) e de "AI" (sigla), reflete o uso do documento de "intelligent" como modificador aplicado a dezenas de domínios distintos (manufatura, sociedade, economia, robôs, sistemas), consistente com a lógica de "intelligentização" (智能化 — "intelligentization", termo que também aparece isoladamente 13 vezes no texto, embora abaixo do corte) como projeto transversal a todos os setores, e não apenas como um campo técnico específico.
- **"research" (104, 6º lugar)** e **"theory" (61, 16º lugar)** — somados aos oito campos teóricos nomeados na Seção III (teoria de dados massivos, percepção e computação intermídia, inteligência híbrida e reforçada, inteligência de enxame, coordenação e controle autônomos, aprendizado de máquina de alto nível, computação inspirada no cérebro, computação quântica inteligente) — indicam um plano com componente teórico-científico proporcionalmente mais denso do que se observaria em um documento voltado primariamente à implementação ou à regulação.
- **"new" (96, 7º lugar)** reforça lexicalmente o enquadramento de "nova geração" que dá nome ao próprio documento.
- **"smart" (82, 8º) e "intelligent" (122, 5º)**, mantidos propositalmente não fundidos (ver metodologia e limitações), sugerem — sem que o gráfico permita afirmar com precisão adicional — que os tradutores empregam as duas palavras inglesas como alternativas para um mesmo campo semântico em chinês; somados, os dois termos ultrapassariam "technology" e ficariam atrás apenas de "AI".
- **"applications" (79, 9º) e "platforms" (69, 10º)** confirmam o peso da infraestrutura de implantação no plano, coerente com a Seção IV ("Resource Allocation"), que nomeia explicitamente diversas categorias de "plataformas de suporte".
- **"promote" (68), "support" (68) e "strengthen" (67)**, entre os verbos mais frequentes, formam um campo semântico de verbos facilitadores/programáticos, distinto do verbo dominante no documento dos EUA ("lead/led", ligado à fórmula de atribuição institucional "Led by [órgão]..."). Nenhuma comparação quantitativa direta entre os dois documentos é feita aqui — isso exigiria a normalização relativa exigida apenas na pasta "Análise Conjunta" —, mas a composição qualitativa do topo do vocabulário verbal já sugere uma diferença de registro entre os dois planos.
- **"innovation" (63, 14º) e "enterprises" (44, 24º)** indicam ênfase recorrente na inovação e no papel do setor empresarial como agente de implementação, ao lado do Estado.
- **"national" (46, 23º)**, isoladamente, é o único termo de enquadramento nacional/soberano a entrar no corte de 26 — "China" (31 ocorrências) fica logo abaixo da linha de corte —, o que indica que, em termos de frequência lexical bruta, o vocabulário do documento é predominantemente técnico-teórico, e não retórico-identitário.

**Ajustes possíveis para aprimorar a visualização.**
- Separar, em um gráfico complementar, os oito campos teóricos nomeados na Seção III (big data intelligence theory, cross-media sensing theory etc.) para uma leitura mais fina da arquitetura conceitual do plano, hoje dissolvida entre os termos genéricos "theory" e "research".
- Construir um segundo gráfico somando, lado a lado (sem fundir), as frequências de "smart" e "intelligent", explicitando a hipótese de equivalência de tradução discutida acima, sem transformá-la em fusão definitiva de termos.
- Segmentar a contagem por seção do documento (I. Situação Estratégica; II. Requisitos Gerais; III. Tarefas-Foco; IV. Alocação de Recursos; V. Medidas de Garantia; VI. Organização e Implementação), permitindo comparar o vocabulário dominante em cada bloco — não realizado aqui porque o pedido do usuário foi por uma visão geral do documento como um todo.
- Reportar, ao lado da frequência absoluta, a frequência relativa (percentual do total de tokens tratados) para leitores que desejem comparar a proeminência relativa dos termos sem depender do tamanho do documento.

**Limitações e fraquezas metodológicas.**
- **Fragmentação por hífen mais acentuada do que no documento comparável dos EUA.** A regra uniforme de tokenização (hífen sempre como separador) fragmenta compostos característicos da prosa deste documento — mais notavelmente "decision-making" (23 das suas ocorrências ficam divididas entre os tokens soltos "decision" e "making"), além de "data-driven"/"knowledge-driven", "high-end"/"high-level" (inflando o token solto "high", 20º lugar) e "brain-inspired". Optou-se por manter a mesma regra do documento dos EUA, por consistência metodológica entre as análises do projeto, em vez de introduzir exceções ad hoc — o preço dessa escolha é declarado aqui e no registro persistente.
- **"smart" e "intelligent" não fundidos**, apesar de indícios de uso intercambiável pelos tradutores — decisão deliberada (ver metodologia) que pode subestimar a proeminência conjunta desse campo semântico no gráfico, mas evita um julgamento de equivalência de tradução não autorizado pela Skill 02_Análise_Vocab_A.
- **"enhanced" mantido fora do agrupamento verbal de "enhance/enhancing"** por corresponder, neste documento, majoritariamente ao termo técnico fixo "hybrid (and) enhanced intelligence" — uma verificação de contexto caso a caso foi necessária para evitar fundir dois referentes distintos sob a mesma forma superficial.
- **Texto de partida é uma tradução do chinês**, não o original — toda leitura de vocabulário aqui descreve as escolhas lexicais dos tradutores da New America/DigiChina, não os termos chineses originais.
- **Contagem lexical, não semântica**: como em qualquer aplicação desta metodologia, termos polissêmicos não identificados manualmente podem persistir na contagem sem desambiguação — apenas os casos explicitamente verificados por leitura de contexto (Seção 5 do registro) foram tratados.

A metodologia completa — incluindo a lista integral de palavras removidas, os 30 agrupamentos de substantivos e as 18 famílias verbais normalizadas, os pares deliberadamente não agrupados com a justificativa de cada um, e as limitações identificadas — está documentada e versionada em [`registro_vocabulario_new_generation_ai_development_plan.md`](./registro_vocabulario_new_generation_ai_development_plan.md).
