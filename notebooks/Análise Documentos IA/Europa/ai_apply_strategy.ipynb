# Análise da Apply AI Strategy — União Europeia

Este notebook integra uma análise acadêmica de rigor, desenvolvida no âmbito de uma pesquisa em Relações Internacionais dedicada ao estudo comparado das estratégias nacionais de inteligência artificial. Seu objeto específico é a Apply AI Strategy, publicada pela Comissão Europeia, que estabelece a estratégia de aceleração da adoção da inteligência artificial pelos setores produtivos e pela administração pública europeia.

A análise aqui conduzida busca compreender, de modo simultaneamente **quantitativo e qualitativo**, a linguagem empregada pelo documento — os termos, categorias, ênfases retóricas, valores e prioridades estratégicas que estruturam o texto —, de modo a identificar os marcos conceituais e os campos semânticos por meio dos quais a União Europeia formula sua política de inteligência artificial.

A partir dessa leitura, pretende-se **posicionar o documento internacionalmente**, comparando a linguagem e as escolhas discursivas de a União Europeia com o vocabulário e as ênfases adotados pelos demais países e blocos contemplados neste projeto (Brasil, China, Estados Unidos, Europa e Índia), de modo a mapear convergências, divergências e posicionamentos estratégicos distintivos no debate internacional sobre desenvolvimento e governança de inteligência artificial.

As análises e visualizações produzidas neste notebook seguem as diretrizes metodológicas da **Skill02** (Análise e Visualização Gráfica de Documentos): baseiam-se exclusivamente no campo `texto_completo` do JSON de extração correspondente a este documento, produzido na etapa anterior (Skill01), adotam rigor acadêmico integral na leitura do texto-fonte, evitam generalizações, simplificações e inferências não fundamentadas no texto original, e cada visualização construída é acompanhada de sua respectiva descrição, leitura, interpretação e eventuais limitações metodológicas.


## Análise de Vocabulário — Termos mais Frequentes

Esta seção aplica a **Skill02** (Análise e Visualização Gráfica de Documentos) em conjunto com seu complemento especializado, a **Skill 02_Análise_Vocab_A** (Análise de Vocabulário e Termos), ao JSON `apply_ai_strategy.json`, produzido na etapa anterior (Skill01).

**Protocolo de uso do JSON.** Conforme exigido pela Skill02, apenas os campos `titulo`, `pais_ou_bloco` e `texto_completo` são utilizados nesta análise. Os campos `elementos_descartados`, `data_extracao`, `data_publicacao` e `fonte` não entram, em nenhuma hipótese, na contagem ou na interpretação do vocabulário.

**Idioma identificado.** O documento está integralmente redigido em **inglês** (a variante britânica do inglês institucional da UE — ex.: *centres*, *organisation*, *specialised*). Todas as listas de palavras funcionais e as regras de normalização morfológica aplicadas abaixo correspondem ao inglês, e não ao português ou a qualquer outro idioma.

**Etapas metodológicas aplicadas** (ver célula de código a seguir para a implementação exata):

1. **Extração bruta** de todos os termos de `texto_completo`, sem qualquer corte prévio de conteúdo.
2. **Pré-processamento do texto corrido, antes da tokenização**, fundamentado em identidade de referente (não em conveniência estatística):
   - `"Artificial Intelligence"` (2 ocorrências) foi unificado ao termo `"AI"` — mesma sigla e forma por extenso designando o mesmo referente. O trigrama **`"Artificial General Intelligence"`** (AGI, 1 ocorrência) — definido no próprio texto como um conceito tecnológico distinto e mais específico ("AI capable of performing any cognitive task that humans can") — foi deliberadamente **preservado fora dessa fusão**, para não conflacionar dois conceitos diferentes sob o mesmo rótulo.
   - `"European Union"` (1 ocorrência) foi unificado a `"EU"` — mesmo referente, forma por extenso e sigla amplamente dominante no restante do documento (53 ocorrências adicionais de "EU").
   - A expressão composta **`"Member States"`** (12 ocorrências) foi tratada como uma única unidade lexical (`member_states`) — termo institucional fixo referente aos Estados-membros da UE, estrutura análoga ao par "United States"/"U.S." já adotado no registro de vocabulário do documento dos Estados Unidos deste mesmo projeto.
   - A expressão técnica composta **`"digital twin(s)"`** (1 + 5 = 6 ocorrências, singular e plural) foi tratada como uma única unidade lexical — conceito tecnológico específico, definido no próprio texto ("virtual replicas of real-world objects or processes"), cuja fragmentação em "digital" + "twin(s)" distorceria tanto a contagem do adjetivo genérico "digital" quanto o significado do termo técnico — mesma lógica do exemplo "aprendizado de máquina" citado pela própria Skill 02_Análise_Vocab_A.
   - A expressão composta **`"AI Act"`** (13 ocorrências) foi tratada como uma única unidade lexical, distinta de outros instrumentos legais nomeados no documento que também contêm a palavra "Act" (`"Cloud and AI Development Act"`, `"Industrial Accelerator Act"`, 1 ocorrência cada) — cada um designa um instrumento legal diferente e não deve ser fundido aos demais.
   - Uma única ocorrência do substantivo comum **"building"** (não o gerúndio do verbo "build") — em *"a building or even of a human body"* — foi identificada por leitura contextual e isolada antes da tokenização, para ser corretamente agrupada ao substantivo plural "buildings" (2 ocorrências), e não à família verbal de "build" (as demais 12 ocorrências de "building" no documento são todas gerúndio/particípio de "build", como em "Building on the AI Continent Action Plan").
3. **Tokenização**: sequências de letras — e do caractere `_`, usado internamente apenas para os termos compostos tratados acima — foram extraídas como tokens; qualquer outro caractere não alfabético (espaço, hífen, apóstrofo, ponto, parênteses, barra) foi tratado como separador. Compostos com hífen não cobertos pelo pré-processamento acima (ex.: *cross-border*, *open-source*, *human-robot*) foram, por consequência, fragmentados em suas palavras componentes — regra única e uniforme, aplicada sem exceção, para evitar qualquer critério ad hoc de "quais hifens preservar" (discutido como limitação ao final).
4. **Remoção (exclusão)** de:
   - Palavras funcionais do inglês (artigos, preposições, conjunções, pronomes, verbos auxiliares/modais e conectores discursivos genéricos como *also*, *including*, *etc.*) — lista declarada integralmente na célula de código e no registro persistente;
   - Tokens de um único caractere, resíduos não semânticos da tokenização de abreviações pontuadas (ex.: "e.g." gera fragmentos soltos "e"/"g") e de possessivos (ex.: "EU's" gera o fragmento solto "s").
5. **Normalização/agrupamento** de variantes que designam exatamente o mesmo referente: **68 grupos** de substantivos singular/plural (ex.: *sector*/*sectors*, *model*/*models*, *development*/*developments*) e **34 famílias** de conjugações verbais (ex.: *support*/*supports*/*supporting*/*supported*). Pares com grafia semelhante mas classe gramatical ou referente distintos foram deliberadamente **mantidos separados** — em particular: **European** (adjetivo) / **EU** (sigla) / **Europe** (topônimo), o análogo direto do caso *America*/*American*/*Americans* já registrado no documento dos Estados Unidos; e ainda os pares derivacionais *development*(substantivo)/*develop*(verbo), *security*/*secure*, *strategy*/*strategic*, *economy*/*economic*, *industry*/*industrial*, *sovereignty*/*sovereign* e *governance*/*government(s)* — todos mantidos separados por pertencerem a classes gramaticais ou lemas distintos, não a flexões de uma mesma palavra.
6. **Critério de corte declarado:** os **25 termos de maior frequência absoluta** após as etapas 1–5, em ordem decrescente. Frequência absoluta é adequada aqui porque esta é uma análise de um único documento — a normalização relativa entre documentos de tamanhos distintos, exigida pela Skill02, aplica-se apenas à pasta "Análise Conjunta", que não é o caso presente.

O registro completo, persistente e auditável de todas as remoções e agrupamentos está salvo em [`registro_vocabulario_apply_ai_strategy.md`](./registro_vocabulario_apply_ai_strategy.md), nesta mesma pasta, e deve ser consultado — e atualizado de forma cumulativa — antes de qualquer nova visualização de vocabulário sobre este mesmo documento.

```python
import json, re
from collections import Counter
import pandas as pd

with open("apply_ai_strategy.json", encoding="utf-8") as f:
    doc = json.load(f)

titulo, pais, texto = doc["titulo"], doc["pais_ou_bloco"], doc["texto_completo"]
print(f"Documento: {titulo} | País/bloco: {pais} | Caracteres em texto_completo: {len(texto)}")

STOPWORDS_EN = set("""
a an the
and or but nor so yet if because while although that which who whom whose when where how than whether as
this these those it its they their them theirs he she his her him himself herself itself themselves
we our ours us you your yours i my mine
is are was were be been being am
has have had having
do does did doing
will would shall should can could may might must
to of in on at by for with from into through across throughout under over without within among between via per about upon toward towards
not no nor
also more most many much very only further therefore thus however moreover
there here
such other others any all both each every some
including include includes included
etc eg ie e g
s
""".split())

def preprocess(text):
    text = re.sub(r'\bArtificial Intelligence\b', 'AI', text, flags=re.I)
    text = re.sub(r'\bEuropean Union\b', 'EU', text, flags=re.I)
    text = re.sub(r'\bMember States\b', 'member_states', text, flags=re.I)
    text = re.sub(r'\bdigital twins?\b', 'digital_twins', text, flags=re.I)
    text = re.sub(r'\bAI Act\b', 'ai_act', text, flags=re.I)
    text = text.replace('a building or even of a human body', 'a buildingnoun or even of a human body')
    return text

NOUN_MERGES = {
    "sectors": ["sector", "sectors"], "models": ["model", "models"], "solutions": ["solution", "solutions"],
    "systems": ["system", "systems"], "tools": ["tool", "tools"], "services": ["service", "services"],
    "needs": ["need", "needs", "needed"], "actions": ["action", "actions"], "initiatives": ["initiative", "initiatives"],
    "challenges": ["challenge", "challenges"], "applications": ["application", "applications"],
    "infrastructures": ["infrastructure", "infrastructures"], "skills": ["skill", "skills"],
    "frameworks": ["framework", "frameworks"], "networks": ["network", "networks"], "platforms": ["platform", "platforms"],
    "organisations": ["organisation", "organisations"], "administrations": ["administration", "administrations"],
    "hubs": ["hub", "hubs"], "practices": ["practice", "practices"], "tasks": ["task", "tasks"],
    "robots": ["robot", "robots"], "vehicles": ["vehicle", "vehicles"], "labs": ["lab", "labs"],
    "benefits": ["benefit", "benefits"], "assets": ["asset", "assets"], "architectures": ["architecture", "architectures"],
    "flagships": ["flagship", "flagships"], "boards": ["board", "boards"], "resources": ["resource", "resources"],
    "threats": ["threat", "threats"], "dialogues": ["dialogue", "dialogues"], "developments": ["development", "developments"],
    "ecosystems": ["ecosystem", "ecosystems"], "capacities": ["capacity", "capacities"],
    "technologies": ["technology", "technologies"], "sciences": ["science", "sciences"],
    "processes": ["process", "processes"], "decisions": ["decision", "decisions"], "areas": ["area", "areas"],
    "ways": ["way", "ways"], "businesses": ["business", "businesses"], "humans": ["human", "humans"],
    "alliances": ["alliance", "alliances"], "purposes": ["purpose", "purposes"], "steps": ["step", "steps"],
    "funds": ["fund", "funds"], "risks": ["risk", "risks"], "interests": ["interest", "interests"],
    "chains": ["chain", "chains"], "startups": ["startup", "startups"], "strengths": ["strength", "strengths"],
    "professionals": ["professional", "professionals"], "drivers": ["driver", "drivers"], "factors": ["factor", "factors"],
    "partners": ["partner", "partners"], "partnerships": ["partnership", "partnerships"],
    "governments": ["government", "governments"], "forms": ["form", "forms"], "results": ["result", "results"],
    "elements": ["element", "elements"], "gaps": ["gap", "gaps"], "values": ["value", "values"],
    "innovations": ["innovation", "innovations"], "strategies": ["strategy", "strategies"],
    "industries": ["industry", "industries"],
    "buildings": ["buildingnoun", "buildings"],
    "states (genérico, não EU)": ["state", "states"],
}

VERB_MERGES = {
    "support": ["support", "supports", "supporting", "supported"],
    "foster": ["foster", "fosters", "fostering", "fostered"],
    "promote": ["promote", "promotes", "promoting", "promoted"],
    "develop (verbo)": ["develop", "develops", "developing", "developed"],
    "build": ["build", "builds", "building", "built"],
    "ensure": ["ensure", "ensures", "ensuring", "ensured"],
    "enable": ["enable", "enables", "enabling", "enabled"],
    "accelerate": ["accelerate", "accelerates", "accelerating", "accelerated"],
    "deploy": ["deploy", "deploys", "deploying", "deployed"],
    "create": ["create", "creates", "creating", "created"],
    "facilitate": ["facilitate", "facilitates", "facilitating", "facilitated"],
    "provide": ["provide", "provides", "providing", "provided"],
    "address": ["address", "addresses", "addressing", "addressed"],
    "improve": ["improve", "improves", "improving", "improved"],
    "launch": ["launch", "launches", "launching", "launched"],
    "establish": ["establish", "establishes", "establishing", "established"],
    "monitor": ["monitor", "monitors", "monitoring", "monitored"],
    "help": ["help", "helps", "helping", "helped"],
    "use": ["use", "uses", "using", "used"],
    "increase": ["increase", "increases", "increasing", "increased"],
    "boost": ["boost", "boosts", "boosting", "boosted"],
    "drive": ["drive", "drives", "driving"],
    "deliver": ["deliver", "delivers", "delivering", "delivered"],
    "leverage": ["leverage", "leverages", "leveraging", "leveraged"],
    "secure (verbo/adj)": ["secure", "secures", "securing", "secured"],
    "strengthen": ["strengthen", "strengthens", "strengthening", "strengthened"],
    "expand": ["expand", "expands", "expanding", "expanded"],
    "integrate": ["integrate", "integrates", "integrating", "integrated"],
    "encourage": ["encourage", "encourages", "encouraging", "encouraged"],
    "protect": ["protect", "protects", "protecting", "protected"],
    "reduce": ["reduce", "reduces", "reducing", "reduced"],
    "adopt": ["adopt", "adopts", "adopting", "adopted"],
    "implement": ["implement", "implements", "implementing", "implemented"],
    "reinforce": ["reinforce", "reinforces", "reinforcing", "reinforced"],
}

texto_tratado = preprocess(texto)
tokens_brutos = re.findall(r"[A-Za-z_]+", texto_tratado)
tokens_lower = [t.lower() for t in tokens_brutos]
tokens_filtrados = [t for t in tokens_lower if len(t) > 1 and t not in STOPWORDS_EN]

canon = {}
for label, variants in NOUN_MERGES.items():
    for v in variants:
        canon[v] = label
for label, variants in VERB_MERGES.items():
    for v in variants:
        canon[v] = label
canon["ai"] = "AI"
canon["ai_act"] = "AI Act"
canon["eu"] = "EU"
canon["member_states"] = "Member States"
canon["digital_twins"] = "digital twins"

tokens_finais = [canon.get(t, t) for t in tokens_filtrados]
freq = Counter(tokens_finais)

print(f"Tokens brutos: {len(tokens_brutos)} | após remoção de funcionais/resíduos: {len(tokens_filtrados)} | termos distintos no vocabulário final: {len(freq)}")

DISPLAY_CAPITALIZE = {"european": "European", "europe": "Europe", "commission": "Commission",
                       "defence": "Defence", "security": "Security", "public": "Public",
                       "data": "Data", "energy": "Energy", "potential": "Potential", "strategic": "Strategic"}

CORTE = 25
top_termos_raw = freq.most_common(CORTE)
top_termos = [(DISPLAY_CAPITALIZE.get(t, t), n) for t, n in top_termos_raw]

tabela_top = pd.DataFrame(top_termos, columns=["termo", "frequência"])
tabela_top.index = tabela_top.index + 1
tabela_top
```

```python
import matplotlib.pyplot as plt

termos_plot = [t for t, _ in top_termos][::-1]
freqs_plot = [n for _, n in top_termos][::-1]

fig, ax = plt.subplots(figsize=(9, 10))
barras = ax.barh(termos_plot, freqs_plot, color="#003399", height=0.68)

ax.set_xlabel("Frequência absoluta (nº de ocorrências em texto_completo)", fontsize=11)
ax.set_title(
    "Termos mais frequentes — Apply AI Strategy (União Europeia, 2025)\n"
    f"Top {CORTE} termos após remoção de palavras funcionais e normalização de variantes",
    fontsize=12.5, fontweight="bold", loc="left", pad=14
)
ax.spines[["top", "right"]].set_visible(False)
ax.tick_params(axis="y", labelsize=10.5)
ax.tick_params(axis="x", labelsize=9.5)
ax.set_axisbelow(True)
ax.xaxis.grid(True, color="#d9d9d9", linewidth=0.8)

for bar, val in zip(barras, freqs_plot):
    ax.text(bar.get_width() + 3, bar.get_y() + bar.get_height() / 2, str(val),
            va="center", ha="left", fontsize=9.5, color="#333333")

ax.set_xlim(0, max(freqs_plot) * 1.12)
fig.text(0.02, -0.01,
         "Fonte: texto_completo de apply_ai_strategy.json (Skill01), tratado segundo a Skill02 / Skill 02_Análise_Vocab_A.",
         fontsize=8, color="#666666")
fig.tight_layout()
plt.show()
```

## Leitura, Interpretação e Limitações do Gráfico

**O que foi construído.** Um gráfico de barras horizontais com os 25 termos de maior frequência absoluta em `texto_completo`, após remoção de palavras funcionais/resíduos e normalização das variantes descritas acima. Cada barra representa a contagem final (já consolidada) de um termo ou família de termos equivalentes.

**Como ler o gráfico.** O eixo vertical lista os termos, em ordem decrescente de frequência (o termo mais frequente no topo); o eixo horizontal mede o número absoluto de ocorrências no documento. O rótulo numérico ao final de cada barra é a contagem exata.

**Interpretação à luz do documento.**
- **"AI" (330 ocorrências)** domina o vocabulário por larga margem — mais de cinco vezes a frequência do segundo termo mais comum —, o que é esperado e consistente com o objeto do documento: uma estratégia inteiramente dedicada à adoção da inteligência artificial. Como no caso já registrado para os EUA, esse resultado isolado tem baixo valor discriminante; o valor analítico do gráfico está principalmente nos termos que disputam a segunda posição em diante, pois revelam **como** — e não apenas que — o documento enquadra a IA.
- **"Sectors" (65)** é o segundo termo mais frequente, e reflete diretamente a arquitetura do próprio documento: a Apply AI Strategy é organizada em torno de **11 flagships setoriais** (saúde, robótica, manufatura, defesa, mobilidade, telecomunicações, energia, clima, agroalimentar, cultura e mídia, setor público), cada um introduzido por uma variação da fórmula "To support the AI first policy in the [setor] sector, the Commission will:". Isso confirma que a lógica organizadora central do texto é **setorial**, não temática ou apenas principiológica — uma leitura relevante para Relações Internacionais, pois indica um modelo de política industrial verticalizada, setor a setor, e não uma diretriz horizontal única.
- **"Support" (59)** é amplamente distribuído e não se reduz a um artefato estrutural: apenas 7 das 59 ocorrências vêm da fórmula fixa de transição setorial ("To support the AI first policy in the X sector..."); as demais 52 aparecem organicamente ao longo do texto corrido — apoio a profissionais de saúde ("support healthcare professionals"), a validações clínicas, a infraestrutura energética, a startups, a pesquisa científica (RAISE), entre outros. Combinado com "Foster" (posição 26, fora do top 25 exibido, mas presente na tabela de frequências completa) e "Promote", "support" compõe um campo semântico de **fomento e capacitação** — a UE se posiciona predominantemente como uma agência de apoio, financiamento e coordenação, e não como reguladora coercitiva nesse documento específico (embora a seção 3.4 sobre o AI Act mostre a dimensão regulatória, ela é proporcionalmente menor no vocabulário geral).
- **"European" (58) e "EU" (54)**, mantidos deliberadamente separados de **"Europe" (21)** — respectivamente adjetivo, sigla institucional e topônimo — juntos ultrapassariam "AI" caso fossem somados, o que evidencia o quanto o documento se autorreferencia constantemente como um projeto europeu/da UE. A escolha metodológica de não fundir esses três termos (análoga à separação America/American/Americans no documento dos EUA) preserva a possibilidade de observar que "EU" (sigla institucional, ligada a atos e instrumentos: "EU Member States", "EU regions", "EU economy") supera ligeiramente "Europe" (topônimo, usado em referências mais retóricas: "Europe's sustainable transformation", "Europe stands at the forefront").
- **"Commission" (46)** no top 6 evidencia a centralidade da Comissão Europeia como agente executor: virtualmente toda ação listada no documento é introduzida por "the Commission will [verbo]", em um padrão análogo — mas não idêntico — ao "Led by [órgão federal]" do plano americano. A diferença estrutural é notável: enquanto o documento dos EUA distribui a execução entre múltiplas agências federais nomeadas (DOC, DOD, NIST...), a Apply AI Strategy concentra a autoria de praticamente todas as ações em um único ator institucional, a Comissão.
- **"Developments" (38)**, resultado da fusão de "development"/"developments", é o termo mais alto que não é imediatamente óbvio a partir do título do documento. Ele aparece predominantemente na formulação "the development of [tecnologia/solução/modelo]" — evidenciando que o verbo-chave do documento não é apenas "usar" ou "adotar" IA, mas **desenvolvê-la** ativamente dentro da UE, consistente com a ênfase em soberania tecnológica europeia.
- **"Use" (37)**, **"Models" (35)**, **"Strategies" (33)**, **"Data" (32)** e **"Solutions" (29)** formam o núcleo do vocabulário técnico-operacional esperado de uma estratégia de adoção de IA, sem surpresas frente ao conteúdo já lido do documento.
- **"Public" (26)** — majoritariamente em "public sector" e "public administrations" — reflete a seção 2.11 dedicada ao setor público, mas também aparece de forma cross-cutting em outras seções (ex.: referências a "public and private AI investment target", "public authorities").
- **"Defence" (21) e "Security" (21)**, empatados, entram no top 25 majoritariamente em função da seção 2.4 (Defence, security and space) e da subseção 3.4 (Ensuring trust), mas "security" também aparece fora dessas seções (ex.: "internal security", "cybersecurity" como termo composto que, por conter hífen, foi fragmentado em "cyber" + "security" pela regra única de tokenização — ver limitações).
- **"Strategic" (19)**, mantido deliberadamente separado de "strategy"/"strategies" (fundidos em "Strategies", 33), fecha o top 25 — mas há um **empate múltiplo na fronteira do corte**: quatro termos empatam em 23 ocorrências (innovations, actions, potential, energy) e seis termos empatam em 21 (technologies, europe, initiatives, systems, defence, security). A posição exata de cada um dentro dessas faixas empatadas é definida apenas pela ordem de primeira ocorrência no texto (comportamento padrão e determinístico do `Counter.most_common()` do Python), não por qualquer critério semântico adicional — um ponto de atenção detalhado nas limitações abaixo.

**Ajustes possíveis para aprimorar a visualização.**
- Segmentar a contagem por seção (2.1 a 2.11 vs. seção 3 "cross-cutting" vs. seção 4 "governança"), permitindo comparar o vocabulário dominante em cada flagship setorial — não realizado aqui porque o pedido do usuário foi por uma visão geral do documento como um todo.
- Construir um segundo gráfico separando explicitamente termos institucionais/nomes próprios (Commission, EU, Member States, AI Act, AI Board, Apply AI Alliance, AI Observatory) dos termos temático-conceituais (sectors, models, data, solutions...), para não misturar "quem executa/qual é o instrumento" com "o que é discutido".
- Reportar, ao lado da frequência absoluta, a frequência relativa (percentual do total de tokens tratados), especialmente útil caso este gráfico venha a ser comparado, em outro momento e sob outra skill, com os documentos de outros países/blocos na pasta "Análise Conjunta".
- Resolver explicitamente os empates na fronteira do corte (23 e 21 ocorrências) exibindo todos os termos empatados na mesma posição, ou ampliando o corte para incluir a faixa completa de empate (ex.: Top 30 em vez de Top 25), evitando a impressão de que a ordem dentro de uma faixa empatada é significativa.

**Limitações e fraquezas metodológicas.**
- **Fragmentação de compostos com hífen não pré-processados:** por aplicar uma regra única de tokenização (hífen sempre como separador, exceto nos cinco compostos tratados explicitamente no pré-processamento), expressões como "cross-border", "open-source", "human-robot", "cost-effective" e "AI-powered" foram fragmentadas em suas palavras componentes, dissolvendo-se em termos genéricos como "cross", "open", "source", "powered" (nenhum dos quais chegou ao top 25, mas que ficam artificialmente diluídos em vez de contados como unidades). Uma análise futura poderia tratar essas expressões como unidades compostas explícitas, à custa de introduzir uma lista de exceções que precisaria ser justificada termo a termo — a mesma limitação já registrada no documento dos Estados Unidos.
- **Polissemia não desambiguada em "actions" e "alliance":** a fusão de "action"/"actions" (23 ocorrências) e de "alliance"/"alliances" (10 ocorrências, fora do top 25) agrega tanto usos genéricos ("policy actions", "the Commission will take actions") quanto componentes de nomes próprios de outras estratégias e iniciativas já existentes (ex.: "AI Continent **Action** Plan", "Automotive **Action** Plan", "Security **Action** for Europe", "European Connected and Autonomous **Alliance**", "Apply AI **Alliance**") — usos distintos contados sob o mesmo rótulo. Optou-se deliberadamente por não desmembrar esses nomes próprios em unidades lexicais compostas específicas, para manter a metodologia consistente e restrita aos poucos compostos institucionais já claramente justificados (Member States, AI Act, digital twins), evitando a proliferação pouco disciplinada de regras ad hoc — mas o efeito de mistura de sentidos permanece e deve ser lido como uma limitação conhecida, não corrigida.
- **Empates extensos na fronteira do corte de 25**, já discutidos acima: a posição relativa entre termos empatados em 23 e em 21 ocorrências não deve ser lida como uma hierarquia de importância entre esses termos — apenas como uma consequência determinística, mas semanticamente arbitrária, da ordem de implementação do `Counter` do Python.
- **Cobertura da normalização morfológica:** os 68 grupos de substantivos e as 34 famílias verbais cobrem os candidatos com frequência combinada relevante para disputar o corte de 25 (identificados sistematicamente a partir de todos os pares singular/plural detectados automaticamente no vocabulário bruto), mas não houve lematização exaustiva de toda a cauda longa do vocabulário (mais de 1.400 termos distintos após a filtragem de palavras funcionais). Isso é adequado para o gráfico produzido, mas não geraria automaticamente o mesmo resultado caso o critério de corte fosse ampliado para, por exemplo, os 100 termos mais frequentes.

A metodologia completa — incluindo a lista integral de palavras removidas, os 68 agrupamentos de substantivos e as 34 famílias verbais normalizadas, os cinco compostos institucionais/técnicos tratados no pré-processamento, e os pares deliberadamente não agrupados — está documentada e versionada em [`registro_vocabulario_apply_ai_strategy.md`](./registro_vocabulario_apply_ai_strategy.md).
