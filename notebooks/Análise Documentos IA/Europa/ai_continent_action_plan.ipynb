# Análise do AI Continent Action Plan — União Europeia

Este notebook integra uma análise acadêmica de rigor, desenvolvida no âmbito de uma pesquisa em Relações Internacionais dedicada ao estudo comparado das estratégias nacionais de inteligência artificial. Seu objeto específico é o AI Continent Action Plan, publicado pela Comissão Europeia, que estabelece o plano de ação voltado a consolidar a União Europeia como polo global de desenvolvimento e infraestrutura de inteligência artificial.

A análise aqui conduzida busca compreender, de modo simultaneamente **quantitativo e qualitativo**, a linguagem empregada pelo documento — os termos, categorias, ênfases retóricas, valores e prioridades estratégicas que estruturam o texto —, de modo a identificar os marcos conceituais e os campos semânticos por meio dos quais a União Europeia formula sua política de inteligência artificial.

A partir dessa leitura, pretende-se **posicionar o documento internacionalmente**, comparando a linguagem e as escolhas discursivas de a União Europeia com o vocabulário e as ênfases adotados pelos demais países e blocos contemplados neste projeto (Brasil, China, Estados Unidos, Europa e Índia), de modo a mapear convergências, divergências e posicionamentos estratégicos distintivos no debate internacional sobre desenvolvimento e governança de inteligência artificial.

As análises e visualizações produzidas neste notebook seguem as diretrizes metodológicas da **Skill02** (Análise e Visualização Gráfica de Documentos): baseiam-se exclusivamente no campo `texto_completo` do JSON de extração correspondente a este documento, produzido na etapa anterior (Skill01), adotam rigor acadêmico integral na leitura do texto-fonte, evitam generalizações, simplificações e inferências não fundamentadas no texto original, e cada visualização construída é acompanhada de sua respectiva descrição, leitura, interpretação e eventuais limitações metodológicas.


## Análise de Vocabulário — Termos mais Frequentes

Esta seção aplica a **Skill02** (Análise e Visualização Gráfica de Documentos) em conjunto com seu complemento especializado, a **Skill 02_Análise_Vocab_A** (Análise de Vocabulário e Termos), ao JSON `ai_continent_action_plan.json`, produzido na etapa anterior (Skill01).

**Protocolo de uso do JSON.** Conforme exigido pela Skill02, apenas os campos `titulo`, `pais_ou_bloco` e `texto_completo` são utilizados nesta análise. Os campos `elementos_descartados`, `data_extracao`, `data_publicacao` e `fonte` não entram, em nenhuma hipótese, na contagem ou na interpretação do vocabulário.

**Idioma identificado.** O documento está integralmente redigido em **inglês**. Todas as listas de palavras funcionais e as regras de normalização morfológica aplicadas abaixo correspondem ao inglês, e não ao português ou a qualquer outro idioma.

**Etapas metodológicas aplicadas** (ver célula de código a seguir para a implementação exata):

1. **Extração bruta** de todos os termos de `texto_completo`, sem qualquer corte prévio de conteúdo.
2. **Remoção do rótulo estrutural** `"Key Commission actions:"` / `"Key Commission / EuroHPC actions:"` (7 ocorrências) — cabeçalho editorial que precede mecanicamente cada lista de ações da Comissão/EuroHPC, sem sentido distinto em nenhuma ocorrência, análogo ao tratamento dado a "Recommended Policy Actions" no documento dos EUA.
3. **Unificação de 17 expressões compostas** que designam um único referente institucional estável ao longo de todo o documento — ex.: `"AI Factory"/"AI Factories"` → **AI Factories** (40 ocorrências); `"AI Gigafactory"/"AI Gigafactories"` → **AI Gigafactories** (14); `"Member States"` → **Member States** (25); `"AI Act"` → **AI Act** (27); `"European Digital Innovation Hub(s)"/"Digital Innovation Hub(s)"` → **Digital Innovation Hubs** (17); `"Data Union Strategy"` → **Data Union Strategy** (7); `"Apply AI Strategy"` → **Apply AI Strategy** (14); `"Data Lab(s)"` → **Data Labs** (8); `"AI Skills Academy"` → **AI Skills Academy** (6); `"Cloud and AI Development Act"` → **Cloud and AI Development Act** (6); `"AI in Science"` → **AI in Science** (6); `"AI Office"` → **AI Office** (6); `"Single Market"` → **Single Market** (9); `"EuroHPC Joint Undertaking"` → **EuroHPC Joint Undertaking** (4); `"public sector"` → **public sector** (15); `"private sector"` → **private sector** (2); `"European Union"` → **EU** (1). Fragmentar essas expressões em palavras isoladas inflaria artificialmente o termo genérico "AI" (já dominante) e dissolveria o sinal analítico de cada programa/instrumento específico — o risco que a própria Skill 02_Análise_Vocab_A adverte ao tratar de expressões técnicas compostas. Importante: `"Artificial Intelligence"` **não** foi fundida a `"AI"` nesta rodada, porque uma de suas duas ocorrências integra `"Artificial General Intelligence (AGI)"`, conceito distinto que o documento diferencia explicitamente da "AI" genérica (ver registro persistente para o detalhamento do caso).
4. **Tokenização**: sequências de letras (tratando `_` dos rótulos compostos acima como parte do token) foram extraídas como tokens; qualquer caractere não alfabético (espaço, hífen, apóstrofo, ponto, parênteses, barra) foi tratado como separador. Compostos com hífen do texto corrido (ex.: *open-source*, *fine-tuning*, *large-scale*) foram, por consequência, fragmentados em suas palavras componentes — regra única e uniforme, aplicada sem exceção, discutida como limitação ao final.
5. **Remoção (exclusão)** de: palavras funcionais do inglês (artigos, preposições, conjunções, pronomes, verbos auxiliares/modais e conectores discursivos genéricos); a partícula de verbo frasal *up* e o resíduo do conector *as well as* (*well*), específicos deste documento; tokens de um único caractere, resíduos de tokenização de possessivos (ex.: "EU's" gera o fragmento solto "s"); e o rótulo estrutural já descrito no item 2.
6. **Normalização/agrupamento** de variantes que designam exatamente o mesmo referente: 30 pares/grupos singular-plural de substantivos (ex.: *model*/*models*, *sector*/*sectors*, *development*/*developments*) e 26 famílias de conjugações verbais (ex.: *support*/*supports*/*supporting*/*supported*; *launch*/*launching*/*launched*). Pares com grafia semelhante mas classe gramatical ou referente distintos foram deliberadamente **mantidos separados** — por exemplo, *AI*/*AGI* (conceitos distintos); *development*/*develop* e *security*/*secure* (substantivo vs. verbo, derivacional); *advance*/*advances*/*advancing*/*advanced* (o particípio "advanced" funciona predominantemente como adjetivo — "advanced AI processors" — e não como forma verbal); *European*/*Europe* (adjetivo vs. topônimo, mesma lógica do par America/American já documentado no registro dos EUA); e *Member States*/*State aid* (referentes institucionais distintos, apesar da raiz lexical comum "state").
7. **Critério de corte declarado:** os **25 termos de maior frequência absoluta** após as etapas 1–6, em ordem decrescente — mesmo corte já aplicado ao documento dos Estados Unidos, por consistência estilística entre os notebooks de país (sem que isso constitua uma comparação normalizada de frequências entre documentos, o que exigiria a Skill 02_Análise_Vocab_B na pasta "Análise Conjunta").

O registro completo, persistente e auditável de todas as remoções e agrupamentos está salvo em [`registro_vocabulario_ai_continent_action_plan.md`](./registro_vocabulario_ai_continent_action_plan.md), nesta mesma pasta, e deve ser consultado — e atualizado de forma cumulativa — antes de qualquer nova visualização de vocabulário sobre este mesmo documento.


```python
import json, re
from collections import Counter
import pandas as pd

with open("ai_continent_action_plan.json", encoding="utf-8") as f:
    doc = json.load(f)

titulo, pais, texto = doc["titulo"], doc["pais_ou_bloco"], doc["texto_completo"]
print(f"Documento: {titulo} | País/bloco: {pais} | Caracteres em texto_completo: {len(texto)}")

STOPWORDS_EN = set("""
a an the
and or but nor so yet if because while although that which who whom whose when where how than whether as
this these those it its they their them theirs he she his her him himself herself itself themselves
we our ours us you your yours i my mine
is are was were be been being am
has have had having
do does did doing
will would shall should can could may might must
to of in on at by for with from into through across throughout under over without within among between via per about upon toward towards
not no nor
also more most many much very only further therefore thus however moreover
there here
such other others any all both each every some
including include includes included
etc eg ie
up well
""".split())

STRUCTURAL_PHRASES_REMOVIDAS = [r'\bKey Commission(?:\s*/\s*EuroHPC)?\s+[Aa]ctions:\s*']

COMPOUND_PATTERNS = [
    (r"\bEuropean Union\b", " EU "),
    (r"\bCloud and AI Development Act\b", " CLOUD_AI_DEVELOPMENT_ACT "),
    (r"\bEuroHPC Joint Undertaking\b", " EUROHPC_JOINT_UNDERTAKING "),
    (r"\bAI Skills Academy\b", " AI_SKILLS_ACADEMY "),
    (r"\bApply AI Strategy\b", " APPLY_AI_STRATEGY "),
    (r"\bData Union Strategy\b", " DATA_UNION_STRATEGY "),
    (r"\bEuropean Digital Innovation Hubs?\b", " DIGITAL_INNOVATION_HUBS "),
    (r"\bDigital Innovation Hubs?\b", " DIGITAL_INNOVATION_HUBS "),
    (r"\bAI in Science\b", " AI_IN_SCIENCE "),
    (r"\bAI Gigafactor(?:y|ies)\b", " AI_GIGAFACTORIES "),
    (r"\bAI Factor(?:y|ies)\b", " AI_FACTORIES "),
    (r"\bAI Act\b", " AI_ACT "),
    (r"\bAI Office\b", " AI_OFFICE "),
    (r"\bMember States\b", " MEMBER_STATES "),
    (r"\bSingle Market\b", " SINGLE_MARKET "),
    (r"\bData Labs?\b", " DATA_LABS "),
    (r"\bpublic sector\b", " PUBLIC_SECTOR "),
    (r"\bprivate sector\b", " PRIVATE_SECTOR "),
]

def preprocess(text):
    for pat in STRUCTURAL_PHRASES_REMOVIDAS:
        text = re.sub(pat, ' ', text)
    for pat, repl in COMPOUND_PATTERNS:
        text = re.sub(pat, repl, text, flags=re.I)
    return text

texto_tratado = preprocess(texto)
PLACEHOLDERS = {repl.strip() for _, repl in COMPOUND_PATTERNS} | {"AI"}

tokens_brutos = re.findall(r"[A-Za-z_]+", texto_tratado)
tokens_lower = [t if t in PLACEHOLDERS else t.lower() for t in tokens_brutos]
tokens_filtrados = [t for t in tokens_lower if (t in PLACEHOLDERS) or (len(t) > 1 and t not in STOPWORDS_EN)]

NOUN_MERGES = {
    "models": ["model", "models"], "services": ["service", "services"], "solutions": ["solution", "solutions"],
    "sectors": ["sector", "sectors"], "initiatives": ["initiative", "initiatives"], "resources": ["resource", "resources"],
    "stakeholders": ["stakeholder", "stakeholders"], "facilities": ["facility", "facilities"],
    "partnerships": ["partnership", "partnerships"], "programmes": ["programme", "programmes"],
    "skills": ["skill", "skills"], "actions": ["action", "actions"], "needs": ["need", "needs"],
    "companies": ["company", "companies"], "startups": ["startup", "startups"], "scaleups": ["scaleup", "scaleups"],
    "technologies": ["technology", "technologies"], "values": ["value", "values"], "tools": ["tool", "tools"],
    "networks": ["network", "networks"], "supercomputers": ["supercomputer", "supercomputers"],
    "talent": ["talent", "talents"], "projects": ["project", "projects"], "funds": ["fund", "funds"],
    "investments": ["investment", "investments"], "gaps": ["gap", "gaps"], "spaces": ["space", "spaces"],
    "risks": ["risk", "risks"], "areas": ["area", "areas"], "chips": ["chip", "chips"],
    "strategy": ["strategy", "strategies"], "development": ["development", "developments"],
    "capacity": ["capacity", "capacities"],
}

VERB_MERGES = {
    "support": ["support", "supports", "supporting", "supported"],
    "launch": ["launch", "launching", "launched"],
    "ensure": ["ensure", "ensures"],
    "use": ["use", "uses", "using", "used"],
    "develop": ["develop", "developing", "developed"],
    "provide": ["provide", "provides", "providing", "provided"],
    "facilitate": ["facilitate", "facilitates", "facilitating", "facilitated"],
    "train": ["train", "training", "trained"],
    "adopt": ["adopt", "adopted"],
    "build": ["build", "building"],
    "create": ["create", "creates", "creating"],
    "strengthen": ["strengthen", "strengthening", "strengthened"],
    "increase": ["increase", "increases", "increasing", "increased"],
    "foster": ["foster", "fosters", "fostering"],
    "attract": ["attract", "attracting"],
    "work": ["work", "works", "working"],
    "set": ["set", "sets"],
    "link": ["link", "links", "linking", "linked"],
    "offer": ["offer", "offers", "offering", "offered"],
    "require": ["require", "requires"],
    "address": ["address", "addresses"],
    "identify": ["identify", "identifying"],
    "continue": ["continue", "continues"],
    "aim": ["aim", "aims", "aimed"],
    "enhance": ["enhance", "enhances"],
    "promote": ["promote", "promoting"],
}

canon = {}
for label, variants in NOUN_MERGES.items():
    for v in variants:
        canon[v] = label
for label, variants in VERB_MERGES.items():
    for v in variants:
        canon[v] = label
canon["eu"] = "EU"

tokens_finais = [canon.get(t, t) for t in tokens_filtrados]
freq = Counter(tokens_finais)

print(f"Tokens brutos: {len(tokens_brutos)} | após remoção de funcionais/resíduos: {len(tokens_filtrados)} | termos distintos no vocabulário final: {len(freq)}")

DISPLAY_LABELS = {
    "AI_FACTORIES": "AI Factories", "AI_GIGAFACTORIES": "AI Gigafactories", "MEMBER_STATES": "Member States",
    "AI_ACT": "AI Act", "DIGITAL_INNOVATION_HUBS": "Digital Innovation Hubs", "DATA_UNION_STRATEGY": "Data Union Strategy",
    "DATA_LABS": "Data Labs", "APPLY_AI_STRATEGY": "Apply AI Strategy", "AI_SKILLS_ACADEMY": "AI Skills Academy",
    "CLOUD_AI_DEVELOPMENT_ACT": "Cloud and AI Development Act", "AI_IN_SCIENCE": "AI in Science",
    "AI_OFFICE": "AI Office", "SINGLE_MARKET": "Single Market", "EUROHPC_JOINT_UNDERTAKING": "EuroHPC Joint Undertaking",
    "PUBLIC_SECTOR": "public sector", "PRIVATE_SECTOR": "private sector", "EU": "EU", "AI": "AI",
}

CORTE = 25
top_termos_raw = freq.most_common(CORTE)
top_termos = [(DISPLAY_LABELS.get(t, t), n) for t, n in top_termos_raw]

tabela_top = pd.DataFrame(top_termos, columns=["termo", "frequência"])
tabela_top.index = tabela_top.index + 1
tabela_top
```

```python
import matplotlib.pyplot as plt

termos_plot = [t for t, _ in top_termos][::-1]
freqs_plot = [n for _, n in top_termos][::-1]

fig, ax = plt.subplots(figsize=(9, 10))
barras = ax.barh(termos_plot, freqs_plot, color="#1f4e79", height=0.68)

ax.set_xlabel("Frequência absoluta (nº de ocorrências em texto_completo)", fontsize=11)
ax.set_title(
    "Termos mais frequentes — AI Continent Action Plan (União Europeia, 2025)\n"
    f"Top {CORTE} termos após remoção de palavras funcionais e normalização de variantes",
    fontsize=12.5, fontweight="bold", loc="left", pad=14
)
ax.spines[["top", "right"]].set_visible(False)
ax.tick_params(axis="y", labelsize=10.5)
ax.tick_params(axis="x", labelsize=9.5)
ax.set_axisbelow(True)
ax.xaxis.grid(True, color="#d9d9d9", linewidth=0.8)

for bar, val in zip(barras, freqs_plot):
    ax.text(bar.get_width() + 3, bar.get_y() + bar.get_height() / 2, str(val),
            va="center", ha="left", fontsize=9.5, color="#333333")

ax.set_xlim(0, max(freqs_plot) * 1.12)
fig.text(0.02, -0.01,
         "Fonte: texto_completo de ai_continent_action_plan.json (Skill01), tratado segundo a Skill02 / Skill 02_Análise_Vocab_A.",
         fontsize=8, color="#666666")
fig.tight_layout()
plt.show()
```

## Leitura, Interpretação e Limitações do Gráfico

**O que foi construído.** Um gráfico de barras horizontais com os 25 termos de maior frequência absoluta em `texto_completo`, após remoção de palavras funcionais/resíduos, unificação das 17 expressões compostas institucionais e normalização das variantes descritas acima. Cada barra representa a contagem final (já consolidada) de um termo, uma expressão composta ou uma família de variantes equivalentes.

**Como ler o gráfico.** O eixo vertical lista os termos, em ordem decrescente de frequência (o termo mais frequente no topo); o eixo horizontal mede o número absoluto de ocorrências no documento. O rótulo numérico ao final de cada barra é a contagem exata.

**Resultado obtido (Top 25):** AI (255), EU (124), data (76), sectors (47), support (45), public (41), european (40), AI Factories (40), commission (38), europe (34), development (34), models (29), research (29), services (28), AI Act (27), access (27), launch (27), computing (25), Member States (25), use (24), needs (23), initiatives (23), train (22), ensure (22), solutions (22).

**Interpretação à luz do documento.**
- **"AI" (255 ocorrências)** domina amplamente o vocabulário, como esperado em um plano de ação inteiramente dedicado à inteligência artificial. Assim como em análises equivalentes de outros documentos deste projeto, esse resultado tem baixo valor discriminante isolado — o valor analítico do gráfico está nos termos que disputam a segunda posição em diante, pois revelam **como**, e não apenas **que**, a União Europeia enquadra a IA.
- **"EU" (124), "european" (40) e "europe" (34)** formam, somados, um campo lexical de autorreferência institucional e geográfica que ultrapassaria 190 ocorrências — evidência textual direta da moldura enunciativa do documento: a União Europeia se define constantemente como sujeito coletivo da ação ("the EU must...", "the EU will..."), e não apenas como pano de fundo geográfico.
- **"data" (76)**, isoladamente o terceiro termo mais frequente, reflete não só a Seção 2 ("Data for AI"), dedicada inteiramente ao tema, mas a presença transversal do vocabulário de dados em praticamente todas as seções do plano (infraestrutura, Gigafactories, Data Labs, Data Union Strategy, Common European Data Spaces) — um indício de que, para a UE, a governança de dados é tratada como pré-condição estrutural da política de IA, não como um pilar isolado entre outros.
- **"sectors" (47)** no topo do ranking evidencia a arquitetura fortemente **setorial** do plano: a Apply AI Strategy é construída em torno de uma lista extensa de setores estratégicos nomeados (manufatura avançada, aeroespacial, defesa, agroalimentar, energia, saúde, cultura, ciência etc.), e a tabela "Key Sectors" das AI Factories reforça essa lógica setorial já na própria infraestrutura computacional.
- **"support" (45)**, o verbo mais frequente do documento, contrasta com um registro de comando direto — o texto se estrutura predominantemente em torno de apoiar, viabilizar e facilitar iniciativas de terceiros (Estados-Membros, startups, cientistas, PME), e não em atribuir ordens executivas a órgãos subordinados. Isso é coerente com a natureza do documento como comunicação de coordenação de políticas em um sistema de governança multinível, e não como um decreto executivo unitário.
- **"public" (41)**, mesmo após a extração à parte da expressão composta "public sector" (15 ocorrências, não incluída neste total), revela quão amplamente o vocabulário do setor público perpassa o documento em outras combinações (administração pública, contratação pública, serviços públicos, consulta pública, autoridades públicas) — o Estado, em suas variadas funções, é tratado tanto como destinatário quanto como instrumento central da política.
- **"AI Factories" (40)**, empatado com "european" na 7ª/8ª posição, mostra como a política de infraestrutura computacional da UE se concretiza, no plano textual, através de um programa nomeado específico — e não apenas de linguagem genérica sobre "computação" ou "infraestrutura" (que aparecem, mas com frequência bem menor: "computing" ocupa apenas a 18ª posição, com 25 ocorrências).
- **"commission" (38)** reflete a arquitetura institucional centralizada do plano: virtualmente toda ação listada nos quadros "Key Commission actions" (após a remoção do próprio rótulo estrutural repetido) é atribuída a um único agente executivo — a Comissão Europeia — o que contrasta com uma estrutura de responsabilidades pulverizada entre múltiplas agências nomeadas individualmente.
- **"development" (34), "models" (29), "research" (29), "services" (28), "computing" (25), "solutions" (22)** confirmam o vocabulário técnico-programático esperado de um plano de infraestrutura e adoção de IA.
- **"AI Act" (27)**, a despeito de ocupar uma seção dedicada (Seção 5) e ser o instrumento regulatório central da estratégia europeia de IA, aparece com frequência textual inferior à de "AI Factories" (40) — um indício quantitativo de que o registro predominante do documento é de **construção de capacidade** (infraestrutura, dados, adoção, competências) mais do que de **regulação**, ainda que esta seja institucionalmente relevante.
- **"Member States" (25)**, na 19ª posição, é um marcador estrutural da natureza da UE como sistema de governança multinível: diferentemente de uma estratégia nacional unitária, o plano precisa invocar constantemente os Estados-Membros como coimplementadores e cofinanciadores de cada iniciativa (AI Factories, Gigafactories, Data Labs, sandboxes regulatórios).
- **"access" (27) e "launch" (27)** completam o quadro: "access" recorre em múltiplos contextos (acesso a computação, a dados, a financiamento, a talentos, a mercados), enquanto "launch" reflete a densidade do documento em itens de ação concretos e datados (os quadros "Key Commission actions" listam dezenas de "Launch..." com trimestre-alvo), um registro de plano de execução, não de declaração de princípios.

**Ajustes possíveis para aprimorar a visualização.**

- Segmentar a contagem pelos cinco domínios do plano (infraestrutura computacional, dados, adoção setorial, competências, simplificação regulatória), permitindo comparar o vocabulário dominante em cada pilar — não realizado aqui porque o pedido do usuário foi por uma visão geral do documento como um todo.
- Reportar, ao lado da frequência absoluta, a frequência relativa (percentual do total de tokens tratados), para leitores que desejem comparar a proeminência relativa dos termos sem depender do tamanho do documento.

**Limitações e fraquezas metodológicas.**
- **Polissemia não desambiguada:** a contagem é lexical, não semântica. Por exemplo, "development" agrega tanto "AI model development" (desenvolvimento técnico de modelos) quanto sentidos mais amplos de "further technological development" ou "regional development" — usos distintos contados como um único termo. O mesmo se aplica a "access" (acesso computacional vs. acesso a mercados/talento) e a "support" (apoio institucional vs. "support services" como substantivo).
- **Fragmentação de compostos com hífen:** a regra única de tokenização (hífen sempre como separador) fragmenta expressões tecnicamente relevantes como "fine-tuning" → "fine"/"tuning" e "large-scale" → "large"/"scale", diluindo esses conceitos técnicos em termos genéricos que não atingiram o corte de 25. Uma análise futura poderia tratar essas expressões como unidades compostas explícitas, à custa de uma lista de exceções que precisaria ser justificada termo a termo — o mesmo trade-off já identificado na análise do documento dos EUA.
- **Cobertura da normalização morfológica:** os 30 grupos nominais e as 26 famílias verbais foram levantados de forma sistemática entre os termos com frequência plausivelmente relevante ao corte de 25 (aproximadamente frequência ≥ 4 no vocabulário já filtrado), não havendo lematização exaustiva de toda a cauda longa de mais de 1.570 termos distintos do vocabulário tratado.
- **Decisões de não-fusão sensíveis a leitura futura:** a separação deliberada entre "AI" e "AGI", e entre "advance"/"advances"/"advancing" (majoritariamente verbo/substantivo) e "advanced" (majoritariamente adjetivo), depende de uma leitura predominante do uso no documento, não de uma verificação individual de cada ocorrência — o registro persistente detalha a base dessa decisão e deve ser revisto caso o corte de termos seja ampliado.
- **17 expressões compostas unificadas** é um número consideravelmente maior do que o aplicado a outros documentos deste projeto (ex.: apenas 2 no caso dos EUA), refletindo uma característica real do texto — sua arquitetura em torno de muitos programas/instrumentos nomeados —, mas também introduzindo mais pontos de decisão metodológica passíveis de auditoria; cada um está justificado individualmente no registro persistente.

A metodologia completa — incluindo a lista integral de palavras removidas, as 17 expressões compostas unificadas, os 30 agrupamentos de substantivos e as 26 famílias verbais normalizadas, e os pares deliberadamente não agrupados — está documentada e versionada em [`registro_vocabulario_ai_continent_action_plan.md`](./registro_vocabulario_ai_continent_action_plan.md).
